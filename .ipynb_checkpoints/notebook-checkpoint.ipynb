```python
import numpy as np
import matplotlib.pyplot as plt
import ipywidgets as widgets
from IPython.display import display

# Generate a single curve starting from the outer radius
def generate_base_curve(num_points, outer_radius, inner_radius, wave_frequency, wave_amplitude):
    """Generate a single curve starting at the outer radius and tapering inward."""
    t = np.linspace(0, 1, num_points)
    radius = outer_radius + (inner_radius - outer_radius) * t  # Linearly taper from outer to inner radius
    angle = 2 * np.pi * wave_frequency * t  # Wave along the curve
    x = radius * np.cos(angle) + wave_amplitude * np.sin(angle)
    y = radius * np.sin(angle)
    return x, y

# Generate rotated and translated copies of the curve
def generate_rotated_copies(x, y, num_copies, radian_range):
    """Rotate and distribute copies of the curve symmetrically within the radian range."""
    curves = []
    angle_step = radian_range / (num_copies - 1) if num_copies > 1 else 0

    for i in range(num_copies):
        angle = -radian_range / 2 + i * angle_step  # Symmetric spread
        rotation_matrix = np.array([
            [np.cos(angle), -np.sin(angle)],
            [np.sin(angle), np.cos(angle)],
        ])
        rotated = np.dot(rotation_matrix, np.vstack((x, y)))
        curves.append(rotated)
    
    return curves

# Plot the curves
def plot_curves(curves):
    """Plot the stack of curves with alternating colors."""
    fig, ax = plt.subplots(figsize=(8, 8))
    colors = ['black'] + ['blue' if i % 2 else 'red' for i in range(1, len(curves) - 1)] + ['black']
    
    for i, curve in enumerate(curves):
        ax.plot(curve[0], curve[1], color=colors[i], lw=2)

    # Outer and inner boundary lines
    ax.plot(curves[0][0], curves[0][1], color='black', lw=3)  # Outer boundary
    ax.plot(curves[-1][0], curves[-1][1], color='black', lw=3)  # Inner boundary

    ax.set_aspect('equal')
    ax.axis('off')
    plt.show()

# Save the design as SVG
def save_curves_as_svg(curves, filename="alternate_greater_than_curve.svg"):
    """Save the design as an SVG."""
    fig, ax = plt.subplots(figsize=(8, 8))
    colors = ['black'] + ['blue' if i % 2 else 'red' for i in range(1, len(curves) - 1)] + ['black']
    
    for i, curve in enumerate(curves):
        ax.plot(curve[0], curve[1], color=colors[i], lw=2)

    # Outer and inner boundary lines
    ax.plot(curves[0][0], curves[0][1], color='black', lw=3)  # Outer boundary
    ax.plot(curves[-1][0], curves[-1][1], color='black', lw=3)  # Inner boundary

    ax.set_aspect('equal')
    ax.axis('off')
    fig.savefig(filename, format="svg", transparent=True)
    plt.close(fig)
    print(f"SVG saved as {filename}")

# Interactive function
def interactive_curve_design(
    num_points, outer_radius, inner_radius, wave_frequency, wave_amplitude, num_copies, radian_range, save
):
    # Generate the base curve
    x, y = generate_base_curve(num_points, outer_radius, inner_radius, wave_frequency, wave_amplitude)
    # Generate rotated copies
    curves = generate_rotated_copies(x, y, num_copies, radian_range)
    
    if save:
        # Save the design
        save_curves_as_svg(curves)
    else:
        # Plot the design
        plot_curves(curves)

# Interactive widgets
interactive_widgets = widgets.interactive(
    interactive_curve_design,
    num_points=widgets.IntSlider(min=100, max=2000, step=100, value=500, description="Points"),
    outer_radius=widgets.FloatSlider(min=5, max=20, step=0.5, value=15, description="Outer Radius"),
    inner_radius=widgets.FloatSlider(min=1, max=10, step=0.5, value=5, description="Inner Radius"),
    wave_frequency=widgets.FloatSlider(min=1, max=10, step=0.5, value=3, description="Wave Frequency"),
    wave_amplitude=widgets.FloatSlider(min=0, max=5, step=0.1, value=1, description="Wave Amplitude"),
    num_copies=widgets.IntSlider(min=2, max=20, step=1, value=5, description="Copies"),
    radian_range=widgets.FloatSlider(min=np.pi/8, max=np.pi, step=np.pi/16, value=np.pi/4, description="Radian"),
    save=widgets.Checkbox(value=False, description="Save SVG"),
)

# Display interactive widgets
display(interactive_widgets)
```

```python
import numpy as np
import matplotlib.pyplot as plt
import ipywidgets as widgets
from IPython.display import display

# Generate a smooth curve with decreasing width and amplitude
def generate_wave_with_smooth_decreasing_width(num_peaks, initial_height, initial_width, amplitude_decay, width_decrement):
    x = np.linspace(0, num_peaks * initial_width, 1000)
    y = np.zeros_like(x)
    for i in range(len(x)):
        progress = x[i] / (num_peaks * initial_width)
        current_amplitude = initial_height * (amplitude_decay ** (progress * num_peaks))
        current_width = max(initial_width - width_decrement * progress * num_peaks, 1)
        y[i] = current_amplitude * np.sin(2 * np.pi * x[i] / current_width)
    valid_indices = y != 0
    return x[valid_indices], y[valid_indices]

# Generate the base curve starting at the outer edge
def generate_curve(num_peaks, initial_height, initial_width, amplitude_decay, width_decrement):
    x, y = generate_wave_with_smooth_decreasing_width(
        num_peaks, initial_height, initial_width, amplitude_decay, width_decrement
    )
    # Reverse the curve so it starts on the outside and ends on the inside
    x, y = x[::-1], y[::-1]
    return x, y

# Stack curves to form a "greater-than" shape ( >c )
def stack_curves_to_form_greater_than(x, y, copies, radian, radius):
    curves = []
    angle_step = radian / (copies - 1)  # Angle increment between curves

    for i in range(copies):
        angle = -radian / 2 + i * angle_step  # Spread curves symmetrically within the radian range
        rotation_matrix = np.array([
            [np.cos(angle), -np.sin(angle)],
            [np.sin(angle), np.cos(angle)],
        ])
        # Apply rotation
        rotated = np.dot(rotation_matrix, np.vstack((x, y)))
        # Translate curves outward, starting at the outer edge, with quadratic tapering for smoother alignment
        rotated[0] += radius * (1 - (i / (copies - 1))**2)  # Quadratic tapering
        rotated[1] += radius * np.sin(angle)
        curves.append(rotated)

    return curves

# Plot the curves
def plot_greater_than_slice(curves, copies):
    fig, ax = plt.subplots(figsize=(8, 8))
    colors = ['black'] + ['blue' if i % 2 else 'red' for i in range(1, copies - 1)] + ['black']
    
    for i, curve in enumerate(curves):
        ax.plot(curve[0], curve[1], color=colors[i], lw=2)  # Increased line width for better visibility

    # Outline the boundaries (first and last curve)
    ax.plot(curves[0][0], curves[0][1], color='black', lw=3)  # Outer boundary
    ax.plot(curves[-1][0], curves[-1][1], color='black', lw=3)  # Inner boundary

    ax.set_aspect('equal')
    ax.axis('off')
    plt.show()

# Save the design as SVG
def save_greater_than_svg(curves, copies):
    colors = ['black'] + ['blue' if i % 2 else 'red' for i in range(1, copies - 1)] + ['black']
    fig, ax = plt.subplots(figsize=(8, 8))
    
    for i, curve in enumerate(curves):
        ax.plot(curve[0], curve[1], color=colors[i], lw=2)  # Increased line width for better visibility

    # Outline the boundaries
    ax.plot(curves[0][0], curves[0][1], color='black', lw=3)  # Outer boundary
    ax.plot(curves[-1][0], curves[-1][1], color='black', lw=3)  # Inner boundary

    ax.set_aspect('equal')
    ax.axis('off')
    
    svg_filename = "greater_than_curve_refined.svg"
    fig.savefig(svg_filename, format="svg", transparent=True)
    plt.close(fig)
    print(f"SVG saved as {svg_filename}")

# Main interactive function
def interactive_greater_than_slice(
    num_peaks, initial_height, initial_width, amplitude_decay, width_decrement, copies, radian, radius, save
):
    # Generate the base curve (starts on the outer edge by default)
    x, y = generate_curve(num_peaks, initial_height, initial_width, amplitude_decay, width_decrement)
    # Stack curves into a '>' shape
    curves = stack_curves_to_form_greater_than(x, y, copies, radian, radius)
    
    if save:
        # Save the design as an SVG
        save_greater_than_svg(curves, copies)
    else:
        # Plot the design
        plot_greater_than_slice(curves, copies)

# Interactive widgets
interact_greater_than_slice = widgets.interactive(
    interactive_greater_than_slice,
    num_peaks=widgets.IntSlider(min=1, max=10, step=1, value=5, description="Peaks"),
    initial_height=widgets.FloatSlider(min=0.5, max=5, step=0.1, value=2, description="Height"),
    initial_width=widgets.FloatSlider(min=1, max=10, step=0.5, value=5, description="Width"),
    amplitude_decay=widgets.FloatSlider(min=0.5, max=1, step=0.05, value=0.8, description="Decay"),
    width_decrement=widgets.FloatSlider(min=0.1, max=2, step=0.1, value=0.5, description="Width Dec."),
    copies=widgets.IntSlider(min=2, max=20, step=1, value=5, description="Copies"),
    radian=widgets.FloatSlider(min=np.pi/8, max=np.pi, step=np.pi/16, value=np.pi/4, description="Radian"),
    radius=widgets.FloatSlider(min=1, max=20, step=1, value=10, description="Radius"),
    save=widgets.Checkbox(value=False, description="Save SVG"),
)

# Display widgets
display(interact_greater_than_slice)
```

```python
import numpy as np
import matplotlib.pyplot as plt
import ipywidgets as widgets
from IPython.display import display

# Generate a smooth curve with decreasing width and amplitude
def generate_wave_with_smooth_decreasing_width(num_peaks, initial_height, initial_width, amplitude_decay, width_decrement):
    x = np.linspace(0, num_peaks * initial_width, 1000)
    y = np.zeros_like(x)
    for i in range(len(x)):
        progress = x[i] / (num_peaks * initial_width)
        current_amplitude = initial_height * (amplitude_decay ** (progress * num_peaks))
        current_width = max(initial_width - width_decrement * progress * num_peaks, 1)
        y[i] = current_amplitude * np.sin(2 * np.pi * x[i] / current_width)
    valid_indices = y != 0
    return x[valid_indices], y[valid_indices]

# Generate the base curve
def generate_curve(num_peaks, initial_height, initial_width, amplitude_decay, width_decrement):
    x, y = generate_wave_with_smooth_decreasing_width(
        num_peaks, initial_height, initial_width, amplitude_decay, width_decrement
    )
    # Reverse the curve so it starts on the opposite end
    x, y = x[::-1], y[::-1]
    return x, y

# Stack curves to form a "greater-than" shape ( >c )
def stack_curves_to_form_greater_than(x, y, copies, radian, radius):
    curves = []
    angle_step = radian / (copies - 1)  # Angle increment between curves

    for i in range(copies):
        angle = radian / 2 - i * angle_step  # Reverse the angle sequence
        rotation_matrix = np.array([
            [np.cos(angle), -np.sin(angle)],
            [np.sin(angle), np.cos(angle)],
        ])
        # Apply rotation
        rotated = np.dot(rotation_matrix, np.vstack((x, y)))
        # Adjust translation to fan out from the opposite end
        rotated[0] += radius * (i / (copies - 1))**2  # Quadratic tapering
        rotated[1] += radius * np.sin(angle)
        curves.append(rotated)

    return curves

# Plot the curves
def plot_greater_than_slice(curves, copies):
    fig, ax = plt.subplots(figsize=(8, 8))
    colors = ['black'] + ['blue' if i % 2 else 'red' for i in range(1, copies - 1)] + ['black']
    
    for i, curve in enumerate(curves):
        ax.plot(curve[0], curve[1], color=colors[i], lw=2)  # Increased line width for better visibility

    # Outline the boundaries (first and last curve)
    ax.plot(curves[0][0], curves[0][1], color='black', lw=3)  # Outer boundary
    ax.plot(curves[-1][0], curves[-1][1], color='black', lw=3)  # Inner boundary

    ax.set_aspect('equal')
    ax.axis('off')
    plt.show()

# Save the design as SVG
def save_greater_than_svg(curves, copies):
    colors = ['black'] + ['blue' if i % 2 else 'red' for i in range(1, copies - 1)] + ['black']
    fig, ax = plt.subplots(figsize=(8, 8))
    
    for i, curve in enumerate(curves):
        ax.plot(curve[0], curve[1], color=colors[i], lw=2)  # Increased line width for better visibility

    # Outline the boundaries
    ax.plot(curves[0][0], curves[0][1], color='black', lw=3)  # Outer boundary
    ax.plot(curves[-1][0], curves[-1][1], color='black', lw=3)  # Inner boundary

    ax.set_aspect('equal')
    ax.axis('off')
    
    svg_filename = "greater_than_curve_refined.svg"
    fig.savefig(svg_filename, format="svg", transparent=True)
    plt.close(fig)
    print(f"SVG saved as {svg_filename}")

# Main interactive function
def interactive_greater_than_slice(
    num_peaks, initial_height, initial_width, amplitude_decay, width_decrement, copies, radian, radius, save
):
    # Generate the base curve (starts on the opposite end)
    x, y = generate_curve(num_peaks, initial_height, initial_width, amplitude_decay, width_decrement)
    # Stack curves into a '>' shape
    curves = stack_curves_to_form_greater_than(x, y, copies, radian, radius)
    
    if save:
        # Save the design as an SVG
        save_greater_than_svg(curves, copies)
    else:
        # Plot the design
        plot_greater_than_slice(curves, copies)

# Interactive widgets
interact_greater_than_slice = widgets.interactive(
    interactive_greater_than_slice,
    num_peaks=widgets.IntSlider(min=1, max=10, step=1, value=5, description="Peaks"),
    initial_height=widgets.FloatSlider(min=0.5, max=5, step=0.1, value=2, description="Height"),
    initial_width=widgets.FloatSlider(min=1, max=10, step=0.5, value=5, description="Width"),
    amplitude_decay=widgets.FloatSlider(min=0.5, max=1, step=0.05, value=0.8, description="Decay"),
    width_decrement=widgets.FloatSlider(min=0.1, max=2, step=0.1, value=0.5, description="Width Dec."),
    copies=widgets.IntSlider(min=2, max=20, step=1, value=5, description="Copies"),
    radian=widgets.FloatSlider(min=np.pi/8, max=np.pi, step=np.pi/16, value=np.pi/4, description="Radian"),
    radius=widgets.FloatSlider(min=1, max=20, step=1, value=10, description="Radius"),
    save=widgets.Checkbox(value=False, description="Save SVG"),
)

# Display widgets
display(interact_greater_than_slice)
```

```python
import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interactive, IntSlider, FloatSlider, Checkbox
from IPython.display import display

# Define the base curve function
def generate_base_curve(num_points=1000, length=10, amplitude=1):
    x = np.linspace(0, length, num_points)
    y = amplitude * np.sin(2 * np.pi * x / length)
    return x, y

# Function to generate and plot the fanned-out curves
def plot_fanned_curves(
    num_curves=5,           # Number of curves to generate
    total_angle=np.pi/2,    # Total angle range for fanning out (in radians)
    length=10,              # Length of the base curve
    amplitude=1,            # Amplitude of the base curve
    amplitude_decrease=False, # Decrease amplitude for subsequent curves
    show_plot=True,         # Whether to display the plot
    save_svg=False          # Whether to save the plot as an SVG file
):
    # Generate the base curve
    x_base, y_base = generate_base_curve(length=length, amplitude=amplitude)

    # Angles at which curves will be rotated (evenly spaced)
    angles = np.linspace(-total_angle / 2, total_angle / 2, num_curves)

    # Create figure and axis
    fig, ax = plt.subplots(figsize=(8, 8))

    # Draw curves
    for i, angle in enumerate(angles):
        # Optional amplitude decrease
        if amplitude_decrease:
            amp_factor = 1 - i / (num_curves - 1)
            x = x_base * amp_factor
            y = y_base * amp_factor
        else:
            x, y = x_base, y_base

        # Rotate curve
        rot_matrix = np.array([
            [np.cos(angle), -np.sin(angle)],
            [np.sin(angle),  np.cos(angle)]
        ])
        curve = np.dot(rot_matrix, np.vstack((x, y)))

        # Translate curve so that all curves start from the origin
        delta_x = +curve[0, 0]
        delta_y = -curve[1, 0]
        curve[0] += delta_x
        curve[1] += delta_y

        # Plot curve
        ax.plot(curve[0], curve[1], lw=2)

    # Set equal aspect ratio and remove axes
    ax.set_aspect('equal')
    ax.axis('off')

    # Show or save the plot
    if show_plot:
        plt.show()
    if save_svg:
        fig.savefig('fanned_curves.svg', format='svg', bbox_inches='tight')
        print('SVG file saved as "fanned_curves.svg"')
    plt.close(fig)

# Interactive widgets for real-time parameter adjustment
def interactive_fanned_curves():
    interactive_plot = interactive(
        plot_fanned_curves,
        num_curves=IntSlider(min=2, max=20, step=1, value=5, description='Number of Curves'),
        total_angle=FloatSlider(min=0, max=2*np.pi, step=np.pi/16, value=np.pi/2, description='Total Angle (rad)'),
        length=FloatSlider(min=1, max=20, step=1, value=10, description='Curve Length'),
        amplitude=FloatSlider(min=0.1, max=5, step=0.1, value=1, description='Amplitude'),
        amplitude_decrease=Checkbox(value=False, description='Decrease Amplitude'),
        save_svg=Checkbox(value=False, description='Save as SVG')
    )
    display(interactive_plot)

# Run the interactive function
interactive_fanned_curves()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interactive, IntSlider, FloatSlider, Checkbox
from IPython.display import display

# Function to generate the base curve
def generate_base_curve(num_points=1000, length=10, amplitude=1, frequency=1):
    x = np.linspace(0, length, num_points)
    y = amplitude * np.sin(2 * np.pi * frequency * x / length)
    return x, y

# Function to generate and plot the fanned-out curves with specified modifications
def plot_modified_fanned_curves(
    num_curves=5,                   # Number of curves
    angle_separation=np.pi/18,      # Separation between curves in radians
    length=10,                      # Base curve length
    base_amplitude=1,               # Base amplitude
    amplitude_growth=0,             # Amplitude growth percentage per curve
    base_frequency=1,               # Base frequency
    frequency_decrease=0,           # Frequency decrease percentage per curve
    show_plot=True,                 # Display the plot
    save_svg=False                  # Save the plot as SVG
):
    # Angles at which curves will be rotated
    total_angle = angle_separation * (num_curves - 1)
    angles = np.linspace(-total_angle / 2, total_angle / 2, num_curves)

    # Create figure and axis
    fig, ax = plt.subplots(figsize=(8, 8))

    # Colors for the curves
    colors = ['black']  # First curve is black
    for i in range(1, num_curves - 1):
        colors.append('blue' if i % 2 == 1 else 'red')
    if num_curves > 1:
        colors.append('black')  # Last curve is black

    curves = []

    # Draw curves
    for i, angle in enumerate(angles):
        # Calculate amplitude and frequency for this curve
        amplitude = base_amplitude * ((1 + amplitude_growth / 100) ** i)
        frequency = base_frequency * ((1 - frequency_decrease / 100) ** i)

        # Generate the base curve with current amplitude and frequency
        x_base, y_base = generate_base_curve(length=length, amplitude=amplitude, frequency=frequency)

        # Rotate curve
        rot_matrix = np.array([
            [np.cos(angle), -np.sin(angle)],
            [np.sin(angle),  np.cos(angle)]
        ])
        curve = np.dot(rot_matrix, np.vstack((x_base, y_base)))

        # Translate curve so that all curves start from the origin
        delta_x = -curve[0, 0]
        delta_y = -curve[1, 0]
        curve[0] += delta_x
        curve[1] += delta_y

        curves.append(curve)

        # Plot curve
        ax.plot(curve[0], curve[1], color=colors[i], lw=2)

    # Outline with arcs at the start and end
    # Calculate start and end points
    start_points = np.array([curve[:, 0] for curve in curves])
    end_points = np.array([curve[:, -1] for curve in curves])

    # Draw arcs at the inner and outer ends
    # Start arc (inner arc)
    start_arc_center = [0, 0]  # All curves start at the origin
    start_arc_radius = 0.0     # No need to draw an arc at the start since all curves start at the same point

    # End arc (outer arc)
    # Calculate the arc that connects the end points of the first and last curves
    end_arc_center = [0, 0]
    end_arc_radius = np.linalg.norm(end_points[0])  # Radius is the distance from origin to the end point of the first curve
    theta_start = np.arctan2(end_points[0, 1], end_points[0, 0])
    theta_end = np.arctan2(end_points[-1, 1], end_points[-1, 0])

    # Ensure theta_start < theta_end
    if theta_start > theta_end:
        theta_start, theta_end = theta_end, theta_start

    theta = np.linspace(theta_start, theta_end, 100)
    arc_x = end_arc_radius * np.cos(theta)
    arc_y = end_arc_radius * np.sin(theta)
    ax.plot(arc_x, arc_y, color='black', lw=2)

    # Set equal aspect ratio and remove axes
    ax.set_aspect('equal')
    ax.axis('off')

    # Show or save the plot
    if show_plot:
        plt.show()
    if save_svg:
        fig.savefig('modified_fanned_curves.svg', format='svg', bbox_inches='tight')
        print('SVG file saved as "modified_fanned_curves.svg"')
    plt.close(fig)

# Interactive widgets
def interactive_modified_fanned_curves():
    interactive_plot = interactive(
        plot_modified_fanned_curves,
        num_curves=IntSlider(min=2, max=20, step=1, value=5, description='Number of Curves'),
        angle_separation=FloatSlider(min=0, max=np.pi, step=np.pi/180, value=np.pi/18, description='Angle Separation (rad)'),
        length=FloatSlider(min=1, max=20, step=1, value=10, description='Curve Length'),
        base_amplitude=FloatSlider(min=0.1, max=5, step=0.1, value=1, description='Base Amplitude'),
        amplitude_growth=FloatSlider(min=-100, max=100, step=1, value=0, description='Amplitude Growth (%)'),
        base_frequency=FloatSlider(min=0.1, max=5, step=0.1, value=1, description='Base Frequency'),
        frequency_decrease=FloatSlider(min=0, max=100, step=1, value=0, description='Frequency Decrease (%)'),
        save_svg=Checkbox(value=False, description='Save as SVG')
    )
    display(interactive_plot)

# Run the interactive function
interactive_modified_fanned_curves()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interactive, IntSlider, FloatSlider, Checkbox
from IPython.display import display

# Function to generate and plot the fanned-out curves with specified modifications
def plot_modified_fanned_curves(
    num_curves=5,                   # Number of curves
    inner_angle=np.pi/6,            # Total angle at the inner end (in radians)
    outer_angle=np.pi/2,            # Total angle at the outer end (in radians)
    length=10,                      # Base curve length
    base_amplitude=1,               # Base amplitude
    amplitude_growth=0,             # Amplitude growth percentage per curve
    base_frequency=1,               # Base frequency
    frequency_decrease=0,           # Frequency decrease percentage per curve
    show_plot=True,                 # Display the plot
    save_svg=False                  # Save the plot as SVG
):
    # Create figure and axis
    fig, ax = plt.subplots(figsize=(8, 8))

    curves = []

    # Angles at which curves will start and end
    start_angles = np.linspace(-inner_angle / 2, inner_angle / 2, num_curves)
    end_angles = np.linspace(-outer_angle / 2, outer_angle / 2, num_curves)

    # Colors for the curves
    colors = ['black']  # First curve is black
    for i in range(1, num_curves - 1):
        colors.append('blue' if i % 2 == 1 else 'red')
    if num_curves > 1:
        colors.append('black')  # Last curve is black

    # Generate the curves
    max_r = 0  # To keep track of the maximum radius for drawing the outer arc
    for i in range(num_curves):
        # Calculate amplitude and frequency for this curve
        amplitude = base_amplitude * ((1 + amplitude_growth / 100) ** i)
        frequency = base_frequency * ((1 - frequency_decrease / 100) ** i)

        # Generate parameter t
        t = np.linspace(0, 1, 1000)

        # Calculate angle for each point along the curve
        angles = start_angles[i] + t * (end_angles[i] - start_angles[i])

        # Calculate radius for each point along the curve
        r = amplitude * np.sin(2 * np.pi * frequency * t * length)

        # Update max_r for outer arc
        max_r = max(max_r, np.max(np.abs(r)))

        # Convert to Cartesian coordinates
        x = r * np.cos(angles)
        y = r * np.sin(angles)

        curves.append(np.vstack((x, y)))

        # Plot curve
        ax.plot(x, y, color=colors[i], lw=2)

    # Outline with arcs at the start and end
    # Start arc
    theta_start_arc = np.linspace(-inner_angle / 2, inner_angle / 2, 100)
    r_start_arc = 0.01  # Small radius to represent the inner arc (since r=0 can't be plotted)
    x_start_arc = r_start_arc * np.cos(theta_start_arc)
    y_start_arc = r_start_arc * np.sin(theta_start_arc)
    # Since the inner arc is essentially a point, we can skip plotting it

    # End arc (outer arc)
    theta_end_arc = np.linspace(-outer_angle / 2, outer_angle / 2, 100)
    r_end_arc = max_r  # Use the maximum radius of the curves
    x_end_arc = r_end_arc * np.cos(theta_end_arc)
    y_end_arc = r_end_arc * np.sin(theta_end_arc)
    ax.plot(x_end_arc, y_end_arc, color='black', lw=2)

    # Set equal aspect ratio and remove axes
    ax.set_aspect('equal')
    ax.axis('off')

    # Show or save the plot
    if show_plot:
        plt.show()
    if save_svg:
        fig.savefig('modified_fanned_curves.svg', format='svg', bbox_inches='tight')
        print('SVG file saved as "modified_fanned_curves.svg"')
    plt.close(fig)

# Interactive widgets
def interactive_modified_fanned_curves():
    interactive_plot = interactive(
        plot_modified_fanned_curves,
        num_curves=IntSlider(min=2, max=20, step=1, value=5, description='Number of Curves'),
        inner_angle=FloatSlider(min=0, max=2*np.pi, step=np.pi/180, value=np.pi/6, description='Inner Angle (rad)'),
        outer_angle=FloatSlider(min=0, max=2*np.pi, step=np.pi/180, value=np.pi/2, description='Outer Angle (rad)'),
        length=FloatSlider(min=1, max=20, step=1, value=10, description='Curve Length'),
        base_amplitude=FloatSlider(min=0.1, max=5, step=0.1, value=1, description='Base Amplitude'),
        amplitude_growth=FloatSlider(min=-100, max=100, step=1, value=0, description='Amplitude Growth (%)'),
        base_frequency=FloatSlider(min=0.1, max=5, step=0.1, value=1, description='Base Frequency'),
        frequency_decrease=FloatSlider(min=0, max=100, step=1, value=0, description='Frequency Decrease (%)'),
        save_svg=Checkbox(value=False, description='Save as SVG')
    )
    display(interactive_plot)

# Run the interactive function
interactive_modified_fanned_curves()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interactive, IntSlider, FloatSlider, Checkbox
from IPython.display import display

# Function to generate the base curve
def generate_base_curve(num_points=1000, length=10, amplitude=1, frequency=1):
    x = np.linspace(0, length, num_points)
    y = amplitude * np.sin(2 * np.pi * frequency * x / length)
    return x, y

# Function to generate and plot the fanned-out curves with specified modifications
def plot_modified_fanned_curves(
    num_curves=5,                   # Number of curves
    angle_separation=np.pi/18,      # Separation between curves in radians
    length=10,                      # Base curve length
    amplitude=1,                    # Amplitude (same for all curves)
    frequency=1,                    # Frequency (same for all curves)
    inner_radius=0,                 # Inner radius (radius of the circle from which curves start)
    show_plot=True,                 # Display the plot
    save_svg=False                  # Save the plot as SVG
):
    # Angles at which curves will be rotated
    total_angle = angle_separation * (num_curves - 1)
    angles = np.linspace(-total_angle / 2, total_angle / 2, num_curves)

    # Create figure and axis
    fig, ax = plt.subplots(figsize=(8, 8))

    # Colors for the curves
    colors = ['black']  # First curve is black
    for i in range(1, num_curves - 1):
        colors.append('blue' if i % 2 == 1 else 'red')
    if num_curves > 1:
        colors.append('black')  # Last curve is black

    curves = []

    # Draw curves
    for i, angle in enumerate(angles):
        # Generate the base curve with specified amplitude and frequency
        x_base, y_base = generate_base_curve(length=length, amplitude=amplitude, frequency=frequency)

        # Rotate curve
        rot_matrix = np.array([
            [np.cos(angle), -np.sin(angle)],
            [np.sin(angle),  np.cos(angle)]
        ])
        curve = np.dot(rot_matrix, np.vstack((x_base, y_base)))

        # Translate curve so that all curves start from the inner circle
        # Calculate the starting point on the inner circle
        start_angle = angle
        delta_x = -curve[0, 0] + inner_radius * np.cos(start_angle)
        delta_y = -curve[1, 0] + inner_radius * np.sin(start_angle)
        curve[0] += delta_x
        curve[1] += delta_y

        curves.append(curve)

        # Plot curve
        ax.plot(curve[0], curve[1], color=colors[i], lw=2)

    # Outline with arcs at the inner and outer ends
    # Calculate start and end points
    start_points = np.array([curve[:, 0] for curve in curves])
    end_points = np.array([curve[:, -1] for curve in curves])

    # Draw inner arc (start of the fan)
    theta_start = angles[0]
    theta_end = angles[-1]
    theta_inner = np.linspace(theta_start, theta_end, 100)
    arc_x_inner = inner_radius * np.cos(theta_inner)
    arc_y_inner = inner_radius * np.sin(theta_inner)
    ax.plot(arc_x_inner, arc_y_inner, color='black', lw=2)

    # Draw outer arc (end of the fan)
    # Calculate the maximum radius of the end points for the outer arc
    end_distances = np.linalg.norm(end_points, axis=1)
    end_arc_radius = np.max(end_distances)
    theta_outer = np.linspace(theta_start, theta_end, 100)
    arc_x_outer = end_arc_radius * np.cos(theta_outer)
    arc_y_outer = end_arc_radius * np.sin(theta_outer)
    ax.plot(arc_x_outer, arc_y_outer, color='black', lw=2)

    # Set equal aspect ratio and remove axes
    ax.set_aspect('equal')
    ax.axis('off')

    # Show or save the plot
    if show_plot:
        plt.show()
    if save_svg:
        fig.savefig('modified_fanned_curves.svg', format='svg', bbox_inches='tight')
        print('SVG file saved as "modified_fanned_curves.svg"')
    plt.close(fig)

# Interactive widgets
def interactive_modified_fanned_curves():
    interactive_plot = interactive(
        plot_modified_fanned_curves,
        num_curves=IntSlider(min=2, max=20, step=1, value=5, description='Number of Curves'),
        angle_separation=FloatSlider(min=0, max=np.pi, step=np.pi/180, value=np.pi/18, description='Angle Separation (rad)'),
        length=FloatSlider(min=1, max=20, step=1, value=10, description='Curve Length'),
        amplitude=FloatSlider(min=0.1, max=5, step=0.1, value=1, description='Amplitude'),
        frequency=FloatSlider(min=0.1, max=5, step=0.1, value=1, description='Frequency'),
        inner_radius=FloatSlider(min=0, max=10, step=0.1, value=2, description='Inner Radius'),
        save_svg=Checkbox(value=False, description='Save as SVG')
    )
    display(interactive_plot)

# Run the interactive function
interactive_modified_fanned_curves()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interactive, IntSlider, FloatSlider, Checkbox
from IPython.display import display

# Function to generate the base curve with decreasing segment sizes
def generate_base_curve_with_decreasing_segments(
    num_points=1000, length=10, amplitude=1, frequency=1, s_max=1, s_min=0.5
):
    # Generate x positions with decreasing segment sizes
    x = [0]
    while x[-1] < length:
        s = s_max - (s_max - s_min) * (x[-1] / length)
        x_next = x[-1] + s
        x.append(x_next)
    x = np.array(x)
    x = x[x <= length]  # Ensure x does not exceed length
    y = amplitude * np.sin(2 * np.pi * frequency * x / length)
    return x, y

# Function to generate and plot the fanned-out curves with transversal folds
def plot_wavy_miura_ori(
    num_curves=5,                   # Number of curves
    angle_separation=np.pi/18,      # Separation between curves in radians
    length=10,                      # Base curve length
    amplitude=1,                    # Amplitude (same for all curves)
    frequency=1,                    # Frequency (same for all curves)
    inner_radius=0,                 # Inner radius (radius of the circle from which curves start)
    s_max=1,                        # Maximum segment size
    s_min=0.5,                      # Minimum segment size
    show_plot=True,                 # Display the plot
    save_svg=False                  # Save the plot as SVG
):
    # Angles at which curves will be rotated
    total_angle = angle_separation * (num_curves - 1)
    angles = np.linspace(-total_angle / 2, total_angle / 2, num_curves)

    # Create figure and axis
    fig, ax = plt.subplots(figsize=(8, 8))

    # Colors for the curves
    colors = ['black']  # First curve is black (outline)
    for i in range(1, num_curves - 1):
        colors.append('red' if i % 2 == 0 else 'blue')  # Alternate mountain and valley folds
    if num_curves > 1:
        colors.append('black')  # Last curve is black (outline)

    curves = []

    # Draw longitudinal folds (curves)
    for i, angle in enumerate(angles):
        # Generate the base curve with decreasing segments
        x_base, y_base = generate_base_curve_with_decreasing_segments(
            length=length, amplitude=amplitude, frequency=frequency, s_max=s_max, s_min=s_min
        )

        # Rotate curve
        rot_matrix = np.array([
            [np.cos(angle), -np.sin(angle)],
            [np.sin(angle),  np.cos(angle)]
        ])
        curve = np.dot(rot_matrix, np.vstack((x_base, y_base)))

        # Translate curve so that all curves start from the inner circle
        start_angle = angle
        delta_x = -curve[0, 0] + inner_radius * np.cos(start_angle)
        delta_y = -curve[1, 0] + inner_radius * np.sin(start_angle)
        curve[0] += delta_x
        curve[1] += delta_y

        curves.append(curve)

        # Plot curve
        ax.plot(curve[0], curve[1], color=colors[i], lw=2)

    # Draw transversal folds
    # For each segment position, draw lines connecting the curves
    num_segments = len(x_base)
    for j in range(num_segments):
        # Collect points from each curve at position x_base[j]
        points = np.array([curve[:, j] for curve in curves])
        # Determine fold type: alternate between mountain and valley
        fold_color = 'red' if j % 2 == 0 else 'blue'
        # Plot lines between adjacent curves
        for k in range(len(points) - 1):
            x_vals = [points[k][0], points[k+1][0]]
            y_vals = [points[k][1], points[k+1][1]]
            ax.plot(x_vals, y_vals, color=fold_color, lw=2)

    # Draw inner and outer arcs (outline)
    theta_start = angles[0]
    theta_end = angles[-1]
    theta_range = np.linspace(theta_start, theta_end, 100)

    # Inner arc
    arc_x_inner = inner_radius * np.cos(theta_range)
    arc_y_inner = inner_radius * np.sin(theta_range)
    ax.plot(arc_x_inner, arc_y_inner, color='black', lw=2)

    # Outer arc
    # Calculate maximum radius from the origin to the outermost point
    max_radius = max(np.hypot(curve[0], curve[1]).max() for curve in curves)
    arc_x_outer = max_radius * np.cos(theta_range)
    arc_y_outer = max_radius * np.sin(theta_range)
    ax.plot(arc_x_outer, arc_y_outer, color='black', lw=2)

    # Set equal aspect ratio and remove axes
    ax.set_aspect('equal')
    ax.axis('off')
    plt.tight_layout()

    # Show or save the plot
    if show_plot:
        plt.show()
    if save_svg:
        fig.savefig('wavy_miura_ori.svg', format='svg', dpi=300, bbox_inches='tight')
        print('SVG file saved as "wavy_miura_ori.svg"')
    plt.close(fig)

# Interactive widgets
def interactive_wavy_miura_ori():
    interactive_plot = interactive(
        plot_wavy_miura_ori,
        num_curves=IntSlider(min=2, max=20, step=1, value=5, description='Number of Curves'),
        angle_separation=FloatSlider(min=0.01, max=np.pi/2, step=0.01, value=np.pi/18, description='Angle Separation (rad)'),
        length=FloatSlider(min=5, max=50, step=1, value=10, description='Curve Length'),
        amplitude=FloatSlider(min=0.1, max=5, step=0.1, value=1, description='Amplitude'),
        frequency=FloatSlider(min=0.1, max=10, step=0.1, value=1, description='Frequency'),
        inner_radius=FloatSlider(min=0, max=10, step=0.1, value=2, description='Inner Radius'),
        s_max=FloatSlider(min=0.1, max=5, step=0.1, value=1, description='Max Segment Size'),
        s_min=FloatSlider(min=0.1, max=5, step=0.1, value=0.5, description='Min Segment Size'),
        save_svg=Checkbox(value=False, description='Save as SVG')
    )
    display(interactive_plot)

# Run the interactive function
interactive_wavy_miura_ori()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interactive, IntSlider, FloatSlider, Checkbox
from IPython.display import display

# Function to generate decreasing vertical positions
def generate_decreasing_vertical_positions(
    total_height=10, num_segments=10, s_max=1, s_min=0.5
):
    y_positions = [0]
    total_s = 0
    for i in range(num_segments - 1):
        s = s_max - (s_max - s_min) * (total_s / total_height)
        total_s += s
        y_positions.append(total_s)
        if total_s >= total_height:
            break
    y_positions = np.array(y_positions)
    y_positions = y_positions[y_positions <= total_height]
    return y_positions

# Function to generate and plot the pattern
def plot_adjustable_miura_ori(
    num_curves=5,                   # Number of curves (longitudinal folds)
    num_segments=10,                # Number of segments (transversal folds)
    length=10,                      # Width of the pattern
    total_height=10,                # Height of the pattern
    amplitude=1,                    # Amplitude of the sine wave
    frequency=1,                    # Frequency of the sine wave
    s_max=1,                        # Maximum segment size (vertical)
    s_min=0.5,                      # Minimum segment size (vertical)
    angle_separation=0,             # Separation between curves in radians
    show_plot=True,                 # Display the plot
    save_svg=False                  # Save the plot as SVG
):
    # Generate decreasing vertical positions
    y_positions = generate_decreasing_vertical_positions(
        total_height=total_height, num_segments=num_segments, s_max=s_max, s_min=s_min
    )

    # Generate horizontal positions along the sine curve
    x_base = np.linspace(0, length, num_curves)
    x_offsets = amplitude * np.sin(2 * np.pi * frequency * y_positions / total_height)

    # Create figure and axis
    fig, ax = plt.subplots(figsize=(8, 8))

    # Calculate angles for rotation of curves
    total_angle = angle_separation * (num_curves - 1)
    angles = np.linspace(-total_angle / 2, total_angle / 2, num_curves)

    # Collect points for longitudinal folds
    curves = []
    for i, angle in enumerate(angles):
        x_curve = np.full_like(y_positions, x_base[i]) + x_offsets
        y_curve = y_positions

        # Rotate curve if needed
        if angle != 0:
            rot_matrix = np.array([
                [np.cos(angle), -np.sin(angle)],
                [np.sin(angle),  np.cos(angle)]
            ])
            curve = np.dot(rot_matrix, np.vstack((x_curve, y_curve)))
            x_curve, y_curve = curve[0], curve[1]

        curves.append((x_curve, y_curve))

    # Plot longitudinal folds
    for i, (x_curve, y_curve) in enumerate(curves):
        if i == 0 or i == num_curves - 1:
            fold_color = 'black'  # Outline
            lw = 2.5
            z = 3
        else:
            fold_color = 'red' if i % 2 == 1 else 'blue'
            lw = 2
            z = 2
        ax.plot(x_curve, y_curve, color=fold_color, lw=lw, zorder=z)

    # Plot transversal folds
    num_points = len(y_positions)
    for j in range(num_points):
        # Collect points at each horizontal level
        points = np.array([[x_curve[j], y_curve[j]] for (x_curve, y_curve) in curves])

        # Draw lines between adjacent curves
        for i in range(len(points) - 1):
            # Determine fold color based on (i + j)
            fold_color = 'red' if (i + j) % 2 == 0 else 'blue'
            lw = 2
            z = 2
            x_vals = [points[i][0], points[i+1][0]]
            y_vals = [points[i][1], points[i+1][1]]
            ax.plot(x_vals, y_vals, color=fold_color, lw=lw, zorder=z)

    # Set equal aspect ratio and remove axes
    ax.set_aspect('equal')
    ax.axis('off')
    plt.tight_layout()

    # Show or save the plot
    if show_plot:
        plt.show()
    if save_svg:
        fig.savefig('adjustable_miura_ori.svg', format='svg', dpi=300, bbox_inches='tight')
        print('SVG file saved as "adjustable_miura_ori.svg')
    plt.close(fig)

# Interactive widgets
def interactive_adjustable_miura_ori():
    interactive_plot = interactive(
        plot_adjustable_miura_ori,
        num_curves=IntSlider(min=3, max=20, step=1, value=5, description='Number of Curves'),
        num_segments=IntSlider(min=5, max=50, step=1, value=10, description='Number of Segments'),
        length=FloatSlider(min=5, max=50, step=1, value=20, description='Pattern Width'),
        total_height=FloatSlider(min=5, max=50, step=1, value=20, description='Pattern Height'),
        amplitude=FloatSlider(min=0, max=5, step=0.1, value=1, description='Amplitude'),
        frequency=FloatSlider(min=0.1, max=30, step=0.1, value=2, description='Frequency'),
        s_max=FloatSlider(min=0.1, max=5, step=0.1, value=1.0, description='Max Segment Size'),
        s_min=FloatSlider(min=0.1, max=5, step=0.1, value=0.5, description='Min Segment Size'),
        angle_separation=FloatSlider(min=-np.pi/4, max=np.pi/4, step=0.01, value=0, description='Angle Separation (rad)'),
        save_svg=Checkbox(value=False, description='Save as SVG')
    )
    display(interactive_plot)

# Run the interactive function
interactive_adjustable_miura_ori()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interactive, FloatSlider, IntSlider
from IPython.display import display

# Function to generate Miura-Ori pattern with column scaling and adjustable angles
def plot_scaled_miura_ori(
    base_width=10,       # Total base width of the pattern
    base_height=10,      # Total base height of the pattern
    num_segments=6,      # Number of segments in each row
    num_columns=5,       # Number of columns
    width_scale=0.9,     # Scaling factor for column width
    height_scale=0.9,    # Scaling factor for column height
    row_angle_deg=30,    # Angle between rows in degrees
    leg_angle_deg=15,    # Incidence angle for legs in degrees
    show_plot=True       # Display the plot
):
    # Convert angles to radians
    row_angle_rad = np.radians(row_angle_deg)
    leg_angle_rad = np.radians(leg_angle_deg)

    # Initialize starting positions
    x_coords = np.zeros((num_columns, num_segments + 1))
    y_coords = np.zeros((num_columns, num_segments + 1))

    # Define initial segment size
    segment_width = base_width / num_segments
    segment_height = base_height

    for col in range(num_columns):
        # Scale width and height for this column
        col_width = segment_width * (width_scale ** col)
        col_height = segment_height * (height_scale ** col)

        for seg in range(num_segments + 1):
            # X-coordinate: linear increment
            x_coords[col, seg] = seg * col_width
            # Y-coordinate: alternating row angles
            if seg % 2 == 0:
                y_coords[col, seg] = col * np.tan(row_angle_rad) * col_width
            else:
                y_coords[col, seg] = col * np.tan(-row_angle_rad) * col_width

    # Plot the Miura-Ori pattern
    plt.figure(figsize=(8, 6))

    # Draw columns
    for col in range(num_columns):
        # Draw horizontal segments
        plt.plot(x_coords[col, :], y_coords[col, :], 'b-')

        # Draw diagonal legs
        for seg in range(num_segments):
            x_leg = [x_coords[col, seg], x_coords[col, seg + 1]]
            if seg % 2 == 0:
                y_leg = [y_coords[col, seg], y_coords[col, seg + 1] + col_height * np.tan(leg_angle_rad)]
            else:
                y_leg = [y_coords[col, seg], y_coords[col, seg + 1] - col_height * np.tan(leg_angle_rad)]
            plt.plot(x_leg, y_leg, 'b-')

    # Draw connecting vertical lines between rows
    for seg in range(num_segments + 1):
        for col in range(num_columns - 1):
            plt.plot(
                [x_coords[col, seg], x_coords[col + 1, seg]],
                [y_coords[col, seg], y_coords[col + 1, seg]],
                'r-'
            )

    # Adjust plot settings
    plt.gca().set_aspect('equal', adjustable='box')
    plt.axis('off')
    plt.title("Scaled Miura-Ori Pattern")
    plt.show()

# Interactive widget
def interactive_scaled_miura_ori():
    interactive_plot = interactive(
        plot_scaled_miura_ori,
        base_width=FloatSlider(value=10, min=5, max=50, step=1, description='Base Width'),
        base_height=FloatSlider(value=10, min=5, max=50, step=1, description='Base Height'),
        num_segments=IntSlider(value=6, min=3, max=20, step=1, description='Segments'),
        num_columns=IntSlider(value=5, min=2, max=10, step=1, description='Columns'),
        width_scale=FloatSlider(value=0.9, min=0.5, max=1.0, step=0.05, description='Width Scale'),
        height_scale=FloatSlider(value=0.9, min=0.5, max=1.0, step=0.05, description='Height Scale'),
        row_angle_deg=FloatSlider(value=30, min=0, max=90, step=1, description='Row Angle (°)'),
        leg_angle_deg=FloatSlider(value=15, min=0, max=90, step=1, description='Leg Angle (°)'),
    )
    display(interactive_plot)

# Run the interactive widget
interactive_scaled_miura_ori()
```

```python


# Function to create the segment shapes
def create_segment(angle1, angle2, base_radius, apex_radius, skew, apex_multiplier, segment_index):
    theta_base = np.linspace(angle1, angle2, 100)
    r_base = np.full_like(theta_base, base_radius)
    r_apex = np.full_like(theta_base, apex_radius * (apex_multiplier ** segment_index))
    
    base_segment = np.column_stack((theta_base, r_base))
    apex_segment = np.column_stack((theta_base + skew, r_apex + base_radius))
    
    segment = np.vstack((base_segment, apex_segment[::-1]))
    return segment

# Interactive plot function
def plot_segments(num_segments, base_radius, apex_radius, skew, apex_multiplier, separation, save_svg=False):
    angles = np.linspace(0, 2 * np.pi, num_segments + 1)
    separation_angle = separation * np.pi / 180  # Convert separation from degrees to radians
    
    fig, ax = plt.subplots(figsize=(8, 8), subplot_kw={'projection': 'polar'})
    ax.set_theta_direction(-1)
    ax.set_theta_offset(np.pi / 2)
    
    for i in range(num_segments):
        segment = create_segment(angles[i] + separation_angle / 2, angles[i + 1] - separation_angle / 2, base_radius, apex_radius, skew, apex_multiplier, i)
        ax.fill(segment[:, 0], segment[:, 1], edgecolor='black')


    plt.title('Interactive Circle with Tangential Segment Apexes')
    if save_svg:
        plt.savefig("segments_plot.svg", format="svg")
        print("SVG file saved as 'segments_plot.svg'")
    else:
        plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

def plot_interpolating_horizontal_zigzags_with_boundaries(num_lines, amplitude, height, angle):
    """
    Plot a graph with vertical zigzag lines with 90-degree angles, where every even line is reversed.
    Horizontal lines are added in the white space between zigzag lines, with interpolating widths.
    A top and bottom black boundary line is added, and the entire shape is rotated by a specified angle.

    :param num_lines: Number of zigzag lines.
    :param amplitude: Amplitude of the zigzag peaks.
    :param height: Height of the plot.
    :param angle: Rotation angle for the entire shape (in degrees).
    :return: None, but displays the plot.
    """
    # Rotation matrix for the specified angle
    theta = np.radians(angle)
    rotation_matrix = np.array([[np.cos(theta), -np.sin(theta)],
                                 [np.sin(theta),  np.cos(theta)]])
    
    fig, ax = plt.subplots(figsize=(12, 6))
    red = "#FF0000"
    blue = "#0000FF"
    multiplier = 2.5

    # Store all points for rotation
    all_points = []

    # Generate and plot each zigzag line
    for i in range(num_lines):
        x_points = []
        y_points = []
        for j in range(height + 1):
            if i % 2 == 0:  # Even line
                x = i * amplitude * multiplier + (amplitude if j % 2 == 0 else 0)
            else:  # Odd line
                x = i * amplitude * multiplier + (amplitude if j % 2 != 0 else 0)
            x_points.append(x)
            y_points.append(j)

        # Add points to the global point list for rotation
        all_points.extend(list(zip(x_points, y_points)))

        color = red if i % 2 == 0 else red
        color = 'black' if i in (0, num_lines - 1) else color
        ax.plot(x_points, y_points, color=color, lw=2)

    # Generate and plot interpolating horizontal lines
    for j in range(height + 1):
        y_horizontal = j
        for k in range(num_lines - 1):
            if (k + j) % 2 == 0:  # Horizontal line in the wider space
                x_start = k * amplitude * multiplier + amplitude
                x_end = x_start + amplitude + 1
            else:  # Horizontal line in the narrower space
                x_start = k * amplitude * multiplier
                x_end = x_start + amplitude * multiplier * 1.4

            color = red if (k + j) % 2 == 0 else blue
            if k <= num_lines - 1:
                all_points.extend([(x_start, y_horizontal), (x_end, y_horizontal)])
                ax.hlines(y_horizontal, x_start, x_end, color=color, linewidth=2)

    # Rotate all points
    rotated_points = np.dot(np.array(all_points), rotation_matrix.T)

    # Extract rotated X and Y coordinates
    x_rotated, y_rotated = rotated_points[:, 0], rotated_points[:, 1]

    # Plot rotated points
    ax.clear()
    ax.plot(x_rotated, y_rotated, color="black", lw=2, alpha=0.7)

    # Set plot limits based on rotated points
    ax.set_xlim(x_rotated.min() - amplitude, x_rotated.max() + amplitude)
    ax.set_ylim(y_rotated.min() - amplitude, y_rotated.max() + amplitude)

    # Remove axes for clarity
    ax.axis('off')

    # Save and show the plot
    plt.savefig("rotated_zigzag.svg", format='svg', bbox_inches='tight', pad_inches=0)
    plt.show()

# Parameters for the plot
num_lines = 5  # Number of zigzag lines
amplitude = 2  # Amplitude of the zigzag peaks
height = 10  # Height of the plot
angle = 30  # Rotation angle in degrees

# Generate the plot
plot_interpolating_horizontal_zigzags_with_boundaries(num_lines, amplitude, height, angle)
```

```python
def plot_tilted_inverse_lines(angle, angle_2, length, length_2, separation):
    """
    Plot two lines where one is tilted positively by a specified angle, and the other is tilted inversely.

    :param angle: Angle in degrees by which the first line is tilted positively and the second inversely.
    :param length: Length of the lines.
    :param separation: Vertical separation between the lines at the origin.
    :return: None, but displays the plot.
    """
    # Line 1: Tilted positively by the given angle
    x1 = np.linspace(0, length, 100)
    y1 = np.tan(np.radians(angle)) * x1

    # Line 2: Tilted inversely (negative angle), offset vertically
    x2 = np.linspace(0, length, 100)
    y2 = separation - np.tan(np.radians(angle)) * x2

     # Line 3: Tilted inversely (negative angle), offset vertically
    x3 = np.linspace(0, length_2, 100)
    y3 = np.tan(np.radians(angle_2)) * x3
    
    x4 = np.linspace(0, length_2, 100)
    y4 = separation - np.tan(np.radians(angle_2)) * x4
    
    # Plot the lines
    fig, ax = plt.subplots(figsize=(10, 6))
    ax.plot(x1, y1, label=f"Line 1 (Tilted by {angle}°)", color="blue", lw=2)
    ax.plot(x2, y2, label=f"Line 2 (Tilted by -{angle}°)", color="red", lw=2)
    ax.plot(x3, y3, label=f"Line 2 (Tilted by -{angle}°)", color="g", lw=2)
    ax.plot(x4, y4, label=f"Line 2 (Tilted by -{angle}°)", color="black", lw=2)

    # Enhance plot appearance
    ax.set_aspect('equal', adjustable='datalim')
    ax.axhline(0, color='black', linewidth=0.5, linestyle='--')
    ax.axvline(0, color='black', linewidth=0.5, linestyle='--')
    ax.set_title("Two Tilted Lines with Inverse Slopes and Vertical Offset")
    ax.legend()
    plt.grid(True)
    plt.show()

# Parameters for the lines
angle = 5  # Angle of tilt for one line (positive) and the other inversely (negative)
angle_2 = 35  # Angle of tilt for one line (positive) and the other inversely (negative)

length = 10  # Length of the lines
length_2 = 2.2  # Length of the lines
separation = 3# Vertical separation between the lines at the origin

# Generate and plot the lines
plot_tilted_inverse_lines(angle, angle_2, length, length_2, separation)
```

```python
import matplotlib.pyplot as plt
import numpy as np

def plot_hexagon_tessellation(rows, cols, size):
    fig, ax = plt.subplots(figsize=(10, 10))

    for row in range(rows):
        for col in range(cols):
            # Calculate the center of the hexagon
            x_offset = size * np.sqrt(3) * (col + 0.5 * (row % 2))
            y_offset = size * 1.5 * row
            hexagon = create_hexagon(x_offset, y_offset, size)
            # Alternate colors for valley (blue) and mountain (red) folds
            if (row + col) % 2 == 0:
                color = 'red'  # Mountain fold
            else:
                color = 'blue'  # Valley fold
            ax.plot(hexagon[:, 0], hexagon[:, 1], color=color)

    ax.set_aspect('equal')
    ax.axis('off')
    plt.show()

def create_hexagon(x_center, y_center, size):
    # Generate points for a hexagon centered at (x_center, y_center)
    angles = np.linspace(0, 2 * np.pi, 7)
    x = x_center + size * np.cos(angles)
    y = y_center + size * np.sin(angles)
    return np.column_stack((x, y))

# Parameters
rows = 2    # Number of rows of hexagons
cols = 5    # Number of columns of hexagons
size = 1    # Size of each hexagon

plot_hexagon_tessellation(rows, cols, size)
```

```python
import matplotlib.pyplot as plt
import numpy as np

def plot_interpolating_horizontal_zigzags_with_boundaries(num_lines, amplitude, height, angle):
    """
    Plot a graph with zigzag lines at a specified angle, where every even line is reversed.
    Horizontal lines are added in the white space between zigzag lines, with interpolating widths.
    A top and bottom black boundary line is added.

    :param num_lines: Number of zigzag lines.
    :param amplitude: Amplitude of the zigzag peaks.
    :param height: Number of vertical steps in the zigzag lines.
    :param angle: Angle in degrees for the zigzag lines.
    :return: None, but displays the plot.
    """
    fig, ax = plt.subplots(figsize=(10, 5))
    angle_rad = np.radians(angle)
    dx = amplitude * np.cos(angle_rad)
    dy = amplitude * np.sin(angle_rad)
    
    # Generate and plot each zigzag line
    for i in range(num_lines):
        x_points = []
        y_points = []
        x_start = i * dx * (height)
        y_start = 0
        x_points.append(x_start)
        y_points.append(y_start)
        
        for j in range(1, height + 1):
            direction = -1 if (i + j) % 2 == 0 else 1  # Alternate direction
            x_new = x_points[-1] + direction * dx
            y_new = y_points[-1] + dy
            x_points.append(x_new)
            y_points.append(y_new)
        
        color = 'red' if i % 2 == 0 else 'blue'
        ax.plot(x_points, y_points, color=color, lw=2)
    
    # Add top and bottom black boundary lines
    x_min = 0
    x_max = num_lines * dx * height
    y_bottom = 0
    y_top = height * dy
    ax.hlines(y_bottom, x_min, x_max, colors='black', linewidth=2)
    ax.hlines(y_top, x_min, x_max, colors='black', linewidth=2)
    
    # Set plot limits
    ax.set_xlim(x_min - amplitude, x_max + amplitude)
    ax.set_ylim(y_bottom - amplitude, y_top + amplitude)
    
    # Remove axes for clarity
    ax.axis('off')
    plt.savefig("zigzag_pattern.svg", format='svg', bbox_inches='tight', pad_inches=0)
    
    # Show the plot
    plt.show()

# Parameters for the plot
num_lines = 5   # Number of zigzag lines
amplitude = 2   # Amplitude of the zigzag peaks
height = 2      # Number of vertical steps in the zigzag lines (2 rows)
angle = 10      # Angle in degrees

# Generate the plot
plot_interpolating_horizontal_zigzags_with_boundaries(num_lines, amplitude, height, angle)
```

```python
import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interactive, IntSlider, FloatSlider, Checkbox
from IPython.display import display

def generate_base_curve_with_decreasing_segments(
    length=10, amplitude=1, frequency=1, s_max=1, s_min=0.5
):
    x = [0]
    while x[-1] < length:
        s = s_max - (s_max - s_min) * (x[-1] / length)
        x_next = x[-1] + s
        x.append(x_next)
    x = np.array(x)
    x = x[x <= length]
    y = amplitude * np.sin(2 * np.pi * frequency * x / length)
    return x, y

def angle_between_vectors(a, b):
    # Computes the angle between two vectors a and b
    cos_angle = np.dot(a, b) / (np.linalg.norm(a)*np.linalg.norm(b))
    # Clamp due to numerical errors
    cos_angle = np.clip(cos_angle, -1.0, 1.0)
    return np.arccos(cos_angle)

def measure_flat_foldability(curves):
    # Pick a vertex in the middle of the pattern (not at edges)
    # For simplicity, pick the midpoint indices:
    if len(curves) < 3 or curves[0].shape[1] < 3:
        return 0  # Not enough data
    i_mid = len(curves)//2
    j_mid = curves[0].shape[1]//2
    # vertex at curves[i_mid][0][j_mid], curves[i_mid][1][j_mid]
    vx = curves[i_mid][0, j_mid]
    vy = curves[i_mid][1, j_mid]
    
    # Adjacent vertices around the chosen vertex:
    # up:    i_mid-1, j_mid
    # down:  i_mid+1, j_mid
    # left:  i_mid, j_mid-1
    # right: i_mid, j_mid+1
    # Check bounds
    if i_mid-1 < 0 or i_mid+1 >= len(curves) or j_mid-1 < 0 or j_mid+1 >= curves[0].shape[1]:
        return 0
    upx, upy = curves[i_mid-1][0, j_mid], curves[i_mid-1][1, j_mid]
    downx, downy = curves[i_mid+1][0, j_mid], curves[i_mid+1][1, j_mid]
    leftx, lefty = curves[i_mid][0, j_mid-1], curves[i_mid][1, j_mid-1]
    rightx, righty = curves[i_mid][0, j_mid+1], curves[i_mid][1, j_mid+1]

    # Vectors from the vertex to neighbors:
    v_up = np.array([upx - vx, upy - vy])
    v_down = np.array([downx - vx, downy - vy])
    v_left = np.array([leftx - vx, lefty - vy])
    v_right = np.array([rightx - vx, righty - vy])

    # Compute angles around the vertex
    # We'll consider the four directions: up, right, down, left.
    # Find angles between successive vectors: up-right, right-down, down-left, left-up
    # Alternating sum condition: angle(up-right) + angle(down-left) should be ~ π
    # and angle(right-down) + angle(left-up) should be ~ π
    angle_ur = angle_between_vectors(v_up, v_right)
    angle_rd = angle_between_vectors(v_right, v_down)
    angle_dl = angle_between_vectors(v_down, v_left)
    angle_lu = angle_between_vectors(v_left, v_up)

    # Alternate sums
    sum1 = angle_ur + angle_dl
    sum2 = angle_rd + angle_lu
    # Ideal: sum1 = π and sum2 = π
    # We'll return how far we are from π in total
    error = abs(sum1 - np.pi) + abs(sum2 - np.pi)
    return error

def plot_adjustable_miura_ori(
    num_curves=5,                   
    angle_separation=0,            
    length=10,                     
    amplitude=1,                   
    frequency=1,                   
    inner_radius=0,                
    s_max=1,                       
    s_min=0.5,                     
    show_plot=True,                
    save_svg=False                 
):
    # We attempt a simple iterative approach to adjust angle_separation for flat-foldability:
    # We'll do a few iterations adjusting angle_separation slightly to minimize the angle error.
    # This is a heuristic approach and may not yield perfect flat-foldability.
    # Initial guess:
    angle_sep_original = angle_separation
    learning_rate = 0.005
    max_iters = 20
    best_error = np.inf
    best_angle_sep = angle_separation

    for _ in range(max_iters):
        # Generate pattern with current angle_separation
        current_angle_sep = angle_separation
        if current_angle_sep > np.pi/6:
            current_angle_sep = np.pi/6
        elif current_angle_sep < -np.pi/6:
            current_angle_sep = -np.pi/6

        x_base, y_base = generate_base_curve_with_decreasing_segments(
            length=length, amplitude=amplitude, frequency=frequency, s_max=s_max, s_min=s_min
        )

        total_angle = current_angle_sep * (num_curves - 1)
        angles = np.linspace(-total_angle / 2, total_angle / 2, num_curves)
        curves = []
        for i, angle in enumerate(angles):
            rot_matrix = np.array([
                [np.cos(angle), -np.sin(angle)],
                [np.sin(angle),  np.cos(angle)]
            ])
            curve = np.dot(rot_matrix, np.vstack((x_base, y_base)))
            start_angle = angle
            delta_x = -curve[0, 0] + inner_radius * np.cos(start_angle)
            delta_y = -curve[1, 0] + inner_radius * np.sin(start_angle)
            curve[0] += delta_x
            curve[1] += delta_y
            curves.append(curve)

        error = measure_flat_foldability(curves)
        if error < best_error:
            best_error = error
            best_angle_sep = current_angle_sep

        # If error is small enough, break
        if best_error < 0.01:
            break

        # Adjust angle_separation to reduce error
        # We'll try increasing/decreasing angle_separation and see which improves
        angle_test_plus = angle_separation + learning_rate
        angle_test_minus = angle_separation - learning_rate

        # Evaluate error for both directions
        test_errors = {}
        for test_angle in [angle_test_plus, angle_test_minus]:
            total_angle = test_angle * (num_curves - 1)
            angles = np.linspace(-total_angle / 2, total_angle / 2, num_curves)
            test_curves = []
            for i, angle in enumerate(angles):
                rot_matrix = np.array([
                    [np.cos(angle), -np.sin(angle)],
                    [np.sin(angle),  np.cos(angle)]
                ])
                curve = np.dot(rot_matrix, np.vstack((x_base, y_base)))
                start_angle = angle
                delta_x = -curve[0, 0] + inner_radius * np.cos(start_angle)
                delta_y = -curve[1, 0] + inner_radius * np.sin(start_angle)
                curve[0] += delta_x
                curve[1] += delta_y
                test_curves.append(curve)
            test_errors[test_angle] = measure_flat_foldability(test_curves)

        # Pick the angle direction that gives lower error
        if test_errors[angle_test_plus] < test_errors[angle_test_minus]:
            angle_separation = angle_test_plus
        else:
            angle_separation = angle_test_minus

    # After iterative adjustment, use best_angle_sep
    angle_separation = best_angle_sep
    # Generate final pattern
    x_base, y_base = generate_base_curve_with_decreasing_segments(
        length=length, amplitude=amplitude, frequency=frequency, s_max=s_max, s_min=s_min
    )
    total_angle = angle_separation * (num_curves - 1)
    angles = np.linspace(-total_angle / 2, total_angle / 2, num_curves)

    fig, ax = plt.subplots(figsize=(8, 8))

    curves = []
    longitudinal_colors = []
    for i, angle in enumerate(angles):
        rot_matrix = np.array([
            [np.cos(angle), -np.sin(angle)],
            [np.sin(angle),  np.cos(angle)]
        ])
        curve = np.dot(rot_matrix, np.vstack((x_base, y_base)))
        start_angle = angle
        delta_x = -curve[0, 0] + inner_radius * np.cos(start_angle)
        delta_y = -curve[1, 0] + inner_radius * np.sin(start_angle)
        curve[0] += delta_x
        curve[1] += delta_y
        curves.append(curve)

        if i == 0 or i == num_curves - 1:
            fold_color = 'black'
            lw = 2.5
            z = 3
            longitudinal_colors.append(fold_color)
        else:
            fold_color = 'red' if (i % 2 == 1) else 'blue'
            lw = 2
            z = 2
            longitudinal_colors.append(fold_color)

        ax.plot(curve[0], curve[1], color=fold_color, lw=lw, zorder=z)

    num_segments = len(x_base)
    for j in range(num_segments):
        points = np.array([curve[:, j] for curve in curves])
        for i in range(len(points) - 1):
            if (i + j) % 2 == 0:
                fold_color = 'red'  
            else:
                fold_color = 'blue' 
            lw = 2
            z = 2
            x_vals = [points[i][0], points[i+1][0]]
            y_vals = [points[i][1], points[i+1][1]]
            ax.plot(x_vals, y_vals, color=fold_color, lw=lw, zorder=z)

    ax.set_aspect('equal')
    ax.axis('off')
    plt.tight_layout()

    if show_plot:
        plt.show()
    if save_svg:
        fig.savefig('adjustable_miura_ori.svg', format='svg', dpi=300, bbox_inches='tight')
    plt.close(fig)

def interactive_adjustable_miura_ori():
    interactive_plot = interactive(
        plot_adjustable_miura_ori,
        num_curves=IntSlider(min=3, max=20, step=1, value=5, description='Number of Curves'),
        angle_separation=FloatSlider(min=-np.pi/4, max=np.pi/4, step=0.01, value=0, description='Angle Separation (rad)'),
        length=FloatSlider(min=5, max=50, step=1, value=20, description='Curve Length'),
        amplitude=FloatSlider(min=0.1, max=5, step=0.1, value=1, description='Amplitude'),
        frequency=FloatSlider(min=0.1, max=10, step=0.1, value=2, description='Frequency'),
        inner_radius=FloatSlider(min=0, max=10, step=0.1, value=0, description='Inner Radius'),
        s_max=FloatSlider(min=0.1, max=5, step=0.1, value=2, description='Max Segment Size'),
        s_min=FloatSlider(min=0.1, max=5, step=0.1, value=1, description='Min Segment Size'),
        save_svg=Checkbox(value=False, description='Save as SVG')
    )
    display(interactive_plot)

interactive_adjustable_miura_ori()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interactive, FloatSlider, IntSlider
from IPython.display import display

def plot_miura_ori(m=5, n=10, a=1.0, b=1.0, alpha=np.pi/3, show_plot=True, save_svg=False):
    # Generate coordinates
    # m rows, n columns of parallelograms -> (m+1) x (n+1) grid of vertices
    i_values = np.arange(m+1)
    j_values = np.arange(n+1)

    # Compute vertex coordinates
    X = np.zeros((m+1, n+1))
    Y = np.zeros((m+1, n+1))
    for i in i_values:
        for j in j_values:
            X[i,j] = j * a + (b * np.cos(alpha) if i % 2 == 1 else 0)
            Y[i,j] = i * b * np.sin(alpha)

    fig, ax = plt.subplots(figsize=(8, 8))

    # Plot longitudinal folds
    # Vertical direction: lines connecting (i,j) to (i+1,j)
    # Alternate coloring for mountain/valley folds
    for j in j_values:
        for i in range(m):
            # Assign folds (just alternate colors for demonstration)
            fold_color = 'red' if (i+j) % 2 == 0 else 'blue'
            ax.plot([X[i,j], X[i+1,j]], [Y[i,j], Y[i+1,j]], color=fold_color, lw=2)

    # Plot transversal folds
    # Horizontal direction: lines connecting (i,j) to (i,j+1)
    # Alternate these as well
    for i in i_values:
        for j in range(n):
            fold_color = 'blue' if (i+j) % 2 == 0 else 'red'
            ax.plot([X[i,j], X[i,j+1]], [Y[i,j], Y[i,j+1]], color=fold_color, lw=2)

    # Plot outer boundary in black for clarity
    # Left and right edges
    for i in range(m):
        ax.plot([X[i,0], X[i+1,0]], [Y[i,0], Y[i+1,0]], color='black', lw=2)
        ax.plot([X[i,n], X[i+1,n]], [Y[i,n], Y[i+1,n]], color='black', lw=2)
    # Top and bottom edges
    for j in range(n):
        ax.plot([X[0,j], X[0,j+1]], [Y[0,j], Y[0,j+1]], color='black', lw=2)
        ax.plot([X[m,j], X[m,j+1]], [Y[m,j], Y[m,j+1]], color='black', lw=2)

    ax.set_aspect('equal')
    ax.axis('off')
    plt.tight_layout()

    if show_plot:
        plt.show()
    if save_svg:
        fig.savefig('miura_ori.svg', format='svg', dpi=300, bbox_inches='tight')
        print('SVG file saved as "miura_ori.svg"')
    plt.close(fig)

def interactive_miura_ori():
    interactive_plot = interactive(
        plot_miura_ori,
        m=IntSlider(min=1, max=20, step=1, value=5, description='Rows (m)'),
        n=IntSlider(min=1, max=20, step=1, value=10, description='Cols (n)'),
        a=FloatSlider(min=0.5, max=5, step=0.1, value=1.0, description='a'),
        b=FloatSlider(min=0.5, max=5, step=0.1, value=1.0, description='b'),
        alpha=FloatSlider(min=0.1, max=np.pi-0.1, step=0.1, value=np.pi/3, description='alpha'),
        show_plot=True,
        save_svg=False
    )
    display(interactive_plot)

interactive_miura_ori()
```

```python
import numpy as np
from scipy.optimize import least_squares

def flat_fold_system(vars, M, N, a_array, b_array):
    # vars is a 1D array containing [X[0,0],Y[0,0], X[0,1],Y[0,1], ... X[M,N],Y[M,N]]
    # Reshape into 2D arrays for convenience
    X = vars[0::2].reshape(M+1, N+1)
    Y = vars[1::2].reshape(M+1, N+1)

    residuals = []

    # Distance constraints: horizontal edges
    for i in range(M+1):
        for j in range(N):
            dx = X[i,j+1] - X[i,j]
            dy = Y[i,j+1] - Y[i,j]
            residuals.append(dx*dx + dy*dy - a_array[i,j]**2)

    # Distance constraints: vertical edges
    for i in range(M):
        for j in range(N+1):
            dx = X[i+1,j] - X[i,j]
            dy = Y[i+1,j] - Y[i,j]
            residuals.append(dx*dx + dy*dy - b_array[i,j]**2)

    # Angle constraints at interior vertices
    # For each interior vertex, compute vectors to neighbors and enforce Kawasaki-Justin
    for i in range(1,M):
        for j in range(1,N):
            vx = X[i,j]
            vy = Y[i,j]
            # Four neighbors: up=(i-1,j), right=(i,j+1), down=(i+1,j), left=(i,j-1)
            # Make sure all neighbors exist
            v_up = np.array([X[i-1,j]-vx, Y[i-1,j]-vy])
            v_down = np.array([X[i+1,j]-vx, Y[i+1,j]-vy])
            v_left = np.array([X[i,j-1]-vx, Y[i,j-1]-vy])
            v_right = np.array([X[i,j+1]-vx, Y[i,j+1]-vy])

            # Compute angles:
            def angle(u,v):
                cos_ = np.dot(u,v)/(np.linalg.norm(u)*np.linalg.norm(v))
                cos_ = np.clip(cos_, -1,1)
                return np.arccos(cos_)

            angle_ur = angle(v_up, v_right)
            angle_rd = angle(v_right, v_down)
            angle_dl = angle(v_down, v_left)
            angle_lu = angle(v_left, v_up)

            # Kawasaki-Justin: angle_ur+angle_dl ~ pi and angle_rd+angle_lu ~ pi
            residuals.append((angle_ur + angle_dl - np.pi))
            residuals.append((angle_rd + angle_lu - np.pi))

    return np.array(residuals)

# Example usage:
M, N = 4, 6
# Define non-uniform a and b
a_array = np.random.uniform(0.9, 1.1, size=(M+1, N))  # horizontal lengths
b_array = np.random.uniform(0.9, 1.1, size=(M, N+1))  # vertical lengths

# Initial guess: start roughly as a grid
X_init = np.zeros((M+1, N+1))
Y_init = np.zeros((M+1, N+1))
for i in range(M+1):
    for j in range(N+1):
        X_init[i,j] = j
        Y_init[i,j] = i

vars_init = np.column_stack((X_init.flatten(), Y_init.flatten())).ravel()

res = least_squares(flat_fold_system, vars_init, args=(M,N,a_array,b_array))

X_sol = res.x[0::2].reshape(M+1, N+1)
Y_sol = res.x[1::2].reshape(M+1, N+1)
```

```python
import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interactive, FloatSlider, IntSlider
from IPython.display import display

def plot_scaled_miura_ori(m=5, n=10, a=1.0, b=1.0, alpha=np.pi/3, delta=0.1, show_plot=True, save_svg=False):
    # Generate per-row a_i and b_i
    # a_i = a*(1-delta)^i
    # b_i = b*(1-delta)^i
    a_values = a * (1 - delta)**np.arange(m+1)
    b_values = b * (1 - delta)**np.arange(m+1)
    
    # Compute vertical offsets: Y[i, *] depends on sum of all b_k for k < i
    # Y[i,j] = (sum_{k=0}^{i-1} b_k)*sin(alpha)
    vertical_offsets = np.zeros(m+1)
    for i in range(1, m+1):
        vertical_offsets[i] = vertical_offsets[i-1] + b_values[i-1]*np.sin(alpha)

    i_values = np.arange(m+1)
    j_values = np.arange(n+1)
    
    X = np.zeros((m+1, n+1))
    Y = np.zeros((m+1, n+1))

    for i in i_values:
        for j in j_values:
            base_x = j * a_values[i]
            # Odd rows shift horizontally by b_i*cos(alpha)
            if i % 2 == 1:
                base_x += b_values[i] * np.cos(alpha)
            X[i,j] = base_x
            Y[i,j] = vertical_offsets[i]

    fig, ax = plt.subplots(figsize=(8, 8))

    # Plot longitudinal folds
    for j in j_values:
        for ii in range(m):
            fold_color = 'blue' if (ii+j) % 2 == 0 else 'red'
            ax.plot([X[ii,j], X[ii+1,j]], [Y[ii,j], Y[ii+1,j]], color=fold_color, lw=2)

    # Plot transversal folds
    for i in i_values:
        for jj in range(n):
            fold_color = 'red' if (i+jj) % 2 == 0 else 'blue'
            ax.plot([X[i,jj], X[i,jj+1]], [Y[i,jj], Y[i,jj+1]], color=fold_color, lw=2)

    # Plot outer boundary in black for clarity
    for i in range(m):
        ax.plot([X[i,0], X[i+1,0]], [Y[i,0], Y[i+1,0]], color='black', lw=2)
        ax.plot([X[i,n], X[i+1,n]], [Y[i,n], Y[i+1,n]], color='black', lw=2)
    for j in range(n):
        ax.plot([X[0,j], X[0,j+1]], [Y[0,j], Y[0,j+1]], color='black', lw=2)
        ax.plot([X[m,j], X[m,j+1]], [Y[m,j], Y[m,j+1]], color='black', lw=2)

    ax.set_aspect('equal')
    ax.axis('off')
    plt.tight_layout()

    if show_plot:
        plt.show()
    if save_svg:
        fig.savefig('scaled_miura_ori.svg', format='svg', dpi=300, bbox_inches='tight')
        print('SVG file saved as "scaled_miura_ori.svg"')
    plt.close(fig)

def interactive_scaled_miura_ori():
    interactive_plot = interactive(
        plot_scaled_miura_ori,
        m=IntSlider(min=1, max=20, step=1, value=5, description='Rows (m)'),
        n=IntSlider(min=1, max=20, step=1, value=10, description='Cols (n)'),
        a=FloatSlider(min=0.5, max=5, step=0.1, value=1.0, description='a'),
        b=FloatSlider(min=0.5, max=5, step=0.1, value=1.0, description='b'),
        alpha=FloatSlider(min=0.1, max=np.pi-0.1, step=0.1, value=np.pi/3, description='alpha'),
        delta=FloatSlider(min=0.0, max=0.5, step=0.05, value=0.1, description='delta (shrink % per row)'),
        show_plot=True,
        save_svg=False
    )
    display(interactive_plot)

interactive_scaled_miura_ori()
```

```python
import numpy as np
from scipy.optimize import minimize

def compute_coordinates(m, n, a, alpha, delta):
    # Compute a_i and b_i values
    a_values = a * (1 - delta)**np.arange(m+1)
    b_values = a_values * np.sin(alpha)  # Maintaining standard Miura condition

    # Compute vertical offsets
    vertical_offsets = np.zeros(m+1)
    for i in range(1, m+1):
        vertical_offsets[i] = vertical_offsets[i-1] + b_values[i-1]*np.sin(alpha)

    i_values = np.arange(m+1)
    j_values = np.arange(n+1)
    X = np.zeros((m+1, n+1))
    Y = np.zeros((m+1, n+1))

    for i in i_values:
        for j in j_values:
            base_x = j * a_values[i]
            # Shift every other row
            if i % 2 == 1:
                base_x += b_values[i] * np.cos(alpha)
            X[i,j] = base_x
            Y[i,j] = vertical_offsets[i]

    # Centering horizontally
    top_min, top_max = X[0,:].min(), X[0,:].max()
    bottom_min, bottom_max = X[m,:].min(), X[m,:].max()
    top_center = 0.5 * (top_min + top_max)
    bottom_center = 0.5 * (bottom_min + bottom_max)
    midpoint = 0.5 * (top_center + bottom_center)
    X = X - midpoint

    return X, Y

def angle_between_vectors(u, v):
    # Compute angle between two vectors u and v in radians
    dot_product = np.dot(u, v)
    norm_u = np.linalg.norm(u)
    norm_v = np.linalg.norm(v)
    # Guard against floating-point errors
    cos_angle = np.clip(dot_product / (norm_u * norm_v), -1.0, 1.0)
    angle = np.degrees(np.arccos(cos_angle))
    return angle

def vertex_angle_deviations(m, n, a, alpha, delta):
    # Compute the coordinates
    X, Y = compute_coordinates(m, n, a, alpha, delta)

    # Collect angle deviations for all internal vertices
    deviations = []
    # Internal vertices: i in [1, m-1], j in [1, n-1]
    for i in range(1, m):
        for j in range(1, n):
            # Neighbors: up (i-1,j), right (i,j+1), down (i+1,j), left (i,j-1)
            v_up = np.array([X[i-1,j]-X[i,j], Y[i-1,j]-Y[i,j]])
            v_right = np.array([X[i,j+1]-X[i,j], Y[i,j+1]-Y[i,j]])
            v_down = np.array([X[i+1,j]-X[i,j], Y[i+1,j]-Y[i,j]])
            v_left = np.array([X[i,j-1]-X[i,j], Y[i,j-1]-Y[i,j]])

            # Order: up -> right -> down -> left -> (back to up)
            # Compute angles between consecutive vectors in clockwise order:
            θ_up_right = angle_between_vectors(v_up, v_right)
            θ_right_down = angle_between_vectors(v_right, v_down)
            θ_down_left = angle_between_vectors(v_down, v_left)
            θ_left_up = angle_between_vectors(v_left, v_up)

            # Kawasaki condition:
            # (θ_up_right + θ_down_left) should be = 180
            # (θ_right_down + θ_left_up) should be = 180
            dev1 = (θ_up_right + θ_down_left) - 180.0
            dev2 = (θ_right_down + θ_left_up) - 180.0

            deviations.append(dev1)
            deviations.append(dev2)

    return np.array(deviations)

def cost_function(params, m, n):
    a, alpha, delta = params
    dev = vertex_angle_deviations(m, n, a, alpha, delta)
    return np.sum(dev**2)

# Example usage:
m, n = 5, 10
initial_guess = [1.0, np.pi/3, 0.05]  # (a, alpha, delta)
result = minimize(cost_function, initial_guess, args=(m, n), method='Nelder-Mead')

a_opt, alpha_opt, delta_opt = result.x
print("Optimized parameters:")
print("a =", a_opt)
print("alpha =", alpha_opt)
print("delta =", delta_opt)
print("Final cost:", result.fun)
```

```python
import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interactive, FloatSlider, IntSlider
from IPython.display import display
from scipy.optimize import minimize

def angle_between_vectors(u, v):
    dot_product = np.dot(u, v)
    norm_u = np.linalg.norm(u)
    norm_v = np.linalg.norm(v)
    cos_angle = np.clip(dot_product / (norm_u * norm_v), -1.0, 1.0)
    angle = np.degrees(np.arccos(cos_angle))
    return angle

def compute_coordinates(m, n, a, alpha, delta):
    # Compute a_i and b_i values assuming b = a * sin(alpha) for a Miura parallelogram
    a_values = a * (1 - delta)**np.arange(m+1)
    b_values = a_values * np.sin(alpha)

    vertical_offsets = np.zeros(m+1)
    for i in range(1, m+1):
        vertical_offsets[i] = vertical_offsets[i-1] + b_values[i-1]*np.sin(alpha)

    i_values = np.arange(m+1)
    j_values = np.arange(n+1)
    X = np.zeros((m+1, n+1))
    Y = np.zeros((m+1, n+1))

    for i in i_values:
        for j in j_values:
            base_x = j * a_values[i]
            if i % 2 == 1:
                base_x += b_values[i] * np.cos(alpha)
            X[i,j] = base_x
            Y[i,j] = vertical_offsets[i]

    top_min, top_max = X[0,:].min(), X[0,:].max()
    bottom_min, bottom_max = X[m,:].min(), X[m,:].max()
    top_center = 0.5 * (top_min + top_max)
    bottom_center = 0.5 * (bottom_min + bottom_max)
    midpoint = 0.5 * (top_center + bottom_center)
    X = X - midpoint

    return X, Y

def vertex_angle_deviations(m, n, a, alpha, delta):
    X, Y = compute_coordinates(m, n, a, alpha, delta)

    deviations = []
    for i in range(1, m):
        for j in range(1, n):
            v_up = np.array([X[i-1,j]-X[i,j], Y[i-1,j]-Y[i,j]])
            v_right = np.array([X[i,j+1]-X[i,j], Y[i,j+1]-Y[i,j]])
            v_down = np.array([X[i+1,j]-X[i,j], Y[i+1,j]-Y[i,j]])
            v_left = np.array([X[i,j-1]-X[i,j], Y[i,j-1]-Y[i,j]])

            θ_up_right = angle_between_vectors(v_up, v_right)
            θ_right_down = angle_between_vectors(v_right, v_down)
            θ_down_left = angle_between_vectors(v_down, v_left)
            θ_left_up = angle_between_vectors(v_left, v_up)

            dev1 = (θ_up_right + θ_down_left) - 180.0
            dev2 = (θ_right_down + θ_left_up) - 180.0

            deviations.append(dev1)
            deviations.append(dev2)

    return np.array(deviations)

def cost_function(a, m, n, alpha, delta):
    dev = vertex_angle_deviations(m, n, a, alpha, delta)
    return np.sum(dev**2)

def plot_scaled_miura_ori(m=5, n=10, a=1.0, b=1.0, alpha=np.pi/3, delta=0.1, show_plot=True, save_svg=False):
    # Optimize only over 'a' to maintain alpha and delta as specified
    result = minimize(cost_function, [a], args=(m, n, alpha, delta), method='Nelder-Mead')
    a_opt = result.x[0]

    i_values = np.arange(m+1)
    a_values = a_opt * (1 - (i_values/m)*delta)
    # Keep b_values tied to a_opt and alpha (Miura condition)
    b_values = (a_opt * np.sin(alpha)) * (1 - (i_values/m)*delta)

    vertical_offsets = np.zeros(m+1)
    for i in range(1, m+1):
        vertical_offsets[i] = vertical_offsets[i-1] + b_values[i-1]*np.sin(alpha)

    j_values = np.arange(n+1)
    X = np.zeros((m+1, n+1))
    Y = np.zeros((m+1, n+1))

    for i in i_values:
        for j in j_values:
            offset_factor = i * b_values[i] * np.cos(alpha) * delta
            if i % 2 == 0:
                base_x = j * a_values[i] + 0.5 * offset_factor
            else:
                base_x = j * a_values[i] + offset_factor
            X[i,j] = base_x
            Y[i,j] = vertical_offsets[i]

    top_min, top_max = X[0,:].min(), X[0,:].max()
    bottom_min, bottom_max = X[m,:].min(), X[m,:].max()
    top_center = 0.5 * (top_min + top_max)
    bottom_center = 0.5 * (bottom_min + bottom_max)
    midpoint = 0.5 * (top_center + bottom_center)
    X = X - midpoint

    fig, ax = plt.subplots(figsize=(8, 8))

    # Longitudinal folds
    for j in j_values:
        for ii in range(m):
            fold_color = 'blue' if j % 2 == 0 else 'red'
            ax.plot([X[ii,j], X[ii+1,j]], [Y[ii,j], Y[ii+1,j]], color=fold_color, lw=2)

    # Transversal folds
    for i in i_values:
        for jj in range(n):
            fold_color = 'red' if (i+jj) % 2 == 0 else 'blue'
            ax.plot([X[i,jj], X[i,jj+1]], [Y[i,jj], Y[i,jj+1]], color=fold_color, lw=2)

    # Outer boundary
    for i in range(m):
        ax.plot([X[i,0], X[i+1,0]], [Y[i,0], Y[i+1,0]], color='black', lw=2)
        ax.plot([X[i,n], X[i+1,n]], [Y[i,n], Y[i+1,n]], color='black', lw=2)
    for j in range(n):
        ax.plot([X[0,j], X[0,j+1]], [Y[0,j], Y[0,j+1]], color='black', lw=2)
        ax.plot([X[m,j], X[m,j+1]], [Y[m,j], Y[m,j+1]], color='black', lw=2)

    ax.set_aspect('equal')
    ax.axis('off')
    plt.tight_layout()

    if show_plot:
        plt.show()
    if save_svg:
        fig.savefig('scaled_miura_ori.svg', format='svg', dpi=300, bbox_inches='tight')
        print('SVG file saved as "scaled_miura_ori.svg"')
    plt.close(fig)

def interactive_scaled_miura_ori():
    interactive_plot = interactive(
        plot_scaled_miura_ori,
        m=IntSlider(min=1, max=20, step=1, value=5, description='Rows (m)'),
        n=IntSlider(min=1, max=20, step=1, value=10, description='Cols (n)'),
        a=FloatSlider(min=0.5, max=5, step=0.1, value=1.0, description='a'),
        b=FloatSlider(min=0.5, max=5, step=0.1, value=1.0, description='b'),
        alpha=FloatSlider(min=0.1, max=np.pi-0.1, step=0.1, value=np.pi/3, description='alpha'),
        delta=FloatSlider(min=0.0, max=0.5, step=0.05, value=0.1, description='delta (shrink % per row)'),
        show_plot=True,
        save_svg=False
    )
    display(interactive_plot)

interactive_scaled_miura_ori()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interactive, FloatSlider, IntSlider
from IPython.display import display
from scipy.optimize import minimize

def angle_between_vectors(u, v):
    dot_product = np.dot(u, v)
    norm_u = np.linalg.norm(u)
    norm_v = np.linalg.norm(v)
    cos_angle = np.clip(dot_product / (norm_u * norm_v), -1.0, 1.0)
    angle = np.degrees(np.arccos(cos_angle))
    return angle

def compute_coordinates_with_params(m, n, a, alpha, delta, params):
    # Extract horizontal (p_x) and vertical (p_y) offsets from params
    p_x = params[:m+1]
    p_y = params[m+1:]

    i_values = np.arange(m+1)
    j_values = np.arange(n+1)

    # Base values
    a_values = a*(1 - (i_values/m)*delta)
    b_values = (a * np.sin(alpha))*(1 - (i_values/m)*delta)

    # Vertical offsets from the standard Miura baseline
    vertical_offsets = np.zeros(m+1)
    for i in range(1, m+1):
        vertical_offsets[i] = vertical_offsets[i-1] + b_values[i-1]*np.sin(alpha)

    X = np.zeros((m+1, n+1))
    Y = np.zeros((m+1, n+1))

    # Introduce a zigzag pattern depending on i and j
    # We use (-1)^(i+j) to alternate sign per vertex and incorporate alpha
    # Example: horizontal zigzag scaled by b_values[i]*cos(alpha),
    # and a smaller vertical modulation as well.
    #
    # Adjust these functions as needed for more/less eccentricity.
    for i in i_values:
        for j in j_values:
            # Baseline horizontal (from Miura)
            base_x = j * a_values[i]
            # Add a zigzag pattern in horizontal direction:
            # For example, alternate every vertex in a checkerboard pattern:
            zigzag_x = ((-1)**(i+j)) * b_values[i] * np.cos(alpha) * 0.5
            # Similarly, add a small vertical zigzag:
            zigzag_y = ((-1)**(j)) * b_values[i] * np.sin(alpha) * 0.1

            # Combine baseline and zigzag
            X[i,j] = base_x + zigzag_x + p_x[i]
            Y[i,j] = vertical_offsets[i] + zigzag_y + p_y[i]

    # Centering horizontally
    top_min, top_max = X[0,:].min(), X[0,:].max()
    bottom_min, bottom_max = X[m,:].min(), X[m,:].max()
    top_center = 0.5 * (top_min + top_max)
    bottom_center = 0.5 * (bottom_min + bottom_max)
    midpoint = 0.5 * (top_center + bottom_center)
    X = X - midpoint

    return X, Y

def vertex_angle_deviations_with_params(m, n, a, alpha, delta, params):
    X, Y = compute_coordinates_with_params(m, n, a, alpha, delta, params)

    deviations = []
    for i in range(1, m):
        for j in range(1, n):
            v_up = np.array([X[i-1,j]-X[i,j], Y[i-1,j]-Y[i,j]])
            v_right = np.array([X[i,j+1]-X[i,j], Y[i,j+1]-Y[i,j]])
            v_down = np.array([X[i+1,j]-X[i,j], Y[i+1,j]-Y[i,j]])
            v_left = np.array([X[i,j-1]-X[i,j], Y[i,j-1]-Y[i,j]])

            θ_up_right = angle_between_vectors(v_up, v_right)
            θ_right_down = angle_between_vectors(v_right, v_down)
            θ_down_left = angle_between_vectors(v_down, v_left)
            θ_left_up = angle_between_vectors(v_left, v_up)

            dev1 = (θ_up_right + θ_down_left) - 180.0
            dev2 = (θ_right_down + θ_left_up) - 180.0

            deviations.append(dev1)
            deviations.append(dev2)
    return np.array(deviations)

def angle_deviation_cost(params, m, n, a, alpha, delta):
    dev = vertex_angle_deviations_with_params(m, n, a, alpha, delta, params)
    return np.sum(dev**2)

def plot_scaled_miura_ori(m=5, n=10, a=1.0, b=1.0, alpha=np.pi/3, delta=0.1, show_plot=True, save_svg=False):
    # Optimize horizontal and vertical offsets (2*(m+1) params)
    initial_guess = np.zeros((m+1)*2)
    result = minimize(angle_deviation_cost, initial_guess, args=(m, n, a, alpha, delta), method='Nelder-Mead')
    params_opt = result.x

    X, Y = compute_coordinates_with_params(m, n, a, alpha, delta, params_opt)
    i_values = np.arange(m+1)
    j_values = np.arange(n+1)

    fig, ax = plt.subplots(figsize=(8, 8))

    # Longitudinal folds
    for j in j_values:
        for ii in range(m):
            fold_color = 'blue' if j % 2 == 0 else 'red'
            ax.plot([X[ii,j], X[ii+1,j]], [Y[ii,j], Y[ii+1,j]], color=fold_color, lw=2)

    # Transversal folds
    for i in i_values:
        for jj in range(n):
            fold_color = 'blue' if (i+jj) % 2 == 0 else 'red'
            ax.plot([X[i,jj], X[i,jj+1]], [Y[i,jj], Y[i,jj+1]], color=fold_color, lw=2)

    # Outer boundary
    for i in range(m):
        ax.plot([X[i,0], X[i+1,0]], [Y[i,0], Y[i+1,0]], color='black', lw=2)
        ax.plot([X[i,n], X[i+1,n]], [Y[i,n], Y[i+1,n]], color='black', lw=2)
    for j in range(n):
        ax.plot([X[0,j], X[0,j+1]], [Y[0,j], Y[0,j+1]], color='black', lw=2)
        ax.plot([X[m,j], X[m,j+1]], [Y[m,j], Y[m,j+1]], color='black', lw=2)

    ax.set_aspect('equal')
    ax.axis('off')
    plt.tight_layout()

    if show_plot:
        plt.show()
    if save_svg:
        fig.savefig('scaled_miura_ori.svg', format='svg', dpi=300, bbox_inches='tight')
        print('SVG file saved as "scaled_miura_ori.svg"')
    plt.close(fig)

def interactive_scaled_miura_ori():
    interactive_plot = interactive(
        plot_scaled_miura_ori,
        m=IntSlider(min=1, max=20, step=1, value=5, description='Rows (m)'),
        n=IntSlider(min=1, max=20, step=1, value=4, description='Cols (n)'),
        a=FloatSlider(min=0.5, max=5, step=0.1, value=4.9, description='a'),
        b=FloatSlider(min=0.5, max=5, step=0.1, value=1.0, description='b'),
        alpha=FloatSlider(min=0.1, max=np.pi-0.1, step=0.9, value=np.pi/3, description='alpha'),
        delta=FloatSlider(min=0.0, max=0.5, step=0.05, value=0.05, description='delta (shrink % per row)'),
        show_plot=True,
        save_svg=True
    )
    display(interactive_plot)

interactive_scaled_miura_ori()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interactive, FloatSlider, IntSlider, Checkbox
from IPython.display import display

def compute_coordinates(m, n, a, alpha, delta):
    """
    Compute the X and Y coordinates of the Miura-ori pattern with scaling.

    Parameters:
    - m (int): Number of rows.
    - n (int): Number of columns.
    - a (float): Initial horizontal side length.
    - alpha (float): Folding angle in radians.
    - delta (float): Scaling factor per row (0 <= delta <= 1).

    Returns:
    - X (2D numpy array): X coordinates of the grid.
    - Y (2D numpy array): Y coordinates of the grid.
    """
    i_values = np.arange(m+1)
    j_values = np.arange(n+1)

    # Compute scaling factors for each row
    # a_i = a * (1 - delta * (i/m))
    a_values = a * (1 - delta * (i_values / m))
    
    # Ensure no negative or zero lengths
    a_values = np.maximum(a_values, 1e-3)
    
    # Compute b_i based on alpha to preserve angles
    # b_i = a_i * tan(alpha)
    b_values = a_values * np.tan(alpha)
    
    # Compute vertical offsets for each row
    vertical_offsets = np.zeros(m+1)
    for i in range(1, m+1):
        vertical_offsets[i] = vertical_offsets[i-1] + b_values[i-1] * np.sin(alpha)
    
    # Initialize coordinate grids
    X = np.zeros((m+1, n+1))
    Y = np.zeros((m+1, n+1))
    
    for i in i_values:
        for j in j_values:
            base_x = j * a_values[i]
            # Shift every other row by b_i * cos(alpha) to create the Miura-ori pattern
            if i % 2 == 1:
                base_x += b_values[i] * np.cos(alpha)
            X[i, j] = base_x
            Y[i, j] = vertical_offsets[i]
    
    # Center the pattern horizontally
    top_center = (X[0, :].min() + X[0, :].max()) / 2
    bottom_center = (X[m, :].min() + X[m, :].max()) / 2
    midpoint = (top_center + bottom_center) / 2
    X -= midpoint
    
    return X, Y

def plot_miura_ori(m=5, n=10, a=1.0, alpha=np.pi/3, delta=0.1, show_plot=True, save_svg=False):
    """
    Plot the scaled Miura-ori pattern ensuring flat foldability.

    Parameters:
    - m (int): Number of rows.
    - n (int): Number of columns.
    - a (float): Initial horizontal side length.
    - alpha (float): Folding angle in radians.
    - delta (float): Scaling factor per row (0 <= delta <= 1).
    - show_plot (bool): Whether to display the plot.
    - save_svg (bool): Whether to save the plot as an SVG file.
    """
    # Compute coordinates
    X, Y = compute_coordinates(m, n, a, alpha, delta)
    
    i_values = np.arange(m+1)
    j_values = np.arange(n+1)
    
    # Initialize plot
    fig, ax = plt.subplots(figsize=(8, 8))
    
    # Plot longitudinal folds (vertical lines)
    for j in j_values:
        for i in range(m):
            fold_color = 'blue' if j % 2 == 0 else 'red'
            ax.plot([X[i, j], X[i+1, j]], [Y[i, j], Y[i+1, j]], color=fold_color, lw=2)
    
    # Plot transversal folds (horizontal lines)
    for i in i_values:
        for j in range(n):
            fold_color = 'red' if (i + j) % 2 == 0 else 'blue'
            ax.plot([X[i, j], X[i, j+1]], [Y[i, j], Y[i, j+1]], color=fold_color, lw=2)
    
    # Plot outer boundaries
    for i in range(m):
        # Left boundary
        ax.plot([X[i, 0], X[i+1, 0]], [Y[i, 0], Y[i+1, 0]], color='black', lw=2)
        # Right boundary
        ax.plot([X[i, n], X[i+1, n]], [Y[i, n], Y[i+1, n]], color='black', lw=2)
    for j in range(n):
        # Top boundary
        ax.plot([X[0, j], X[0, j+1]], [Y[0, j], Y[0, j+1]], color='black', lw=2)
        # Bottom boundary
        ax.plot([X[m, j], X[m, j+1]], [Y[m, j], Y[m, j+1]], color='black', lw=2)
    
    # Final plot adjustments
    ax.set_aspect('equal')
    ax.axis('off')
    plt.tight_layout()
    
    if show_plot:
        plt.show()
    if save_svg:
        fig.savefig('scaled_miura_ori.svg', format='svg', dpi=300, bbox_inches='tight')
        print('SVG file saved as "scaled_miura_ori.svg"')
    plt.close(fig)

def interactive_scaled_miura_ori():
    """
    Create an interactive widget to adjust Miura-ori parameters and visualize the pattern.
    """
    interactive_plot = interactive(
        plot_miura_ori,
        m=IntSlider(min=1, max=20, step=1, value=5, description='Rows (m)'),
        n=IntSlider(min=1, max=20, step=1, value=4, description='Cols (n)'),
        a=FloatSlider(min=0.5, max=5.0, step=0.1, value=1.0, description='a'),
        alpha=FloatSlider(min=0.1, max=np.pi-0.1, step=0.05, value=1.05, description='alpha (rad)'),
        delta=FloatSlider(min=0.0, max=0.5, step=0.05, value=0.25, description='delta'),
        show_plot=Checkbox(value=True, description='Show Plot'),
        save_svg=Checkbox(value=False, description='Save SVG')
    )
    display(interactive_plot)

# Run the interactive plot
interactive_scaled_miura_ori()
```

```python
import numpy as np
import pandas as pd

def generate_flat_foldable_data(m=5, n=4, num_samples=10000):
    """
    Generate a dataset of Miura-ori patterns ensuring flat foldability.
    
    Parameters:
    - m (int): Number of rows.
    - n (int): Number of columns.
    - num_samples (int): Number of samples to generate.
    
    Returns:
    - DataFrame: Contains parameters and corresponding y-coordinates.
    """
    data = []
    
    for _ in range(num_samples):
        # Randomly sample parameters within specified ranges
        a = np.random.uniform(0.5, 5.0)
        alpha = np.random.uniform(0.1, np.pi - 0.1)
        delta = np.random.uniform(0.0, 0.5)
        
        # Compute a_i and b_i
        i_values = np.arange(m+1)
        a_values = a * (1 - delta * (i_values / m))
        a_values = np.maximum(a_values, 1e-3)  # Prevent zero or negative lengths
        b_values = a_values * np.tan(alpha)
        
        # Compute y-coordinates
        y_offsets = [0.0]
        for i in range(1, m+1):
            y_new = y_offsets[i-1] + b_values[i-1] * np.sin(alpha)
            y_offsets.append(y_new)
        
        # Flatten y_offsets for the dataset
        y_coords = y_offsets
        
        # Append to data
        sample = {
            'a': a,
            'alpha': alpha,
            'delta': delta,
            'y0': y_coords[0],
            'y1': y_coords[1],
            'y2': y_coords[2],
            'y3': y_coords[3],
            'y4': y_coords[4],
            'y5': y_coords[5],
        }
        data.append(sample)
    
    df = pd.DataFrame(data)
    return df
```

```python
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split

def build_model():
    """
    Build and compile the neural network model.
    
    Returns:
    - model (keras.Model): Compiled neural network model.
    """
    model = keras.Sequential([
        layers.Input(shape=(3,)),  # Inputs: a, alpha, delta
        layers.Dense(64, activation='relu'),
        layers.Dense(64, activation='relu'),
        layers.Dense(6)  # Outputs: y0 to y5
    ])
    
    model.compile(optimizer='adam', loss='mse', metrics=['mae'])
    return model

# Generate dataset
df = generate_flat_foldable_data(m=5, n=4, num_samples=100000)

# Split into inputs and outputs
X = df[['a', 'alpha', 'delta']].values
y = df[['y0', 'y1', 'y2', 'y3', 'y4', 'y5']].values

# Normalize inputs and outputs for better training
scaler_X = StandardScaler()
scaler_y = StandardScaler()

X_scaled = scaler_X.fit_transform(X)
y_scaled = scaler_y.fit_transform(y)

# Split into training and validation sets
X_train, X_val, y_train, y_val = train_test_split(X_scaled, y_scaled, test_size=0.3, random_state=42)

# Build and train the model
model = build_model()

history = model.fit(
    X_train, y_train,
    epochs=100,
    batch_size=32,
    validation_data=(X_val, y_val),
    verbose=1
)
```

```python
def predict_y_coords(a, alpha, delta, model, scaler_X, scaler_y):
    """
    Predict y-coordinates using the trained model.
    
    Parameters:
    - a (float): Initial horizontal side length.
    - alpha (float): Folding angle in radians.
    - delta (float): Scaling factor per row.
    - model (keras.Model): Trained Keras model.
    - scaler_X (StandardScaler): Scaler for inputs.
    - scaler_y (StandardScaler): Scaler for outputs.
    
    Returns:
    - y_coords (list): Predicted y-coordinates [y0, y1, ..., ym]
    """
    input_data = np.array([[a, alpha, delta]])
    input_scaled = scaler_X.transform(input_data)
    y_pred_scaled = model.predict(input_scaled)
    y_pred = scaler_y.inverse_transform(y_pred_scaled)
    y_coords = y_pred[0].tolist()
    return y_coords
```

```python
def plot_scaled_miura_ori_with_model(m=5, n=4, a=1.0, alpha=1.05, delta=0.25, model=None, scaler_X=None, scaler_y=None, show_plot=True, save_svg=False):
    """
    Plot the scaled Miura-ori pattern using predicted y-coordinates from the model.
    
    Parameters:
    - m (int): Number of rows.
    - n (int): Number of columns.
    - a (float): Initial horizontal side length.
    - alpha (float): Folding angle in radians.
    - delta (float): Scaling factor per row.
    - model (keras.Model): Trained Keras model.
    - scaler_X (StandardScaler): Scaler for inputs.
    - scaler_y (StandardScaler): Scaler for outputs.
    - show_plot (bool): Whether to display the plot.
    - save_svg (bool): Whether to save the plot as an SVG file.
    """
    if model is None or scaler_X is None or scaler_y is None:
        print("Model and scalers must be provided.")
        return
    
    # Predict y-coordinates using the model
    y_coords = predict_y_coords(a, alpha, delta, model, scaler_X, scaler_y)
    
    # Compute a_i and b_i
    i_values = np.arange(m+1)
    a_values = a * (1 - delta * (i_values / m))
    a_values = np.maximum(a_values, 1e-3)  # Prevent zero or negative lengths
    b_values = a_values * np.tan(alpha)
    
    # Initialize coordinate grids
    X = np.zeros((m+1, n+1))
    Y = np.zeros((m+1, n+1))
    
    for i in i_values:
        for j in range(n+1):
            base_x = j * a_values[i]
            # Shift every other row by b_i * cos(alpha) to create the Miura-ori pattern
            if i % 2 == 1:
                base_x += b_values[i] * np.cos(alpha)
            X[i, j] = base_x
            Y[i, j] = y_coords[i]
    
    # Center the pattern horizontally
    top_center = (X[0, :].min() + X[0, :].max()) / 2
    bottom_center = (X[m, :].min() + X[m, :].max()) / 2
    midpoint = (top_center + bottom_center) / 2
    X -= midpoint
    
    # Initialize plot
    fig, ax = plt.subplots(figsize=(8, 8))
    
    # Plot longitudinal folds (vertical lines)
    for j in range(n+1):
        for i in range(m):
            fold_color = 'blue' if j % 2 == 0 else 'red'
            ax.plot([X[i, j], X[i+1, j]], [Y[i, j], Y[i+1, j]], color=fold_color, lw=2)
    
    # Plot transversal folds (horizontal lines)
    for i in i_values:
        for j in range(n):
            fold_color = 'red' if (i + j) % 2 == 0 else 'blue'
            ax.plot([X[i, j], X[i, j+1]], [Y[i, j], Y[i, j+1]], color=fold_color, lw=2)
    
    # Plot outer boundaries
    for i in range(m):
        # Left boundary
        ax.plot([X[i, 0], X[i+1, 0]], [Y[i, 0], Y[i+1, 0]], color='black', lw=2)
        # Right boundary
        ax.plot([X[i, n], X[i+1, n]], [Y[i, n], Y[i+1, n]], color='black', lw=2)
    for j in range(n):
        # Top boundary
        ax.plot([X[0, j], X[0, j+1]], [Y[0, j], Y[0, j+1]], color='black', lw=2)
        # Bottom boundary
        ax.plot([X[m, j], X[m, j+1]], [Y[m, j], Y[m, j+1]], color='black', lw=2)
    
    # Final plot adjustments
    ax.set_aspect('equal')
    ax.axis('off')
    plt.tight_layout()
    
    if show_plot:
        plt.show()
    if save_svg:
        fig.savefig('scaled_miura_ori.svg', format='svg', dpi=300, bbox_inches='tight')
        print('SVG file saved as "scaled_miura_ori.svg"')
    plt.close(fig)
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import minimize
from ipywidgets import interactive, FloatSlider, IntSlider, Checkbox
from IPython.display import display

def compute_coordinates(m, n, a, alpha, delta, y_coords):
    """
    Compute the X and Y coordinates of the Miura-ori pattern.

    Parameters:
    - m (int): Number of rows.
    - n (int): Number of columns.
    - a (float): Initial horizontal side length.
    - alpha (float): Folding angle in radians.
    - delta (float): Scaling factor per row (0 <= delta <= 1).
    - y_coords (array-like): Optimized y-coordinates for each row.

    Returns:
    - X (2D numpy array): X coordinates of the grid.
    - Y (2D numpy array): Y coordinates of the grid.
    """
    i_values = np.arange(m+1)
    j_values = np.arange(n+1)

    # Compute scaling factors for each row
    a_values = a * (1 - delta * (i_values / m))
    a_values = np.maximum(a_values, 1e-3)  # Prevent zero or negative lengths

    # Compute b_i based on alpha to preserve angles
    b_values = a_values * np.tan(alpha)

    # Initialize coordinate grids
    X = np.zeros((m+1, n+1))
    Y = np.zeros((m+1, n+1))

    for i in i_values:
        for j in j_values:
            base_x = j * a_values[i]
            # Shift every other row by b_i * cos(alpha) to create the Miura-ori pattern
            if i % 2 == 1:
                base_x += b_values[i] * np.cos(alpha)
            X[i, j] = base_x
            Y[i, j] = y_coords[i]

    # Center the pattern horizontally
    top_center = (X[0, :].min() + X[0, :].max()) / 2
    bottom_center = (X[m, :].min() + X[m, :].max()) / 2
    midpoint = (top_center + bottom_center) / 2
    X -= midpoint

    return X, Y

def angle_between_vectors(u, v):
    """
    Compute the angle between two vectors in degrees.

    Parameters:
    - u (array-like): First vector.
    - v (array-like): Second vector.

    Returns:
    - angle (float): Angle between u and v in degrees.
    """
    dot_product = np.dot(u, v)
    norm_u = np.linalg.norm(u)
    norm_v = np.linalg.norm(v)
    # Prevent division by zero
    if norm_u == 0 or norm_v == 0:
        return 0.0
    # Compute cosine of the angle and clip to avoid numerical errors
    cos_angle = np.clip(dot_product / (norm_u * norm_v), -1.0, 1.0)
    angle = np.degrees(np.arccos(cos_angle))
    return angle

def compute_angle_deviations(m, n, X, Y):
    """
    Compute the angle deviations from flat foldability at each internal vertex.

    Parameters:
    - m (int): Number of rows.
    - n (int): Number of columns.
    - X (2D numpy array): X coordinates of the grid.
    - Y (2D numpy array): Y coordinates of the grid.

    Returns:
    - deviations (1D numpy array): Array of angle deviations.
    """
    deviations = []
    # Iterate over internal vertices
    for i in range(1, m):
        for j in range(1, n):
            # Get coordinates of the four neighbors
            p = np.array([X[i, j], Y[i, j]])
            p_up = np.array([X[i-1, j], Y[i-1, j]])
            p_right = np.array([X[i, j+1], Y[i, j+1]])
            p_down = np.array([X[i+1, j], Y[i+1, j]])
            p_left = np.array([X[i, j-1], Y[i, j-1]])

            # Vectors from the center to neighbors
            v_up = p_up - p
            v_right = p_right - p
            v_down = p_down - p
            v_left = p_left - p

            # Compute angles between consecutive vectors in clockwise order
            theta_up_right = angle_between_vectors(v_up, v_right)
            theta_right_down = angle_between_vectors(v_right, v_down)
            theta_down_left = angle_between_vectors(v_down, v_left)
            theta_left_up = angle_between_vectors(v_left, v_up)

            # Kawasaki's condition: sum of alternating angles = 180 degrees
            dev1 = (theta_up_right + theta_down_left) - 180.0
            dev2 = (theta_right_down + theta_left_up) - 180.0

            deviations.extend([dev1, dev2])

    return np.array(deviations)

def cost_function(y_coords, m, n, a, alpha, delta):
    """
    Cost function to minimize deviations from flat foldability.

    Parameters:
    - y_coords (array-like): Current y-coordinates for each row.
    - m (int): Number of rows.
    - n (int): Number of columns.
    - a (float): Initial horizontal side length.
    - alpha (float): Folding angle in radians.
    - delta (float): Scaling factor per row.

    Returns:
    - cost (float): Sum of squared angle deviations.
    """
    X, Y = compute_coordinates(m, n, a, alpha, delta, y_coords)
    deviations = compute_angle_deviations(m, n, X, Y)
    # Sum of squared deviations
    cost = np.sum(deviations**2)
    return cost

def optimize_y_coordinates(m, n, a, alpha, delta, initial_guess=None):
    """
    Optimize y-coordinates to achieve flat foldability with debug information.
    
    Parameters:
    - m (int): Number of rows.
    - n (int): Number of columns.
    - a (float): Initial horizontal side length.
    - alpha (float): Folding angle in radians.
    - delta (float): Scaling factor per row.
    - initial_guess (array-like, optional): Initial guess for y-coordinates.
    
    Returns:
    - result.x (1D numpy array): Optimized y-coordinates.
    - result (OptimizeResult): Full optimization result.
    """
    if initial_guess is None:
        # Initialize y-coordinates based on standard Miura-ori
        i_values = np.arange(m+1)
        a_values = a * (1 - delta * (i_values / m))
        a_values = np.maximum(a_values, 1e-3)
        b_values = a_values * np.tan(alpha)
        y_offsets = [0.0]
        for i in range(1, m+1):
            y_new = y_offsets[i-1] + b_values[i-1] * np.sin(alpha)
            y_offsets.append(y_new)
        initial_guess = np.array(y_offsets)
    
    # Define constraints: y[i] > y[i-1] for all i
    constraints = []
    for i in range(1, m+1):
        constraints.append({
            'type': 'ineq',
            'fun': lambda y, i=i: y[i] - y[i-1] - 1e-3  # Ensure y[i] > y[i-1] + small epsilon
        })
    
    # Define bounds: y-coordinates should be positive and increasing
    bounds = [(None, None)] * (m+1)
    
    # Define a callback function to monitor optimization
    def callback(xk):
        current_cost = cost_function(xk, m, n, a, alpha, delta)
#         print(f"Current cost: {current_cost:.6f}")
    
    # Optimization using SLSQP with callback
    result = minimize(
        cost_function,
        initial_guess,
        args=(m, n, a, alpha, delta),
        method='SLSQP',
        constraints=constraints,
        bounds=bounds,
        options={'maxiter': 10000, 'ftol': 1e-9, 'disp': False},
        callback=callback
    )
    
    return result.x, result

def plot_miura_ori(m, n, a, alpha, delta, y_coords, show_plot=True, save_svg=False):
    """
    Plot the Miura-ori crease pattern.

    Parameters:
    - m (int): Number of rows.
    - n (int): Number of columns.
    - a (float): Initial horizontal side length.
    - alpha (float): Folding angle in radians.
    - delta (float): Scaling factor per row.
    - y_coords (array-like): Optimized y-coordinates for each row.
    - show_plot (bool): Whether to display the plot.
    - save_svg (bool): Whether to save the plot as an SVG file.
    """
    X, Y = compute_coordinates(m, n, a, alpha, delta, y_coords)
    i_values = np.arange(m+1)
    j_values = np.arange(n+1)
    
    fig, ax = plt.subplots(figsize=(8, 8))
    
    # Plot longitudinal folds (vertical lines)
    for j in j_values:
        for i in range(m):
            fold_color = 'blue' if j % 2 == 0 else 'red'
            ax.plot([X[i, j], X[i+1, j]], [Y[i, j], Y[i+1, j]], color=fold_color, lw=2)
    
    # Plot transversal folds (horizontal lines)
    for i in i_values:
        for j in range(n):
            fold_color = 'red' if (i + j) % 2 == 0 else 'blue'
            ax.plot([X[i, j], X[i, j+1]], [Y[i, j], Y[i, j+1]], color=fold_color, lw=2)
    
    # Plot outer boundaries
    for i in range(m):
        # Left boundary
        ax.plot([X[i, 0], X[i+1, 0]], [Y[i, 0], Y[i+1, 0]], color='black', lw=2)
        # Right boundary
        ax.plot([X[i, n], X[i+1, n]], [Y[i, n], Y[i+1, n]], color='black', lw=2)
    for j in range(n):
        # Top boundary
        ax.plot([X[0, j], X[0, j+1]], [Y[0, j], Y[0, j+1]], color='black', lw=2)
        # Bottom boundary
        ax.plot([X[m, j], X[m, j+1]], [Y[m, j], Y[m, j+1]], color='black', lw=2)
    
    # Final plot adjustments
    ax.set_aspect('equal')
    ax.axis('off')
    plt.tight_layout()
    
    if show_plot:
        plt.show()
    if save_svg:
        fig.savefig('scaled_miura_ori.svg', format='svg', dpi=300, bbox_inches='tight')
        print('SVG file saved as "scaled_miura_ori.svg"')
    plt.close(fig)

def interactive_miura_ori():
    """
    Create an interactive widget to adjust Miura-ori parameters and visualize the pattern.
    """
    def update_plot(m=5, n=4, a=1.0, alpha=1.05, delta=0.25, show_plot=True, save_svg=False):
        # Optimize y-coordinates with improved optimization function
        y_opt, result = optimize_y_coordinates(m, n, a, alpha, delta)
        
        # Display optimization status
        print(f"Optimization successful: {result.success}")
        print(f"Message: {result.message}")
        print(f"Final cost (sum of squared deviations): {result.fun:.6f}")
        
        if result.success:
            # Plot the crease pattern
            plot_miura_ori(m, n, a, alpha, delta, y_opt, show_plot, save_svg)
        else:
            print("Optimization did not converge. Please try adjusting the parameters.")
    
    interactive_plot = interactive(
        update_plot,
        m=IntSlider(min=1, max=20, step=1, value=5, description='Rows (m)'),
        n=IntSlider(min=1, max=20, step=1, value=4, description='Cols (n)'),
        a=FloatSlider(min=0.5, max=5.0, step=0.1, value=1.0, description='a'),
        alpha=FloatSlider(min=0.1, max=np.pi-0.1, step=0.05, value=1.05, description='alpha (rad)'),
        delta=FloatSlider(min=0.0, max=0.5, step=0.05, value=0.25, description='delta'),
        show_plot=Checkbox(value=True, description='Show Plot'),
        save_svg=Checkbox(value=False, description='Save SVG')
    )
    
    display(interactive_plot)

interactive_miura_ori()
```
