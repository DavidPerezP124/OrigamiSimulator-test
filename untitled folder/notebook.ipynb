```python
import numpy as np
import matplotlib.pyplot as plt
import ipywidgets as widgets
from IPython.display import display

# Generate a single curve starting from the outer radius
def generate_base_curve(num_points, outer_radius, inner_radius, wave_frequency, wave_amplitude):
    """Generate a single curve starting at the outer radius and tapering inward."""
    t = np.linspace(0, 1, num_points)
    radius = outer_radius + (inner_radius - outer_radius) * t  # Linearly taper from outer to inner radius
    angle = 2 * np.pi * wave_frequency * t  # Wave along the curve
    x = radius * np.cos(angle) + wave_amplitude * np.sin(angle)
    y = radius * np.sin(angle)
    return x, y

# Generate rotated and translated copies of the curve
def generate_rotated_copies(x, y, num_copies, radian_range):
    """Rotate and distribute copies of the curve symmetrically within the radian range."""
    curves = []
    angle_step = radian_range / (num_copies - 1) if num_copies > 1 else 0

    for i in range(num_copies):
        angle = -radian_range / 2 + i * angle_step  # Symmetric spreadc
        rotation_matrix = np.array([
            [np.cos(angle), -np.sin(angle)],
            [np.sin(angle), np.cos(angle)],
        ])
        rotated = np.dot(rotation_matrix, np.vstack((x, y)))
        curves.append(rotated)
    
    return curves

# Plot the curves
def plot_curves(curves):
    """Plot the stack of curves with alternating colors."""
    fig, ax = plt.subplots(figsize=(8, 8))
    colors = ['black'] + ['blue' if i % 2 else 'red' for i in range(1, len(curves) - 1)] + ['black']
    
    for i, curve in enumerate(curves):
        ax.plot(curve[0], curve[1], color=colors[i], lw=2)

    # Outer and inner boundary lines
    ax.plot(curves[0][0], curves[0][1], color='black', lw=3)  # Outer boundary
    ax.plot(curves[-1][0], curves[-1][1], color='black', lw=3)  # Inner boundary

    ax.set_aspect('equal')
    ax.axis('off')
    plt.show()

# Save the design as SVG
def save_curves_as_svg(curves, filename="alternate_greater_than_curve.svg"):
    """Save the design as an SVG."""
    fig, ax = plt.subplots(figsize=(8, 8))
    colors = ['black'] + ['blue' if i % 2 else 'red' for i in range(1, len(curves) - 1)] + ['black']
    
    for i, curve in enumerate(curves):
        ax.plot(curve[0], curve[1], color=colors[i], lw=2)

    # Outer and inner boundary lines
    ax.plot(curves[0][0], curves[0][1], color='black', lw=3)  # Outer boundary
    ax.plot(curves[-1][0], curves[-1][1], color='black', lw=3)  # Inner boundary

    ax.set_aspect('equal')
    ax.axis('off')
    fig.savefig(filename, format="svg", transparent=True)
    plt.close(fig)
    print(f"SVG saved as {filename}")

# Interactive function
def interactive_curve_design(
    num_points, outer_radius, inner_radius, wave_frequency, wave_amplitude, num_copies, radian_range, save
):
    # Generate the base curve
    x, y = generate_base_curve(num_points, outer_radius, inner_radius, wave_frequency, wave_amplitude)
    # Generate rotated copies
    curves = generate_rotated_copies(x, y, num_copies, radian_range)
    
    if save:
        # Save the design
        save_curves_as_svg(curves)
    else:
        # Plot the design
        plot_curves(curves)

# Interactive widgets
interactive_widgets = widgets.interactive(
    interactive_curve_design,
    num_points=widgets.IntSlider(min=100, max=2000, step=100, value=500, description="Points"),
    outer_radius=widgets.FloatSlider(min=5, max=20, step=0.5, value=15, description="Outer Radius"),
    inner_radius=widgets.FloatSlider(min=1, max=10, step=0.5, value=5, description="Inner Radius"),
    wave_frequency=widgets.FloatSlider(min=1, max=10, step=0.5, value=3, description="Wave Frequency"),
    wave_amplitude=widgets.FloatSlider(min=0, max=5, step=0.1, value=1, description="Wave Amplitude"),
    num_copies=widgets.IntSlider(min=2, max=20, step=1, value=5, description="Copies"),
    radian_range=widgets.FloatSlider(min=np.pi/8, max=np.pi, step=np.pi/16, value=np.pi/4, description="Radian"),
    save=widgets.Checkbox(value=False, description="Save SVG"),
)

# Display interactive widgets
display(interactive_widgets)
```

```python
import numpy as np
import matplotlib.pyplot as plt
import ipywidgets as widgets
from IPython.display import display

# Generate a smooth curve with decreasing width and amplitude
def generate_wave_with_smooth_decreasing_width(num_peaks, initial_height, initial_width, amplitude_decay, width_decrement):
    x = np.linspace(0, num_peaks * initial_width, 1000)
    y = np.zeros_like(x)
    for i in range(len(x)):
        progress = x[i] / (num_peaks * initial_width)
        current_amplitude = initial_height * (amplitude_decay ** (progress * num_peaks))
        current_width = max(initial_width - width_decrement * progress * num_peaks, 1)
        y[i] = current_amplitude * np.sin(2 * np.pi * x[i] / current_width)
    valid_indices = y != 0
    return x[valid_indices], y[valid_indices]

# Generate the base curve starting at the outer edge
def generate_curve(num_peaks, initial_height, initial_width, amplitude_decay, width_decrement):
    x, y = generate_wave_with_smooth_decreasing_width(
        num_peaks, initial_height, initial_width, amplitude_decay, width_decrement
    )
    # Reverse the curve so it starts on the outside and ends on the inside
    x, y = x[::-1], y[::-1]
    return x, y

# Stack curves to form a "greater-than" shape ( >c )
def stack_curves_to_form_greater_than(x, y, copies, radian, radius):
    curves = []
    angle_step = radian / (copies - 1)  # Angle increment between curves

    for i in range(copies):
        angle = -radian / 2 + i * angle_step  # Spread curves symmetrically within the radian range
        rotation_matrix = np.array([
            [np.cos(angle), -np.sin(angle)],
            [np.sin(angle), np.cos(angle)],
        ])
        # Apply rotation
        rotated = np.dot(rotation_matrix, np.vstack((x, y)))
        # Translate curves outward, starting at the outer edge, with quadratic tapering for smoother alignment
        rotated[0] += radius * (1 - (i / (copies - 1))**2)  # Quadratic tapering
        rotated[1] += radius * np.sin(angle)
        curves.append(rotated)

    return curves

# Plot the curves
def plot_greater_than_slice(curves, copies):
    fig, ax = plt.subplots(figsize=(8, 8))
    colors = ['black'] + ['blue' if i % 2 else 'red' for i in range(1, copies - 1)] + ['black']
    
    for i, curve in enumerate(curves):
        ax.plot(curve[0], curve[1], color=colors[i], lw=2)  # Increased line width for better visibility

    # Outline the boundaries (first and last curve)
    ax.plot(curves[0][0], curves[0][1], color='black', lw=3)  # Outer boundary
    ax.plot(curves[-1][0], curves[-1][1], color='black', lw=3)  # Inner boundary

    ax.set_aspect('equal')
    ax.axis('off')
    plt.show()

# Save the design as SVG
def save_greater_than_svg(curves, copies):
    colors = ['black'] + ['blue' if i % 2 else 'red' for i in range(1, copies - 1)] + ['black']
    fig, ax = plt.subplots(figsize=(8, 8))
    
    for i, curve in enumerate(curves):
        ax.plot(curve[0], curve[1], color=colors[i], lw=2)  # Increased line width for better visibility

    # Outline the boundaries
    ax.plot(curves[0][0], curves[0][1], color='black', lw=3)  # Outer boundary
    ax.plot(curves[-1][0], curves[-1][1], color='black', lw=3)  # Inner boundary

    ax.set_aspect('equal')
    ax.axis('off')
    
    svg_filename = "greater_than_curve_refined.svg"
    fig.savefig(svg_filename, format="svg", transparent=True)
    plt.close(fig)
    print(f"SVG saved as {svg_filename}")

# Main interactive function
def interactive_greater_than_slice(
    num_peaks, initial_height, initial_width, amplitude_decay, width_decrement, copies, radian, radius, save
):
    # Generate the base curve (starts on the outer edge by default)
    x, y = generate_curve(num_peaks, initial_height, initial_width, amplitude_decay, width_decrement)
    # Stack curves into a '>' shape
    curves = stack_curves_to_form_greater_than(x, y, copies, radian, radius)
    
    if save:
        # Save the design as an SVG
        save_greater_than_svg(curves, copies)
    else:
        # Plot the design
        plot_greater_than_slice(curves, copies)

# Interactive widgets
interact_greater_than_slice = widgets.interactive(
    interactive_greater_than_slice,
    num_peaks=widgets.IntSlider(min=1, max=10, step=1, value=5, description="Peaks"),
    initial_height=widgets.FloatSlider(min=0.5, max=5, step=0.1, value=2, description="Height"),
    initial_width=widgets.FloatSlider(min=1, max=10, step=0.5, value=5, description="Width"),
    amplitude_decay=widgets.FloatSlider(min=0.5, max=1, step=0.05, value=0.8, description="Decay"),
    width_decrement=widgets.FloatSlider(min=0.1, max=2, step=0.1, value=0.5, description="Width Dec."),
    copies=widgets.IntSlider(min=2, max=20, step=1, value=5, description="Copies"),
    radian=widgets.FloatSlider(min=np.pi/8, max=np.pi, step=np.pi/16, value=np.pi/4, description="Radian"),
    radius=widgets.FloatSlider(min=1, max=20, step=1, value=10, description="Radius"),
    save=widgets.Checkbox(value=False, description="Save SVG"),
)

# Display widgets
display(interact_greater_than_slice)
```

```python
import numpy as np
import matplotlib.pyplot as plt
import ipywidgets as widgets
from IPython.display import display

# Generate a smooth curve with decreasing width and amplitude
def generate_wave_with_smooth_decreasing_width(num_peaks, initial_height, initial_width, amplitude_decay, width_decrement):
    x = np.linspace(0, num_peaks * initial_width, 1000)
    y = np.zeros_like(x)
    for i in range(len(x)):
        progress = x[i] / (num_peaks * initial_width)
        current_amplitude = initial_height * (amplitude_decay ** (progress * num_peaks))
        current_width = max(initial_width - width_decrement * progress * num_peaks, 1)
        y[i] = current_amplitude * np.sin(2 * np.pi * x[i] / current_width)
    valid_indices = y != 0
    return x[valid_indices], y[valid_indices]

# Generate the base curve
def generate_curve(num_peaks, initial_height, initial_width, amplitude_decay, width_decrement):
    x, y = generate_wave_with_smooth_decreasing_width(
        num_peaks, initial_height, initial_width, amplitude_decay, width_decrement
    )
    # Reverse the curve so it starts on the opposite end
    x, y = x[::-1], y[::-1]
    return x, y

# Stack curves to form a "greater-than" shape ( >c )
def stack_curves_to_form_greater_than(x, y, copies, radian, radius):
    curves = []
    angle_step = radian / (copies - 1)  # Angle increment between curves

    for i in range(copies):
        angle = radian / 2 - i * angle_step  # Reverse the angle sequence
        rotation_matrix = np.array([
            [np.cos(angle), -np.sin(angle)],
            [np.sin(angle), np.cos(angle)],
        ])
        # Apply rotation
        rotated = np.dot(rotation_matrix, np.vstack((x, y)))
        # Adjust translation to fan out from the opposite end
        rotated[0] += radius * (i / (copies - 1))**2  # Quadratic tapering
        rotated[1] += radius * np.sin(angle)
        curves.append(rotated)

    return curves

# Plot the curves
def plot_greater_than_slice(curves, copies):
    fig, ax = plt.subplots(figsize=(8, 8))
    colors = ['black'] + ['blue' if i % 2 else 'red' for i in range(1, copies - 1)] + ['black']
    
    for i, curve in enumerate(curves):
        ax.plot(curve[0], curve[1], color=colors[i], lw=2)  # Increased line width for better visibility

    # Outline the boundaries (first and last curve)
    ax.plot(curves[0][0], curves[0][1], color='black', lw=3)  # Outer boundary
    ax.plot(curves[-1][0], curves[-1][1], color='black', lw=3)  # Inner boundary

    ax.set_aspect('equal')
    ax.axis('off')
    plt.show()

# Save the design as SVG
def save_greater_than_svg(curves, copies):
    colors = ['black'] + ['blue' if i % 2 else 'red' for i in range(1, copies - 1)] + ['black']
    fig, ax = plt.subplots(figsize=(8, 8))
    
    for i, curve in enumerate(curves):
        ax.plot(curve[0], curve[1], color=colors[i], lw=2)  # Increased line width for better visibility

    # Outline the boundaries
    ax.plot(curves[0][0], curves[0][1], color='black', lw=3)  # Outer boundary
    ax.plot(curves[-1][0], curves[-1][1], color='black', lw=3)  # Inner boundary

    ax.set_aspect('equal')
    ax.axis('off')
    
    svg_filename = "greater_than_curve_refined.svg"
    fig.savefig(svg_filename, format="svg", transparent=True)
    plt.close(fig)
    print(f"SVG saved as {svg_filename}")

# Main interactive function
def interactive_greater_than_slice(
    num_peaks, initial_height, initial_width, amplitude_decay, width_decrement, copies, radian, radius, save
):
    # Generate the base curve (starts on the opposite end)
    x, y = generate_curve(num_peaks, initial_height, initial_width, amplitude_decay, width_decrement)
    # Stack curves into a '>' shape
    curves = stack_curves_to_form_greater_than(x, y, copies, radian, radius)
    
    if save:
        # Save the design as an SVG
        save_greater_than_svg(curves, copies)
    else:
        # Plot the design
        plot_greater_than_slice(curves, copies)

# Interactive widgets
interact_greater_than_slice = widgets.interactive(
    interactive_greater_than_slice,
    num_peaks=widgets.IntSlider(min=1, max=10, step=1, value=5, description="Peaks"),
    initial_height=widgets.FloatSlider(min=0.5, max=5, step=0.1, value=2, description="Height"),
    initial_width=widgets.FloatSlider(min=1, max=10, step=0.5, value=5, description="Width"),
    amplitude_decay=widgets.FloatSlider(min=0.5, max=1, step=0.05, value=0.8, description="Decay"),
    width_decrement=widgets.FloatSlider(min=0.1, max=2, step=0.1, value=0.5, description="Width Dec."),
    copies=widgets.IntSlider(min=2, max=20, step=1, value=5, description="Copies"),
    radian=widgets.FloatSlider(min=np.pi/8, max=np.pi, step=np.pi/16, value=np.pi/4, description="Radian"),
    radius=widgets.FloatSlider(min=1, max=20, step=1, value=10, description="Radius"),
    save=widgets.Checkbox(value=False, description="Save SVG"),
)

# Display widgets
display(interact_greater_than_slice)
```

```python
import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interactive, IntSlider, FloatSlider, Checkbox
from IPython.display import display

# Define the base curve function
def generate_base_curve(num_points=1000, length=10, amplitude=1):
    x = np.linspace(0, length, num_points)
    y = amplitude * np.sin(2 * np.pi * x / length)
    return x, y

# Function to generate and plot the fanned-out curves
def plot_fanned_curves(
    num_curves=5,           # Number of curves to generate
    total_angle=np.pi/2,    # Total angle range for fanning out (in radians)
    length=10,              # Length of the base curve
    amplitude=1,            # Amplitude of the base curve
    amplitude_decrease=False, # Decrease amplitude for subsequent curves
    show_plot=True,         # Whether to display the plot
    save_svg=False          # Whether to save the plot as an SVG file
):
    # Generate the base curve
    x_base, y_base = generate_base_curve(length=length, amplitude=amplitude)

    # Angles at which curves will be rotated (evenly spaced)
    angles = np.linspace(-total_angle / 2, total_angle / 2, num_curves)

    # Create figure and axis
    fig, ax = plt.subplots(figsize=(8, 8))

    # Draw curves
    for i, angle in enumerate(angles):
        # Optional amplitude decrease
        if amplitude_decrease:
            amp_factor = 1 - i / (num_curves - 1)
            x = x_base * amp_factor
            y = y_base * amp_factor
        else:
            x, y = x_base, y_base

        # Rotate curve
        rot_matrix = np.array([
            [np.cos(angle), -np.sin(angle)],
            [np.sin(angle),  np.cos(angle)]
        ])
        curve = np.dot(rot_matrix, np.vstack((x, y)))

        # Translate curve so that all curves start from the origin
        delta_x = +curve[0, 0]
        delta_y = -curve[1, 0]
        curve[0] += delta_x
        curve[1] += delta_y

        # Plot curve
        ax.plot(curve[0], curve[1], lw=2)

    # Set equal aspect ratio and remove axes
    ax.set_aspect('equal')
    ax.axis('off')

    # Show or save the plot
    if show_plot:
        plt.show()
    if save_svg:
        fig.savefig('fanned_curves.svg', format='svg', bbox_inches='tight')
        print('SVG file saved as "fanned_curves.svg"')
    plt.close(fig)

# Interactive widgets for real-time parameter adjustment
def interactive_fanned_curves():
    interactive_plot = interactive(
        plot_fanned_curves,
        num_curves=IntSlider(min=2, max=20, step=1, value=5, description='Number of Curves'),
        total_angle=FloatSlider(min=0, max=2*np.pi, step=np.pi/16, value=np.pi/2, description='Total Angle (rad)'),
        length=FloatSlider(min=1, max=20, step=1, value=10, description='Curve Length'),
        amplitude=FloatSlider(min=0.1, max=5, step=0.1, value=1, description='Amplitude'),
        amplitude_decrease=Checkbox(value=False, description='Decrease Amplitude'),
        save_svg=Checkbox(value=False, description='Save as SVG')
    )
    display(interactive_plot)

# Run the interactive function
interactive_fanned_curves()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interactive, IntSlider, FloatSlider, Checkbox
from IPython.display import display

# Function to generate the base curve
def generate_base_curve(num_points=1000, length=10, amplitude=1, frequency=1):
    x = np.linspace(0, length, num_points)
    y = amplitude * np.sin(2 * np.pi * frequency * x / length)
    return x, y

# Function to generate and plot the fanned-out curves with specified modifications
def plot_modified_fanned_curves(
    num_curves=5,                   # Number of curves
    angle_separation=np.pi/18,      # Separation between curves in radians
    length=10,                      # Base curve length
    base_amplitude=1,               # Base amplitude
    amplitude_growth=0,             # Amplitude growth percentage per curve
    base_frequency=1,               # Base frequency
    frequency_decrease=0,           # Frequency decrease percentage per curve
    show_plot=True,                 # Display the plot
    save_svg=False                  # Save the plot as SVG
):
    # Angles at which curves will be rotated
    total_angle = angle_separation * (num_curves - 1)
    angles = np.linspace(-total_angle / 2, total_angle / 2, num_curves)

    # Create figure and axis
    fig, ax = plt.subplots(figsize=(8, 8))

    # Colors for the curves
    colors = ['black']  # First curve is black
    for i in range(1, num_curves - 1):
        colors.append('blue' if i % 2 == 1 else 'red')
    if num_curves > 1:
        colors.append('black')  # Last curve is black

    curves = []

    # Draw curves
    for i, angle in enumerate(angles):
        # Calculate amplitude and frequency for this curve
        amplitude = base_amplitude * ((1 + amplitude_growth / 100) ** i)
        frequency = base_frequency * ((1 - frequency_decrease / 100) ** i)

        # Generate the base curve with current amplitude and frequency
        x_base, y_base = generate_base_curve(length=length, amplitude=amplitude, frequency=frequency)

        # Rotate curve
        rot_matrix = np.array([
            [np.cos(angle), -np.sin(angle)],
            [np.sin(angle),  np.cos(angle)]
        ])
        curve = np.dot(rot_matrix, np.vstack((x_base, y_base)))

        # Translate curve so that all curves start from the origin
        delta_x = -curve[0, 0]
        delta_y = -curve[1, 0]
        curve[0] += delta_x
        curve[1] += delta_y

        curves.append(curve)

        # Plot curve
        ax.plot(curve[0], curve[1], color=colors[i], lw=2)

    # Outline with arcs at the start and end
    # Calculate start and end points
    start_points = np.array([curve[:, 0] for curve in curves])
    end_points = np.array([curve[:, -1] for curve in curves])

    # Draw arcs at the inner and outer ends
    # Start arc (inner arc)
    start_arc_center = [0, 0]  # All curves start at the origin
    start_arc_radius = 0.0     # No need to draw an arc at the start since all curves start at the same point

    # End arc (outer arc)
    # Calculate the arc that connects the end points of the first and last curves
    end_arc_center = [0, 0]
    end_arc_radius = np.linalg.norm(end_points[0])  # Radius is the distance from origin to the end point of the first curve
    theta_start = np.arctan2(end_points[0, 1], end_points[0, 0])
    theta_end = np.arctan2(end_points[-1, 1], end_points[-1, 0])

    # Ensure theta_start < theta_end
    if theta_start > theta_end:
        theta_start, theta_end = theta_end, theta_start

    theta = np.linspace(theta_start, theta_end, 100)
    arc_x = end_arc_radius * np.cos(theta)
    arc_y = end_arc_radius * np.sin(theta)
    ax.plot(arc_x, arc_y, color='black', lw=2)

    # Set equal aspect ratio and remove axes
    ax.set_aspect('equal')
    ax.axis('off')

    # Show or save the plot
    if show_plot:
        plt.show()
    if save_svg:
        fig.savefig('modified_fanned_curves.svg', format='svg', bbox_inches='tight')
        print('SVG file saved as "modified_fanned_curves.svg"')
    plt.close(fig)

# Interactive widgets
def interactive_modified_fanned_curves():
    interactive_plot = interactive(
        plot_modified_fanned_curves,
        num_curves=IntSlider(min=2, max=20, step=1, value=5, description='Number of Curves'),
        angle_separation=FloatSlider(min=0, max=np.pi, step=np.pi/180, value=np.pi/18, description='Angle Separation (rad)'),
        length=FloatSlider(min=1, max=20, step=1, value=10, description='Curve Length'),
        base_amplitude=FloatSlider(min=0.1, max=5, step=0.1, value=1, description='Base Amplitude'),
        amplitude_growth=FloatSlider(min=-100, max=100, step=1, value=0, description='Amplitude Growth (%)'),
        base_frequency=FloatSlider(min=0.1, max=5, step=0.1, value=1, description='Base Frequency'),
        frequency_decrease=FloatSlider(min=0, max=100, step=1, value=0, description='Frequency Decrease (%)'),
        save_svg=Checkbox(value=False, description='Save as SVG')
    )
    display(interactive_plot)

# Run the interactive function
interactive_modified_fanned_curves()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interactive, IntSlider, FloatSlider, Checkbox
from IPython.display import display

# Function to generate and plot the fanned-out curves with specified modifications
def plot_modified_fanned_curves(
    num_curves=5,                   # Number of curves
    inner_angle=np.pi/6,            # Total angle at the inner end (in radians)
    outer_angle=np.pi/2,            # Total angle at the outer end (in radians)
    length=10,                      # Base curve length
    base_amplitude=1,               # Base amplitude
    amplitude_growth=0,             # Amplitude growth percentage per curve
    base_frequency=1,               # Base frequency
    frequency_decrease=0,           # Frequency decrease percentage per curve
    show_plot=True,                 # Display the plot
    save_svg=False                  # Save the plot as SVG
):
    # Create figure and axis
    fig, ax = plt.subplots(figsize=(8, 8))

    curves = []

    # Angles at which curves will start and end
    start_angles = np.linspace(-inner_angle / 2, inner_angle / 2, num_curves)
    end_angles = np.linspace(-outer_angle / 2, outer_angle / 2, num_curves)

    # Colors for the curves
    colors = ['black']  # First curve is black
    for i in range(1, num_curves - 1):
        colors.append('blue' if i % 2 == 1 else 'red')
    if num_curves > 1:
        colors.append('black')  # Last curve is black

    # Generate the curves
    max_r = 0  # To keep track of the maximum radius for drawing the outer arc
    for i in range(num_curves):
        # Calculate amplitude and frequency for this curve
        amplitude = base_amplitude * ((1 + amplitude_growth / 100) ** i)
        frequency = base_frequency * ((1 - frequency_decrease / 100) ** i)

        # Generate parameter t
        t = np.linspace(0, 1, 1000)

        # Calculate angle for each point along the curve
        angles = start_angles[i] + t * (end_angles[i] - start_angles[i])

        # Calculate radius for each point along the curve
        r = amplitude * np.sin(2 * np.pi * frequency * t * length)

        # Update max_r for outer arc
        max_r = max(max_r, np.max(np.abs(r)))

        # Convert to Cartesian coordinates
        x = r * np.cos(angles)
        y = r * np.sin(angles)

        curves.append(np.vstack((x, y)))

        # Plot curve
        ax.plot(x, y, color=colors[i], lw=2)

    # Outline with arcs at the start and end
    # Start arc
    theta_start_arc = np.linspace(-inner_angle / 2, inner_angle / 2, 100)
    r_start_arc = 0.01  # Small radius to represent the inner arc (since r=0 can't be plotted)
    x_start_arc = r_start_arc * np.cos(theta_start_arc)
    y_start_arc = r_start_arc * np.sin(theta_start_arc)
    # Since the inner arc is essentially a point, we can skip plotting it

    # End arc (outer arc)
    theta_end_arc = np.linspace(-outer_angle / 2, outer_angle / 2, 100)
    r_end_arc = max_r  # Use the maximum radius of the curves
    x_end_arc = r_end_arc * np.cos(theta_end_arc)
    y_end_arc = r_end_arc * np.sin(theta_end_arc)
    ax.plot(x_end_arc, y_end_arc, color='black', lw=2)

    # Set equal aspect ratio and remove axes
    ax.set_aspect('equal')
    ax.axis('off')

    # Show or save the plot
    if show_plot:
        plt.show()
    if save_svg:
        fig.savefig('modified_fanned_curves.svg', format='svg', bbox_inches='tight')
        print('SVG file saved as "modified_fanned_curves.svg"')
    plt.close(fig)

# Interactive widgets
def interactive_modified_fanned_curves():
    interactive_plot = interactive(
        plot_modified_fanned_curves,
        num_curves=IntSlider(min=2, max=20, step=1, value=5, description='Number of Curves'),
        inner_angle=FloatSlider(min=0, max=2*np.pi, step=np.pi/180, value=np.pi/6, description='Inner Angle (rad)'),
        outer_angle=FloatSlider(min=0, max=2*np.pi, step=np.pi/180, value=np.pi/2, description='Outer Angle (rad)'),
        length=FloatSlider(min=1, max=20, step=1, value=10, description='Curve Length'),
        base_amplitude=FloatSlider(min=0.1, max=5, step=0.1, value=1, description='Base Amplitude'),
        amplitude_growth=FloatSlider(min=-100, max=100, step=1, value=0, description='Amplitude Growth (%)'),
        base_frequency=FloatSlider(min=0.1, max=5, step=0.1, value=1, description='Base Frequency'),
        frequency_decrease=FloatSlider(min=0, max=100, step=1, value=0, description='Frequency Decrease (%)'),
        save_svg=Checkbox(value=False, description='Save as SVG')
    )
    display(interactive_plot)

# Run the interactive function
interactive_modified_fanned_curves()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interactive, IntSlider, FloatSlider, Checkbox
from IPython.display import display

# Function to generate the base curve
def generate_base_curve(num_points=1000, length=10, amplitude=1, frequency=1):
    x = np.linspace(0, length, num_points)
    y = amplitude * np.sin(2 * np.pi * frequency * x / length)
    return x, y

# Function to generate and plot the fanned-out curves with specified modifications
def plot_modified_fanned_curves(
    num_curves=5,                   # Number of curves
    angle_separation=np.pi/18,      # Separation between curves in radians
    length=10,                      # Base curve length
    base_amplitude=1,               # Base amplitude
    amplitude_growth=0,             # Amplitude growth percentage per curve
    base_frequency=1,               # Base frequency
    frequency_decrease=0,           # Frequency decrease percentage per curve
    inner_radius=0,                 # Inner radius (radius of the circle from which curves start)
    show_plot=True,                 # Display the plot
    save_svg=False                  # Save the plot as SVG
):
    # Angles at which curves will be rotated
    total_angle = angle_separation * (num_curves - 1)
    angles = np.linspace(-total_angle / 2, total_angle / 2, num_curves)

    # Create figure and axis
    fig, ax = plt.subplots(figsize=(8, 8))

    # Colors for the curves
    colors = ['black']  # First curve is black
    for i in range(1, num_curves - 1):
        colors.append('blue' if i % 2 == 1 else 'red')
    if num_curves > 1:
        colors.append('black')  # Last curve is black

    curves = []

    # Draw curves
    for i, angle in enumerate(angles):
        # Calculate amplitude and frequency for this curve
        amplitude = base_amplitude * ((1 + amplitude_growth / 100) ** i)
        frequency = base_frequency * ((1 - frequency_decrease / 100) ** i)

        # Generate the base curve with current amplitude and frequency
        x_base, y_base = generate_base_curve(length=length, amplitude=amplitude, frequency=frequency)

        # Rotate curve
        rot_matrix = np.array([
            [np.cos(angle), -np.sin(angle)],
            [np.sin(angle),  np.cos(angle)]
        ])
        curve = np.dot(rot_matrix, np.vstack((x_base, y_base)))

        # Translate curve so that all curves start from the inner circle
        # Calculate the starting point on the inner circle
        start_angle = angle
        delta_x = -curve[0, 0] + inner_radius * np.cos(start_angle)
        delta_y = -curve[1, 0] + inner_radius * np.sin(start_angle)
        curve[0] += delta_x
        curve[1] += delta_y

        curves.append(curve)

        # Plot curve
        ax.plot(curve[0], curve[1], color=colors[i], lw=2)

    # Outline with arcs at the inner and outer ends
    # Calculate start and end points
    start_points = np.array([curve[:, 0] for curve in curves])
    end_points = np.array([curve[:, -1] for curve in curves])

    # Draw inner arc (start of the fan)
    theta_start = angles[0]
    theta_end = angles[-1]
    theta_inner = np.linspace(theta_start, theta_end, 100)
    arc_x_inner = inner_radius * np.cos(theta_inner)
    arc_y_inner = inner_radius * np.sin(theta_inner)
    ax.plot(arc_x_inner, arc_y_inner, color='black', lw=2)

    # Draw outer arc (end of the fan)
    # Calculate the maximum radius of the end points for the outer arc
    end_distances = np.linalg.norm(end_points, axis=1)
    end_arc_radius = np.max(end_distances)
    theta_outer = np.linspace(theta_start, theta_end, 100)
    arc_x_outer = end_arc_radius * np.cos(theta_outer)
    arc_y_outer = end_arc_radius * np.sin(theta_outer)
    ax.plot(arc_x_outer, arc_y_outer, color='black', lw=2)

    # Set equal aspect ratio and remove axes
    ax.set_aspect('equal')
    ax.axis('off')

    # Show or save the plot
    if show_plot:
        plt.show()
    if save_svg:
        fig.savefig('modified_fanned_curves.svg', format='svg', bbox_inches='tight')
        print('SVG file saved as "modified_fanned_curves.svg"')
    plt.close(fig)

# Interactive widgets
def interactive_modified_fanned_curves():
    interactive_plot = interactive(
        plot_modified_fanned_curves,
        num_curves=IntSlider(min=2, max=20, step=1, value=5, description='Number of Curves'),
        angle_separation=FloatSlider(min=0, max=np.pi, step=np.pi/180, value=np.pi/18, description='Angle Separation (rad)'),
        length=FloatSlider(min=1, max=20, step=1, value=10, description='Curve Length'),
        base_amplitude=FloatSlider(min=0.1, max=5, step=0.1, value=1, description='Base Amplitude'),
        amplitude_growth=FloatSlider(min=-100, max=100, step=1, value=0, description='Amplitude Growth (%)'),
        base_frequency=FloatSlider(min=0.1, max=5, step=0.1, value=1, description='Base Frequency'),
        frequency_decrease=FloatSlider(min=0, max=100, step=1, value=0, description='Frequency Decrease (%)'),
        inner_radius=FloatSlider(min=0, max=10, step=0.1, value=2, description='Inner Radius'),
        save_svg=Checkbox(value=False, description='Save as SVG')
    )
    display(interactive_plot)

# Run the interactive function
interactive_modified_fanned_curves()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interactive, IntSlider, FloatSlider, Checkbox
from IPython.display import display

# Function to generate the base curve
def generate_base_curve(num_points=1000, length=10, amplitude=1, frequency=1):
    x = np.linspace(0, length, num_points)
    y = amplitude * np.sin(2 * np.pi * frequency * x / length)
    return x, y

# Function to generate and plot the fanned-out curves with specified modifications
def plot_modified_fanned_curves(
    num_curves=5,                   # Number of curves
    angle_separation=np.pi/18,      # Separation between curves in radians
    length=10,                      # Base curve length
    amplitude=1,                    # Amplitude (same for all curves)
    frequency=1,                    # Frequency (same for all curves)
    inner_radius=0,                 # Inner radius (radius of the circle from which curves start)
    show_plot=True,                 # Display the plot
    save_svg=False                  # Save the plot as SVG
):
    # Angles at which curves will be rotated
    total_angle = angle_separation * (num_curves - 1)
    angles = np.linspace(-total_angle / 2, total_angle / 2, num_curves)

    # Create figure and axis
    fig, ax = plt.subplots(figsize=(8, 8))

    # Colors for the curves
    colors = ['black']  # First curve is black
    for i in range(1, num_curves - 1):
        colors.append('blue' if i % 2 == 1 else 'red')
    if num_curves > 1:
        colors.append('black')  # Last curve is black

    curves = []

    # Draw curves
    for i, angle in enumerate(angles):
        # Generate the base curve with specified amplitude and frequency
        x_base, y_base = generate_base_curve(length=length, amplitude=amplitude, frequency=frequency)

        # Rotate curve
        rot_matrix = np.array([
            [np.cos(angle), -np.sin(angle)],
            [np.sin(angle),  np.cos(angle)]
        ])
        curve = np.dot(rot_matrix, np.vstack((x_base, y_base)))

        # Translate curve so that all curves start from the inner circle
        # Calculate the starting point on the inner circle
        start_angle = angle
        delta_x = -curve[0, 0] + inner_radius * np.cos(start_angle)
        delta_y = -curve[1, 0] + inner_radius * np.sin(start_angle)
        curve[0] += delta_x
        curve[1] += delta_y

        curves.append(curve)

        # Plot curve
        ax.plot(curve[0], curve[1], color=colors[i], lw=2)

    # Outline with arcs at the inner and outer ends
    # Calculate start and end points
    start_points = np.array([curve[:, 0] for curve in curves])
    end_points = np.array([curve[:, -1] for curve in curves])

    # Draw inner arc (start of the fan)
    theta_start = angles[0]
    theta_end = angles[-1]
    theta_inner = np.linspace(theta_start, theta_end, 100)
    arc_x_inner = inner_radius * np.cos(theta_inner)
    arc_y_inner = inner_radius * np.sin(theta_inner)
    ax.plot(arc_x_inner, arc_y_inner, color='black', lw=2)

    # Draw outer arc (end of the fan)
    # Calculate the maximum radius of the end points for the outer arc
    end_distances = np.linalg.norm(end_points, axis=1)
    end_arc_radius = np.max(end_distances)
    theta_outer = np.linspace(theta_start, theta_end, 100)
    arc_x_outer = end_arc_radius * np.cos(theta_outer)
    arc_y_outer = end_arc_radius * np.sin(theta_outer)
    ax.plot(arc_x_outer, arc_y_outer, color='black', lw=2)

    # Set equal aspect ratio and remove axes
    ax.set_aspect('equal')
    ax.axis('off')

    # Show or save the plot
    if show_plot:
        plt.show()
    if save_svg:
        fig.savefig('modified_fanned_curves.svg', format='svg', bbox_inches='tight')
        print('SVG file saved as "modified_fanned_curves.svg"')
    plt.close(fig)

# Interactive widgets
def interactive_modified_fanned_curves():
    interactive_plot = interactive(
        plot_modified_fanned_curves,
        num_curves=IntSlider(min=2, max=20, step=1, value=5, description='Number of Curves'),
        angle_separation=FloatSlider(min=0, max=np.pi, step=np.pi/180, value=np.pi/18, description='Angle Separation (rad)'),
        length=FloatSlider(min=1, max=20, step=1, value=10, description='Curve Length'),
        amplitude=FloatSlider(min=0.1, max=5, step=0.1, value=1, description='Amplitude'),
        frequency=FloatSlider(min=0.1, max=5, step=0.1, value=1, description='Frequency'),
        inner_radius=FloatSlider(min=0, max=10, step=0.1, value=2, description='Inner Radius'),
        save_svg=Checkbox(value=False, description='Save as SVG')
    )
    display(interactive_plot)

# Run the interactive function
interactive_modified_fanned_curves()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interactive, IntSlider, FloatSlider, Checkbox
from IPython.display import display

# Function to generate the base curve with decreasing segment sizes
def generate_base_curve_with_decreasing_segments(
    num_points=1000, length=10, amplitude=1, frequency=1, s_max=1, s_min=0.5
):
    # Generate x positions with decreasing segment sizes
    x = [0]
    while x[-1] < length:
        s = s_max - (s_max - s_min) * (x[-1] / length)
        x_next = x[-1] + s
        x.append(x_next)
    x = np.array(x)
    x = x[x <= length]  # Ensure x does not exceed length
    y = amplitude * np.sin(2 * np.pi * frequency * x / length)
    return x, y

# Function to generate and plot the fanned-out curves with transversal folds
def plot_wavy_miura_ori(
    num_curves=5,                   # Number of curves
    angle_separation=np.pi/18,      # Separation between curves in radians
    length=10,                      # Base curve length
    amplitude=1,                    # Amplitude (same for all curves)
    frequency=1,                    # Frequency (same for all curves)
    inner_radius=0,                 # Inner radius (radius of the circle from which curves start)
    s_max=1,                        # Maximum segment size
    s_min=0.5,                      # Minimum segment size
    show_plot=True,                 # Display the plot
    save_svg=False                  # Save the plot as SVG
):
    # Angles at which curves will be rotated
    total_angle = angle_separation * (num_curves - 1)
    angles = np.linspace(-total_angle / 2, total_angle / 2, num_curves)

    # Create figure and axis
    fig, ax = plt.subplots(figsize=(8, 8))

    # Colors for the curves
    colors = ['black']  # First curve is black (outline)
    for i in range(1, num_curves - 1):
        colors.append('red' if i % 2 == 0 else 'blue')  # Alternate mountain and valley folds
    if num_curves > 1:
        colors.append('black')  # Last curve is black (outline)

    curves = []

    # Draw longitudinal folds (curves)
    for i, angle in enumerate(angles):
        # Generate the base curve with decreasing segments
        x_base, y_base = generate_base_curve_with_decreasing_segments(
            length=length, amplitude=amplitude, frequency=frequency, s_max=s_max, s_min=s_min
        )

        # Rotate curve
        rot_matrix = np.array([
            [np.cos(angle), -np.sin(angle)],
            [np.sin(angle),  np.cos(angle)]
        ])
        curve = np.dot(rot_matrix, np.vstack((x_base, y_base)))

        # Translate curve so that all curves start from the inner circle
        start_angle = angle
        delta_x = -curve[0, 0] + inner_radius * np.cos(start_angle)
        delta_y = -curve[1, 0] + inner_radius * np.sin(start_angle)
        curve[0] += delta_x
        curve[1] += delta_y

        curves.append(curve)

        # Plot curve
        ax.plot(curve[0], curve[1], color=colors[i], lw=2)

    # Draw transversal folds
    # For each segment position, draw lines connecting the curves
    num_segments = len(x_base)
    for j in range(num_segments):
        # Collect points from each curve at position x_base[j]
        points = np.array([curve[:, j] for curve in curves])
        # Determine fold type: alternate between mountain and valley
        fold_color = 'red' if j % 2 == 0 else 'blue'
        # Plot lines between adjacent curves
        for k in range(len(points) - 1):
            x_vals = [points[k][0], points[k+1][0]]
            y_vals = [points[k][1], points[k+1][1]]
            ax.plot(x_vals, y_vals, color=fold_color, lw=2)

    # Draw inner and outer arcs (outline)
    theta_start = angles[0]
    theta_end = angles[-1]
    theta_range = np.linspace(theta_start, theta_end, 100)

    # Inner arc
    arc_x_inner = inner_radius * np.cos(theta_range)
    arc_y_inner = inner_radius * np.sin(theta_range)
    ax.plot(arc_x_inner, arc_y_inner, color='black', lw=2)

    # Outer arc
    # Calculate maximum radius from the origin to the outermost point
    max_radius = max(np.hypot(curve[0], curve[1]).max() for curve in curves)
    arc_x_outer = max_radius * np.cos(theta_range)
    arc_y_outer = max_radius * np.sin(theta_range)
    ax.plot(arc_x_outer, arc_y_outer, color='black', lw=2)

    # Set equal aspect ratio and remove axes
    ax.set_aspect('equal')
    ax.axis('off')
    plt.tight_layout()

    # Show or save the plot
    if show_plot:
        plt.show()
    if save_svg:
        fig.savefig('wavy_miura_ori.svg', format='svg', dpi=300, bbox_inches='tight')
        print('SVG file saved as "wavy_miura_ori.svg"')
    plt.close(fig)

# Interactive widgets
def interactive_wavy_miura_ori():
    interactive_plot = interactive(
        plot_wavy_miura_ori,
        num_curves=IntSlider(min=2, max=20, step=1, value=5, description='Number of Curves'),
        angle_separation=FloatSlider(min=0.01, max=np.pi/2, step=0.01, value=np.pi/18, description='Angle Separation (rad)'),
        length=FloatSlider(min=5, max=50, step=1, value=10, description='Curve Length'),
        amplitude=FloatSlider(min=0.1, max=5, step=0.1, value=1, description='Amplitude'),
        frequency=FloatSlider(min=0.1, max=10, step=0.1, value=1, description='Frequency'),
        inner_radius=FloatSlider(min=0, max=10, step=0.1, value=2, description='Inner Radius'),
        s_max=FloatSlider(min=0.1, max=5, step=0.1, value=1, description='Max Segment Size'),
        s_min=FloatSlider(min=0.1, max=5, step=0.1, value=0.5, description='Min Segment Size'),
        save_svg=Checkbox(value=False, description='Save as SVG')
    )
    display(interactive_plot)

# Run the interactive function
interactive_wavy_miura_ori()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interactive, IntSlider, FloatSlider, Checkbox
from IPython.display import display

# Function to get the opposite color
def opposite_color(color):
    if color == 'red':
        return 'blue'
    elif color == 'blue':
        return 'red'
    else:
        return 'black'  # For 'black' or any other color

# Function to generate the base curve with decreasing segment sizes
def generate_base_curve_with_decreasing_segments(
    length=10, amplitude=1, frequency=1, s_max=1, s_min=0.5
):
    # Generate x positions with decreasing segment sizes
    x = [0]
    while x[-1] < length:
        s = s_max - (s_max - s_min) * (x[-1] / length)
        x_next = x[-1] + s
        x.append(x_next)
    x = np.array(x)
    x = x[x <= length]  # Ensure x does not exceed length
    y = amplitude * np.sin(2 * np.pi * frequency * x / length)
    return x, y

# Function to generate and plot the fanned-out curves with transversal folds
def plot_wavy_miura_ori(
    num_curves=5,                   # Number of curves
    angle_separation=np.pi/18,      # Separation between curves in radians
    length=10,                      # Base curve length
    amplitude=1,                    # Amplitude (same for all curves)
    frequency=1,                    # Frequency (same for all curves)
    inner_radius=0,                 # Inner radius (radius of the circle from which curves start)
    s_max=1,                        # Maximum segment size
    s_min=0.5,                      # Minimum segment size
    show_plot=True,                 # Display the plot
    save_svg=False                  # Save the plot as SVG
):
    # Angles at which curves will be rotated
    total_angle = angle_separation * (num_curves - 1)
    angles = np.linspace(-total_angle / 2, total_angle / 2, num_curves)

    # Create figure and axis
    fig, ax = plt.subplots(figsize=(8, 8))

    curves = []
    longitudinal_colors = []

    # Draw longitudinal folds (curves)
    for i, angle in enumerate(angles):
        # Generate the base curve with decreasing segments
        x_base, y_base = generate_base_curve_with_decreasing_segments(
            length=length, amplitude=amplitude, frequency=frequency, s_max=s_max, s_min=s_min
        )

        # Rotate curve
        rot_matrix = np.array([
            [np.cos(angle), -np.sin(angle)],
            [np.sin(angle),  np.cos(angle)]
        ])
        curve = np.dot(rot_matrix, np.vstack((x_base, y_base)))

        # Translate curve so that all curves start from the inner circle
        start_angle = angle
        delta_x = -curve[0, 0] + inner_radius * np.cos(start_angle)
        delta_y = -curve[1, 0] + inner_radius * np.sin(start_angle)
        curve[0] += delta_x
        curve[1] += delta_y

        curves.append(curve)

        # Determine fold type: first and last curves are black outlines
        if i == 0 or i == num_curves - 1:
            fold_color = 'black'  # Outline
            lw = 2.5  # Slightly thicker for outline
            z = 3     # Higher z-order to plot on top
            longitudinal_colors.append(fold_color)
        else:
            # Alternate between red and blue
            fold_color = 'red' if (i % 2 == 1) else 'blue'
            lw = 2
            z = 2
            longitudinal_colors.append(fold_color)

        # Plot curve
        ax.plot(curve[0], curve[1], color=fold_color, lw=lw, zorder=z)

    # Draw transversal folds
    # For each segment position, draw lines connecting the curves
    num_segments = len(x_base)
    for j in range(num_segments):
        # Collect points from each curve at position x_base[j]
        points = np.array([curve[:, j] for curve in curves])
        # Draw lines between adjacent curves
        for i in range(len(points) - 1):
            # Determine fold color
            if longitudinal_colors[i+1] == 'black':
                # Assign opposite color of the longitudinal fold to the left
                fold_color = opposite_color(longitudinal_colors[i])
                lw = 2
                z = 2
            else:
                fold_color = longitudinal_colors[i+1]
                lw = 2
                z = 2
            # Plot line between curve i and i+1
            x_vals = [points[i][0], points[i+1][0]]
            y_vals = [points[i][1], points[i+1][1]]
            ax.plot(x_vals, y_vals, color=fold_color, lw=lw, zorder=z)

    # Set equal aspect ratio and remove axes
    ax.set_aspect('equal')
    ax.axis('off')
    plt.tight_layout()

    # Show or save the plot
    if show_plot:
        plt.show()
    if save_svg:
        fig.savefig('wavy_miura_ori.svg', format='svg', dpi=300, bbox_inches='tight')
        print('SVG file saved as "wavy_miura_ori.svg')
    plt.close(fig)

# Interactive widgets
def interactive_wavy_miura_ori():
    interactive_plot = interactive(
        plot_wavy_miura_ori,
        num_curves=IntSlider(min=3, max=20, step=1, value=5, description='Number of Curves'),
        angle_separation=FloatSlider(min=0.01, max=np.pi/2, step=0.01, value=np.pi/18, description='Angle Separation (rad)'),
        length=FloatSlider(min=5, max=50, step=1, value=10, description='Curve Length'),
        amplitude=FloatSlider(min=0.1, max=5, step=0.1, value=1, description='Amplitude'),
        frequency=FloatSlider(min=0.1, max=10, step=0.1, value=1, description='Frequency'),
        inner_radius=FloatSlider(min=0, max=10, step=0.1, value=2, description='Inner Radius'),
        s_max=FloatSlider(min=0.1, max=5, step=0.1, value=1, description='Max Segment Size'),
        s_min=FloatSlider(min=0.1, max=5, step=0.1, value=0.5, description='Min Segment Size'),
        save_svg=Checkbox(value=False, description='Save as SVG')
    )
    display(interactive_plot)

# Run the interactive function
interactive_wavy_miura_ori()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interactive, IntSlider, FloatSlider, Checkbox
from IPython.display import display

# Function to generate the base curve with decreasing segment sizes
def generate_base_curve_with_decreasing_segments(
    length=10, amplitude=1, frequency=1, s_max=1, s_min=0.5
):
    # Generate x positions with decreasing segment sizes
    x = [0]
    while x[-1] < length:
        s = s_max - (s_max - s_min) * (x[-1] / length)
        x_next = x[-1] + s
        x.append(x_next)
    x = np.array(x)
    x = x[x <= length]  # Ensure x does not exceed length
    y = amplitude * np.sin(2 * np.pi * frequency * x / length)
    return x, y

# Function to get the opposite color
def opposite_color(color):
    if color == 'red':
        return 'blue'
    elif color == 'blue':
        return 'red'
    else:
        return 'black'  # For 'black' or any other color

# Function to generate and plot the pattern
def plot_adjustable_miura_ori(
    num_curves=5,                   # Number of curves
    angle_separation=0,             # Separation between curves in radians
    length=10,                      # Base curve length
    amplitude=1,                    # Amplitude
    frequency=1,                    # Frequency
    inner_radius=0,                 # Inner radius
    s_max=1,                        # Maximum segment size
    s_min=0.5,                      # Minimum segment size
    show_plot=True,                 # Display the plot
    save_svg=False                  # Save the plot as SVG
):
    # Angles at which curves will be rotated
    total_angle = angle_separation * (num_curves - 1)
    angles = np.linspace(-total_angle / 2, total_angle / 2, num_curves)

    # Create figure and axis
    fig, ax = plt.subplots(figsize=(8, 8))

    curves = []
    longitudinal_colors = []

    # Draw longitudinal folds (curves)
    for i, angle in enumerate(angles):
        # Generate the base curve with decreasing segments
        x_base, y_base = generate_base_curve_with_decreasing_segments(
            length=length, amplitude=amplitude, frequency=frequency, s_max=s_max, s_min=s_min
        )

        # Rotate curve
        rot_matrix = np.array([
            [np.cos(angle), -np.sin(angle)],
            [np.sin(angle),  np.cos(angle)]
        ])
        curve = np.dot(rot_matrix, np.vstack((x_base, y_base)))

        # Translate curve so that all curves start from the inner circle
        start_angle = angle
        delta_x = -curve[0, 0] + inner_radius * np.cos(start_angle)
        delta_y = -curve[1, 0] + inner_radius * np.sin(start_angle)
        curve[0] += delta_x
        curve[1] += delta_y

        curves.append(curve)

        # Determine fold type: first and last curves are black outlines
        if i == 0 or i == num_curves - 1:
            fold_color = 'black'  # Outline
            lw = 2.5  # Thicker for outline
            z = 3     # Higher z-order to plot on top
            longitudinal_colors.append(fold_color)
        else:
            # Alternate between red and blue
            fold_color = 'red' if (i % 2 == 1) else 'blue'
            lw = 2
            z = 2
            longitudinal_colors.append(fold_color)

        # Plot curve
        ax.plot(curve[0], curve[1], color=fold_color, lw=lw, zorder=z)

    # Draw transversal folds
    num_segments = len(x_base)
    for j in range(num_segments):
        # Collect points from each curve at position x_base[j]
        points = np.array([curve[:, j] for curve in curves])
        # Draw lines between adjacent curves
        for i in range(len(points) - 1):
            # Determine fold color
            if longitudinal_colors[i+1] == 'black':
                # Assign opposite color of the longitudinal fold to the left
                fold_color = opposite_color(longitudinal_colors[i])
                lw = 2
                z = 2
            else:
                fold_color = longitudinal_colors[i+1]
                lw = 2
                z = 2
            # Plot line between curve i and i+1
            x_vals = [points[i][0], points[i+1][0]]
            y_vals = [points[i][1], points[i+1][1]]
            ax.plot(x_vals, y_vals, color=fold_color, lw=lw, zorder=z)

    # Optionally, add control points or markers
    # For example, plot a circle at the base to represent a control area
    # ax.scatter(curves[0][0, 0], curves[0][1, 0], color='green', s=50, zorder=4)

    # Set equal aspect ratio and remove axes
    ax.set_aspect('equal')
    ax.axis('off')
    plt.tight_layout()

    # Show or save the plot
    if show_plot:
        plt.show()
    if save_svg:
        fig.savefig('adjustable_miura_ori.svg', format='svg', dpi=300, bbox_inches='tight')
        print('SVG file saved as "adjustable_miura_ori.svg')
    plt.close(fig)

# Interactive widgets
def interactive_adjustable_miura_ori():
    interactive_plot = interactive(
        plot_adjustable_miura_ori,
        num_curves=IntSlider(min=3, max=20, step=1, value=5, description='Number of Curves'),
        angle_separation=FloatSlider(min=-np.pi/4, max=np.pi/4, step=0.01, value=0, description='Angle Separation (rad)'),
        length=FloatSlider(min=5, max=50, step=1, value=20, description='Curve Length'),
        amplitude=FloatSlider(min=0.1, max=5, step=0.1, value=1, description='Amplitude'),
        frequency=FloatSlider(min=0.1, max=10, step=0.1, value=2, description='Frequency'),
        inner_radius=FloatSlider(min=0, max=10, step=0.1, value=0, description='Inner Radius'),
        s_max=FloatSlider(min=0.1, max=5, step=0.1, value=2, description='Max Segment Size'),
        s_min=FloatSlider(min=0.1, max=5, step=0.1, value=1, description='Min Segment Size'),
        save_svg=Checkbox(value=False, description='Save as SVG')
    )
    display(interactive_plot)

# Run the interactive function
interactive_adjustable_miura_ori()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interactive, IntSlider, FloatSlider, Checkbox
from IPython.display import display

# Function to generate decreasing vertical positions
def generate_decreasing_vertical_positions(
    total_height=10, num_segments=10, s_max=1, s_min=0.5
):
    y_positions = [0]
    total_s = 0
    for i in range(num_segments - 1):
        s = s_max - (s_max - s_min) * (total_s / total_height)
        total_s += s
        y_positions.append(total_s)
        if total_s >= total_height:
            break
    y_positions = np.array(y_positions)
    y_positions = y_positions[y_positions <= total_height]
    return y_positions

# Function to generate and plot the pattern
def plot_adjustable_miura_ori(
    num_curves=5,                   # Number of curves (longitudinal folds)
    num_segments=10,                # Number of segments (transversal folds)
    length=10,                      # Width of the pattern
    total_height=10,                # Height of the pattern
    amplitude=1,                    # Amplitude of the sine wave
    frequency=1,                    # Frequency of the sine wave
    s_max=1,                        # Maximum segment size (vertical)
    s_min=0.5,                      # Minimum segment size (vertical)
    angle_separation=0,             # Separation between curves in radians
    show_plot=True,                 # Display the plot
    save_svg=False                  # Save the plot as SVG
):
    # Generate decreasing vertical positions
    y_positions = generate_decreasing_vertical_positions(
        total_height=total_height, num_segments=num_segments, s_max=s_max, s_min=s_min
    )

    # Generate horizontal positions along the sine curve
    x_base = np.linspace(0, length, num_curves)
    x_offsets = amplitude * np.sin(2 * np.pi * frequency * y_positions / total_height)

    # Create figure and axis
    fig, ax = plt.subplots(figsize=(8, 8))

    # Calculate angles for rotation of curves
    total_angle = angle_separation * (num_curves - 1)
    angles = np.linspace(-total_angle / 2, total_angle / 2, num_curves)

    # Collect points for longitudinal folds
    curves = []
    for i, angle in enumerate(angles):
        x_curve = np.full_like(y_positions, x_base[i]) + x_offsets
        y_curve = y_positions

        # Rotate curve if needed
        if angle != 0:
            rot_matrix = np.array([
                [np.cos(angle), -np.sin(angle)],
                [np.sin(angle),  np.cos(angle)]
            ])
            curve = np.dot(rot_matrix, np.vstack((x_curve, y_curve)))
            x_curve, y_curve = curve[0], curve[1]

        curves.append((x_curve, y_curve))

    # Plot longitudinal folds
    for i, (x_curve, y_curve) in enumerate(curves):
        if i == 0 or i == num_curves - 1:
            fold_color = 'black'  # Outline
            lw = 2.5
            z = 3
        else:
            fold_color = 'red' if i % 2 == 1 else 'blue'
            lw = 2
            z = 2
        ax.plot(x_curve, y_curve, color=fold_color, lw=lw, zorder=z)

    # Plot transversal folds
    num_points = len(y_positions)
    for j in range(num_points):
        # Collect points at each horizontal level
        points = np.array([[x_curve[j], y_curve[j]] for (x_curve, y_curve) in curves])

        # Draw lines between adjacent curves
        for i in range(len(points) - 1):
            # Determine fold color based on (i + j)
            fold_color = 'red' if (i + j) % 2 == 0 else 'blue'
            lw = 2
            z = 2
            x_vals = [points[i][0], points[i+1][0]]
            y_vals = [points[i][1], points[i+1][1]]
            ax.plot(x_vals, y_vals, color=fold_color, lw=lw, zorder=z)

    # Set equal aspect ratio and remove axes
    ax.set_aspect('equal')
    ax.axis('off')
    plt.tight_layout()

    # Show or save the plot
    if show_plot:
        plt.show()
    if save_svg:
        fig.savefig('adjustable_miura_ori.svg', format='svg', dpi=300, bbox_inches='tight')
        print('SVG file saved as "adjustable_miura_ori.svg')
    plt.close(fig)

# Interactive widgets
def interactive_adjustable_miura_ori():
    interactive_plot = interactive(
        plot_adjustable_miura_ori,
        num_curves=IntSlider(min=3, max=20, step=1, value=5, description='Number of Curves'),
        num_segments=IntSlider(min=5, max=50, step=1, value=10, description='Number of Segments'),
        length=FloatSlider(min=5, max=50, step=1, value=20, description='Pattern Width'),
        total_height=FloatSlider(min=5, max=50, step=1, value=20, description='Pattern Height'),
        amplitude=FloatSlider(min=0, max=5, step=0.1, value=1, description='Amplitude'),
        frequency=FloatSlider(min=0.1, max=30, step=0.1, value=2, description='Frequency'),
        s_max=FloatSlider(min=0.1, max=5, step=0.1, value=1.0, description='Max Segment Size'),
        s_min=FloatSlider(min=0.1, max=5, step=0.1, value=0.5, description='Min Segment Size'),
        angle_separation=FloatSlider(min=-np.pi/4, max=np.pi/4, step=0.01, value=0, description='Angle Separation (rad)'),
        save_svg=Checkbox(value=False, description='Save as SVG')
    )
    display(interactive_plot)

# Run the interactive function
interactive_adjustable_miura_ori()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interactive, FloatSlider, IntSlider
from IPython.display import display

# Function to generate Miura-Ori pattern with column scaling and adjustable angles
def plot_scaled_miura_ori(
    base_width=10,       # Total base width of the pattern
    base_height=10,      # Total base height of the pattern
    num_segments=6,      # Number of segments in each row
    num_columns=5,       # Number of columns
    width_scale=0.9,     # Scaling factor for column width
    height_scale=0.9,    # Scaling factor for column height
    row_angle_deg=30,    # Angle between rows in degrees
    leg_angle_deg=15,    # Incidence angle for legs in degrees
    show_plot=True       # Display the plot
):
    # Convert angles to radians
    row_angle_rad = np.radians(row_angle_deg)
    leg_angle_rad = np.radians(leg_angle_deg)

    # Initialize starting positions
    x_coords = np.zeros((num_columns, num_segments + 1))
    y_coords = np.zeros((num_columns, num_segments + 1))

    # Define initial segment size
    segment_width = base_width / num_segments
    segment_height = base_height

    for col in range(num_columns):
        # Scale width and height for this column
        col_width = segment_width * (width_scale ** col)
        col_height = segment_height * (height_scale ** col)

        for seg in range(num_segments + 1):
            # X-coordinate: linear increment
            x_coords[col, seg] = seg * col_width
            # Y-coordinate: alternating row angles
            if seg % 2 == 0:
                y_coords[col, seg] = col * np.tan(row_angle_rad) * col_width
            else:
                y_coords[col, seg] = col * np.tan(-row_angle_rad) * col_width

    # Plot the Miura-Ori pattern
    plt.figure(figsize=(8, 6))

    # Draw columns
    for col in range(num_columns):
        # Draw horizontal segments
        plt.plot(x_coords[col, :], y_coords[col, :], 'b-')

        # Draw diagonal legs
        for seg in range(num_segments):
            x_leg = [x_coords[col, seg], x_coords[col, seg + 1]]
            if seg % 2 == 0:
                y_leg = [y_coords[col, seg], y_coords[col, seg + 1] + col_height * np.tan(leg_angle_rad)]
            else:
                y_leg = [y_coords[col, seg], y_coords[col, seg + 1] - col_height * np.tan(leg_angle_rad)]
            plt.plot(x_leg, y_leg, 'b-')

    # Draw connecting vertical lines between rows
    for seg in range(num_segments + 1):
        for col in range(num_columns - 1):
            plt.plot(
                [x_coords[col, seg], x_coords[col + 1, seg]],
                [y_coords[col, seg], y_coords[col + 1, seg]],
                'r-'
            )

    # Adjust plot settings
    plt.gca().set_aspect('equal', adjustable='box')
    plt.axis('off')
    plt.title("Scaled Miura-Ori Pattern")
    plt.show()

# Interactive widget
def interactive_scaled_miura_ori():
    interactive_plot = interactive(
        plot_scaled_miura_ori,
        base_width=FloatSlider(value=10, min=5, max=50, step=1, description='Base Width'),
        base_height=FloatSlider(value=10, min=5, max=50, step=1, description='Base Height'),
        num_segments=IntSlider(value=6, min=3, max=20, step=1, description='Segments'),
        num_columns=IntSlider(value=5, min=2, max=10, step=1, description='Columns'),
        width_scale=FloatSlider(value=0.9, min=0.5, max=1.0, step=0.05, description='Width Scale'),
        height_scale=FloatSlider(value=0.9, min=0.5, max=1.0, step=0.05, description='Height Scale'),
        row_angle_deg=FloatSlider(value=30, min=0, max=90, step=1, description='Row Angle (°)'),
        leg_angle_deg=FloatSlider(value=15, min=0, max=90, step=1, description='Leg Angle (°)'),
    )
    display(interactive_plot)

# Run the interactive widget
interactive_scaled_miura_ori()
```

```python


# Function to create the segment shapes
def create_segment(angle1, angle2, base_radius, apex_radius, skew, apex_multiplier, segment_index):
    theta_base = np.linspace(angle1, angle2, 100)
    r_base = np.full_like(theta_base, base_radius)
    r_apex = np.full_like(theta_base, apex_radius * (apex_multiplier ** segment_index))
    
    base_segment = np.column_stack((theta_base, r_base))
    apex_segment = np.column_stack((theta_base + skew, r_apex + base_radius))
    
    segment = np.vstack((base_segment, apex_segment[::-1]))
    return segment

# Interactive plot function
def plot_segments(num_segments, base_radius, apex_radius, skew, apex_multiplier, separation, save_svg=False):
    angles = np.linspace(0, 2 * np.pi, num_segments + 1)
    separation_angle = separation * np.pi / 180  # Convert separation from degrees to radians
    
    fig, ax = plt.subplots(figsize=(8, 8), subplot_kw={'projection': 'polar'})
    ax.set_theta_direction(-1)
    ax.set_theta_offset(np.pi / 2)
    
    for i in range(num_segments):
        segment = create_segment(angles[i] + separation_angle / 2, angles[i + 1] - separation_angle / 2, base_radius, apex_radius, skew, apex_multiplier, i)
        ax.fill(segment[:, 0], segment[:, 1], edgecolor='black')


    plt.title('Interactive Circle with Tangential Segment Apexes')
    if save_svg:
        plt.savefig("segments_plot.svg", format="svg")
        print("SVG file saved as 'segments_plot.svg'")
    else:
        plt.show()
```

```python
def plot_interpolating_horizontal_zigzags_with_boundaries(num_lines, amplitude, height, angle):
    """
    Plot a graph with vertical zigzag lines with 90-degree angles, where every even line is reversed.
    Horizontal lines are added in the white space between zigzag lines, with interpolating widths.
    A top and bottom black boundary line is added.

    :param num_lines: Number of zigzag lines.
    :param amplitude: Amplitude of the zigzag peaks.
    :param height: Height of the plot.
    :return: None, but displays the plot.
    """
    fig, ax = plt.subplots(figsize=(60, 30))
    red = "#FF0000"
    blue = "#0000FF"
    multiplier = 2.5
    # Generate and plot each zigzag line
    for i in range(num_lines):
        x_points = []
        y_points = []
        for j in range(height + 1):
            if i % 2 == 0:  # Even line
                x = i * amplitude * multiplier + (amplitude if j % 2 == 0 else 0)
            else:  # Odd line
                x = i * amplitude * multiplier + (amplitude if j % 2 != 0 else 0)
            x_points.append(x)
            y_points.append(j)

        color = red if i % 2 == 0 else red
        color = 'black' if i in (0, num_lines - 1) else color
        ax.plot(x_points, y_points, color=red, lw=2)

    # Generate and plot interpolating horizontal lines
    for j in range(height + 1):
        y_horizontal = j
        for k in range(num_lines - 1):
            if (k + j) % 2 == 0:  # Horizontal line in the wider space
                x_start = k * amplitude * multiplier + amplitude
                x_end = x_start + amplitude + 1
            else:  # Horizontal line in the narrower space
                x_start = k * amplitude * multiplier
                x_end = x_start + amplitude * multiplier * 1.4
            
            color = red if (k + j) % 2 == 0 else blue
            if (k <= num_lines - 1):
                ax.hlines(y_horizontal, x_start, x_end, color=color, linewidth=2)

    # Add top and bottom black boundary lines

    # Set plot limits
    ax.set_xlim(0, num_lines * amplitude * multiplier)
    ax.set_ylim(0, height)

    # Remove axes for clarity
    ax.axis('off')
    plt.savefig("miura.svg", format='svg', bbox_inches='tight', pad_inches=0)

    # Show the plot
    plt.show()

# Parameters for the plot
num_lines = 5  # Number of zigzag lines
amplitude = 2 # Amplitude of the zigzag peaks
height = 2# Height of the plot
angle = pi*0.2
# Generate the plot
plot_interpolating_horizontal_zigzags_with_boundaries(num_lines, amplitude, height, angle)
```

```python
import matplotlib.pyplot as plt
import numpy as np

def plot_interpolating_horizontal_zigzags_with_boundaries(num_lines, amplitude, height, angle):
    """
    Plot a graph with vertical zigzag lines with 90-degree angles, where every even line is reversed.
    Horizontal lines are added in the white space between zigzag lines, with interpolating widths.
    A top and bottom black boundary line is added, and the entire shape is rotated by a specified angle.

    :param num_lines: Number of zigzag lines.
    :param amplitude: Amplitude of the zigzag peaks.
    :param height: Height of the plot.
    :param angle: Rotation angle for the entire shape (in degrees).
    :return: None, but displays the plot.
    """
    # Rotation matrix for the specified angle
    theta = np.radians(angle)
    rotation_matrix = np.array([[np.cos(theta), -np.sin(theta)],
                                 [np.sin(theta),  np.cos(theta)]])
    
    fig, ax = plt.subplots(figsize=(12, 6))
    red = "#FF0000"
    blue = "#0000FF"
    multiplier = 2.5

    # Store all points for rotation
    all_points = []

    # Generate and plot each zigzag line
    for i in range(num_lines):
        x_points = []
        y_points = []
        for j in range(height + 1):
            if i % 2 == 0:  # Even line
                x = i * amplitude * multiplier + (amplitude if j % 2 == 0 else 0)
            else:  # Odd line
                x = i * amplitude * multiplier + (amplitude if j % 2 != 0 else 0)
            x_points.append(x)
            y_points.append(j)

        # Add points to the global point list for rotation
        all_points.extend(list(zip(x_points, y_points)))

        color = red if i % 2 == 0 else red
        color = 'black' if i in (0, num_lines - 1) else color
        ax.plot(x_points, y_points, color=color, lw=2)

    # Generate and plot interpolating horizontal lines
    for j in range(height + 1):
        y_horizontal = j
        for k in range(num_lines - 1):
            if (k + j) % 2 == 0:  # Horizontal line in the wider space
                x_start = k * amplitude * multiplier + amplitude
                x_end = x_start + amplitude + 1
            else:  # Horizontal line in the narrower space
                x_start = k * amplitude * multiplier
                x_end = x_start + amplitude * multiplier * 1.4

            color = red if (k + j) % 2 == 0 else blue
            if k <= num_lines - 1:
                all_points.extend([(x_start, y_horizontal), (x_end, y_horizontal)])
                ax.hlines(y_horizontal, x_start, x_end, color=color, linewidth=2)

    # Rotate all points
    rotated_points = np.dot(np.array(all_points), rotation_matrix.T)

    # Extract rotated X and Y coordinates
    x_rotated, y_rotated = rotated_points[:, 0], rotated_points[:, 1]

    # Plot rotated points
    ax.clear()
    ax.plot(x_rotated, y_rotated, color="black", lw=2, alpha=0.7)

    # Set plot limits based on rotated points
    ax.set_xlim(x_rotated.min() - amplitude, x_rotated.max() + amplitude)
    ax.set_ylim(y_rotated.min() - amplitude, y_rotated.max() + amplitude)

    # Remove axes for clarity
    ax.axis('off')

    # Save and show the plot
    plt.savefig("rotated_zigzag.svg", format='svg', bbox_inches='tight', pad_inches=0)
    plt.show()

# Parameters for the plot
num_lines = 5  # Number of zigzag lines
amplitude = 2  # Amplitude of the zigzag peaks
height = 10  # Height of the plot
angle = 30  # Rotation angle in degrees

# Generate the plot
plot_interpolating_horizontal_zigzags_with_boundaries(num_lines, amplitude, height, angle)
```

```python
def plot_tilted_inverse_lines(angle, angle_2, length, length_2, separation):
    """
    Plot two lines where one is tilted positively by a specified angle, and the other is tilted inversely.

    :param angle: Angle in degrees by which the first line is tilted positively and the second inversely.
    :param length: Length of the lines.
    :param separation: Vertical separation between the lines at the origin.
    :return: None, but displays the plot.
    """
    # Line 1: Tilted positively by the given angle
    x1 = np.linspace(0, length, 100)
    y1 = np.tan(np.radians(angle)) * x1

    # Line 2: Tilted inversely (negative angle), offset vertically
    x2 = np.linspace(0, length, 100)
    y2 = separation - np.tan(np.radians(angle)) * x2

     # Line 3: Tilted inversely (negative angle), offset vertically
    x3 = np.linspace(0, length_2, 100)
    y3 = np.tan(np.radians(angle_2)) * x3
    
    x4 = np.linspace(0, length_2, 100)
    y4 = separation - np.tan(np.radians(angle_2)) * x4
    
    # Plot the lines
    fig, ax = plt.subplots(figsize=(10, 6))
    ax.plot(x1, y1, label=f"Line 1 (Tilted by {angle}°)", color="blue", lw=2)
    ax.plot(x2, y2, label=f"Line 2 (Tilted by -{angle}°)", color="red", lw=2)
    ax.plot(x3, y3, label=f"Line 2 (Tilted by -{angle}°)", color="g", lw=2)
    ax.plot(x4, y4, label=f"Line 2 (Tilted by -{angle}°)", color="black", lw=2)

    # Enhance plot appearance
    ax.set_aspect('equal', adjustable='datalim')
    ax.axhline(0, color='black', linewidth=0.5, linestyle='--')
    ax.axvline(0, color='black', linewidth=0.5, linestyle='--')
    ax.set_title("Two Tilted Lines with Inverse Slopes and Vertical Offset")
    ax.legend()
    plt.grid(True)
    plt.show()

# Parameters for the lines
angle = 5  # Angle of tilt for one line (positive) and the other inversely (negative)
angle_2 = 35  # Angle of tilt for one line (positive) and the other inversely (negative)

length = 10  # Length of the lines
length_2 = 2.2  # Length of the lines
separation = 3# Vertical separation between the lines at the origin

# Generate and plot the lines
plot_tilted_inverse_lines(angle, angle_2, length, length_2, separation)
```

```python
import matplotlib.pyplot as plt
import numpy as np

def plot_hexagon_tessellation(rows, cols, size):
    fig, ax = plt.subplots(figsize=(10, 10))

    for row in range(rows):
        for col in range(cols):
            # Calculate the center of the hexagon
            x_offset = size * np.sqrt(3) * (col + 0.5 * (row % 2))
            y_offset = size * 1.5 * row
            hexagon = create_hexagon(x_offset, y_offset, size)
            # Alternate colors for valley (blue) and mountain (red) folds
            if (row + col) % 2 == 0:
                color = 'red'  # Mountain fold
            else:
                color = 'blue'  # Valley fold
            ax.plot(hexagon[:, 0], hexagon[:, 1], color=color)

    ax.set_aspect('equal')
    ax.axis('off')
    plt.show()

def create_hexagon(x_center, y_center, size):
    # Generate points for a hexagon centered at (x_center, y_center)
    angles = np.linspace(0, 2 * np.pi, 7)
    x = x_center + size * np.cos(angles)
    y = y_center + size * np.sin(angles)
    return np.column_stack((x, y))

# Parameters
rows = 2    # Number of rows of hexagons
cols = 5    # Number of columns of hexagons
size = 1    # Size of each hexagon

plot_hexagon_tessellation(rows, cols, size)
```

```python
import matplotlib.pyplot as plt
import numpy as np

def plot_interpolating_horizontal_zigzags_with_boundaries(num_lines, amplitude, height, angle):
    """
    Plot a graph with zigzag lines at a specified angle, where every even line is reversed.
    Horizontal lines are added in the white space between zigzag lines, with interpolating widths.
    A top and bottom black boundary line is added.

    :param num_lines: Number of zigzag lines.
    :param amplitude: Amplitude of the zigzag peaks.
    :param height: Number of vertical steps in the zigzag lines.
    :param angle: Angle in degrees for the zigzag lines.
    :return: None, but displays the plot.
    """
    fig, ax = plt.subplots(figsize=(10, 5))
    angle_rad = np.radians(angle)
    dx = amplitude * np.cos(angle_rad)
    dy = amplitude * np.sin(angle_rad)
    
    # Generate and plot each zigzag line
    for i in range(num_lines):
        x_points = []
        y_points = []
        x_start = i * dx * (height)
        y_start = 0
        x_points.append(x_start)
        y_points.append(y_start)
        
        for j in range(1, height + 1):
            direction = -1 if (i + j) % 2 == 0 else 1  # Alternate direction
            x_new = x_points[-1] + direction * dx
            y_new = y_points[-1] + dy
            x_points.append(x_new)
            y_points.append(y_new)
        
        color = 'red' if i % 2 == 0 else 'blue'
        ax.plot(x_points, y_points, color=color, lw=2)
    
    # Add top and bottom black boundary lines
    x_min = 0
    x_max = num_lines * dx * height
    y_bottom = 0
    y_top = height * dy
    ax.hlines(y_bottom, x_min, x_max, colors='black', linewidth=2)
    ax.hlines(y_top, x_min, x_max, colors='black', linewidth=2)
    
    # Set plot limits
    ax.set_xlim(x_min - amplitude, x_max + amplitude)
    ax.set_ylim(y_bottom - amplitude, y_top + amplitude)
    
    # Remove axes for clarity
    ax.axis('off')
    plt.savefig("zigzag_pattern.svg", format='svg', bbox_inches='tight', pad_inches=0)
    
    # Show the plot
    plt.show()

# Parameters for the plot
num_lines = 5   # Number of zigzag lines
amplitude = 2   # Amplitude of the zigzag peaks
height = 2      # Number of vertical steps in the zigzag lines (2 rows)
angle = 10      # Angle in degrees

# Generate the plot
plot_interpolating_horizontal_zigzags_with_boundaries(num_lines, amplitude, height, angle)
```

```python
import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interactive, IntSlider, FloatSlider, Checkbox
from IPython.display import display

def generate_base_curve_with_decreasing_segments(
    length=10, amplitude=1, frequency=1, s_max=1, s_min=0.5
):
    x = [0]
    while x[-1] < length:
        s = s_max - (s_max - s_min) * (x[-1] / length)
        x_next = x[-1] + s
        x.append(x_next)
    x = np.array(x)
    x = x[x <= length]
    y = amplitude * np.sin(2 * np.pi * frequency * x / length)
    return x, y

def angle_between_vectors(a, b):
    # Computes the angle between two vectors a and b
    cos_angle = np.dot(a, b) / (np.linalg.norm(a)*np.linalg.norm(b))
    # Clamp due to numerical errors
    cos_angle = np.clip(cos_angle, -1.0, 1.0)
    return np.arccos(cos_angle)

def measure_flat_foldability(curves):
    # Pick a vertex in the middle of the pattern (not at edges)
    # For simplicity, pick the midpoint indices:
    if len(curves) < 3 or curves[0].shape[1] < 3:
        return 0  # Not enough data
    i_mid = len(curves)//2
    j_mid = curves[0].shape[1]//2
    # vertex at curves[i_mid][0][j_mid], curves[i_mid][1][j_mid]
    vx = curves[i_mid][0, j_mid]
    vy = curves[i_mid][1, j_mid]
    
    # Adjacent vertices around the chosen vertex:
    # up:    i_mid-1, j_mid
    # down:  i_mid+1, j_mid
    # left:  i_mid, j_mid-1
    # right: i_mid, j_mid+1
    # Check bounds
    if i_mid-1 < 0 or i_mid+1 >= len(curves) or j_mid-1 < 0 or j_mid+1 >= curves[0].shape[1]:
        return 0
    upx, upy = curves[i_mid-1][0, j_mid], curves[i_mid-1][1, j_mid]
    downx, downy = curves[i_mid+1][0, j_mid], curves[i_mid+1][1, j_mid]
    leftx, lefty = curves[i_mid][0, j_mid-1], curves[i_mid][1, j_mid-1]
    rightx, righty = curves[i_mid][0, j_mid+1], curves[i_mid][1, j_mid+1]

    # Vectors from the vertex to neighbors:
    v_up = np.array([upx - vx, upy - vy])
    v_down = np.array([downx - vx, downy - vy])
    v_left = np.array([leftx - vx, lefty - vy])
    v_right = np.array([rightx - vx, righty - vy])

    # Compute angles around the vertex
    # We'll consider the four directions: up, right, down, left.
    # Find angles between successive vectors: up-right, right-down, down-left, left-up
    # Alternating sum condition: angle(up-right) + angle(down-left) should be ~ π
    # and angle(right-down) + angle(left-up) should be ~ π
    angle_ur = angle_between_vectors(v_up, v_right)
    angle_rd = angle_between_vectors(v_right, v_down)
    angle_dl = angle_between_vectors(v_down, v_left)
    angle_lu = angle_between_vectors(v_left, v_up)

    # Alternate sums
    sum1 = angle_ur + angle_dl
    sum2 = angle_rd + angle_lu
    # Ideal: sum1 = π and sum2 = π
    # We'll return how far we are from π in total
    error = abs(sum1 - np.pi) + abs(sum2 - np.pi)
    return error

def plot_adjustable_miura_ori(
    num_curves=5,                   
    angle_separation=0,            
    length=10,                     
    amplitude=1,                   
    frequency=1,                   
    inner_radius=0,                
    s_max=1,                       
    s_min=0.5,                     
    show_plot=True,                
    save_svg=False                 
):
    # We attempt a simple iterative approach to adjust angle_separation for flat-foldability:
    # We'll do a few iterations adjusting angle_separation slightly to minimize the angle error.
    # This is a heuristic approach and may not yield perfect flat-foldability.
    # Initial guess:
    angle_sep_original = angle_separation
    learning_rate = 0.005
    max_iters = 20
    best_error = np.inf
    best_angle_sep = angle_separation

    for _ in range(max_iters):
        # Generate pattern with current angle_separation
        current_angle_sep = angle_separation
        if current_angle_sep > np.pi/6:
            current_angle_sep = np.pi/6
        elif current_angle_sep < -np.pi/6:
            current_angle_sep = -np.pi/6

        x_base, y_base = generate_base_curve_with_decreasing_segments(
            length=length, amplitude=amplitude, frequency=frequency, s_max=s_max, s_min=s_min
        )

        total_angle = current_angle_sep * (num_curves - 1)
        angles = np.linspace(-total_angle / 2, total_angle / 2, num_curves)
        curves = []
        for i, angle in enumerate(angles):
            rot_matrix = np.array([
                [np.cos(angle), -np.sin(angle)],
                [np.sin(angle),  np.cos(angle)]
            ])
            curve = np.dot(rot_matrix, np.vstack((x_base, y_base)))
            start_angle = angle
            delta_x = -curve[0, 0] + inner_radius * np.cos(start_angle)
            delta_y = -curve[1, 0] + inner_radius * np.sin(start_angle)
            curve[0] += delta_x
            curve[1] += delta_y
            curves.append(curve)

        error = measure_flat_foldability(curves)
        if error < best_error:
            best_error = error
            best_angle_sep = current_angle_sep

        # If error is small enough, break
        if best_error < 0.01:
            break

        # Adjust angle_separation to reduce error
        # We'll try increasing/decreasing angle_separation and see which improves
        angle_test_plus = angle_separation + learning_rate
        angle_test_minus = angle_separation - learning_rate

        # Evaluate error for both directions
        test_errors = {}
        for test_angle in [angle_test_plus, angle_test_minus]:
            total_angle = test_angle * (num_curves - 1)
            angles = np.linspace(-total_angle / 2, total_angle / 2, num_curves)
            test_curves = []
            for i, angle in enumerate(angles):
                rot_matrix = np.array([
                    [np.cos(angle), -np.sin(angle)],
                    [np.sin(angle),  np.cos(angle)]
                ])
                curve = np.dot(rot_matrix, np.vstack((x_base, y_base)))
                start_angle = angle
                delta_x = -curve[0, 0] + inner_radius * np.cos(start_angle)
                delta_y = -curve[1, 0] + inner_radius * np.sin(start_angle)
                curve[0] += delta_x
                curve[1] += delta_y
                test_curves.append(curve)
            test_errors[test_angle] = measure_flat_foldability(test_curves)

        # Pick the angle direction that gives lower error
        if test_errors[angle_test_plus] < test_errors[angle_test_minus]:
            angle_separation = angle_test_plus
        else:
            angle_separation = angle_test_minus

    # After iterative adjustment, use best_angle_sep
    angle_separation = best_angle_sep
    # Generate final pattern
    x_base, y_base = generate_base_curve_with_decreasing_segments(
        length=length, amplitude=amplitude, frequency=frequency, s_max=s_max, s_min=s_min
    )
    total_angle = angle_separation * (num_curves - 1)
    angles = np.linspace(-total_angle / 2, total_angle / 2, num_curves)

    fig, ax = plt.subplots(figsize=(8, 8))

    curves = []
    longitudinal_colors = []
    for i, angle in enumerate(angles):
        rot_matrix = np.array([
            [np.cos(angle), -np.sin(angle)],
            [np.sin(angle),  np.cos(angle)]
        ])
        curve = np.dot(rot_matrix, np.vstack((x_base, y_base)))
        start_angle = angle
        delta_x = -curve[0, 0] + inner_radius * np.cos(start_angle)
        delta_y = -curve[1, 0] + inner_radius * np.sin(start_angle)
        curve[0] += delta_x
        curve[1] += delta_y
        curves.append(curve)

        if i == 0 or i == num_curves - 1:
            fold_color = 'black'
            lw = 2.5
            z = 3
            longitudinal_colors.append(fold_color)
        else:
            fold_color = 'red' if (i % 2 == 1) else 'blue'
            lw = 2
            z = 2
            longitudinal_colors.append(fold_color)

        ax.plot(curve[0], curve[1], color=fold_color, lw=lw, zorder=z)

    num_segments = len(x_base)
    for j in range(num_segments):
        points = np.array([curve[:, j] for curve in curves])
        for i in range(len(points) - 1):
            if (i + j) % 2 == 0:
                fold_color = 'red'  
            else:
                fold_color = 'blue' 
            lw = 2
            z = 2
            x_vals = [points[i][0], points[i+1][0]]
            y_vals = [points[i][1], points[i+1][1]]
            ax.plot(x_vals, y_vals, color=fold_color, lw=lw, zorder=z)

    ax.set_aspect('equal')
    ax.axis('off')
    plt.tight_layout()

    if show_plot:
        plt.show()
    if save_svg:
        fig.savefig('adjustable_miura_ori.svg', format='svg', dpi=300, bbox_inches='tight')
    plt.close(fig)

def interactive_adjustable_miura_ori():
    interactive_plot = interactive(
        plot_adjustable_miura_ori,
        num_curves=IntSlider(min=3, max=20, step=1, value=5, description='Number of Curves'),
        angle_separation=FloatSlider(min=-np.pi/4, max=np.pi/4, step=0.01, value=0, description='Angle Separation (rad)'),
        length=FloatSlider(min=5, max=50, step=1, value=20, description='Curve Length'),
        amplitude=FloatSlider(min=0.1, max=5, step=0.1, value=1, description='Amplitude'),
        frequency=FloatSlider(min=0.1, max=10, step=0.1, value=2, description='Frequency'),
        inner_radius=FloatSlider(min=0, max=10, step=0.1, value=0, description='Inner Radius'),
        s_max=FloatSlider(min=0.1, max=5, step=0.1, value=2, description='Max Segment Size'),
        s_min=FloatSlider(min=0.1, max=5, step=0.1, value=1, description='Min Segment Size'),
        save_svg=Checkbox(value=False, description='Save as SVG')
    )
    display(interactive_plot)

interactive_adjustable_miura_ori()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interactive, FloatSlider, IntSlider
from IPython.display import display

def plot_miura_ori(m=5, n=10, a=1.0, b=1.0, alpha=np.pi/3, show_plot=True, save_svg=False):
    # Generate coordinates
    # m rows, n columns of parallelograms -> (m+1) x (n+1) grid of vertices
    i_values = np.arange(m+1)
    j_values = np.arange(n+1)

    # Compute vertex coordinates
    X = np.zeros((m+1, n+1))
    Y = np.zeros((m+1, n+1))
    for i in i_values:
        for j in j_values:
            X[i,j] = j * a + (b * np.cos(alpha) if i % 2 == 1 else 0)
            Y[i,j] = i * b * np.sin(alpha)

    fig, ax = plt.subplots(figsize=(8, 8))

    # Plot longitudinal folds
    # Vertical direction: lines connecting (i,j) to (i+1,j)
    # Alternate coloring for mountain/valley folds
    for j in j_values:
        for i in range(m):
            # Assign folds (just alternate colors for demonstration)
            fold_color = 'red' if (i+j) % 2 == 0 else 'blue'
            ax.plot([X[i,j], X[i+1,j]], [Y[i,j], Y[i+1,j]], color=fold_color, lw=2)

    # Plot transversal folds
    # Horizontal direction: lines connecting (i,j) to (i,j+1)
    # Alternate these as well
    for i in i_values:
        for j in range(n):
            fold_color = 'blue' if (i+j) % 2 == 0 else 'red'
            ax.plot([X[i,j], X[i,j+1]], [Y[i,j], Y[i,j+1]], color=fold_color, lw=2)

    # Plot outer boundary in black for clarity
    # Left and right edges
    for i in range(m):
        ax.plot([X[i,0], X[i+1,0]], [Y[i,0], Y[i+1,0]], color='black', lw=2)
        ax.plot([X[i,n], X[i+1,n]], [Y[i,n], Y[i+1,n]], color='black', lw=2)
    # Top and bottom edges
    for j in range(n):
        ax.plot([X[0,j], X[0,j+1]], [Y[0,j], Y[0,j+1]], color='black', lw=2)
        ax.plot([X[m,j], X[m,j+1]], [Y[m,j], Y[m,j+1]], color='black', lw=2)

    ax.set_aspect('equal')
    ax.axis('off')
    plt.tight_layout()

    if show_plot:
        plt.show()
    if save_svg:
        fig.savefig('miura_ori.svg', format='svg', dpi=300, bbox_inches='tight')
        print('SVG file saved as "miura_ori.svg"')
    plt.close(fig)

def interactive_miura_ori():
    interactive_plot = interactive(
        plot_miura_ori,
        m=IntSlider(min=1, max=20, step=1, value=5, description='Rows (m)'),
        n=IntSlider(min=1, max=20, step=1, value=10, description='Cols (n)'),
        a=FloatSlider(min=0.5, max=5, step=0.1, value=1.0, description='a'),
        b=FloatSlider(min=0.5, max=5, step=0.1, value=1.0, description='b'),
        alpha=FloatSlider(min=0.1, max=np.pi-0.1, step=0.1, value=np.pi/3, description='alpha'),
        show_plot=True,
        save_svg=False
    )
    display(interactive_plot)

interactive_miura_ori()
```

```python
import numpy as np
from scipy.optimize import least_squares

def flat_fold_system(vars, M, N, a_array, b_array):
    # vars is a 1D array containing [X[0,0],Y[0,0], X[0,1],Y[0,1], ... X[M,N],Y[M,N]]
    # Reshape into 2D arrays for convenience
    X = vars[0::2].reshape(M+1, N+1)
    Y = vars[1::2].reshape(M+1, N+1)

    residuals = []

    # Distance constraints: horizontal edges
    for i in range(M+1):
        for j in range(N):
            dx = X[i,j+1] - X[i,j]
            dy = Y[i,j+1] - Y[i,j]
            residuals.append(dx*dx + dy*dy - a_array[i,j]**2)

    # Distance constraints: vertical edges
    for i in range(M):
        for j in range(N+1):
            dx = X[i+1,j] - X[i,j]
            dy = Y[i+1,j] - Y[i,j]
            residuals.append(dx*dx + dy*dy - b_array[i,j]**2)

    # Angle constraints at interior vertices
    # For each interior vertex, compute vectors to neighbors and enforce Kawasaki-Justin
    for i in range(1,M):
        for j in range(1,N):
            vx = X[i,j]
            vy = Y[i,j]
            # Four neighbors: up=(i-1,j), right=(i,j+1), down=(i+1,j), left=(i,j-1)
            # Make sure all neighbors exist
            v_up = np.array([X[i-1,j]-vx, Y[i-1,j]-vy])
            v_down = np.array([X[i+1,j]-vx, Y[i+1,j]-vy])
            v_left = np.array([X[i,j-1]-vx, Y[i,j-1]-vy])
            v_right = np.array([X[i,j+1]-vx, Y[i,j+1]-vy])

            # Compute angles:
            def angle(u,v):
                cos_ = np.dot(u,v)/(np.linalg.norm(u)*np.linalg.norm(v))
                cos_ = np.clip(cos_, -1,1)
                return np.arccos(cos_)

            angle_ur = angle(v_up, v_right)
            angle_rd = angle(v_right, v_down)
            angle_dl = angle(v_down, v_left)
            angle_lu = angle(v_left, v_up)

            # Kawasaki-Justin: angle_ur+angle_dl ~ pi and angle_rd+angle_lu ~ pi
            residuals.append((angle_ur + angle_dl - np.pi))
            residuals.append((angle_rd + angle_lu - np.pi))

    return np.array(residuals)

# Example usage:
M, N = 4, 6
# Define non-uniform a and b
a_array = np.random.uniform(0.9, 1.1, size=(M+1, N))  # horizontal lengths
b_array = np.random.uniform(0.9, 1.1, size=(M, N+1))  # vertical lengths

# Initial guess: start roughly as a grid
X_init = np.zeros((M+1, N+1))
Y_init = np.zeros((M+1, N+1))
for i in range(M+1):
    for j in range(N+1):
        X_init[i,j] = j
        Y_init[i,j] = i

vars_init = np.column_stack((X_init.flatten(), Y_init.flatten())).ravel()

res = least_squares(flat_fold_system, vars_init, args=(M,N,a_array,b_array))

X_sol = res.x[0::2].reshape(M+1, N+1)
Y_sol = res.x[1::2].reshape(M+1, N+1)
```

```python
import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interactive, FloatSlider, IntSlider
from IPython.display import display

def plot_scaled_miura_ori(m=5, n=10, a=1.0, b=1.0, alpha=np.pi/3, delta=0.1, show_plot=True, save_svg=False):
    # Generate per-row a_i and b_i
    # a_i = a*(1-delta)^i
    # b_i = b*(1-delta)^i
    a_values = a * (1 - delta)**np.arange(m+1)
    b_values = b * (1 - delta)**np.arange(m+1)
    
    # Compute vertical offsets: Y[i, *] depends on sum of all b_k for k < i
    # Y[i,j] = (sum_{k=0}^{i-1} b_k)*sin(alpha)
    vertical_offsets = np.zeros(m+1)
    for i in range(1, m+1):
        vertical_offsets[i] = vertical_offsets[i-1] + b_values[i-1]*np.sin(alpha)

    i_values = np.arange(m+1)
    j_values = np.arange(n+1)
    
    X = np.zeros((m+1, n+1))
    Y = np.zeros((m+1, n+1))

    for i in i_values:
        for j in j_values:
            base_x = j * a_values[i]
            # Odd rows shift horizontally by b_i*cos(alpha)
            if i % 2 == 1:
                base_x += b_values[i] * np.cos(alpha)
            X[i,j] = base_x
            Y[i,j] = vertical_offsets[i]

    fig, ax = plt.subplots(figsize=(8, 8))

    # Plot longitudinal folds
    for j in j_values:
        for ii in range(m):
            fold_color = 'blue' if (ii+j) % 2 == 0 else 'red'
            ax.plot([X[ii,j], X[ii+1,j]], [Y[ii,j], Y[ii+1,j]], color=fold_color, lw=2)

    # Plot transversal folds
    for i in i_values:
        for jj in range(n):
            fold_color = 'red' if (i+jj) % 2 == 0 else 'blue'
            ax.plot([X[i,jj], X[i,jj+1]], [Y[i,jj], Y[i,jj+1]], color=fold_color, lw=2)

    # Plot outer boundary in black for clarity
    for i in range(m):
        ax.plot([X[i,0], X[i+1,0]], [Y[i,0], Y[i+1,0]], color='black', lw=2)
        ax.plot([X[i,n], X[i+1,n]], [Y[i,n], Y[i+1,n]], color='black', lw=2)
    for j in range(n):
        ax.plot([X[0,j], X[0,j+1]], [Y[0,j], Y[0,j+1]], color='black', lw=2)
        ax.plot([X[m,j], X[m,j+1]], [Y[m,j], Y[m,j+1]], color='black', lw=2)

    ax.set_aspect('equal')
    ax.axis('off')
    plt.tight_layout()

    if show_plot:
        plt.show()
    if save_svg:
        fig.savefig('scaled_miura_ori.svg', format='svg', dpi=300, bbox_inches='tight')
        print('SVG file saved as "scaled_miura_ori.svg"')
    plt.close(fig)

def interactive_scaled_miura_ori():
    interactive_plot = interactive(
        plot_scaled_miura_ori,
        m=IntSlider(min=1, max=20, step=1, value=5, description='Rows (m)'),
        n=IntSlider(min=1, max=20, step=1, value=10, description='Cols (n)'),
        a=FloatSlider(min=0.5, max=5, step=0.1, value=1.0, description='a'),
        b=FloatSlider(min=0.5, max=5, step=0.1, value=1.0, description='b'),
        alpha=FloatSlider(min=0.1, max=np.pi-0.1, step=0.1, value=np.pi/3, description='alpha'),
        delta=FloatSlider(min=0.0, max=0.5, step=0.05, value=0.1, description='delta (shrink % per row)'),
        show_plot=True,
        save_svg=False
    )
    display(interactive_plot)

interactive_scaled_miura_ori()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interactive, FloatSlider, IntSlider
from IPython.display import display

def plot_scaled_miura_ori(m=5, n=10, a=1.0, b=1.0, alpha=np.pi/3, delta=0.1, show_plot=True, save_svg=False):
    # Generate per-row a_i and b_i
    a_values = a * (1 - delta)**np.arange(m+1)
    b_values = b * (1 - delta)**np.arange(m+1)
    
    # Compute vertical offsets
    vertical_offsets = np.zeros(m+1)
    for i in range(1, m+1):
        vertical_offsets[i] = vertical_offsets[i-1] + b_values[i-1]*np.sin(alpha)

    i_values = np.arange(m+1)
    j_values = np.arange(n+1)
    
    X = np.zeros((m+1, n+1))
    Y = np.zeros((m+1, n+1))

    for i in i_values:
        for j in j_values:
            base_x = j * a_values[i]
            # Odd rows shift horizontally by b_i*cos(alpha)
            if i % 2 == 1:
                base_x += b_values[i] * np.cos(alpha)
            X[i,j] = base_x
            Y[i,j] = vertical_offsets[i]

    fig, ax = plt.subplots(figsize=(8, 8))

    # Plot longitudinal folds (all one color, e.g. blue)
    for j in j_values:
        for ii in range(m):
            fold_color = 'blue' if j % 2 == 0 else 'red'
            ax.plot([X[ii,j], X[ii+1,j]], [Y[ii,j], Y[ii+1,j]], color=fold_color, lw=2)

    # Plot transversal folds (keep alternating colors)
    for i in i_values:
        for jj in range(n):
            fold_color = 'red' if (i+jj) % 2 == 0 else 'blue'
            ax.plot([X[i,jj], X[i,jj+1]], [Y[i,jj], Y[i,jj+1]], color=fold_color, lw=2)

    # Plot outer boundary in black for clarity
    for i in range(m):
        ax.plot([X[i,0], X[i+1,0]], [Y[i,0], Y[i+1,0]], color='black', lw=2)
        ax.plot([X[i,n], X[i+1,n]], [Y[i,n], Y[i+1,n]], color='black', lw=2)
    for j in range(n):
        ax.plot([X[0,j], X[0,j+1]], [Y[0,j], Y[0,j+1]], color='black', lw=2)
        ax.plot([X[m,j], X[m,j+1]], [Y[m,j], Y[m,j+1]], color='black', lw=2)

    ax.set_aspect('equal')
    ax.axis('off')
    plt.tight_layout()

    if show_plot:
        plt.show()
    if save_svg:
        fig.savefig('scaled_miura_ori.svg', format='svg', dpi=300, bbox_inches='tight')
        print('SVG file saved as "scaled_miura_ori.svg"')
    plt.close(fig)

def interactive_scaled_miura_ori():
    interactive_plot = interactive(
        plot_scaled_miura_ori,
        m=IntSlider(min=1, max=20, step=1, value=5, description='Rows (m)'),
        n=IntSlider(min=1, max=20, step=1, value=10, description='Cols (n)'),
        a=FloatSlider(min=0.5, max=5, step=0.1, value=1.0, description='a'),
        b=FloatSlider(min=0.5, max=5, step=0.1, value=1.0, description='b'),
        alpha=FloatSlider(min=0.1, max=np.pi-0.1, step=0.1, value=np.pi/3, description='alpha'),
        delta=FloatSlider(min=0.0, max=0.5, step=0.05, value=0.1, description='delta (shrink % per row)'),
        show_plot=True,
        save_svg=False
    )
    display(interactive_plot)

interactive_scaled_miura_ori()
```

```python
import numpy as np
from scipy.optimize import minimize

def compute_coordinates(m, n, a, alpha, delta):
    # Compute a_i and b_i values
    a_values = a * (1 - delta)**np.arange(m+1)
    b_values = a_values * np.sin(alpha)  # Maintaining standard Miura condition

    # Compute vertical offsets
    vertical_offsets = np.zeros(m+1)
    for i in range(1, m+1):
        vertical_offsets[i] = vertical_offsets[i-1] + b_values[i-1]*np.sin(alpha)

    i_values = np.arange(m+1)
    j_values = np.arange(n+1)
    X = np.zeros((m+1, n+1))
    Y = np.zeros((m+1, n+1))

    for i in i_values:
        for j in j_values:
            base_x = j * a_values[i]
            # Shift every other row
            if i % 2 == 1:
                base_x += b_values[i] * np.cos(alpha)
            X[i,j] = base_x
            Y[i,j] = vertical_offsets[i]

    # Centering horizontally
    top_min, top_max = X[0,:].min(), X[0,:].max()
    bottom_min, bottom_max = X[m,:].min(), X[m,:].max()
    top_center = 0.5 * (top_min + top_max)
    bottom_center = 0.5 * (bottom_min + bottom_max)
    midpoint = 0.5 * (top_center + bottom_center)
    X = X - midpoint

    return X, Y

def angle_between_vectors(u, v):
    # Compute angle between two vectors u and v in radians
    dot_product = np.dot(u, v)
    norm_u = np.linalg.norm(u)
    norm_v = np.linalg.norm(v)
    # Guard against floating-point errors
    cos_angle = np.clip(dot_product / (norm_u * norm_v), -1.0, 1.0)
    angle = np.degrees(np.arccos(cos_angle))
    return angle

def vertex_angle_deviations(m, n, a, alpha, delta):
    # Compute the coordinates
    X, Y = compute_coordinates(m, n, a, alpha, delta)

    # Collect angle deviations for all internal vertices
    deviations = []
    # Internal vertices: i in [1, m-1], j in [1, n-1]
    for i in range(1, m):
        for j in range(1, n):
            # Neighbors: up (i-1,j), right (i,j+1), down (i+1,j), left (i,j-1)
            v_up = np.array([X[i-1,j]-X[i,j], Y[i-1,j]-Y[i,j]])
            v_right = np.array([X[i,j+1]-X[i,j], Y[i,j+1]-Y[i,j]])
            v_down = np.array([X[i+1,j]-X[i,j], Y[i+1,j]-Y[i,j]])
            v_left = np.array([X[i,j-1]-X[i,j], Y[i,j-1]-Y[i,j]])

            # Order: up -> right -> down -> left -> (back to up)
            # Compute angles between consecutive vectors in clockwise order:
            θ_up_right = angle_between_vectors(v_up, v_right)
            θ_right_down = angle_between_vectors(v_right, v_down)
            θ_down_left = angle_between_vectors(v_down, v_left)
            θ_left_up = angle_between_vectors(v_left, v_up)

            # Kawasaki condition:
            # (θ_up_right + θ_down_left) should be = 180
            # (θ_right_down + θ_left_up) should be = 180
            dev1 = (θ_up_right + θ_down_left) - 180.0
            dev2 = (θ_right_down + θ_left_up) - 180.0

            deviations.append(dev1)
            deviations.append(dev2)

    return np.array(deviations)

def cost_function(params, m, n):
    a, alpha, delta = params
    dev = vertex_angle_deviations(m, n, a, alpha, delta)
    return np.sum(dev**2)

# Example usage:
m, n = 5, 10
initial_guess = [1.0, np.pi/3, 0.05]  # (a, alpha, delta)
result = minimize(cost_function, initial_guess, args=(m, n), method='Nelder-Mead')

a_opt, alpha_opt, delta_opt = result.x
print("Optimized parameters:")
print("a =", a_opt)
print("alpha =", alpha_opt)
print("delta =", delta_opt)
print("Final cost:", result.fun)
```

```python
import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interactive, FloatSlider, IntSlider
from IPython.display import display
from scipy.optimize import minimize

def angle_between_vectors(u, v):
    dot_product = np.dot(u, v)
    norm_u = np.linalg.norm(u)
    norm_v = np.linalg.norm(v)
    cos_angle = np.clip(dot_product / (norm_u * norm_v), -1.0, 1.0)
    angle = np.degrees(np.arccos(cos_angle))
    return angle

def compute_coordinates(m, n, a, alpha, delta):
    # Compute a_i and b_i values assuming b = a * sin(alpha) for a Miura parallelogram
    a_values = a * (1 - delta)**np.arange(m+1)
    b_values = a_values * np.sin(alpha)

    vertical_offsets = np.zeros(m+1)
    for i in range(1, m+1):
        vertical_offsets[i] = vertical_offsets[i-1] + b_values[i-1]*np.sin(alpha)

    i_values = np.arange(m+1)
    j_values = np.arange(n+1)
    X = np.zeros((m+1, n+1))
    Y = np.zeros((m+1, n+1))

    for i in i_values:
        for j in j_values:
            base_x = j * a_values[i]
            if i % 2 == 1:
                base_x += b_values[i] * np.cos(alpha)
            X[i,j] = base_x
            Y[i,j] = vertical_offsets[i]

    top_min, top_max = X[0,:].min(), X[0,:].max()
    bottom_min, bottom_max = X[m,:].min(), X[m,:].max()
    top_center = 0.5 * (top_min + top_max)
    bottom_center = 0.5 * (bottom_min + bottom_max)
    midpoint = 0.5 * (top_center + bottom_center)
    X = X - midpoint

    return X, Y

def vertex_angle_deviations(m, n, a, alpha, delta):
    X, Y = compute_coordinates(m, n, a, alpha, delta)

    deviations = []
    for i in range(1, m):
        for j in range(1, n):
            v_up = np.array([X[i-1,j]-X[i,j], Y[i-1,j]-Y[i,j]])
            v_right = np.array([X[i,j+1]-X[i,j], Y[i,j+1]-Y[i,j]])
            v_down = np.array([X[i+1,j]-X[i,j], Y[i+1,j]-Y[i,j]])
            v_left = np.array([X[i,j-1]-X[i,j], Y[i,j-1]-Y[i,j]])

            θ_up_right = angle_between_vectors(v_up, v_right)
            θ_right_down = angle_between_vectors(v_right, v_down)
            θ_down_left = angle_between_vectors(v_down, v_left)
            θ_left_up = angle_between_vectors(v_left, v_up)

            dev1 = (θ_up_right + θ_down_left) - 180.0
            dev2 = (θ_right_down + θ_left_up) - 180.0

            deviations.append(dev1)
            deviations.append(dev2)

    return np.array(deviations)

def cost_function(a, m, n, alpha, delta):
    dev = vertex_angle_deviations(m, n, a, alpha, delta)
    return np.sum(dev**2)

def plot_scaled_miura_ori(m=5, n=10, a=1.0, b=1.0, alpha=np.pi/3, delta=0.1, show_plot=True, save_svg=False):
    # Optimize only over 'a' to maintain alpha and delta as specified
    result = minimize(cost_function, [a], args=(m, n, alpha, delta), method='Nelder-Mead')
    a_opt = result.x[0]

    i_values = np.arange(m+1)
    a_values = a_opt * (1 - (i_values/m)*delta)
    # Keep b_values tied to a_opt and alpha (Miura condition)
    b_values = (a_opt * np.sin(alpha)) * (1 - (i_values/m)*delta)

    vertical_offsets = np.zeros(m+1)
    for i in range(1, m+1):
        vertical_offsets[i] = vertical_offsets[i-1] + b_values[i-1]*np.sin(alpha)

    j_values = np.arange(n+1)
    X = np.zeros((m+1, n+1))
    Y = np.zeros((m+1, n+1))

    for i in i_values:
        for j in j_values:
            offset_factor = i * b_values[i] * np.cos(alpha) * delta
            if i % 2 == 0:
                base_x = j * a_values[i] + 0.5 * offset_factor
            else:
                base_x = j * a_values[i] + offset_factor
            X[i,j] = base_x
            Y[i,j] = vertical_offsets[i]

    top_min, top_max = X[0,:].min(), X[0,:].max()
    bottom_min, bottom_max = X[m,:].min(), X[m,:].max()
    top_center = 0.5 * (top_min + top_max)
    bottom_center = 0.5 * (bottom_min + bottom_max)
    midpoint = 0.5 * (top_center + bottom_center)
    X = X - midpoint

    fig, ax = plt.subplots(figsize=(8, 8))

    # Longitudinal folds
    for j in j_values:
        for ii in range(m):
            fold_color = 'blue' if j % 2 == 0 else 'red'
            ax.plot([X[ii,j], X[ii+1,j]], [Y[ii,j], Y[ii+1,j]], color=fold_color, lw=2)

    # Transversal folds
    for i in i_values:
        for jj in range(n):
            fold_color = 'red' if (i+jj) % 2 == 0 else 'blue'
            ax.plot([X[i,jj], X[i,jj+1]], [Y[i,jj], Y[i,jj+1]], color=fold_color, lw=2)

    # Outer boundary
    for i in range(m):
        ax.plot([X[i,0], X[i+1,0]], [Y[i,0], Y[i+1,0]], color='black', lw=2)
        ax.plot([X[i,n], X[i+1,n]], [Y[i,n], Y[i+1,n]], color='black', lw=2)
    for j in range(n):
        ax.plot([X[0,j], X[0,j+1]], [Y[0,j], Y[0,j+1]], color='black', lw=2)
        ax.plot([X[m,j], X[m,j+1]], [Y[m,j], Y[m,j+1]], color='black', lw=2)

    ax.set_aspect('equal')
    ax.axis('off')
    plt.tight_layout()

    if show_plot:
        plt.show()
    if save_svg:
        fig.savefig('scaled_miura_ori.svg', format='svg', dpi=300, bbox_inches='tight')
        print('SVG file saved as "scaled_miura_ori.svg"')
    plt.close(fig)

def interactive_scaled_miura_ori():
    interactive_plot = interactive(
        plot_scaled_miura_ori,
        m=IntSlider(min=1, max=20, step=1, value=5, description='Rows (m)'),
        n=IntSlider(min=1, max=20, step=1, value=10, description='Cols (n)'),
        a=FloatSlider(min=0.5, max=5, step=0.1, value=1.0, description='a'),
        b=FloatSlider(min=0.5, max=5, step=0.1, value=1.0, description='b'),
        alpha=FloatSlider(min=0.1, max=np.pi-0.1, step=0.1, value=np.pi/3, description='alpha'),
        delta=FloatSlider(min=0.0, max=0.5, step=0.05, value=0.1, description='delta (shrink % per row)'),
        show_plot=True,
        save_svg=False
    )
    display(interactive_plot)

interactive_scaled_miura_ori()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interactive, FloatSlider, IntSlider
from IPython.display import display
from scipy.optimize import minimize

def angle_between_vectors(u, v):
    dot_product = np.dot(u, v)
    norm_u = np.linalg.norm(u)
    norm_v = np.linalg.norm(v)
    cos_angle = np.clip(dot_product / (norm_u * norm_v), -1.0, 1.0)
    angle = np.degrees(np.arccos(cos_angle))
    return angle

def compute_coordinates_with_params(m, n, a, alpha, delta, params):
    # a_values and b_values based on alpha and delta
    i_values = np.arange(m+1)
    a_values = a*(1 - (i_values/m)*delta)
    b_values = (a * np.sin(alpha))*(1 - (i_values/m)*delta)

    # Compute vertical offsets
    vertical_offsets = np.zeros(m+1)
    for i in range(1, m+1):
        vertical_offsets[i] = vertical_offsets[i-1] + b_values[i-1]*np.sin(alpha)

    j_values = np.arange(n+1)
    X = np.zeros((m+1, n+1))
    Y = np.zeros((m+1, n+1))

    # Incorporate row offsets: here params[i] is a horizontal shift for row i
    for i in i_values:
        for j in j_values:
            offset_factor = i * b_values[i] * np.cos(alpha) * delta
            # Apply base geometry
            if i % 2 == 0:
                base_x = j * a_values[i] + 0.5 * offset_factor
            else:
                base_x = j * a_values[i] + offset_factor

            # Add the optimization-derived offset params[i] to shift row i horizontally
            X[i,j] = base_x + params[i]  
            Y[i,j] = vertical_offsets[i]

    # Centering horizontally as before
    top_min, top_max = X[0,:].min(), X[0,:].max()
    bottom_min, bottom_max = X[m,:].min(), X[m,:].max()
    top_center = 0.5 * (top_min + top_max)
    bottom_center = 0.5 * (bottom_min + bottom_max)
    midpoint = 0.5 * (top_center + bottom_center)
    X = X - midpoint

    return X, Y

def vertex_angle_deviations_with_params(m, n, a, alpha, delta, params):
    X, Y = compute_coordinates_with_params(m, n, a, alpha, delta, params)

    deviations = []
    for i in range(1, m):
        for j in range(1, n):
            v_up = np.array([X[i-1,j]-X[i,j], Y[i-1,j]-Y[i,j]])
            v_right = np.array([X[i,j+1]-X[i,j], Y[i,j+1]-Y[i,j]])
            v_down = np.array([X[i+1,j]-X[i,j], Y[i+1,j]-Y[i,j]])
            v_left = np.array([X[i,j-1]-X[i,j], Y[i,j-1]-Y[i,j]])

            θ_up_right = angle_between_vectors(v_up, v_right)
            θ_right_down = angle_between_vectors(v_right, v_down)
            θ_down_left = angle_between_vectors(v_down, v_left)
            θ_left_up = angle_between_vectors(v_left, v_up)

            dev1 = (θ_up_right + θ_down_left) - 180.0
            dev2 = (θ_right_down + θ_left_up) - 180.0

            deviations.append(dev1)
            deviations.append(dev2)

    return np.array(deviations)

def angle_deviation_cost(params, m, n, a, alpha, delta):
    dev = vertex_angle_deviations_with_params(m, n, a, alpha, delta, params)
    return np.sum(dev**2)

def plot_scaled_miura_ori(m=5, n=10, a=1.0, b=1.0, alpha=np.pi/3, delta=0.1, show_plot=True, save_svg=False):
    # We'll optimize the offsets (one per row)
    initial_guess = np.zeros(m+1)
    result = minimize(angle_deviation_cost, initial_guess, args=(m, n, a, alpha, delta), method='Nelder-Mead')
    params_opt = result.x

    # Compute final coordinates with optimized parameters
    X, Y = compute_coordinates_with_params(m, n, a, alpha, delta, params_opt)
    i_values = np.arange(m+1)
    j_values = np.arange(n+1)

    fig, ax = plt.subplots(figsize=(8, 8))

    # Longitudinal folds
    for j in j_values:
        for ii in range(m):
            fold_color = 'blue' if j % 2 == 0 else 'red'
            ax.plot([X[ii,j], X[ii+1,j]], [Y[ii,j], Y[ii+1,j]], color=fold_color, lw=2)

    # Transversal folds
    for i in i_values:
        for jj in range(n):
            fold_color = 'blue' if (i+jj) % 2 == 0 else 'red'
            ax.plot([X[i,jj], X[i,jj+1]], [Y[i,jj], Y[i,jj+1]], color=fold_color, lw=2)

    # Outer boundary
    for i in range(m):
        ax.plot([X[i,0], X[i+1,0]], [Y[i,0], Y[i+1,0]], color='black', lw=2)
        ax.plot([X[i,n], X[i+1,n]], [Y[i,n], Y[i+1,n]], color='black', lw=2)
    for j in range(n):
        ax.plot([X[0,j], X[0,j+1]], [Y[0,j], Y[0,j+1]], color='black', lw=2)
        ax.plot([X[m,j], X[m,j+1]], [Y[m,j], Y[m,j+1]], color='black', lw=2)

#     ax.set_aspect('equal')
    ax.axis('off')
    plt.tight_layout()

    if show_plot:
        plt.show()
    if save_svg:
        fig.savefig(f'scaled_miura_ori_{m}_{n}_{a}_{b}_{alpha}_{delta}.svg', format='svg', dpi=300, bbox_inches='tight')
        print('SVG file saved as "scaled_miura_ori.svg"')
    plt.close(fig)

def interactive_scaled_miura_ori():
    interactive_plot = interactive(
        plot_scaled_miura_ori,
        m=IntSlider(min=1, max=20, step=1, value=7, description='Rows (m)'),
        n=IntSlider(min=1, max=20, step=1, value=4, description='Cols (n)'),
        a=FloatSlider(min=0.5, max=5, step=0.1, value=2.0, description='a'),
        b=FloatSlider(min=0.5, max=5, step=0.1, value=2.10, description='b'), # 'b' is retained from original UI but not directly used in optimization here
        alpha=FloatSlider(min=0.1, max=np.pi-0.1, step=0.1, value=1.50, description='alpha'),
        delta=FloatSlider(min=0.0, max=0.5, step=0.05, value=0.15, description='delta (shrink % per row)'),
        show_plot=True,
        save_svg=True
    )
    display(interactive_plot)

interactive_scaled_miura_ori()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interactive, FloatSlider, IntSlider
from IPython.display import display
from scipy.optimize import minimize

def angle_between_vectors(u, v):
    dot_product = np.dot(u, v)
    norm_u = np.linalg.norm(u)
    norm_v = np.linalg.norm(v)
    cos_angle = np.clip(dot_product / (norm_u * norm_v), -1.0, 1.0)
    angle = np.degrees(np.arccos(cos_angle))
    return angle

def compute_coordinates_with_params(m, n, a, alpha, delta, params):
    # Same extraction of p_x, p_y
    p_x = params[:m+1]
    p_y = params[m+1:]
    
    i_values = np.arange(m+1)
    a_values = a*(1 - (i_values/m)*delta)
    b_values = (a * np.sin(alpha))*(1 - (i_values/m)*delta)

    vertical_offsets = np.zeros(m+1)
    for i in range(1, m+1):
        vertical_offsets[i] = vertical_offsets[i-1] + b_values[i-1]*np.sin(alpha)

    X = np.zeros((m+1, n+1))
    Y = np.zeros((m+1, n+1))
    j_values = np.arange(n+1)

    # Example zigzag baseline using alpha:
    # For each row i:
    # - If i is even, row shifts slightly to the left by b_i*cos(alpha)
    # - If i is odd, row shifts to the right by b_i*cos(alpha)
    # You can scale these shifts by delta or another factor.
    for i in i_values:
        zigzag_shift = ((-1)**i) * b_values[i] * np.cos(alpha)

        for j in j_values:
            base_x = j * a_values[i] + zigzag_shift
            # Add the offsets found by the solver
            X[i,j] = base_x + p_x[i]
            Y[i,j] = vertical_offsets[i] + p_y[i]

    # Horizontal centering as before
    top_min, top_max = X[0,:].min(), X[0,:].max()
    bottom_min, bottom_max = X[m,:].min(), X[m,:].max()
    top_center = 0.5 * (top_min + top_max)
    bottom_center = 0.5 * (bottom_min + bottom_max)
    midpoint = 0.5 * (top_center + bottom_center)
    X = X - midpoint

    return X, Y

def vertex_angle_deviations_with_params(m, n, a, alpha, delta, params):
    X, Y = compute_coordinates_with_params(m, n, a, alpha, delta, params)

    deviations = []
    for i in range(1, m):
        for j in range(1, n):
            v_up = np.array([X[i-1,j]-X[i,j], Y[i-1,j]-Y[i,j]])
            v_right = np.array([X[i,j+1]-X[i,j], Y[i,j+1]-Y[i,j]])
            v_down = np.array([X[i+1,j]-X[i,j], Y[i+1,j]-Y[i,j]])
            v_left = np.array([X[i,j-1]-X[i,j], Y[i,j-1]-Y[i,j]])

            θ_up_right = angle_between_vectors(v_up, v_right)
            θ_right_down = angle_between_vectors(v_right, v_down)
            θ_down_left = angle_between_vectors(v_down, v_left)
            θ_left_up = angle_between_vectors(v_left, v_up)

            dev1 = (θ_up_right + θ_down_left) - 180.0
            dev2 = (θ_right_down + θ_left_up) - 180.0

            deviations.append(dev1)
            deviations.append(dev2)

    return np.array(deviations)

def angle_deviation_cost(params, m, n, a, alpha, delta):
    dev = vertex_angle_deviations_with_params(m, n, a, alpha, delta, params)
    return np.sum(dev**2)

def plot_scaled_miura_ori(m=5, n=10, a=1.0, b=1.0, alpha=np.pi/3, delta=0.1, show_plot=True, save_svg=False):
    # We'll now optimize both horizontal and vertical offsets (2*(m+1) parameters)
    initial_guess = np.zeros((m+1)*2)  # first m+1 for horizontal offsets, next m+1 for vertical
    result = minimize(angle_deviation_cost, initial_guess, args=(m, n, a, alpha, delta), method='Nelder-Mead')
    params_opt = result.x

    # Compute final coordinates with optimized parameters
    X, Y = compute_coordinates_with_params(m, n, a, alpha, delta, params_opt)
    i_values = np.arange(m+1)
    j_values = np.arange(n+1)

    fig, ax = plt.subplots(figsize=(8, 8))

    # Longitudinal folds
    for j in j_values:
        for ii in range(m):
            fold_color = 'blue' if j % 2 == 0 else 'red'
            ax.plot([X[ii,j], X[ii+1,j]], [Y[ii,j], Y[ii+1,j]], color=fold_color, lw=2)

    # Transversal folds
    for i in i_values:
        for jj in range(n):
            fold_color = 'red' if (i+jj) % 2 == 0 else 'blue'
            ax.plot([X[i,jj], X[i,jj+1]], [Y[i,jj], Y[i,jj+1]], color=fold_color, lw=2)

    # Outer boundary
    for i in range(m):
        ax.plot([X[i,0], X[i+1,0]], [Y[i,0], Y[i+1,0]], color='black', lw=2)
        ax.plot([X[i,n], X[i+1,n]], [Y[i,n], Y[i+1,n]], color='black', lw=2)
    for j in range(n):
        ax.plot([X[0,j], X[0,j+1]], [Y[0,j], Y[0,j+1]], color='black', lw=2)
        ax.plot([X[m,j], X[m,j+1]], [Y[m,j], Y[m,j+1]], color='black', lw=2)

    ax.set_aspect('equal')
    ax.axis('off')
    plt.tight_layout()

    if show_plot:
        plt.show()
    if save_svg:
        fig.savefig('scaled_miura_ori.svg', format='svg', dpi=300, bbox_inches='tight')
        print('SVG file saved as "scaled_miura_ori.svg"')
    plt.close(fig)

def interactive_scaled_miura_ori():
    interactive_plot = interactive(
        plot_scaled_miura_ori,
        m=IntSlider(min=1, max=20, step=1, value=5, description='Rows (m)'),
        n=IntSlider(min=1, max=20, step=1, value=10, description='Cols (n)'),
        a=FloatSlider(min=0.5, max=5, step=0.1, value=1.0, description='a'),
        b=FloatSlider(min=0.5, max=5, step=0.1, value=1.0, description='b'),
        alpha=FloatSlider(min=0.1, max=np.pi-0.1, step=0.1, value=np.pi/3, description='alpha'),
        delta=FloatSlider(min=0.0, max=0.5, step=0.05, value=0.1, description='delta'),
        show_plot=True,
        save_svg=False
    )
    display(interactive_plot)

interactive_scaled_miura_ori()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interactive, FloatSlider, IntSlider
from IPython.display import display
from scipy.optimize import minimize

def angle_between_vectors(u, v):
    dot_product = np.dot(u, v)
    norm_u = np.linalg.norm(u)
    norm_v = np.linalg.norm(v)
    cos_angle = np.clip(dot_product / (norm_u * norm_v), -1.0, 1.0)
    angle = np.degrees(np.arccos(cos_angle))
    return angle

def compute_coordinates_with_params(m, n, a, alpha, delta, params):
    # Extract horizontal (p_x) and vertical (p_y) offsets from params
    p_x = params[:m+1]
    p_y = params[m+1:]

    i_values = np.arange(m+1)
    j_values = np.arange(n+1)

    # Base values
    a_values = a*(1 - (i_values/m)*delta)
    b_values = (a * np.sin(alpha))*(1 - (i_values/m)*delta)

    # Vertical offsets from the standard Miura baseline
    vertical_offsets = np.zeros(m+1)
    for i in range(1, m+1):
        vertical_offsets[i] = vertical_offsets[i-1] + b_values[i-1]*np.sin(alpha)

    X = np.zeros((m+1, n+1))
    Y = np.zeros((m+1, n+1))

    # Introduce a zigzag pattern depending on i and j
    # We use (-1)^(i+j) to alternate sign per vertex and incorporate alpha
    # Example: horizontal zigzag scaled by b_values[i]*cos(alpha),
    # and a smaller vertical modulation as well.
    #
    # Adjust these functions as needed for more/less eccentricity.
    for i in i_values:
        for j in j_values:
            # Baseline horizontal (from Miura)
            base_x = j * a_values[i]
            # Add a zigzag pattern in horizontal direction:
            # For example, alternate every vertex in a checkerboard pattern:
            zigzag_x = ((-1)**(i+j)) * b_values[i] * np.cos(alpha) * 0.5
            # Similarly, add a small vertical zigzag:
            zigzag_y = ((-1)**(j)) * b_values[i] * np.sin(alpha) * 0.1

            # Combine baseline and zigzag
            X[i,j] = base_x + zigzag_x + p_x[i]
            Y[i,j] = vertical_offsets[i] + zigzag_y + p_y[i]

    # Centering horizontally
    top_min, top_max = X[0,:].min(), X[0,:].max()
    bottom_min, bottom_max = X[m,:].min(), X[m,:].max()
    top_center = 0.5 * (top_min + top_max)
    bottom_center = 0.5 * (bottom_min + bottom_max)
    midpoint = 0.5 * (top_center + bottom_center)
    X = X - midpoint

    return X, Y

def vertex_angle_deviations_with_params(m, n, a, alpha, delta, params):
    X, Y = compute_coordinates_with_params(m, n, a, alpha, delta, params)

    deviations = []
    for i in range(1, m):
        for j in range(1, n):
            v_up = np.array([X[i-1,j]-X[i,j], Y[i-1,j]-Y[i,j]])
            v_right = np.array([X[i,j+1]-X[i,j], Y[i,j+1]-Y[i,j]])
            v_down = np.array([X[i+1,j]-X[i,j], Y[i+1,j]-Y[i,j]])
            v_left = np.array([X[i,j-1]-X[i,j], Y[i,j-1]-Y[i,j]])

            θ_up_right = angle_between_vectors(v_up, v_right)
            θ_right_down = angle_between_vectors(v_right, v_down)
            θ_down_left = angle_between_vectors(v_down, v_left)
            θ_left_up = angle_between_vectors(v_left, v_up)

            dev1 = (θ_up_right + θ_down_left) - 180.0
            dev2 = (θ_right_down + θ_left_up) - 180.0

            deviations.append(dev1)
            deviations.append(dev2)
    return np.array(deviations)

def angle_deviation_cost(params, m, n, a, alpha, delta):
    dev = vertex_angle_deviations_with_params(m, n, a, alpha, delta, params)
    return np.sum(dev**2)

def plot_scaled_miura_ori(m=5, n=10, a=1.0, b=1.0, alpha=np.pi/3, delta=0.1, show_plot=True, save_svg=False):
    # Optimize horizontal and vertical offsets (2*(m+1) params)
    initial_guess = np.zeros((m+1)*2)
    result = minimize(angle_deviation_cost, initial_guess, args=(m, n, a, alpha, delta), method='Nelder-Mead')
    params_opt = result.x

    X, Y = compute_coordinates_with_params(m, n, a, alpha, delta, params_opt)
    i_values = np.arange(m+1)
    j_values = np.arange(n+1)

    fig, ax = plt.subplots(figsize=(8, 8))

    # Longitudinal folds
    for j in j_values:
        for ii in range(m):
            fold_color = 'blue' if j % 2 == 0 else 'red'
            ax.plot([X[ii,j], X[ii+1,j]], [Y[ii,j], Y[ii+1,j]], color=fold_color, lw=2)

    # Transversal folds
    for i in i_values:
        for jj in range(n):
            fold_color = 'blue' if (i+jj) % 2 == 0 else 'red'
            ax.plot([X[i,jj], X[i,jj+1]], [Y[i,jj], Y[i,jj+1]], color=fold_color, lw=2)

    # Outer boundary
    for i in range(m):
        ax.plot([X[i,0], X[i+1,0]], [Y[i,0], Y[i+1,0]], color='black', lw=2)
        ax.plot([X[i,n], X[i+1,n]], [Y[i,n], Y[i+1,n]], color='black', lw=2)
    for j in range(n):
        ax.plot([X[0,j], X[0,j+1]], [Y[0,j], Y[0,j+1]], color='black', lw=2)
        ax.plot([X[m,j], X[m,j+1]], [Y[m,j], Y[m,j+1]], color='black', lw=2)

    ax.set_aspect('equal')
    ax.axis('off')
    plt.tight_layout()

    if show_plot:
        plt.show()
    if save_svg:
        fig.savefig('scaled_miura_ori_asd.svg', format='svg', dpi=300, bbox_inches='tight')
        print('SVG file saved as "scaled_miura_ori.svg"')
    plt.close(fig)

def interactive_scaled_miura_ori():
    interactive_plot = interactive(
        plot_scaled_miura_ori,
        m=IntSlider(min=1, max=20, step=1, value=5, description='Rows (m)'),
        n=IntSlider(min=1, max=20, step=1, value=4, description='Cols (n)'),
        a=FloatSlider(min=0.5, max=5, step=0.1, value=4.9, description='a'),
        b=FloatSlider(min=0.5, max=5, step=0.1, value=1.0, description='b'),
        alpha=FloatSlider(min=0.1, max=np.pi-0.1, step=0.9, value=np.pi/3, description='alpha'),
        delta=FloatSlider(min=0.0, max=0.5, step=0.05, value=0.05, description='delta (shrink % per row)'),
        show_plot=True,
        save_svg=True
    )
    display(interactive_plot)

interactive_scaled_miura_ori()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interactive, FloatSlider, IntSlider, Checkbox
from IPython.display import display

def compute_coordinates(m, n, a, alpha, delta):
    """
    Compute the X and Y coordinates of the Miura-ori pattern with scaling.

    Parameters:
    - m (int): Number of rows.
    - n (int): Number of columns.
    - a (float): Initial horizontal side length.
    - alpha (float): Folding angle in radians.
    - delta (float): Scaling factor per row (0 <= delta <= 1).

    Returns:
    - X (2D numpy array): X coordinates of the grid.
    - Y (2D numpy array): Y coordinates of the grid.
    """
    i_values = np.arange(m+1)
    j_values = np.arange(n+1)

    # Compute scaling factors for each row
    # a_i = a * (1 - delta * (i/m))
    a_values = a * (1 - delta * (i_values / m))
    
    # Ensure no negative or zero lengths
    a_values = np.maximum(a_values, 1e-3)
    
    # Compute b_i based on alpha to preserve angles
    # b_i = a_i * tan(alpha)
    b_values = a_values * np.tan(alpha)
    
    # Compute vertical offsets for each row
    vertical_offsets = np.zeros(m+1)
    for i in range(1, m+1):
        vertical_offsets[i] = vertical_offsets[i-1] + b_values[i-1] * np.sin(alpha)
    
    # Initialize coordinate grids
    X = np.zeros((m+1, n+1))
    Y = np.zeros((m+1, n+1))
    
    for i in i_values:
        for j in j_values:
            base_x = j * a_values[i]
            # Shift every other row by b_i * cos(alpha) to create the Miura-ori pattern
            if i % 2 == 1:
                base_x += b_values[i] * np.cos(alpha)
            X[i, j] = base_x
            Y[i, j] = vertical_offsets[i]
    
    # Center the pattern horizontally
    top_center = (X[0, :].min() + X[0, :].max()) / 2
    bottom_center = (X[m, :].min() + X[m, :].max()) / 2
    midpoint = (top_center + bottom_center) / 2
    X -= midpoint
    
    return X, Y

def plot_miura_ori(m=5, n=10, a=1.0, alpha=np.pi/3, delta=0.1, show_plot=True, save_svg=False):
    """
    Plot the scaled Miura-ori pattern ensuring flat foldability.

    Parameters:
    - m (int): Number of rows.
    - n (int): Number of columns.
    - a (float): Initial horizontal side length.
    - alpha (float): Folding angle in radians.
    - delta (float): Scaling factor per row (0 <= delta <= 1).
    - show_plot (bool): Whether to display the plot.
    - save_svg (bool): Whether to save the plot as an SVG file.
    """
    # Compute coordinates
    X, Y = compute_coordinates(m, n, a, alpha, delta)
    
    i_values = np.arange(m+1)
    j_values = np.arange(n+1)
    
    # Initialize plot
    fig, ax = plt.subplots(figsize=(8, 8))
    
    # Plot longitudinal folds (vertical lines)
    for j in j_values:
        for i in range(m):
            fold_color = 'blue' if j % 2 == 0 else 'red'
            ax.plot([X[i, j], X[i+1, j]], [Y[i, j], Y[i+1, j]], color=fold_color, lw=2)
    
    # Plot transversal folds (horizontal lines)
    for i in i_values:
        for j in range(n):
            fold_color = 'red' if (i + j) % 2 == 0 else 'blue'
            ax.plot([X[i, j], X[i, j+1]], [Y[i, j], Y[i, j+1]], color=fold_color, lw=2)
    
    # Plot outer boundaries
    for i in range(m):
        # Left boundary
        ax.plot([X[i, 0], X[i+1, 0]], [Y[i, 0], Y[i+1, 0]], color='black', lw=2)
        # Right boundary
        ax.plot([X[i, n], X[i+1, n]], [Y[i, n], Y[i+1, n]], color='black', lw=2)
    for j in range(n):
        # Top boundary
        ax.plot([X[0, j], X[0, j+1]], [Y[0, j], Y[0, j+1]], color='black', lw=2)
        # Bottom boundary
        ax.plot([X[m, j], X[m, j+1]], [Y[m, j], Y[m, j+1]], color='black', lw=2)
    
    # Final plot adjustments
    ax.set_aspect('equal')
    ax.axis('off')
    plt.tight_layout()
    
    if show_plot:
        plt.show()
    if save_svg:
        fig.savefig('scaled_miura_ori.svg', format='svg', dpi=300, bbox_inches='tight')
        print('SVG file saved as "scaled_miura_ori.svg"')
    plt.close(fig)

def interactive_scaled_miura_ori():
    """
    Create an interactive widget to adjust Miura-ori parameters and visualize the pattern.
    """
    interactive_plot = interactive(
        plot_miura_ori,
        m=IntSlider(min=1, max=20, step=1, value=5, description='Rows (m)'),
        n=IntSlider(min=1, max=20, step=1, value=4, description='Cols (n)'),
        a=FloatSlider(min=0.5, max=5.0, step=0.1, value=1.0, description='a'),
        alpha=FloatSlider(min=0.1, max=np.pi-0.1, step=0.05, value=1.05, description='alpha (rad)'),
        delta=FloatSlider(min=0.0, max=0.5, step=0.05, value=0.25, description='delta'),
        show_plot=Checkbox(value=True, description='Show Plot'),
        save_svg=Checkbox(value=False, description='Save SVG')
    )
    display(interactive_plot)

# Run the interactive plot
interactive_scaled_miura_ori()
```

```python
import numpy as np
import pandas as pd

def generate_flat_foldable_data(m=5, n=4, num_samples=10000):
    """
    Generate a dataset of Miura-ori patterns ensuring flat foldability.
    
    Parameters:
    - m (int): Number of rows.
    - n (int): Number of columns.
    - num_samples (int): Number of samples to generate.
    
    Returns:
    - DataFrame: Contains parameters and corresponding y-coordinates.
    """
    data = []
    
    for _ in range(num_samples):
        # Randomly sample parameters within specified ranges
        a = np.random.uniform(0.5, 5.0)
        alpha = np.random.uniform(0.1, np.pi - 0.1)
        delta = np.random.uniform(0.0, 0.5)
        
        # Compute a_i and b_i
        i_values = np.arange(m+1)
        a_values = a * (1 - delta * (i_values / m))
        a_values = np.maximum(a_values, 1e-3)  # Prevent zero or negative lengths
        b_values = a_values * np.tan(alpha)
        
        # Compute y-coordinates
        y_offsets = [0.0]
        for i in range(1, m+1):
            y_new = y_offsets[i-1] + b_values[i-1] * np.sin(alpha)
            y_offsets.append(y_new)
        
        # Flatten y_offsets for the dataset
        y_coords = y_offsets
        
        # Append to data
        sample = {
            'a': a,
            'alpha': alpha,
            'delta': delta,
            'y0': y_coords[0],
            'y1': y_coords[1],
            'y2': y_coords[2],
            'y3': y_coords[3],
            'y4': y_coords[4],
            'y5': y_coords[5],
        }
        data.append(sample)
    
    df = pd.DataFrame(data)
    return df
```

```python
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split

def build_model():
    """
    Build and compile the neural network model.
    
    Returns:
    - model (keras.Model): Compiled neural network model.
    """
    model = keras.Sequential([
        layers.Input(shape=(3,)),  # Inputs: a, alpha, delta
        layers.Dense(64, activation='relu'),
        layers.Dense(64, activation='relu'),
        layers.Dense(6)  # Outputs: y0 to y5
    ])
    
    model.compile(optimizer='adam', loss='mse', metrics=['mae'])
    return model

# Generate dataset
df = generate_flat_foldable_data(m=5, n=4, num_samples=100000)

# Split into inputs and outputs
X = df[['a', 'alpha', 'delta']].values
y = df[['y0', 'y1', 'y2', 'y3', 'y4', 'y5']].values

# Normalize inputs and outputs for better training
scaler_X = StandardScaler()
scaler_y = StandardScaler()

X_scaled = scaler_X.fit_transform(X)
y_scaled = scaler_y.fit_transform(y)

# Split into training and validation sets
X_train, X_val, y_train, y_val = train_test_split(X_scaled, y_scaled, test_size=0.3, random_state=42)

# Build and train the model
model = build_model()

history = model.fit(
    X_train, y_train,
    epochs=100,
    batch_size=32,
    validation_data=(X_val, y_val),
    verbose=1
)
```

```python
def predict_y_coords(a, alpha, delta, model, scaler_X, scaler_y):
    """
    Predict y-coordinates using the trained model.
    
    Parameters:
    - a (float): Initial horizontal side length.
    - alpha (float): Folding angle in radians.
    - delta (float): Scaling factor per row.
    - model (keras.Model): Trained Keras model.
    - scaler_X (StandardScaler): Scaler for inputs.
    - scaler_y (StandardScaler): Scaler for outputs.
    
    Returns:
    - y_coords (list): Predicted y-coordinates [y0, y1, ..., ym]
    """
    input_data = np.array([[a, alpha, delta]])
    input_scaled = scaler_X.transform(input_data)
    y_pred_scaled = model.predict(input_scaled)
    y_pred = scaler_y.inverse_transform(y_pred_scaled)
    y_coords = y_pred[0].tolist()
    return y_coords
```

```python
def plot_scaled_miura_ori_with_model(m=5, n=4, a=1.0, alpha=1.05, delta=0.25, model=None, scaler_X=None, scaler_y=None, show_plot=True, save_svg=False):
    """
    Plot the scaled Miura-ori pattern using predicted y-coordinates from the model.
    
    Parameters:
    - m (int): Number of rows.
    - n (int): Number of columns.
    - a (float): Initial horizontal side length.
    - alpha (float): Folding angle in radians.
    - delta (float): Scaling factor per row.
    - model (keras.Model): Trained Keras model.
    - scaler_X (StandardScaler): Scaler for inputs.
    - scaler_y (StandardScaler): Scaler for outputs.
    - show_plot (bool): Whether to display the plot.
    - save_svg (bool): Whether to save the plot as an SVG file.
    """
    if model is None or scaler_X is None or scaler_y is None:
        print("Model and scalers must be provided.")
        return
    
    # Predict y-coordinates using the model
    y_coords = predict_y_coords(a, alpha, delta, model, scaler_X, scaler_y)
    
    # Compute a_i and b_i
    i_values = np.arange(m+1)
    a_values = a * (1 - delta * (i_values / m))
    a_values = np.maximum(a_values, 1e-3)  # Prevent zero or negative lengths
    b_values = a_values * np.tan(alpha)
    
    # Initialize coordinate grids
    X = np.zeros((m+1, n+1))
    Y = np.zeros((m+1, n+1))
    
    for i in i_values:
        for j in range(n+1):
            base_x = j * a_values[i]
            # Shift every other row by b_i * cos(alpha) to create the Miura-ori pattern
            if i % 2 == 1:
                base_x += b_values[i] * np.cos(alpha)
            X[i, j] = base_x
            Y[i, j] = y_coords[i]
    
    # Center the pattern horizontally
    top_center = (X[0, :].min() + X[0, :].max()) / 2
    bottom_center = (X[m, :].min() + X[m, :].max()) / 2
    midpoint = (top_center + bottom_center) / 2
    X -= midpoint
    
    # Initialize plot
    fig, ax = plt.subplots(figsize=(8, 8))
    
    # Plot longitudinal folds (vertical lines)
    for j in range(n+1):
        for i in range(m):
            fold_color = 'blue' if j % 2 == 0 else 'red'
            ax.plot([X[i, j], X[i+1, j]], [Y[i, j], Y[i+1, j]], color=fold_color, lw=2)
    
    # Plot transversal folds (horizontal lines)
    for i in i_values:
        for j in range(n):
            fold_color = 'red' if (i + j) % 2 == 0 else 'blue'
            ax.plot([X[i, j], X[i, j+1]], [Y[i, j], Y[i, j+1]], color=fold_color, lw=2)
    
    # Plot outer boundaries
    for i in range(m):
        # Left boundary
        ax.plot([X[i, 0], X[i+1, 0]], [Y[i, 0], Y[i+1, 0]], color='black', lw=2)
        # Right boundary
        ax.plot([X[i, n], X[i+1, n]], [Y[i, n], Y[i+1, n]], color='black', lw=2)
    for j in range(n):
        # Top boundary
        ax.plot([X[0, j], X[0, j+1]], [Y[0, j], Y[0, j+1]], color='black', lw=2)
        # Bottom boundary
        ax.plot([X[m, j], X[m, j+1]], [Y[m, j], Y[m, j+1]], color='black', lw=2)
    
    # Final plot adjustments
    ax.set_aspect('equal')
    ax.axis('off')
    plt.tight_layout()
    
    if show_plot:
        plt.show()
    if save_svg:
        fig.savefig('scaled_miura_ori.svg', format='svg', dpi=300, bbox_inches='tight')
        print('SVG file saved as "scaled_miura_ori.svg"')
    plt.close(fig)
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import minimize
from ipywidgets import interactive, FloatSlider, IntSlider, Checkbox
from IPython.display import display

def compute_coordinates(m, n, a, alpha, delta, y_coords):
    """
    Compute the X and Y coordinates of the Miura-ori pattern.

    Parameters:
    - m (int): Number of rows.
    - n (int): Number of columns.
    - a (float): Initial horizontal side length.
    - alpha (float): Folding angle in radians.
    - delta (float): Scaling factor per row (0 <= delta <= 1).
    - y_coords (array-like): Optimized y-coordinates for each row.

    Returns:
    - X (2D numpy array): X coordinates of the grid.
    - Y (2D numpy array): Y coordinates of the grid.
    """
    i_values = np.arange(m+1)
    j_values = np.arange(n+1)

    # Compute scaling factors for each row
    a_values = a * (1 - delta * (i_values / m))
    a_values = np.maximum(a_values, 1e-3)  # Prevent zero or negative lengths

    # Compute b_i based on alpha to preserve angles
    b_values = a_values * np.tan(alpha)

    # Initialize coordinate grids
    X = np.zeros((m+1, n+1))
    Y = np.zeros((m+1, n+1))

    for i in i_values:
        for j in j_values:
            base_x = j * a_values[i]
            # Shift every other row by b_i * cos(alpha) to create the Miura-ori pattern
            if i % 2 == 1:
                base_x += b_values[i] * np.cos(alpha)
            X[i, j] = base_x
            Y[i, j] = y_coords[i]

    # Center the pattern horizontally
    top_center = (X[0, :].min() + X[0, :].max()) / 2
    bottom_center = (X[m, :].min() + X[m, :].max()) / 2
    midpoint = (top_center + bottom_center) / 2
    X -= midpoint

    return X, Y

def angle_between_vectors(u, v):
    """
    Compute the angle between two vectors in degrees.

    Parameters:
    - u (array-like): First vector.
    - v (array-like): Second vector.

    Returns:
    - angle (float): Angle between u and v in degrees.
    """
    dot_product = np.dot(u, v)
    norm_u = np.linalg.norm(u)
    norm_v = np.linalg.norm(v)
    # Prevent division by zero
    if norm_u == 0 or norm_v == 0:
        return 0.0
    # Compute cosine of the angle and clip to avoid numerical errors
    cos_angle = np.clip(dot_product / (norm_u * norm_v), -1.0, 1.0)
    angle = np.degrees(np.arccos(cos_angle))
    return angle

def compute_angle_deviations(m, n, X, Y):
    """
    Compute the angle deviations from flat foldability at each internal vertex.

    Parameters:
    - m (int): Number of rows.
    - n (int): Number of columns.
    - X (2D numpy array): X coordinates of the grid.
    - Y (2D numpy array): Y coordinates of the grid.

    Returns:
    - deviations (1D numpy array): Array of angle deviations.
    """
    deviations = []
    # Iterate over internal vertices
    for i in range(1, m):
        for j in range(1, n):
            # Get coordinates of the four neighbors
            p = np.array([X[i, j], Y[i, j]])
            p_up = np.array([X[i-1, j], Y[i-1, j]])
            p_right = np.array([X[i, j+1], Y[i, j+1]])
            p_down = np.array([X[i+1, j], Y[i+1, j]])
            p_left = np.array([X[i, j-1], Y[i, j-1]])

            # Vectors from the center to neighbors
            v_up = p_up - p
            v_right = p_right - p
            v_down = p_down - p
            v_left = p_left - p

            # Compute angles between consecutive vectors in clockwise order
            theta_up_right = angle_between_vectors(v_up, v_right)
            theta_right_down = angle_between_vectors(v_right, v_down)
            theta_down_left = angle_between_vectors(v_down, v_left)
            theta_left_up = angle_between_vectors(v_left, v_up)

            # Kawasaki's condition: sum of alternating angles = 180 degrees
            dev1 = (theta_up_right + theta_down_left) - 180.0
            dev2 = (theta_right_down + theta_left_up) - 180.0

            deviations.extend([dev1, dev2])

    return np.array(deviations)

def cost_function(y_coords, m, n, a, alpha, delta):
    """
    Cost function to minimize deviations from flat foldability.

    Parameters:
    - y_coords (array-like): Current y-coordinates for each row.
    - m (int): Number of rows.
    - n (int): Number of columns.
    - a (float): Initial horizontal side length.
    - alpha (float): Folding angle in radians.
    - delta (float): Scaling factor per row.

    Returns:
    - cost (float): Sum of squared angle deviations.
    """
    X, Y = compute_coordinates(m, n, a, alpha, delta, y_coords)
    deviations = compute_angle_deviations(m, n, X, Y)
    # Sum of squared deviations
    cost = np.sum(deviations**2)
    return cost

def optimize_y_coordinates(m, n, a, alpha, delta, initial_guess=None):
    """
    Optimize y-coordinates to achieve flat foldability with debug information.
    
    Parameters:
    - m (int): Number of rows.
    - n (int): Number of columns.
    - a (float): Initial horizontal side length.
    - alpha (float): Folding angle in radians.
    - delta (float): Scaling factor per row.
    - initial_guess (array-like, optional): Initial guess for y-coordinates.
    
    Returns:
    - result.x (1D numpy array): Optimized y-coordinates.
    - result (OptimizeResult): Full optimization result.
    """
    if initial_guess is None:
        # Initialize y-coordinates based on standard Miura-ori
        i_values = np.arange(m+1)
        a_values = a * (1 - delta * (i_values / m))
        a_values = np.maximum(a_values, 1e-3)
        b_values = a_values * np.tan(alpha)
        y_offsets = [0.0]
        for i in range(1, m+1):
            y_new = y_offsets[i-1] + b_values[i-1] * np.sin(alpha)
            y_offsets.append(y_new)
        initial_guess = np.array(y_offsets)
    
    # Define constraints: y[i] > y[i-1] for all i
    constraints = []
    for i in range(1, m+1):
        constraints.append({
            'type': 'ineq',
            'fun': lambda y, i=i: y[i] - y[i-1] - 1e-3  # Ensure y[i] > y[i-1] + small epsilon
        })
    
    # Define bounds: y-coordinates should be positive and increasing
    bounds = [(None, None)] * (m+1)
    
    # Define a callback function to monitor optimization
    def callback(xk):
        current_cost = cost_function(xk, m, n, a, alpha, delta)
#         print(f"Current cost: {current_cost:.6f}")
    
    # Optimization using SLSQP with callback
    result = minimize(
        cost_function,
        initial_guess,
        args=(m, n, a, alpha, delta),
        method='SLSQP',
        constraints=constraints,
        bounds=bounds,
        options={'maxiter': 10000, 'ftol': 1e-9, 'disp': False},
        callback=callback
    )
    
    return result.x, result

def plot_miura_ori(m, n, a, alpha, delta, y_coords, show_plot=True, save_svg=False):
    """
    Plot the Miura-ori crease pattern.

    Parameters:
    - m (int): Number of rows.
    - n (int): Number of columns.
    - a (float): Initial horizontal side length.
    - alpha (float): Folding angle in radians.
    - delta (float): Scaling factor per row.
    - y_coords (array-like): Optimized y-coordinates for each row.
    - show_plot (bool): Whether to display the plot.
    - save_svg (bool): Whether to save the plot as an SVG file.
    """
    X, Y = compute_coordinates(m, n, a, alpha, delta, y_coords)
    i_values = np.arange(m+1)
    j_values = np.arange(n+1)
    
    fig, ax = plt.subplots(figsize=(8, 8))
    
    # Plot longitudinal folds (vertical lines)
    for j in j_values:
        for i in range(m):
            fold_color = 'blue' if j % 2 == 0 else 'red'
            ax.plot([X[i, j], X[i+1, j]], [Y[i, j], Y[i+1, j]], color=fold_color, lw=2)
    
    # Plot transversal folds (horizontal lines)
    for i in i_values:
        for j in range(n):
            fold_color = 'red' if (i + j) % 2 == 0 else 'blue'
            ax.plot([X[i, j], X[i, j+1]], [Y[i, j], Y[i, j+1]], color=fold_color, lw=2)
    
    # Plot outer boundaries
    for i in range(m):
        # Left boundary
        ax.plot([X[i, 0], X[i+1, 0]], [Y[i, 0], Y[i+1, 0]], color='black', lw=2)
        # Right boundary
        ax.plot([X[i, n], X[i+1, n]], [Y[i, n], Y[i+1, n]], color='black', lw=2)
    for j in range(n):
        # Top boundary
        ax.plot([X[0, j], X[0, j+1]], [Y[0, j], Y[0, j+1]], color='black', lw=2)
        # Bottom boundary
        ax.plot([X[m, j], X[m, j+1]], [Y[m, j], Y[m, j+1]], color='black', lw=2)
    
    # Final plot adjustments
    ax.set_aspect('equal')
    ax.axis('off')
    plt.tight_layout()
    
    if show_plot:
        plt.show()
    if save_svg:
        fig.savefig(f'scaled_miura_ori_{m}_{n}_{a}_{alpha}_{delta}.svg', format='svg', dpi=300, bbox_inches='tight')
        print('SVG file saved as "scaled_miura_ori.svg"')
    plt.close(fig)

def interactive_miura_ori():
    """
    Create an interactive widget to adjust Miura-ori parameters and visualize the pattern.
    """
    def update_plot(m=5, n=4, a=1.0, alpha=1.05, delta=0.25, show_plot=True, save_svg=False):
        # Optimize y-coordinates with improved optimization function
        y_opt, result = optimize_y_coordinates(m, n, a, alpha, delta)
        
        # Display optimization status
        print(f"Optimization successful: {result.success}")
        print(f"Message: {result.message}")
        print(f"Final cost (sum of squared deviations): {result.fun:.6f}")
        
        if result.success:
            # Plot the crease pattern
            plot_miura_ori(m, n, a, alpha, delta, y_opt, show_plot, save_svg)
        else:
            print("Optimization did not converge. Please try adjusting the parameters.")
    
    interactive_plot = interactive(
        update_plot,
        m=IntSlider(min=1, max=20, step=1, value=5, description='Rows (m)'),
        n=IntSlider(min=1, max=20, step=1, value=4, description='Cols (n)'),
        a=FloatSlider(min=0.5, max=5.0, step=0.1, value=1.0, description='a'),
        alpha=FloatSlider(min=0.1, max=np.pi-0.1, step=0.05, value=1.05, description='alpha (rad)'),
        delta=FloatSlider(min=0.0, max=0.5, step=0.05, value=0.25, description='delta'),
        show_plot=Checkbox(value=True, description='Show Plot'),
        save_svg=Checkbox(value=False, description='Save SVG')
    )
    
    display(interactive_plot)

interactive_miura_ori()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import minimize
from ipywidgets import interactive, FloatSlider, IntSlider, Checkbox
from IPython.display import display

def compute_coordinates(m, n, a, alpha, delta, y_coords):
    """
    Compute the X and Y coordinates of the Miura-ori pattern.

    Parameters:
    - m (int): Number of rows.
    - n (int): Number of columns.
    - a (float): Initial horizontal side length.
    - alpha (float): Folding angle in radians.
    - delta (float): Scaling factor per row (0 <= delta <= 1).
    - y_coords (array-like): Optimized y-coordinates for each row.

    Returns:
    - X (2D numpy array): X coordinates of the grid.
    - Y (2D numpy array): Y coordinates of the grid.
    """
    i_values = np.arange(m+1)
    j_values = np.arange(n+1)

    # Compute scaling factors for each row
    a_values = a * (1 - delta * (i_values / m))
    a_values = np.maximum(a_values, 1e-3)  # Prevent zero or negative lengths

    # Compute b_i based on alpha to preserve angles
    b_values = a_values * np.tan(alpha)

    # Initialize coordinate grids
    X = np.zeros((m+1, n+1))
    Y = np.zeros((m+1, n+1))

    for i in i_values:
        for j in j_values:
            base_x = j * a_values[i]
            # Shift every other row horizontally by b_i * cos(alpha) to create the Miura-ori pattern
            if i % 2 == 1:
                base_x += b_values[i] * np.cos(alpha)
            X[i, j] = base_x
            Y[i, j] = y_coords[i]

    # Center the pattern horizontally for better visualization
    top_center = (X[0, :].min() + X[0, :].max()) / 2
    bottom_center = (X[m, :].min() + X[m, :].max()) / 2
    midpoint = (top_center + bottom_center) / 2
    X -= midpoint

    return X, Y

def angle_between_vectors(u, v):
    """
    Compute the angle between two vectors in degrees.

    Parameters:
    - u (array-like): First vector.
    - v (array-like): Second vector.

    Returns:
    - angle (float): Angle between u and v in degrees.
    """
    dot_product = np.dot(u, v)
    norm_u = np.linalg.norm(u)
    norm_v = np.linalg.norm(v)
    # Prevent division by zero
    if norm_u == 0 or norm_v == 0:
        return 0.0
    # Compute cosine of the angle and clip to avoid numerical errors
    cos_angle = np.clip(dot_product / (norm_u * norm_v), -1.0, 1.0)
    angle = np.degrees(np.arccos(cos_angle))
    return angle

def compute_angle_deviations(m, n, X, Y):
    """
    Compute the angle deviations from flat foldability at each internal vertex.

    Parameters:
    - m (int): Number of rows.
    - n (int): Number of columns.
    - X (2D numpy array): X coordinates of the grid.
    - Y (2D numpy array): Y coordinates of the grid.

    Returns:
    - deviations (1D numpy array): Array of angle deviations.
    """
    deviations = []
    # Iterate over internal vertices
    for i in range(1, m):
        for j in range(1, n):
            # Get coordinates of the four neighbors
            p = np.array([X[i, j], Y[i, j]])
            p_up = np.array([X[i-1, j], Y[i-1, j]])
            p_right = np.array([X[i, j+1], Y[i, j+1]])
            p_down = np.array([X[i+1, j], Y[i+1, j]])
            p_left = np.array([X[i, j-1], Y[i, j-1]])

            # Vectors from the center to neighbors
            v_up = p_up - p
            v_right = p_right - p
            v_down = p_down - p
            v_left = p_left - p

            # Compute angles between consecutive vectors in clockwise order
            theta_up_right = angle_between_vectors(v_up, v_right)
            theta_right_down = angle_between_vectors(v_right, v_down)
            theta_down_left = angle_between_vectors(v_down, v_left)
            theta_left_up = angle_between_vectors(v_left, v_up)

            # Kawasaki's condition: sum of alternating angles = 180 degrees
            dev1 = (theta_up_right + theta_down_left) - 180.0
            dev2 = (theta_right_down + theta_left_up) - 180.0

            deviations.extend([dev1, dev2])

    return np.array(deviations)

def cost_function(y_coords, m, n, a, alpha, delta):
    """
    Cost function to minimize deviations from flat foldability.

    Parameters:
    - y_coords (array-like): Current y-coordinates for each row.
    - m (int): Number of rows.
    - n (int): Number of columns.
    - a (float): Initial horizontal side length.
    - alpha (float): Folding angle in radians.
    - delta (float): Scaling factor per row.

    Returns:
    - cost (float): Sum of squared angle deviations.
    """
    X, Y = compute_coordinates(m, n, a, alpha, delta, y_coords)
    deviations = compute_angle_deviations(m, n, X, Y)
    # Sum of squared deviations
    cost = np.sum(deviations**2)
    return cost

def optimize_y_coordinates(m, n, a, alpha, delta, initial_guess=None):
    """
    Optimize y-coordinates to achieve flat foldability with enhanced constraints and validation.

    Parameters:
    - m (int): Number of rows.
    - n (int): Number of columns.
    - a (float): Initial horizontal side length.
    - alpha (float): Folding angle in radians.
    - delta (float): Scaling factor per row.
    - initial_guess (array-like, optional): Initial guess for y-coordinates.

    Returns:
    - result.x (1D numpy array): Optimized y-coordinates.
    - result (OptimizeResult): Full optimization result.
    """
    if initial_guess is None:
        # Initialize y-coordinates based on standard Miura-ori
        i_values = np.arange(m+1)
        a_values = a * (1 - delta * (i_values / m))
        a_values = np.maximum(a_values, 1e-3)
        b_values = a_values * np.tan(alpha)
        y_offsets = [0.0]
        for i in range(1, m+1):
            y_new = y_offsets[i-1] + b_values[i-1] * np.sin(alpha)
            y_offsets.append(y_new)
        initial_guess = np.array(y_offsets)

    # Define constraints: y[i] > y[i-1] for all i
    constraints = []
    for i in range(1, m+1):
        constraints.append({
            'type': 'ineq',
            'fun': lambda y, i=i: y[i] - y[i-1] - 1e-3  # Ensure y[i] > y[i-1] + small epsilon
        })

    # Define bounds: y-coordinates should be positive and increasing
    bounds = [(0, None)] * (m+1)

    # Lists to store optimization history
    cost_history = []
    y_history = []

    # Define a callback function to monitor optimization
    def callback(xk):
        current_cost = cost_function(xk, m, n, a, alpha, delta)
        cost_history.append(current_cost)
        y_history.append(np.copy(xk))
        # print(f"Current cost: {current_cost:.6f}")

    # Optimization using SLSQP with enhanced constraints
    result = minimize(
        cost_function,
        initial_guess,
        args=(m, n, a, alpha, delta),
        method='SLSQP',
        constraints=constraints,
        bounds=bounds,
        options={'maxiter': 10000, 'ftol': 1e-9, 'disp': False},
        callback=callback
    )

    # After optimization, plot the cost history
    plt.figure(figsize=(10, 4))
    plt.plot(cost_history, label='Cost Function')
    plt.xlabel('Iteration')
    plt.ylabel('Cost')
    plt.title('Optimization Progress')
    plt.legend()
    plt.grid(True)
    plt.show()

    # Plot the progression of y-coordinates
    if y_history:
        y_history = np.array(y_history)
        plt.figure(figsize=(10, 4))
        for i in range(len(y_history[0])):
            plt.plot(y_history[:, i], label=f'y{i}')
        plt.xlabel('Iteration')
        plt.ylabel('y-coordinate')
        plt.title('Y-coordinate Progression')
        plt.legend()
        plt.grid(True)
        plt.show()

    # Validate constraints
    def validate_constraints(y_coords, m):
        violations = []
        for i in range(1, m+1):
            if y_coords[i] <= y_coords[i-1] + 1e-3:
                violations.append((i, y_coords[i]))
        if violations:
            print("Constraint violations detected:")
            for violation in violations:
                print(f"Row {violation[0]}: y = {violation[1]:.6f}")
        else:
            print("All constraints satisfied.")

    validate_constraints(result.x, m)

    return result.x, result

def plot_miura_ori(m, n, a, alpha, delta, y_coords, show_plot=True, save_svg=False):
    """
    Plot the Miura-ori crease pattern.

    Parameters:
    - m (int): Number of rows.
    - n (int): Number of columns.
    - a (float): Initial horizontal side length.
    - alpha (float): Folding angle in radians.
    - delta (float): Scaling factor per row.
    - y_coords (array-like): Optimized y-coordinates for each row.
    - show_plot (bool): Whether to display the plot.
    - save_svg (bool): Whether to save the plot as an SVG file.
    """
    X, Y = compute_coordinates(m, n, a, alpha, delta, y_coords)
    i_values = np.arange(m+1)
    j_values = np.arange(n+1)
    
    fig, ax = plt.subplots(figsize=(8, 8))
    
    # Plot longitudinal folds (vertical lines)
    for j in j_values:
        for i in range(m):
            fold_color = 'blue' if j % 2 == 0 else 'red'
            ax.plot([X[i, j], X[i+1, j]], [Y[i, j], Y[i+1, j]], color=fold_color, lw=2)
    
    # Plot transversal folds (horizontal lines)
    for i in i_values:
        for j in range(n):
            fold_color = 'red' if (i + j) % 2 == 0 else 'blue'
            ax.plot([X[i, j], X[i, j+1]], [Y[i, j], Y[i, j+1]], color=fold_color, lw=2)
    
    # Plot outer boundaries
    for i in range(m):
        # Left boundary
        ax.plot([X[i, 0], X[i+1, 0]], [Y[i, 0], Y[i+1, 0]], color='black', lw=2)
        # Right boundary
        ax.plot([X[i, n], X[i+1, n]], [Y[i, n], Y[i+1, n]], color='black', lw=2)
    for j in range(n):
        # Top boundary
        ax.plot([X[0, j], X[0, j+1]], [Y[0, j], Y[0, j+1]], color='black', lw=2)
        # Bottom boundary
        ax.plot([X[m, j], X[m, j+1]], [Y[m, j], Y[m, j+1]], color='black', lw=2)
    
    # Final plot adjustments
    ax.set_aspect('equal')
    ax.axis('off')
    plt.tight_layout()
    
    if show_plot:
        plt.show()
    if save_svg:
        fig.savefig(f'scaled_miura_ori_{m}_{n}_{a}_{alpha}_{delta}.svg', format='svg', dpi=300)
        print('SVG file saved as "scaled_miura_ori.svg"')
    plt.close(fig)

def interactive_miura_ori():
    """
    Create an interactive widget to adjust Miura-ori parameters and visualize the pattern.
    """
    def update_plot(m=5, n=4, a=1.0, alpha=1.05, delta=0.25, show_plot=True, save_svg=False):
        # Optimize y-coordinates with enhanced validation
        y_opt, result = optimize_y_coordinates(m, n, a, alpha, delta)
        
        # Display optimization status
        print(f"Optimization successful: {result.success}")
        print(f"Message: {result.message}")
        print(f"Final cost (sum of squared deviations): {result.fun:.6f}")
        
        if result.success:
            # Plot the crease pattern
            plot_miura_ori(m, n, a, alpha, delta, y_opt, show_plot, save_svg)
        else:
            print("Optimization did not converge. Please try adjusting the parameters.")
    
    interactive_plot = interactive(
        update_plot,
        m=IntSlider(min=1, max=20, step=1, value=5, description='Rows (m)'),
        n=IntSlider(min=1, max=20, step=1, value=4, description='Cols (n)'),
        a=FloatSlider(min=0.5, max=5.0, step=0.1, value=1.0, description='a'),
        alpha=FloatSlider(min=0.1, max=np.pi-0.1, step=0.05, value=1.05, description='alpha (rad)'),
        delta=FloatSlider(min=0.0, max=0.5, step=0.05, value=0.25, description='delta'),
        show_plot=Checkbox(value=True, description='Show Plot'),
        save_svg=Checkbox(value=False, description='Save SVG')
    )
    
    display(interactive_plot)

# Initialize the interactive interface
interactive_miura_ori()
```

```python
# Import necessary libraries
import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import minimize
from ipywidgets import interactive, FloatSlider, IntSlider, Checkbox
from IPython.display import display

# Enable inline plotting
def compute_coordinates_with_deviation(m, n, a, alpha, delta, y_deviation):
    """
    Compute the X and Y coordinates of the Miura-ori pattern with Y-deviation.
    
    Parameters:
    - m (int): Number of rows.
    - n (int): Number of columns.
    - a (float): Initial horizontal side length.
    - alpha (float): Folding angle in radians.
    - delta (float): Scaling factor per row (0 <= delta <= 1).
    - y_deviation (array-like): Y-deviations for each row.
    
    Returns:
    - X (2D numpy array): X coordinates of the grid.
    - Y (2D numpy array): Y coordinates of the grid.
    """
    i_values = np.arange(m+1)
    j_values = np.arange(n+1)

    # Compute scaling factors for each row
    a_values = a * (1 - delta * (i_values / m))
    a_values = np.maximum(a_values, 1e-3)  # Prevent zero or negative lengths

    # Compute b_i based on alpha to preserve angles
    b_values = a_values * np.tan(alpha)

    # Initialize coordinate grids
    X = np.zeros((m+1, n+1))
    Y = np.zeros((m+1, n+1))

    for i in i_values:
        for j in j_values:
            base_x = j * a_values[i]
            # Shift every other row horizontally by b_i * cos(alpha) to create the Miura-ori pattern
            if i % 2 == 1:
                base_x += b_values[i] * np.cos(alpha)
            X[i, j] = base_x
            Y[i, j] = y_deviation[i]

    # Center the pattern horizontally for better visualization
    top_center = (X[0, :].min() + X[0, :].max()) / 2
    bottom_center = (X[m, :].min() + X[m, :].max()) / 2
    midpoint = (top_center + bottom_center) / 2
    X -= midpoint

    return X, Y

def angle_between_vectors(u, v):
    """
    Compute the angle between two vectors in degrees.

    Parameters:
    - u (array-like): First vector.
    - v (array-like): Second vector.

    Returns:
    - angle (float): Angle between u and v in degrees.
    """
    dot_product = np.dot(u, v)
    norm_u = np.linalg.norm(u)
    norm_v = np.linalg.norm(v)
    # Prevent division by zero
    if norm_u == 0 or norm_v == 0:
        return 0.0
    # Compute cosine of the angle and clip to avoid numerical errors
    cos_angle = np.clip(dot_product / (norm_u * norm_v), -1.0, 1.0)
    angle = np.degrees(np.arccos(cos_angle))
    return angle

def compute_angle_deviations(m, n, X, Y):
    """
    Compute the angle deviations from flat foldability at each internal vertex.

    Parameters:
    - m (int): Number of rows.
    - n (int): Number of columns.
    - X (2D numpy array): X coordinates of the grid.
    - Y (2D numpy array): Y coordinates of the grid.

    Returns:
    - deviations (1D numpy array): Array of angle deviations.
    """
    deviations = []
    # Iterate over internal vertices
    for i in range(1, m):
        for j in range(1, n):
            # Get coordinates of the four neighbors
            p = np.array([X[i, j], Y[i, j]])
            p_up = np.array([X[i-1, j], Y[i-1, j]])
            p_right = np.array([X[i, j+1], Y[i, j+1]])
            p_down = np.array([X[i+1, j], Y[i+1, j]])
            p_left = np.array([X[i, j-1], Y[i, j-1]])

            # Vectors from the center to neighbors
            v_up = p_up - p
            v_right = p_right - p
            v_down = p_down - p
            v_left = p_left - p

            # Compute angles between consecutive vectors in clockwise order
            theta_up_right = angle_between_vectors(v_up, v_right)
            theta_right_down = angle_between_vectors(v_right, v_down)
            theta_down_left = angle_between_vectors(v_down, v_left)
            theta_left_up = angle_between_vectors(v_left, v_up)

            # Kawasaki's condition: sum of alternating angles = 180 degrees
            dev1 = (theta_up_right + theta_down_left) - 180.0
            dev2 = (theta_right_down + theta_left_up) - 180.0

            deviations.extend([dev1, dev2])

    return np.array(deviations)

def cost_function_with_deviation(y_deviation, m, n, a, alpha, delta, lambda_reg=1.0):
    """
    Cost function to minimize deviations from flat foldability and Y-deviation magnitude.

    Parameters:
    - y_deviation (array-like): Current Y-deviations for each row.
    - m (int): Number of rows.
    - n (int): Number of columns.
    - a (float): Initial horizontal side length.
    - alpha (float): Folding angle in radians.
    - delta (float): Scaling factor per row.
    - lambda_reg (float): Regularization coefficient for Y-deviation.

    Returns:
    - cost (float): Combined cost of angle deviations and Y-deviation magnitude.
    """
    X, Y = compute_coordinates_with_deviation(m, n, a, alpha, delta, y_deviation)
    deviations = compute_angle_deviations(m, n, X, Y)
    # Sum of squared deviations from flat foldability
    angle_cost = np.sum(deviations**2)
    # Regularization to minimize Y-deviation magnitude
    reg_cost = lambda_reg * np.sum(y_deviation**2)
    # Total cost
    cost = angle_cost + reg_cost
    return cost

def cost_function_with_deviation(y_deviation, m, n, a, alpha, delta, lambda_reg=1.0):
    """
    Cost function to minimize deviations from flat foldability and Y-deviation magnitude.

    Parameters:
    - y_deviation (array-like): Current Y-deviations for each row.
    - m (int): Number of rows.
    - n (int): Number of columns.
    - a (float): Initial horizontal side length.
    - alpha (float): Folding angle in radians.
    - delta (float): Scaling factor per row.
    - lambda_reg (float): Regularization coefficient for Y-deviation.

    Returns:
    - cost (float): Combined cost of angle deviations and Y-deviation magnitude.
    """
    X, Y = compute_coordinates_with_deviation(m, n, a, alpha, delta, y_deviation)
    deviations = compute_angle_deviations(m, n, X, Y)
    # Sum of squared deviations from flat foldability
    angle_cost = np.sum(deviations**2)
    # Regularization to minimize Y-deviation magnitude
    reg_cost = lambda_reg * np.sum(y_deviation**2)
    # Total cost
    cost = angle_cost + reg_cost
    return cost

def optimize_y_deviation(m, n, a, alpha, delta, delta_max=0.1, lambda_reg=1.0, initial_guess=None):
    """
    Optimize Y-deviations to achieve flat foldability with controlled skew.

    Parameters:
    - m (int): Number of rows.
    - n (int): Number of columns.
    - a (float): Initial horizontal side length.
    - alpha (float): Folding angle in radians.
    - delta (float): Scaling factor per row.
    - delta_max (float): Maximum allowed Y-deviation.
    - lambda_reg (float): Regularization coefficient for Y-deviation.
    - initial_guess (array-like, optional): Initial guess for Y-deviations.

    Returns:
    - result.x (1D numpy array): Optimized Y-deviations.
    - result (OptimizeResult): Full optimization result.
    """
    if initial_guess is None:
        # Initialize Y-deviations to zero
        initial_guess = np.zeros(m+1)

    # Define constraints: y_deviation[i] <= delta_max and y_deviation[i] >= -delta_max
    bounds = [(-delta_max, delta_max) for _ in range(m+1)]

    # Define constraints: y_new[i] > y_new[i-1] + epsilon
    # Since y_new[i] = y_base[i] + y_deviation[i], and y_base[i] is strictly increasing,
    # ensure that y_deviation[i] does not invert the order.
    # For simplicity, we assume y_base[i] already ensures y[i] > y[i-1], so additional constraints might be redundant.
    # However, to be safe, we can enforce:
    constraints = []
    epsilon = 1e-3  # Small positive value to ensure strict inequality
    for i in range(1, m+1):
        constraints.append({
            'type': 'ineq',
            'fun': lambda y, i=i: y[i] - y[i-1] - 1e-4  # Ensures y_deviation[i] >= y_deviation[i-1] + delta
        })

    # Optimization using SLSQP
    result = minimize(
        cost_function_with_deviation,
        initial_guess,
        args=(m, n, a, alpha, delta, lambda_reg),
        method='SLSQP',
        bounds=bounds,
        constraints=constraints,
        options={'maxiter': 10000, 'ftol': 1e-9, 'disp': False}
    )

    # After optimization, plot the cost history
    # Note: Since we're not storing the history in this setup, consider removing the plot
    # or implementing a history tracking mechanism if needed.

    # Validate constraints
    def validate_constraints(y_deviation, m, epsilon=1e-4):
        violations = []
        for i in range(1, m+1):
            if y_deviation[i] - y_deviation[i-1] < epsilon:
                violations.append((i, y_deviation[i]))
        if violations:
            print("Constraint violations detected:")
            for violation in violations:
                print(f"Row {violation[0]}: y_deviation = {violation[1]:.6f}")
        else:
            print("All constraints satisfied.")

    validate_constraints(result.x, m)

    return result.x, result

def plot_miura_ori_with_deviation(m, n, a, alpha, delta, y_deviation, show_plot=True, save_svg=False):
    """
    Plot the Miura-ori crease pattern with Y-deviation.

    Parameters:
    - m (int): Number of rows.
    - n (int): Number of columns.
    - a (float): Initial horizontal side length.
    - alpha (float): Folding angle in radians.
    - delta (float): Scaling factor per row.
    - y_deviation (array-like): Y-deviations for each row.
    - show_plot (bool): Whether to display the plot.
    - save_svg (bool): Whether to save the plot as an SVG file.
    """
    X, Y = compute_coordinates_with_deviation(m, n, a, alpha, delta, y_deviation)
    i_values = np.arange(m+1)
    j_values = np.arange(n+1)
    
    fig, ax = plt.subplots(figsize=(8, 8))
    
    # Plot longitudinal folds (vertical lines)
    for j in j_values:
        for i in range(m):
            fold_color = 'blue' if j % 2 == 0 else 'red'
            ax.plot([X[i, j], X[i+1, j]], [Y[i, j], Y[i+1, j]], color=fold_color, lw=2)
    
    # Plot transversal folds (horizontal lines)
    for i in i_values:
        for j in range(n):
            fold_color = 'red' if (i + j) % 2 == 0 else 'blue'
            ax.plot([X[i, j], X[i, j+1]], [Y[i, j], Y[i, j+1]], color=fold_color, lw=2)
    
    # Plot outer boundaries
    for i in range(m):
        # Left boundary
        ax.plot([X[i, 0], X[i+1, 0]], [Y[i, 0], Y[i+1, 0]], color='black', lw=2)
        # Right boundary
        ax.plot([X[i, n], X[i+1, n]], [Y[i, n], Y[i+1, n]], color='black', lw=2)
    for j in range(n):
        # Top boundary
        ax.plot([X[0, j], X[0, j+1]], [Y[0, j], Y[0, j+1]], color='black', lw=2)
        # Bottom boundary
        ax.plot([X[m, j], X[m, j+1]], [Y[m, j], Y[m, j+1]], color='black', lw=2)
    
    # Final plot adjustments
    # ax.set_aspect('equal')
    ax.axis('off')
    # plt.tight_layout()
    
    if show_plot:
        plt.show()
    if save_svg:
        fig.savefig(f'skewed_miura_ori_{m}_{n}_{a}_{alpha}_{delta}.svg', format='svg', dpi=300)
        print(f'SVG file saved as "skewed_miura_ori_{m}_{n}_{a}_{alpha}_{delta}.svg"')
    # plt.close(fig)

def interactive_miura_ori_with_deviation():
    """
    Create an interactive widget to adjust Miura-ori parameters with Y-deviation and visualize the pattern.
    """
    def update_plot(m=5, n=4, a=1.0, alpha=1.05, delta=0.25, delta_max=0.1, lambda_reg=1.0, show_plot=True, save_svg=False):
        # Optimize Y-deviations
        y_deviation_opt, result = optimize_y_deviation(m, n, a, alpha, delta, delta_max, lambda_reg)
        
        # Display optimization status
        print(f"Optimization successful: {result.success}")
        print(f"Message: {result.message}")
        print(f"Final cost (angle deviations + regularization): {result.fun:.6f}")
        
        if result.success:
            # Plot the crease pattern with Y-deviation
            plot_miura_ori_with_deviation(m, n, a, alpha, delta, y_deviation_opt, show_plot, save_svg)
        else:
            print("Optimization did not converge. Please try adjusting the parameters.")

    interactive_plot = interactive(
        update_plot,
        m=IntSlider(min=3, max=20, step=1, value=5, description='Rows (m)'),
        n=IntSlider(min=2, max=20, step=1, value=4, description='Cols (n)'),
        a=FloatSlider(min=0.5, max=5.0, step=0.1, value=1.0, description='a'),
        alpha=FloatSlider(min=0.1, max=np.pi-0.1, step=0.05, value=1.05, description='alpha (rad)'),
        delta=FloatSlider(min=0.0, max=0.5, step=0.05, value=0.25, description='delta'),
        delta_max=FloatSlider(min=0.0, max=0.5, step=0.05, value=0.1, description='Max Y-Deviation'),
        lambda_reg=FloatSlider(min=0.0, max=10.0, step=0.1, value=1.0, description='Reg. Coef'),
        show_plot=Checkbox(value=True, description='Show Plot'),
        save_svg=Checkbox(value=False, description='Save SVG')
    )
    
    display(interactive_plot)

    # Initialize the interactive interface
interactive_miura_ori_with_deviation()
```

```python
def compute_coordinates_with_deviation_and_rotation(m, n, a, alpha, delta, y_deviation, theta_max):
    """
    Compute the X and Y coordinates of the Miura-ori pattern with Y-deviation and rotation.
    
    Parameters:
    - m (int): Number of rows.
    - n (int): Number of columns.
    - a (float): Initial horizontal side length.
    - alpha (float): Folding angle in radians.
    - delta (float): Scaling factor per row (0 <= delta <= 1).
    - y_deviation (array-like): Y-deviations for each row.
    - theta_max (float): Maximum rotation angle in radians.
    
    Returns:
    - X (2D numpy array): X coordinates of the grid.
    - Y (2D numpy array): Y coordinates of the grid.
    """
    i_values = np.arange(m+1)
    j_values = np.arange(n+1)

    # Compute scaling factors for each row
    a_values = a * (1 - delta * (i_values / m))
    a_values = np.maximum(a_values, 1e-3)  # Prevent zero or negative lengths

    # Compute b_i based on alpha to preserve angles
    b_values = a_values * np.tan(alpha)

    # Initialize coordinate grids
    X = np.zeros((m+1, n+1))
    Y = np.zeros((m+1, n+1))

    for i in i_values:
        for j in j_values:
            base_x = j * a_values[i]
            # Shift every other row horizontally by b_i * cos(alpha) to create the Miura-ori pattern
            if i % 2 == 1:
                base_x += b_values[i] * np.cos(alpha)
            # Apply rotation (skew) to the x-coordinate based on the column index
            # Alternating the direction of rotation per parallelogram
            rotation_angle = theta_max * (-1)**j
            skew_offset = y_deviation[i] * np.tan(rotation_angle)
            X[i, j] = base_x + skew_offset
            Y[i, j] = y_deviation[i]

    # Center the pattern horizontally for better visualization
    top_center = (X[0, :].min() + X[0, :].max()) / 2
    bottom_center = (X[m, :].min() + X[m, :].max()) / 2
    midpoint = (top_center + bottom_center) / 2
    X -= midpoint

    return X, Y
```

```python
def angle_between_vectors(u, v):
    """
    Compute the angle between two vectors in degrees.

    Parameters:
    - u (array-like): First vector.
    - v (array-like): Second vector.

    Returns:
    - angle (float): Angle between u and v in degrees.
    """
    dot_product = np.dot(u, v)
    norm_u = np.linalg.norm(u)
    norm_v = np.linalg.norm(v)
    # Prevent division by zero
    if norm_u == 0 or norm_v == 0:
        return 0.0
    # Compute cosine of the angle and clip to avoid numerical errors
    cos_angle = np.clip(dot_product / (norm_u * norm_v), -1.0, 1.0)
    angle = np.degrees(np.arccos(cos_angle))
    return angle

def compute_angle_deviations(m, n, X, Y):
    """
    Compute the angle deviations from flat foldability at each internal vertex.

    Parameters:
    - m (int): Number of rows.
    - n (int): Number of columns.
    - X (2D numpy array): X coordinates of the grid.
    - Y (2D numpy array): Y coordinates of the grid.

    Returns:
    - deviations (1D numpy array): Array of angle deviations.
    """
    deviations = []
    # Iterate over internal vertices
    for i in range(1, m):
        for j in range(1, n):
            # Get coordinates of the four neighbors
            p = np.array([X[i, j], Y[i, j]])
            p_up = np.array([X[i-1, j], Y[i-1, j]])
            p_right = np.array([X[i, j+1], Y[i, j+1]])
            p_down = np.array([X[i+1, j], Y[i+1, j]])
            p_left = np.array([X[i, j-1], Y[i, j-1]])

            # Vectors from the center to neighbors
            v_up = p_up - p
            v_right = p_right - p
            v_down = p_down - p
            v_left = p_left - p

            # Compute angles between consecutive vectors in clockwise order
            theta_up_right = angle_between_vectors(v_up, v_right)
            theta_right_down = angle_between_vectors(v_right, v_down)
            theta_down_left = angle_between_vectors(v_down, v_left)
            theta_left_up = angle_between_vectors(v_left, v_up)

            # Kawasaki's condition: sum of alternating angles = 180 degrees
            dev1 = (theta_up_right + theta_down_left) - 180.0
            dev2 = (theta_right_down + theta_left_up) - 180.0

            deviations.extend([dev1, dev2])

    return np.array(deviations)
```

```python
def cost_function_with_deviation_and_rotation(y_deviation, m, n, a, alpha, delta, theta_max, lambda_reg=1.0):
    """
    Cost function to minimize deviations from flat foldability and Y-deviation magnitude.

    Parameters:
    - y_deviation (array-like): Current Y-deviations for each row.
    - m (int): Number of rows.
    - n (int): Number of columns.
    - a (float): Initial horizontal side length.
    - alpha (float): Folding angle in radians.
    - delta (float): Scaling factor per row.
    - theta_max (float): Maximum rotation angle in radians.
    - lambda_reg (float): Regularization coefficient for Y-deviation.

    Returns:
    - cost (float): Combined cost of angle deviations and Y-deviation magnitude.
    """
    X, Y = compute_coordinates_with_deviation_and_rotation(m, n, a, alpha, delta, y_deviation, theta_max)
    deviations = compute_angle_deviations(m, n, X, Y)
    # Sum of squared deviations from flat foldability
    angle_cost = np.sum(deviations**2)
    # Regularization to minimize Y-deviation magnitude
    reg_cost = lambda_reg * np.sum(y_deviation**2)
    # Total cost
    cost = angle_cost + reg_cost
    return cost
```

```python
def optimize_y_deviation_and_rotation(m, n, a, alpha, delta, theta_max, lambda_reg=1.0, delta_max=0.1, initial_guess=None):
    """
    Optimize Y-deviations to achieve flat foldability with controlled skew.

    Parameters:
    - m (int): Number of rows.
    - n (int): Number of columns.
    - a (float): Initial horizontal side length.
    - alpha (float): Folding angle in radians.
    - delta (float): Scaling factor per row.
    - theta_max (float): Maximum rotation angle in radians.
    - lambda_reg (float): Regularization coefficient for Y-deviation.
    - delta_max (float): Maximum allowed Y-deviation.
    - initial_guess (array-like, optional): Initial guess for Y-deviations.

    Returns:
    - result.x (1D numpy array): Optimized Y-deviations.
    - result (OptimizeResult): Full optimization result.
    """
    if initial_guess is None:
        # Initialize Y-deviations based on standard Miura-ori without skew
        i_values = np.arange(m+1)
        a_values = a * (1 - delta * (i_values / m))
        a_values = np.maximum(a_values, 1e-3)
        b_values = a_values * np.tan(alpha)
        y_offsets = [0.0]
        for i in range(1, m+1):
            y_new = y_offsets[i-1] + b_values[i-1] * np.sin(alpha)
            y_offsets.append(y_new)
        initial_guess = np.array(y_offsets)

    # Define constraints: y_deviation[i] <= delta_max and y_deviation[i] >= -delta_max
    bounds = [(-delta_max, delta_max) for _ in range(m+1)]

    # Define constraints: y_deviation[i] - y_deviation[i-1] >= epsilon
    # To ensure monotonicity in Y-deviations
    constraints = []
    epsilon = 1e-4  # Small positive value to ensure strict inequality
    for i in range(1, m+1):
        constraints.append({
            'type': 'ineq',
            'fun': lambda y, i=i: y[i] - y[i-1] - epsilon  # y[i] > y[i-1] + epsilon
        })

    # Optimization using SLSQP
    result = minimize(
        cost_function_with_deviation_and_rotation,
        initial_guess,
        args=(m, n, a, alpha, delta, theta_max, lambda_reg),
        method='SLSQP',
        bounds=bounds,
        constraints=constraints,
        options={'maxiter': 10000, 'ftol': 1e-9, 'disp': False}
    )

    # Validate constraints
    def validate_constraints(y_deviation, m, epsilon=1e-4):
        violations = []
        for i in range(1, m+1):
            if y_deviation[i] - y_deviation[i-1] < epsilon:
                violations.append((i, y_deviation[i]))
        if violations:
            print("Constraint violations detected:")
            for violation in violations:
                print(f"Row {violation[0]}: y_deviation = {violation[1]:.6f}")
        else:
            print("All constraints satisfied.")

    validate_constraints(result.x, m)

    return result.x, result
```

```python
def plot_miura_ori_with_deviation_and_rotation(m, n, a, alpha, delta, y_deviation, theta_max, show_plot=True, save_svg=False):
    """
    Plot the Miura-ori crease pattern with Y-deviation and rotation.

    Parameters:
    - m (int): Number of rows.
    - n (int): Number of columns.
    - a (float): Initial horizontal side length.
    - alpha (float): Folding angle in radians.
    - delta (float): Scaling factor per row.
    - y_deviation (array-like): Y-deviations for each row.
    - theta_max (float): Maximum rotation angle in radians.
    - show_plot (bool): Whether to display the plot.
    - save_svg (bool): Whether to save the plot as an SVG file.
    """
    X, Y = compute_coordinates_with_deviation_and_rotation(m, n, a, alpha, delta, y_deviation, theta_max)
    i_values = np.arange(m+1)
    j_values = np.arange(n+1)
    
    fig, ax = plt.subplots(figsize=(8, 8))
    
    # Plot longitudinal folds (vertical lines)
    for j in j_values:
        for i in range(m):
            fold_color = 'blue' if j % 2 == 0 else 'red'
            ax.plot([X[i, j], X[i+1, j]], [Y[i, j], Y[i+1, j]], color=fold_color, lw=2)
    
    # Plot transversal folds (horizontal lines)
    for i in i_values:
        for j in range(n):
            fold_color = 'red' if (i + j) % 2 == 0 else 'blue'
            ax.plot([X[i, j], X[i, j+1]], [Y[i, j], Y[i, j+1]], color=fold_color, lw=2)
    
    # Plot outer boundaries
    for i in range(m):
        # Left boundary
        ax.plot([X[i, 0], X[i+1, 0]], [Y[i, 0], Y[i+1, 0]], color='black', lw=2)
        # Right boundary
        ax.plot([X[i, n], X[i+1, n]], [Y[i, n], Y[i+1, n]], color='black', lw=2)
    for j in range(n):
        # Top boundary
        ax.plot([X[0, j], X[0, j+1]], [Y[0, j], Y[0, j+1]], color='black', lw=2)
        # Bottom boundary
        ax.plot([X[m, j], X[m, j+1]], [Y[m, j], Y[m, j+1]], color='black', lw=2)
    
    # Final plot adjustments
    # ax.set_aspect('equal')
    ax.axis('off')
    # plt.tight_layout()
    
    if show_plot:
        plt.show()
    if save_svg:
        fig.savefig(f'skewed_miura_ori_{m}_{n}_{a}_{alpha}_{delta}.svg', format='svg', dpi=300)
        print(f'SVG file saved as "skewed_miura_ori_{m}_{n}_{a}_{alpha}_{delta}.svg"')
    plt.close(fig)
```

```python
def interactive_miura_ori_with_deviation_and_rotation():
    """
    Create an interactive widget to adjust Miura-ori parameters with Y-deviation and rotation and visualize the pattern.
    """
    def update_plot(m=5, n=4, a=1.0, alpha=1.05, delta=0.25, theta_max=0.05, delta_max=0.1, lambda_reg=1.0, show_plot=True, save_svg=False):
        # Optimize Y-deviations with the given theta_max
        y_deviation_opt, result = optimize_y_deviation_and_rotation(m, n, a, alpha, delta, theta_max, lambda_reg, delta_max)
        
        # Display optimization status
        print(f"Optimization successful: {result.success}")
        print(f"Message: {result.message}")
        print(f"Final cost (angle deviations + regularization): {result.fun:.6f}")
        
        if result.success:
            # Plot the crease pattern with Y-deviation and rotation
            plot_miura_ori_with_deviation_and_rotation(m, n, a, alpha, delta, y_deviation_opt, theta_max, show_plot, save_svg)
        else:
            print("Optimization did not converge. Please try adjusting the parameters.")
    
    interactive_plot = interactive(
        update_plot,
        m=IntSlider(min=3, max=20, step=1, value=5, description='Rows (m)'),
        n=IntSlider(min=2, max=20, step=1, value=4, description='Cols (n)'),
        a=FloatSlider(min=0.5, max=5.0, step=0.1, value=1.0, description='a'),
        alpha=FloatSlider(min=0.1, max=np.pi-0.1, step=0.05, value=1.05, description='alpha (rad)'),
        delta=FloatSlider(min=0.0, max=0.5, step=0.05, value=0.25, description='delta'),
        theta_max=FloatSlider(min=0.0, max=np.pi/4, step=0.01, value=0.05, description='Max Rotation (rad)'),
        delta_max=FloatSlider(min=0.0, max=0.5, step=0.05, value=0.1, description='Max Y-Deviation'),
        lambda_reg=FloatSlider(min=0.0, max=10.0, step=0.1, value=1.0, description='Reg. Coef'),
        show_plot=Checkbox(value=True, description='Show Plot'),
        save_svg=Checkbox(value=False, description='Save SVG')
    )
    
    display(interactive_plot)

# Initialize the interactive interface
interactive_miura_ori_with_deviation_and_rotation()
```

```python
# Import necessary libraries
import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import minimize
from ipywidgets import interactive, FloatSlider, IntSlider, Checkbox
from IPython.display import display

# Enable inline plotting
%matplotlib inline

def compute_coordinates_with_deviation_and_rotation(m, n, alpha, delta, y_deviation, theta_max):
    """
    Compute the X and Y coordinates of the Miura-ori pattern with Y-deviation and alternating rotation per row.
    
    Parameters:
    - m (int): Number of rows.
    - n (int): Number of columns.
    - alpha (float): Folding angle in radians.
    - delta (float): Scaling factor per row (0 <= delta <= 1).
    - y_deviation (array-like): Y-deviations for each row.
    - theta_max (float): Maximum rotation angle in radians.
    
    Returns:
    - X (2D numpy array): X coordinates of the grid.
    - Y (2D numpy array): Y coordinates of the grid.
    """
    i_values = np.arange(m+1)
    j_values = np.arange(n+1)

    # Compute scaling factors for each row
    a = 1.0  # Fixed initial horizontal side length
    a_values = a * (1 - delta * (i_values / m))
    a_values = np.maximum(a_values, 1e-3)  # Prevent zero or negative lengths

    # Compute b_i based on alpha to preserve angles
    b_values = a_values * np.tan(alpha)

    # Initialize coordinate grids
    X = np.zeros((m+1, n+1))
    Y = np.zeros((m+1, n+1))

    for i in i_values:
        for j in j_values:
            base_x = j * a_values[i]
            # Shift every other row horizontally by b_i * cos(alpha) to create the Miura-ori pattern
            if i % 2 == 1:
                base_x += b_values[i] * np.cos(alpha)
            # Apply rotation (skew) to the x-coordinate based on the row index
            # Alternating the direction of rotation per row
            rotation_angle = theta_max * (-1)**i  # Alternates sign per row
            skew_offset = y_deviation[i] * np.tan(rotation_angle)
            X[i, j] = base_x + skew_offset
            Y[i, j] = y_deviation[i]

    # Center the pattern horizontally for better visualization
    top_center = (X[0, :].min() + X[0, :].max()) / 2
    bottom_center = (X[m, :].min() + X[m, :].max()) / 2
    midpoint = (top_center + bottom_center) / 2
    X -= midpoint

    return X, Y

def angle_between_vectors(u, v):
    """
    Compute the angle between two vectors in degrees.

    Parameters:
    - u (array-like): First vector.
    - v (array-like): Second vector.

    Returns:
    - angle (float): Angle between u and v in degrees.
    """
    dot_product = np.dot(u, v)
    norm_u = np.linalg.norm(u)
    norm_v = np.linalg.norm(v)
    # Prevent division by zero
    if norm_u == 0 or norm_v == 0:
        return 0.0
    # Compute cosine of the angle and clip to avoid numerical errors
    cos_angle = np.clip(dot_product / (norm_u * norm_v), -1.0, 1.0)
    angle = np.degrees(np.arccos(cos_angle))
    return angle

def compute_angle_deviations(m, n, X, Y):
    """
    Compute the angle deviations from flat foldability at each internal vertex.

    Parameters:
    - m (int): Number of rows.
    - n (int): Number of columns.
    - X (2D numpy array): X coordinates of the grid.
    - Y (2D numpy array): Y coordinates of the grid.

    Returns:
    - deviations (1D numpy array): Array of angle deviations.
    """
    deviations = []
    # Iterate over internal vertices
    for i in range(1, m):
        for j in range(1, n):
            # Get coordinates of the four neighbors
            p = np.array([X[i, j], Y[i, j]])
            p_up = np.array([X[i-1, j], Y[i-1, j]])
            p_right = np.array([X[i, j+1], Y[i, j+1]])
            p_down = np.array([X[i+1, j], Y[i+1, j]])
            p_left = np.array([X[i, j-1], Y[i, j-1]])

            # Vectors from the center to neighbors
            v_up = p_up - p
            v_right = p_right - p
            v_down = p_down - p
            v_left = p_left - p

            # Compute angles between consecutive vectors in clockwise order
            theta_up_right = angle_between_vectors(v_up, v_right)
            theta_right_down = angle_between_vectors(v_right, v_down)
            theta_down_left = angle_between_vectors(v_down, v_left)
            theta_left_up = angle_between_vectors(v_left, v_up)

            # Kawasaki's condition: sum of alternating angles = 180 degrees
            dev1 = (theta_up_right + theta_down_left) - 180.0
            dev2 = (theta_right_down + theta_left_up) - 180.0

            deviations.extend([dev1, dev2])

    return np.array(deviations)

def cost_function_with_deviation_and_rotation(y_deviation, m, n, alpha, delta, theta_max, lambda_reg=1.0):
    """
    Cost function to minimize deviations from flat foldability and Y-deviation magnitude.

    Parameters:
    - y_deviation (array-like): Current Y-deviations for each row.
    - m (int): Number of rows.
    - n (int): Number of columns.
    - alpha (float): Folding angle in radians.
    - delta (float): Scaling factor per row.
    - theta_max (float): Maximum rotation angle in radians.
    - lambda_reg (float): Regularization coefficient for Y-deviation.

    Returns:
    - cost (float): Combined cost of angle deviations and Y-deviation magnitude.
    """
    X, Y = compute_coordinates_with_deviation_and_rotation(m, n, alpha, delta, y_deviation, theta_max)
    deviations = compute_angle_deviations(m, n, X, Y)
    # Sum of squared deviations from flat foldability
    angle_cost = np.sum(deviations**2)
    # Regularization to minimize Y-deviation magnitude
    reg_cost = lambda_reg * np.sum(y_deviation**2)
    # Total cost
    cost = angle_cost + reg_cost
    return cost

def optimize_y_deviation_and_rotation(m, n, alpha, delta, theta_max, lambda_reg=1.0, delta_max=0.1, initial_guess=None):
    """
    Optimize Y-deviations to achieve flat foldability with controlled skew.

    Parameters:
    - m (int): Number of rows.
    - n (int): Number of columns.
    - alpha (float): Folding angle in radians.
    - delta (float): Scaling factor per row.
    - theta_max (float): Maximum rotation angle in radians.
    - lambda_reg (float): Regularization coefficient for Y-deviation.
    - delta_max (float): Maximum allowed Y-deviation.
    - initial_guess (array-like, optional): Initial guess for Y-deviations.

    Returns:
    - result.x (1D numpy array): Optimized Y-deviations.
    - result (OptimizeResult): Full optimization result.
    """
    if initial_guess is None:
        # Initialize Y-deviations based on standard Miura-ori without skew
        y_offsets = [0.0]
        a = 1.0  # Fixed initial horizontal side length
        for i in range(1, m+1):
            a_i = a * (1 - delta * (i / m))
            a_i = max(a_i, 1e-3)
            b_i = a_i * np.tan(alpha)
            y_new = y_offsets[i-1] + b_i * np.sin(theta_max)
            y_offsets.append(y_new)
        initial_guess = np.array(y_offsets)

    # Define bounds: y_deviation[i] should be within [-delta_max, delta_max]
    bounds = [(-delta_max, delta_max) for _ in range(m+1)]

    # Define constraints: y_deviation[i] - y_deviation[i-1] >= epsilon
    # To ensure that y_deviation is monotonically increasing
    constraints = []
    epsilon = 1e-4  # Small positive value to ensure strict inequality
    for i in range(1, m+1):
        constraints.append({
            'type': 'ineq',
            'fun': lambda y, i=i: y[i] - y[i-1] - epsilon  # y[i] > y[i-1] + epsilon
        })

    # Optimization using SLSQP
    result = minimize(
        cost_function_with_deviation_and_rotation,
        initial_guess,
        args=(m, n, alpha, delta, theta_max, lambda_reg),
        method='SLSQP',
        bounds=bounds,
        constraints=constraints,
        options={'maxiter': 10000, 'ftol': 1e-9, 'disp': False}
    )

    # Validate constraints
    def validate_constraints(y_deviation, m, epsilon=1e-4):
        violations = []
        for i in range(1, m+1):
            if y_deviation[i] - y_deviation[i-1] < epsilon:
                violations.append((i, y_deviation[i]))
        if violations:
            print("Constraint violations detected:")
            for violation in violations:
                print(f"Row {violation[0]}: y_deviation = {violation[1]:.6f}")
        else:
            print("All constraints satisfied.")

    validate_constraints(result.x, m)

    return result.x, result

def plot_miura_ori_with_deviation_and_rotation(m, n, alpha, delta, y_deviation, theta_max, show_plot=True, save_svg=False):
    """
    Plot the Miura-ori crease pattern with Y-deviation and alternating rotation per row.

    Parameters:
    - m (int): Number of rows.
    - n (int): Number of columns.
    - alpha (float): Folding angle in radians.
    - delta (float): Scaling factor per row.
    - y_deviation (array-like): Y-deviations for each row.
    - theta_max (float): Maximum rotation angle in radians.
    - show_plot (bool): Whether to display the plot.
    - save_svg (bool): Whether to save the plot as an SVG file.
    """
    X, Y = compute_coordinates_with_deviation_and_rotation(m, n, alpha, delta, y_deviation, theta_max)
    i_values = np.arange(m+1)
    j_values = np.arange(n+1)
    
    fig, ax = plt.subplots(figsize=(8, 8))
    
    # Plot longitudinal folds (vertical lines)
    for j in j_values:
        for i in range(m):
            fold_color = 'blue' if j % 2 == 0 else 'red'
            ax.plot([X[i, j], X[i+1, j]], [Y[i, j], Y[i+1, j]], color=fold_color, lw=2)
    
    # Plot transversal folds (horizontal lines)
    for i in i_values:
        for j in range(n):
            fold_color = 'red' if (i + j) % 2 == 0 else 'blue'
            ax.plot([X[i, j], X[i, j+1]], [Y[i, j], Y[i, j+1]], color=fold_color, lw=2)
    
    # Plot outer boundaries
    for i in range(m):
        # Left boundary
        ax.plot([X[i, 0], X[i+1, 0]], [Y[i, 0], Y[i+1, 0]], color='black', lw=2)
        # Right boundary
        ax.plot([X[i, n], X[i+1, n]], [Y[i, n], Y[i+1, n]], color='black', lw=2)
    for j in range(n):
        # Top boundary
        ax.plot([X[0, j], X[0, j+1]], [Y[0, j], Y[0, j+1]], color='black', lw=2)
        # Bottom boundary
        ax.plot([X[m, j], X[m, j+1]], [Y[m, j], Y[m, j+1]], color='black', lw=2)
    
    # Final plot adjustments
    # ax.set_aspect('equal')
    ax.axis('off')
    # plt.tight_layout()
    
    if show_plot:
        plt.show()
    if save_svg:
        fig.savefig(f'skewed_miura_ori_{m}_{n}_{alpha}_{delta}.svg', format='svg', dpi=300)
        print(f'SVG file saved as "skewed_miura_ori_{m}_{n}_{alpha}_{delta}.svg"')
    plt.close(fig)

def interactive_miura_ori_simplified():
    """
    Create an interactive widget to adjust Miura-ori parameters with Y-deviation and alternating rotation per row and visualize the pattern.
    """
    def update_plot(m=5, n=4, alpha=1.05, delta=0.25, theta_max=0.05, show_plot=True, save_svg=False):
        # Optimize Y-deviations with the given theta_max
        y_deviation_opt, result = optimize_y_deviation_and_rotation(
            m=m,
            n=n,
            alpha=alpha,
            delta=delta,
            theta_max=theta_max,
            lambda_reg=1.0,
            delta_max=0.1
        )
        
        # Display optimization status
        print(f"Optimization successful: {result.success}")
        print(f"Message: {result.message}")
        print(f"Final cost (angle deviations + regularization): {result.fun:.6f}")
        
        if result.success:
            # Plot the crease pattern with Y-deviation and rotation
            plot_miura_ori_with_deviation_and_rotation(
                m=m,
                n=n,
                alpha=alpha,
                delta=delta,
                y_deviation=y_deviation_opt,
                theta_max=theta_max,
                show_plot=show_plot,
                save_svg=save_svg
            )
        else:
            print("Optimization did not converge. Please try adjusting the parameters.")

    interactive_plot = interactive(
        update_plot,
        m=IntSlider(min=3, max=20, step=1, value=5, description='Rows (m)'),
        n=IntSlider(min=2, max=20, step=1, value=4, description='Cols (n)'),
        alpha=FloatSlider(min=0.1, max=np.pi-0.1, step=0.05, value=1.05, description='alpha (rad)'),
        delta=FloatSlider(min=0.0, max=0.8, step=0.05, value=0.25, description='delta'),
        theta_max=FloatSlider(min=0.0, max=np.pi/4, step=0.01, value=0.05, description='X Rotation (rad)'),
        show_plot=Checkbox(value=True, description='Show Plot'),
        save_svg=Checkbox(value=False, description='Save SVG')
    )
    
    display(interactive_plot)

# Initialize the simplified interactive interface
interactive_miura_ori_simplified()
```

```python
# Import necessary libraries
import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import minimize
from ipywidgets import interactive, FloatSlider, IntSlider, Checkbox
from IPython.display import display

# Enable inline plotting
%matplotlib inline

def compute_coordinates_with_deviation_and_rotation(m, n, alpha, delta, y_deviation, theta_max):
    """
    Compute the X and Y coordinates of the Miura-ori pattern with Y-deviation and alternating rotation per column.
    
    Parameters:
    - m (int): Number of rows.
    - n (int): Number of columns.
    - alpha (float): Folding angle in radians.
    - delta (float): Scaling factor per row (0 <= delta <= 1).
    - y_deviation (array-like): Y-deviations for each row.
    - theta_max (float): Maximum rotation angle in radians.
    
    Returns:
    - X (2D numpy array): X coordinates of the grid.
    - Y (2D numpy array): Y coordinates of the grid.
    """
    i_values = np.arange(m+1)
    j_values = np.arange(n+1)

    # Compute scaling factors for each row
    a = 1.0  # Fixed initial horizontal side length
    a_values = a * (1 - delta * (i_values / m))
    a_values = np.maximum(a_values, 1e-3)  # Prevent zero or negative lengths

    # Compute b_i based on alpha to preserve angles
    b_values = a_values * np.tan(alpha)

    # Initialize coordinate grids
    X = np.zeros((m+1, n+1))
    Y = np.zeros((m+1, n+1))

    for i in i_values:
        for j in j_values:
            base_x = j * a_values[i]
            # Shift every other row horizontally by b_i * cos(alpha) to create the Miura-ori pattern
            if i % 2 == 1:
                base_x += b_values[i] * np.cos(alpha)
            # Apply rotation (skew) to the x-coordinate based on the column index
            # Alternating the direction of rotation per column
            rotation_angle = theta_max * (-1)**j  # Alternates sign per column
            skew_offset = y_deviation[i] * np.tan(rotation_angle)
            X[i, j] = base_x + skew_offset
            Y[i, j] = y_deviation[i]

    # Center the pattern horizontally for better visualization
    top_center = (X[0, :].min() + X[0, :].max()) / 2
    bottom_center = (X[m, :].min() + X[m, :].max()) / 2
    midpoint = (top_center + bottom_center) / 2
    X -= midpoint

    return X, Y

def angle_between_vectors(u, v):
    """
    Compute the angle between two vectors in degrees.

    Parameters:
    - u (array-like): First vector.
    - v (array-like): Second vector.

    Returns:
    - angle (float): Angle between u and v in degrees.
    """
    dot_product = np.dot(u, v)
    norm_u = np.linalg.norm(u)
    norm_v = np.linalg.norm(v)
    # Prevent division by zero
    if norm_u == 0 or norm_v == 0:
        return 0.0
    # Compute cosine of the angle and clip to avoid numerical errors
    cos_angle = np.clip(dot_product / (norm_u * norm_v), -1.0, 1.0)
    angle = np.degrees(np.arccos(cos_angle))
    return angle

def compute_angle_deviations(m, n, X, Y):
    """
    Compute the angle deviations from flat foldability at each internal vertex.

    Parameters:
    - m (int): Number of rows.
    - n (int): Number of columns.
    - X (2D numpy array): X coordinates of the grid.
    - Y (2D numpy array): Y coordinates of the grid.

    Returns:
    - deviations (1D numpy array): Array of angle deviations.
    """
    deviations = []
    # Iterate over internal vertices
    for i in range(1, m):
        for j in range(1, n):
            # Get coordinates of the four neighbors
            p = np.array([X[i, j], Y[i, j]])
            p_up = np.array([X[i-1, j], Y[i-1, j]])
            p_right = np.array([X[i, j+1], Y[i, j+1]])
            p_down = np.array([X[i+1, j], Y[i+1, j]])
            p_left = np.array([X[i, j-1], Y[i, j-1]])

            # Vectors from the center to neighbors
            v_up = p_up - p
            v_right = p_right - p
            v_down = p_down - p
            v_left = p_left - p

            # Compute angles between consecutive vectors in clockwise order
            theta_up_right = angle_between_vectors(v_up, v_right)
            theta_right_down = angle_between_vectors(v_right, v_down)
            theta_down_left = angle_between_vectors(v_down, v_left)
            theta_left_up = angle_between_vectors(v_left, v_up)

            # Kawasaki's condition: sum of alternating angles = 180 degrees
            dev1 = (theta_up_right + theta_down_left) - 180.0
            dev2 = (theta_right_down + theta_left_up) - 180.0

            deviations.extend([dev1, dev2])

    return np.array(deviations)

def cost_function_with_deviation_and_rotation(y_deviation, m, n, alpha, delta, theta_max, lambda_reg=1.0):
    """
    Cost function to minimize deviations from flat foldability and Y-deviation magnitude.

    Parameters:
    - y_deviation (array-like): Current Y-deviations for each row.
    - m (int): Number of rows.
    - n (int): Number of columns.
    - alpha (float): Folding angle in radians.
    - delta (float): Scaling factor per row.
    - theta_max (float): Maximum rotation angle in radians.
    - lambda_reg (float): Regularization coefficient for Y-deviation.

    Returns:
    - cost (float): Combined cost of angle deviations and Y-deviation magnitude.
    """
    X, Y = compute_coordinates_with_deviation_and_rotation(m, n, alpha, delta, y_deviation, theta_max)
    deviations = compute_angle_deviations(m, n, X, Y)
    # Sum of squared deviations from flat foldability
    angle_cost = np.sum(deviations**2)
    # Regularization to minimize Y-deviation magnitude
    reg_cost = lambda_reg * np.sum(y_deviation**2)
    # Total cost
    cost = angle_cost + reg_cost
    return cost

def optimize_y_deviation_and_rotation(m, n, alpha, delta, theta_max, lambda_reg=1.0, delta_max=0.1, initial_guess=None):
    """
    Optimize Y-deviations to achieve flat foldability with controlled skew.

    Parameters:
    - m (int): Number of rows.
    - n (int): Number of columns.
    - alpha (float): Folding angle in radians.
    - delta (float): Scaling factor per row.
    - theta_max (float): Maximum rotation angle in radians.
    - lambda_reg (float): Regularization coefficient for Y-deviation.
    - delta_max (float): Maximum allowed Y-deviation.
    - initial_guess (array-like, optional): Initial guess for Y-deviations.

    Returns:
    - result.x (1D numpy array): Optimized Y-deviations.
    - result (OptimizeResult): Full optimization result.
    """
    if initial_guess is None:
        # Initialize Y-deviations based on standard Miura-ori without skew
        y_offsets = [0.0]
        a = 1.0  # Fixed initial horizontal side length
        for i in range(1, m+1):
            a_i = a * (1 - delta * (i / m))
            a_i = max(a_i, 1e-3)
            b_i = a_i * np.tan(alpha)
            y_new = y_offsets[i-1] + b_i * np.sin(alpha)
            y_offsets.append(y_new)
        initial_guess = np.array(y_offsets)

    # Define bounds: y_deviation[i] should be within [-delta_max, delta_max]
    bounds = [(-delta_max, delta_max) for _ in range(m+1)]

    # Define constraints: y_deviation[i] - y_deviation[i-1] >= epsilon
    # To ensure that y_deviation is monotonically increasing
    constraints = []
    epsilon = 1e-4  # Small positive value to ensure strict inequality
    for i in range(1, m+1):
        constraints.append({
            'type': 'ineq',
            'fun': lambda y, i=i: y[i] - y[i-1] - epsilon  # y[i] > y[i-1] + epsilon
        })

    # Optimization using SLSQP
    result = minimize(
        cost_function_with_deviation_and_rotation,
        initial_guess,
        args=(m, n, alpha, delta, theta_max, lambda_reg),
        method='SLSQP',
        bounds=bounds,
        constraints=constraints,
        options={'maxiter': 10000, 'ftol': 1e-9, 'disp': False}
    )

    # Validate constraints
    def validate_constraints(y_deviation, m, epsilon=1e-4):
        violations = []
        for i in range(1, m+1):
            if y_deviation[i] - y_deviation[i-1] < epsilon:
                violations.append((i, y_deviation[i]))
        if violations:
            print("Constraint violations detected:")
            for violation in violations:
                print(f"Row {violation[0]}: y_deviation = {violation[1]:.6f}")
        else:
            print("All constraints satisfied.")

    validate_constraints(result.x, m)

    return result.x, result

def plot_miura_ori_with_deviation_and_rotation(m, n, alpha, delta, y_deviation, theta_max, show_plot=True, save_svg=False):
    """
    Plot the Miura-ori crease pattern with Y-deviation and alternating rotation per column.

    Parameters:
    - m (int): Number of rows.
    - n (int): Number of columns.
    - alpha (float): Folding angle in radians.
    - delta (float): Scaling factor per row.
    - y_deviation (array-like): Y-deviations for each row.
    - theta_max (float): Maximum rotation angle in radians.
    - show_plot (bool): Whether to display the plot.
    - save_svg (bool): Whether to save the plot as an SVG file.
    """
    X, Y = compute_coordinates_with_deviation_and_rotation(m, n, alpha, delta, y_deviation, theta_max)
    i_values = np.arange(m+1)
    j_values = np.arange(n+1)
    
    fig, ax = plt.subplots(figsize=(8, 8))
    
    # Plot longitudinal folds (vertical lines)
    for j in j_values:
        for i in range(m):
            fold_color = 'blue' if j % 2 == 0 else 'red'
            ax.plot([X[i, j], X[i+1, j]], [Y[i, j], Y[i+1, j]], color=fold_color, lw=2)
    
    # Plot transversal folds (horizontal lines)
    for i in i_values:
        for j in range(n):
            fold_color = 'red' if (i + j) % 2 == 0 else 'blue'
            ax.plot([X[i, j], X[i, j+1]], [Y[i, j], Y[i, j+1]], color=fold_color, lw=2)
    
    # Plot outer boundaries
    for i in range(m):
        # Left boundary
        ax.plot([X[i, 0], X[i+1, 0]], [Y[i, 0], Y[i+1, 0]], color='black', lw=2)
        # Right boundary
        ax.plot([X[i, n], X[i+1, n]], [Y[i, n], Y[i+1, n]], color='black', lw=2)
    for j in range(n):
        # Top boundary
        ax.plot([X[0, j], X[0, j+1]], [Y[0, j], Y[0, j+1]], color='black', lw=2)
        # Bottom boundary
        ax.plot([X[m, j], X[m, j+1]], [Y[m, j], Y[m, j+1]], color='black', lw=2)
    
    # Final plot adjustments
    # ax.set_aspect('equal')
    ax.axis('off')
    # plt.tight_layout()
    
    if show_plot:
        plt.show()
    if save_svg:
        fig.savefig(f'skewed_miura_ori_{m}_{n}_{alpha}_{delta}.svg', format='svg', dpi=300)
        print(f'SVG file saved as "skewed_miura_ori_{m}_{n}_{alpha}_{delta}.svg"')
    plt.close(fig)

def interactive_miura_ori_simplified():
    """
    Create an interactive widget to adjust Miura-ori parameters with Y-deviation and alternating rotation per column and visualize the pattern.
    """
    def update_plot(m=5, n=4, alpha=1.05, delta=0.25, theta_max=0.05, show_plot=True, save_svg=False):
        # Optimize Y-deviations with the given theta_max
        y_deviation_opt, result = optimize_y_deviation_and_rotation(
            m=m,
            n=n,
            alpha=alpha,
            delta=delta,
            theta_max=theta_max,
            lambda_reg=1.0,
            delta_max=0.1
        )
        
        # Display optimization status
        print(f"Optimization successful: {result.success}")
        print(f"Message: {result.message}")
        print(f"Final cost (angle deviations + regularization): {result.fun:.6f}")
        
        if result.success:
            # Plot the crease pattern with Y-deviation and rotation
            plot_miura_ori_with_deviation_and_rotation(
                m=m,
                n=n,
                alpha=alpha,
                delta=delta,
                y_deviation=y_deviation_opt,
                theta_max=theta_max,
                show_plot=show_plot,
                save_svg=save_svg
            )
        else:
            print("Optimization did not converge. Please try adjusting the parameters.")

    interactive_plot = interactive(
        update_plot,
        m=IntSlider(min=3, max=20, step=1, value=5, description='Rows (m)'),
        n=IntSlider(min=2, max=20, step=1, value=4, description='Cols (n)'),
        alpha=FloatSlider(min=0.1, max=np.pi-0.1, step=0.05, value=1.05, description='alpha (rad)'),
        delta=FloatSlider(min=0.0, max=0.8, step=0.05, value=0.25, description='delta'),
        theta_max=FloatSlider(min=0.0, max=np.pi/4, step=0.01, value=0.05, description='X Rotation (rad)'),
        show_plot=Checkbox(value=True, description='Show Plot'),
        save_svg=Checkbox(value=False, description='Save SVG')
    )
    
    display(interactive_plot)

# Initialize the simplified interactive interface
interactive_miura_ori_simplified()
```

```python
# Import necessary libraries
import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import minimize
from ipywidgets import interactive, FloatSlider, IntSlider, Checkbox
from IPython.display import display

# Enable inline plotting
%matplotlib inline

def compute_coordinates_with_deviation_and_rotation(m, n, alpha, delta, y_deviation, theta_max):
    """
    Compute the X and Y coordinates of the Miura-ori pattern with Y-deviation and alternating rotation per column.
    
    Parameters:
    - m (int): Number of rows.
    - n (int): Number of columns.
    - alpha (float): Folding angle in radians (applied to horizontal parallels).
    - delta (float): Scaling factor per row (0 <= delta <= 1).
    - y_deviation (array-like): Y-deviations for each row.
    - theta_max (float): Maximum rotation angle in radians (alternates per column).
    
    Returns:
    - X (2D numpy array): X coordinates of the grid.
    - Y (2D numpy array): Y coordinates of the grid.
    """
    i_values = np.arange(m+1)
    j_values = np.arange(n+1)

    # Compute scaling factors for each row
    a = 1.0  # Fixed initial horizontal side length
    a_values = a * (1 - delta * (i_values / m))
    a_values = np.maximum(a_values, 1e-3)  # Prevent zero or negative lengths

    # Compute b_i based on alpha to preserve angles
    b_values = a_values * np.tan(alpha)

    # Initialize coordinate grids
    X = np.zeros((m+1, n+1))
    Y = np.zeros((m+1, n+1))

    for i in i_values:
        for j in j_values:
            base_y = j * a_values[i]
            # Shift every other column horizontally by b_i * cos(alpha) to create the Miura-ori pattern
            if j % 2 == 1:
                base_y += b_values[i] * np.cos(alpha)
            # Apply rotation (skew) to the y-coordinate based on the column index
            # Alternating the direction of rotation per column
            rotation_angle = theta_max * (-1)**j  # Alternates sign per column
            skew_offset = y_deviation[i] * np.tan(rotation_angle)
            Y[i, j] = base_y + skew_offset
            X[i, j] = y_deviation[i]

    # Center the pattern vertically for better visualization
    left_center = (X[:,0].min() + X[:,0].max()) / 2
    right_center = (X[:,n].min() + X[:,n].max()) / 2
    midpoint = (left_center + right_center) / 2
    X -= midpoint

    return X, Y

def angle_between_vectors(u, v):
    """
    Compute the angle between two vectors in degrees.

    Parameters:
    - u (array-like): First vector.
    - v (array-like): Second vector.

    Returns:
    - angle (float): Angle between u and v in degrees.
    """
    dot_product = np.dot(u, v)
    norm_u = np.linalg.norm(u)
    norm_v = np.linalg.norm(v)
    # Prevent division by zero
    if norm_u == 0 or norm_v == 0:
        return 0.0
    # Compute cosine of the angle and clip to avoid numerical errors
    cos_angle = np.clip(dot_product / (norm_u * norm_v), -1.0, 1.0)
    angle = np.degrees(np.arccos(cos_angle))
    return angle

def compute_angle_deviations(m, n, X, Y):
    """
    Compute the angle deviations from flat foldability at each internal vertex.

    Parameters:
    - m (int): Number of rows.
    - n (int): Number of columns.
    - X (2D numpy array): X coordinates of the grid.
    - Y (2D numpy array): Y coordinates of the grid.

    Returns:
    - deviations (1D numpy array): Array of angle deviations.
    """
    deviations = []
    # Iterate over internal vertices
    for i in range(1, m):
        for j in range(1, n):
            # Get coordinates of the four neighbors
            p = np.array([X[i, j], Y[i, j]])
            p_up = np.array([X[i-1, j], Y[i-1, j]])
            p_right = np.array([X[i, j+1], Y[i, j+1]])
            p_down = np.array([X[i+1, j], Y[i+1, j]])
            p_left = np.array([X[i, j-1], Y[i, j-1]])

            # Vectors from the center to neighbors
            v_up = p_up - p
            v_right = p_right - p
            v_down = p_down - p
            v_left = p_left - p

            # Compute angles between consecutive vectors in clockwise order
            theta_up_right = angle_between_vectors(v_up, v_right)
            theta_right_down = angle_between_vectors(v_right, v_down)
            theta_down_left = angle_between_vectors(v_down, v_left)
            theta_left_up = angle_between_vectors(v_left, v_up)

            # Kawasaki's condition: sum of alternating angles = 180 degrees
            dev1 = (theta_up_right + theta_down_left) - 180.0
            dev2 = (theta_right_down + theta_left_up) - 180.0

            deviations.extend([dev1, dev2])

    return np.array(deviations)

def cost_function_with_deviation_and_rotation(y_deviation, m, n, alpha, delta, theta_max, lambda_reg=1.0):
    """
    Cost function to minimize deviations from flat foldability and Y-deviation magnitude.

    Parameters:
    - y_deviation (array-like): Current Y-deviations for each row.
    - m (int): Number of rows.
    - n (int): Number of columns.
    - alpha (float): Folding angle in radians.
    - delta (float): Scaling factor per row.
    - theta_max (float): Maximum rotation angle in radians.
    - lambda_reg (float): Regularization coefficient for Y-deviation.

    Returns:
    - cost (float): Combined cost of angle deviations and Y-deviation magnitude.
    """
    X, Y = compute_coordinates_with_deviation_and_rotation(m, n, alpha, delta, y_deviation, theta_max)
    deviations = compute_angle_deviations(m, n, X, Y)
    # Sum of squared deviations from flat foldability
    angle_cost = np.sum(deviations**2)
    # Regularization to minimize Y-deviation magnitude
    reg_cost = lambda_reg * np.sum(y_deviation**2)
    # Total cost
    cost = angle_cost + reg_cost
    return cost

def optimize_y_deviation_and_rotation(m, n, alpha, delta, theta_max, lambda_reg=1.0, delta_max=0.1, initial_guess=None):
    """
    Optimize Y-deviations to achieve flat foldability with controlled skew.

    Parameters:
    - m (int): Number of rows.
    - n (int): Number of columns.
    - alpha (float): Folding angle in radians.
    - delta (float): Scaling factor per row.
    - theta_max (float): Maximum rotation angle in radians.
    - lambda_reg (float): Regularization coefficient for Y-deviation.
    - delta_max (float): Maximum allowed Y-deviation.
    - initial_guess (array-like, optional): Initial guess for Y-deviations.

    Returns:
    - result.x (1D numpy array): Optimized Y-deviations.
    - result (OptimizeResult): Full optimization result.
    """
    if initial_guess is None:
        # Initialize Y-deviations based on standard Miura-ori without skew
        y_offsets = [0.0]
        a = 1.0  # Fixed initial horizontal side length
        for i in range(1, m+1):
            a_i = a * (1 - delta * (i / m))
            a_i = max(a_i, 1e-3)
            b_i = a_i * np.tan(alpha)
            y_new = y_offsets[i-1] + b_i * np.sin(alpha)
            y_offsets.append(y_new)
        initial_guess = np.array(y_offsets)

    # Define bounds: y_deviation[i] should be within [-delta_max, delta_max]
    bounds = [(-delta_max, delta_max) for _ in range(m+1)]

    # Define constraints: y_deviation[i] - y_deviation[i-1] >= epsilon
    # To ensure that y_deviation is monotonically increasing
    constraints = []
    epsilon = 1e-4  # Small positive value to ensure strict inequality
    for i in range(1, m+1):
        constraints.append({
            'type': 'ineq',
            'fun': lambda y, i=i: y[i] - y[i-1] - epsilon  # y[i] > y[i-1] + epsilon
        })

    # Optimization using SLSQP
    result = minimize(
        cost_function_with_deviation_and_rotation,
        initial_guess,
        args=(m, n, alpha, delta, theta_max, lambda_reg),
        method='SLSQP',
        bounds=bounds,
        constraints=constraints,
        options={'maxiter': 10000, 'ftol': 1e-9, 'disp': False}
    )

    # Validate constraints
    def validate_constraints(y_deviation, m, epsilon=1e-4):
        violations = []
        for i in range(1, m+1):
            if y_deviation[i] - y_deviation[i-1] < epsilon:
                violations.append((i, y_deviation[i]))
        if violations:
            print("Constraint violations detected:")
            for violation in violations:
                print(f"Row {violation[0]}: y_deviation = {violation[1]:.6f}")
        else:
            print("All constraints satisfied.")

    validate_constraints(result.x, m)

    return result.x, result

def plot_miura_ori_with_deviation_and_rotation(m, n, alpha, delta, y_deviation, theta_max, show_plot=True, save_svg=False):
    """
    Plot the Miura-ori crease pattern with Y-deviation and alternating rotation per column.

    Parameters:
    - m (int): Number of rows.
    - n (int): Number of columns.
    - alpha (float): Folding angle in radians (applied to horizontal parallels).
    - delta (float): Scaling factor per row.
    - y_deviation (array-like): Y-deviations for each row.
    - theta_max (float): Maximum rotation angle in radians.
    - show_plot (bool): Whether to display the plot.
    - save_svg (bool): Whether to save the plot as an SVG file.
    """
    X, Y = compute_coordinates_with_deviation_and_rotation(m, n, alpha, delta, y_deviation, theta_max)
    i_values = np.arange(m+1)
    j_values = np.arange(n+1)
    
    fig, ax = plt.subplots(figsize=(8, 8))
    
    # Plot longitudinal folds (vertical lines)
    for j in j_values:
        for i in range(m):
            fold_color = 'blue' if j % 2 == 0 else 'red'
            ax.plot([X[i, j], X[i+1, j]], [Y[i, j], Y[i+1, j]], color=fold_color, lw=2)
    
    # Plot transversal folds (horizontal lines)
    for i in i_values:
        for j in range(n):
            fold_color = 'red' if (i + j) % 2 == 0 else 'blue'
            ax.plot([X[i, j], X[i, j+1]], [Y[i, j], Y[i, j+1]], color=fold_color, lw=2)
    
    # Plot outer boundaries
    for i in range(m):
        # Left boundary
        ax.plot([X[i, 0], X[i+1, 0]], [Y[i, 0], Y[i+1, 0]], color='black', lw=2)
        # Right boundary
        ax.plot([X[i, n], X[i+1, n]], [Y[i, n], Y[i+1, n]], color='black', lw=2)
    for j in range(n):
        # Top boundary
        ax.plot([X[0, j], X[0, j+1]], [Y[0, j], Y[0, j+1]], color='black', lw=2)
        # Bottom boundary
        ax.plot([X[m, j], X[m, j+1]], [Y[m, j], Y[m, j+1]], color='black', lw=2)
    
    # Final plot adjustments
    ax.axis('off')

    
    if show_plot:
        plt.show()
    if save_svg:
        fig.savefig(f'skewed_miura_ori_{m}_{n}_{alpha}_{delta}.svg', format='svg', dpi=300)
        print(f'SVG file saved as "skewed_miura_ori_{m}_{n}_{alpha}_{delta}.svg"')
    plt.close(fig)

def interactive_miura_ori_simplified():
    """
    Create an interactive widget to adjust Miura-ori parameters with Y-deviation and alternating rotation per column and visualize the pattern.
    """
    def update_plot(m=5, n=4, alpha=1.05, delta=0.25, theta_max=0.05, show_plot=True, save_svg=False):
        # Optimize Y-deviations with the given theta_max
        y_deviation_opt, result = optimize_y_deviation_and_rotation(
            m=m,
            n=n,
            alpha=alpha,
            delta=delta,
            theta_max=theta_max,
            lambda_reg=1.0,
            delta_max=0.1
        )
        
        # Display optimization status
        print(f"Optimization successful: {result.success}")
        print(f"Message: {result.message}")
        print(f"Final cost (angle deviations + regularization): {result.fun:.6f}")
        
        if result.success:
            # Plot the crease pattern with Y-deviation and rotation
            plot_miura_ori_with_deviation_and_rotation(
                m=m,
                n=n,
                alpha=alpha,
                delta=delta,
                y_deviation=y_deviation_opt,
                theta_max=theta_max,
                show_plot=show_plot,
                save_svg=save_svg
            )
        else:
            print("Optimization did not converge. Please try adjusting the parameters.")

    interactive_plot = interactive(
        update_plot,
        m=IntSlider(min=3, max=20, step=1, value=5, description='Rows (m)'),
        n=IntSlider(min=2, max=20, step=1, value=4, description='Cols (n)'),
        alpha=FloatSlider(min=0.1, max=np.pi-0.1, step=0.05, value=1.05, description='alpha (rad)'),
        delta=FloatSlider(min=0.0, max=0.8, step=0.05, value=0.25, description='delta'),
        theta_max=FloatSlider(min=0.0, max=np.pi/4, step=0.01, value=0.05, description='X Rotation (rad)'),
        show_plot=Checkbox(value=True, description='Show Plot'),
        save_svg=Checkbox(value=False, description='Save SVG')
    )
    
    display(interactive_plot)

# Initialize the simplified interactive interface
interactive_miura_ori_simplified()
```

```python
# Import necessary libraries
import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import minimize
from ipywidgets import interactive, FloatSlider, IntSlider, Checkbox
from IPython.display import display

# Enable inline plotting
%matplotlib inline

def compute_coordinates_with_deviation_and_rotation(m, n, alpha, delta, y_deviation, theta_max):
    """
    Compute the X and Y coordinates of the Miura-ori pattern with Y-deviation and alternating rotation per row.
    
    Parameters:
    - m (int): Number of rows.
    - n (int): Number of columns.
    - alpha (float): Folding angle in radians (applied to vertical parallels).
    - delta (float): Scaling factor per row (0 <= delta <= 1).
    - y_deviation (array-like): Y-deviations for each row.
    - theta_max (float): Maximum rotation angle in radians (alternates per row).
    
    Returns:
    - X (2D numpy array): X coordinates of the grid.
    - Y (2D numpy array): Y coordinates of the grid.
    """
    # Initialize grids
    X = np.zeros((m+1, n+1))
    Y = np.zeros((m+1, n+1))
    
    # Base vertical side length
    a = 1.0  # You can adjust this as needed
    
    for i in range(m+1):
        # Scaling factor for the current row
        a_i = a * (1 - delta * (i / m)) if m != 0 else a
        a_i = max(a_i, 1e-3)  # Prevent zero or negative lengths
        
        # Horizontal side length based on alpha
        b_i = a_i * np.tan(alpha)
        
        for j in range(n+1):
            # Base X-coordinate
            base_x = j * b_i
            
            # Every other row is shifted horizontally to create the Miura-ori pattern
            if i % 2 == 1:
                base_x += b_i / 2  # Shift by half the horizontal side length
                
            # Apply alternating rotation per row
            rotation_angle = theta_max * (-1)**i  # Alternates sign per row
            skew_offset = y_deviation[i] * np.tan(rotation_angle)
            
            # Assign coordinates
            X[i, j] = base_x + skew_offset
            Y[i, j] = y_deviation[i]
    
    # Center the pattern horizontally for better visualization
    X -= np.mean(X)
    
    return X, Y

def angle_between_vectors(u, v):
    """
    Compute the angle between two vectors in degrees.

    Parameters:
    - u (array-like): First vector.
    - v (array-like): Second vector.

    Returns:
    - angle (float): Angle between u and v in degrees.
    """
    dot_product = np.dot(u, v)
    norm_u = np.linalg.norm(u)
    norm_v = np.linalg.norm(v)
    # Prevent division by zero
    if norm_u == 0 or norm_v == 0:
        return 0.0
    # Compute cosine of the angle and clip to avoid numerical errors
    cos_angle = np.clip(dot_product / (norm_u * norm_v), -1.0, 1.0)
    angle = np.degrees(np.arccos(cos_angle))
    return angle

def compute_angle_deviations(m, n, X, Y):
    """
    Compute the angle deviations from flat foldability at each internal vertex.

    Parameters:
    - m (int): Number of rows.
    - n (int): Number of columns.
    - X (2D numpy array): X coordinates of the grid.
    - Y (2D numpy array): Y coordinates of the grid.

    Returns:
    - deviations (1D numpy array): Array of angle deviations.
    """
    deviations = []
    # Iterate over internal vertices
    for i in range(1, m):
        for j in range(1, n):
            # Get coordinates of the four neighbors
            p = np.array([X[i, j], Y[i, j]])
            p_up = np.array([X[i-1, j], Y[i-1, j]])
            p_right = np.array([X[i, j+1], Y[i, j+1]])
            p_down = np.array([X[i+1, j], Y[i+1, j]])
            p_left = np.array([X[i, j-1], Y[i, j-1]])

            # Vectors from the center to neighbors
            v_up = p_up - p
            v_right = p_right - p
            v_down = p_down - p
            v_left = p_left - p

            # Compute angles between consecutive vectors in clockwise order
            theta_up_right = angle_between_vectors(v_up, v_right)
            theta_right_down = angle_between_vectors(v_right, v_down)
            theta_down_left = angle_between_vectors(v_down, v_left)
            theta_left_up = angle_between_vectors(v_left, v_up)

            # Kawasaki's condition: sum of alternating angles = 180 degrees
            dev1 = (theta_up_right + theta_down_left) - 180.0
            dev2 = (theta_right_down + theta_left_up) - 180.0

            deviations.extend([dev1, dev2])

    return np.array(deviations)

def cost_function_with_deviation_and_rotation(y_deviation, m, n, alpha, delta, theta_max, lambda_reg=1.0):
    """
    Cost function to minimize deviations from flat foldability and Y-deviation magnitude.

    Parameters:
    - y_deviation (array-like): Current Y-deviations for each row.
    - m (int): Number of rows.
    - n (int): Number of columns.
    - alpha (float): Folding angle in radians.
    - delta (float): Scaling factor per row.
    - theta_max (float): Maximum rotation angle in radians.
    - lambda_reg (float): Regularization coefficient for Y-deviation.

    Returns:
    - cost (float): Combined cost of angle deviations and Y-deviation magnitude.
    """
    X, Y = compute_coordinates_with_deviation_and_rotation(m, n, alpha, delta, y_deviation, theta_max)
    deviations = compute_angle_deviations(m, n, X, Y)
    # Sum of squared deviations from flat foldability
    angle_cost = np.sum(deviations**2)
    # Regularization to minimize Y-deviation magnitude
    reg_cost = lambda_reg * np.sum(y_deviation**2)
    # Total cost
    cost = angle_cost + reg_cost
    return cost

def optimize_y_deviation_and_rotation(m, n, alpha, delta, theta_max, lambda_reg=1.0, delta_max=0.1, initial_guess=None):
    """
    Optimize Y-deviations to achieve flat foldability with controlled skew.

    Parameters:
    - m (int): Number of rows.
    - n (int): Number of columns.
    - alpha (float): Folding angle in radians.
    - delta (float): Scaling factor per row.
    - theta_max (float): Maximum rotation angle in radians.
    - lambda_reg (float): Regularization coefficient for Y-deviation.
    - delta_max (float): Maximum allowed Y-deviation.
    - initial_guess (array-like, optional): Initial guess for Y-deviations.

    Returns:
    - result.x (1D numpy array): Optimized Y-deviations.
    - result (OptimizeResult): Full optimization result.
    """
    if initial_guess is None:
        # Initialize Y-deviations based on standard Miura-ori without skew
        y_offsets = [0.0]
        a = 1.0  # Fixed initial vertical side length
        for i in range(1, m+1):
            a_i = a * (1 - delta * (i / m))
            a_i = max(a_i, 1e-3)
            b_i = a_i * np.tan(alpha)
            y_new = y_offsets[i-1] + b_i * np.sin(alpha)
            y_offsets.append(y_new)
        initial_guess = np.array(y_offsets)

    # Define bounds: y_deviation[i] should be within [-delta_max, delta_max]
    bounds = [(-delta_max, delta_max) for _ in range(m+1)]

    # Define constraints: y_deviation[i] - y_deviation[i-1] >= epsilon
    # To ensure that y_deviation is monotonically increasing
    constraints = []
    epsilon = 1e-4  # Small positive value to ensure strict inequality
    for i in range(1, m+1):
        constraints.append({
            'type': 'ineq',
            'fun': lambda y, i=i: y[i] - y[i-1] - epsilon  # y[i] > y[i-1] + epsilon
        })

    # Optimization using SLSQP
    result = minimize(
        cost_function_with_deviation_and_rotation,
        initial_guess,
        args=(m, n, alpha, delta, theta_max, lambda_reg),
        method='SLSQP',
        bounds=bounds,
        constraints=constraints,
        options={'maxiter': 10000, 'ftol': 1e-9, 'disp': False}
    )

    # Validate constraints
    def validate_constraints(y_deviation, m, epsilon=1e-4):
        violations = []
        for i in range(1, m+1):
            if y_deviation[i] - y_deviation[i-1] < epsilon:
                violations.append((i, y_deviation[i]))
        if violations:
            print("Constraint violations detected:")
            for violation in violations:
                print(f"Row {violation[0]}: y_deviation = {violation[1]:.6f}")
        else:
            print("All constraints satisfied.")

    validate_constraints(result.x, m)

    return result.x, result

def plot_miura_ori_with_deviation_and_rotation(m, n, alpha, delta, y_deviation, theta_max, show_plot=True, save_svg=False):
    """
    Plot the Miura-ori crease pattern with Y-deviation and alternating rotation per row.

    Parameters:
    - m (int): Number of rows.
    - n (int): Number of columns.
    - alpha (float): Folding angle in radians (applied to vertical parallels).
    - delta (float): Scaling factor per row.
    - y_deviation (array-like): Y-deviations for each row.
    - theta_max (float): Maximum rotation angle in radians.
    - show_plot (bool): Whether to display the plot.
    - save_svg (bool): Whether to save the plot as an SVG file.
    """
    X, Y = compute_coordinates_with_deviation_and_rotation(m, n, alpha, delta, y_deviation, theta_max)
    i_values = np.arange(m+1)
    j_values = np.arange(n+1)
    
    fig, ax = plt.subplots(figsize=(8, 8))
    
    # Plot longitudinal folds (vertical lines)
    for j in j_values:
        for i in range(m):
            fold_color = 'blue' if j % 2 == 0 else 'red'
            ax.plot([X[i, j], X[i+1, j]], [Y[i, j], Y[i+1, j]], color=fold_color, lw=2)
    
    # Plot transversal folds (horizontal lines)
    for i in i_values:
        for j in range(n):
            fold_color = 'red' if (i + j) % 2 == 0 else 'blue'
            ax.plot([X[i, j], X[i, j+1]], [Y[i, j], Y[i, j+1]], color=fold_color, lw=2)
    
    # Plot outer boundaries
    for i in range(m):
        # Left boundary
        ax.plot([X[i, 0], X[i+1, 0]], [Y[i, 0], Y[i+1, 0]], color='black', lw=2)
        # Right boundary
        ax.plot([X[i, n], X[i+1, n]], [Y[i, n], Y[i+1, n]], color='black', lw=2)
    for j in range(n):
        # Top boundary
        ax.plot([X[0, j], X[0, j+1]], [Y[0, j], Y[0, j+1]], color='black', lw=2)
        # Bottom boundary
        ax.plot([X[m, j], X[m, j+1]], [Y[m, j], Y[m, j+1]], color='black', lw=2)
    
    # Final plot adjustments
    ax.axis('off')
    
    if show_plot:
        plt.show()
    if save_svg:
        fig.savefig(f'skewed_miura_ori_{m}_{n}_{alpha:.2f}_{delta:.2f}.svg', format='svg', dpi=300)
        print(f'SVG file saved as "skewed_miura_ori_{m}_{n}_{alpha:.2f}_{delta:.2f}.svg"')
    plt.close(fig)

def interactive_miura_ori_simplified():
    """
    Create an interactive widget to adjust Miura-ori parameters with Y-deviation and alternating rotation per row and visualize the pattern.
    """
    def update_plot(m=5, n=4, alpha=1.05, delta=0.25, theta_max=0.05, show_plot=True, save_svg=False):
        # Optimize Y-deviations with the given theta_max
        y_deviation_opt, result = optimize_y_deviation_and_rotation(
            m=m,
            n=n,
            alpha=alpha,
            delta=delta,
            theta_max=theta_max,
            lambda_reg=1.0,
            delta_max=0.1
        )
        
        # Display optimization status
        print(f"Optimization successful: {result.success}")
        print(f"Message: {result.message}")
        print(f"Final cost (angle deviations + regularization): {result.fun:.6f}")
        
        if result.success:
            # Plot the crease pattern with Y-deviation and rotation
            plot_miura_ori_with_deviation_and_rotation(
                m=m,
                n=n,
                alpha=alpha,
                delta=delta,
                y_deviation=y_deviation_opt,
                theta_max=theta_max,
                show_plot=show_plot,
                save_svg=save_svg
            )
        else:
            print("Optimization did not converge. Please try adjusting the parameters.")

    interactive_plot = interactive(
        update_plot,
        m=IntSlider(min=3, max=20, step=1, value=5, description='Rows (m)'),
        n=IntSlider(min=2, max=20, step=1, value=4, description='Cols (n)'),
        alpha=FloatSlider(min=0.1, max=np.pi-0.1, step=0.05, value=1.05, description='alpha (rad)'),
        delta=FloatSlider(min=0.0, max=0.8, step=0.05, value=0.25, description='delta'),
        theta_max=FloatSlider(min=0.0, max=np.pi/4, step=0.01, value=0.05, description='X Rotation (rad)'),
        show_plot=Checkbox(value=True, description='Show Plot'),
        save_svg=Checkbox(value=False, description='Save SVG')
    )
    
    display(interactive_plot)

# Initialize the simplified interactive interface
interactive_miura_ori_simplified()
```

```python
# Import necessary libraries
import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import minimize
from ipywidgets import interactive, FloatSlider, IntSlider, Checkbox
from IPython.display import display

# Enable inline plotting
%matplotlib inline

def compute_coordinates_with_deviation_and_rotation(m, n, alpha, delta, y_deviation, theta_max):
    """
    Compute the X and Y coordinates of the Miura-ori pattern with Y-deviation and alternating rotation per column.
    
    Parameters:
    - m (int): Number of rows.
    - n (int): Number of columns.
    - alpha (float): Folding angle in radians (applied to vertical parallels).
    - delta (float): Scaling factor per column (0 <= delta <= 1).
    - y_deviation (array-like): Y-deviations for each column.
    - theta_max (float): Maximum rotation angle in radians (alternates per column).
    
    Returns:
    - X (2D numpy array): X coordinates of the grid.
    - Y (2D numpy array): Y coordinates of the grid.
    """
    # Initialize grids
    X = np.zeros((m+1, n+1))
    Y = np.zeros((m+1, n+1))
    
    # Base vertical side length
    a = 1.0  # You can adjust this as needed
    
    for j in range(n+1):
        # Scaling factor for the current column
        a_j = a * (1 - delta * (j / n)) if n != 0 else a
        a_j = max(a_j, 1e-3)  # Prevent zero or negative lengths
        
        # Horizontal side length based on alpha
        b_j = a_j * np.tan(alpha)
        
        for i in range(m+1):
            # Base Y-coordinate
            base_y = i * a_j
            
            # Every other column is shifted vertically to create the Miura-ori pattern
            if j % 2 == 1:
                base_y += b_j / 2  # Shift by half the horizontal side length
            
            # Apply alternating rotation per column
            rotation_angle = theta_max * (-1)**j  # Alternates sign per column
            skew_offset = y_deviation[j] * np.tan(rotation_angle)
            
            # Assign coordinates
            X[i, j] = y_deviation[j] + skew_offset
            Y[i, j] = base_y
    
    # Center the pattern vertically for better visualization
    Y -= np.mean(Y)
    
    return X, Y

def angle_between_vectors(u, v):
    """
    Compute the angle between two vectors in degrees.

    Parameters:
    - u (array-like): First vector.
    - v (array-like): Second vector.

    Returns:
    - angle (float): Angle between u and v in degrees.
    """
    dot_product = np.dot(u, v)
    norm_u = np.linalg.norm(u)
    norm_v = np.linalg.norm(v)
    # Prevent division by zero
    if norm_u == 0 or norm_v == 0:
        return 0.0
    # Compute cosine of the angle and clip to avoid numerical errors
    cos_angle = np.clip(dot_product / (norm_u * norm_v), -1.0, 1.0)
    angle = np.degrees(np.arccos(cos_angle))
    return angle

def compute_angle_deviations(m, n, X, Y):
    """
    Compute the angle deviations from flat foldability at each internal vertex.

    Parameters:
    - m (int): Number of rows.
    - n (int): Number of columns.
    - X (2D numpy array): X coordinates of the grid.
    - Y (2D numpy array): Y coordinates of the grid.

    Returns:
    - deviations (1D numpy array): Array of angle deviations.
    """
    deviations = []
    # Iterate over internal vertices
    for i in range(1, m):
        for j in range(1, n):
            # Get coordinates of the four neighbors
            p = np.array([X[i, j], Y[i, j]])
            p_up = np.array([X[i-1, j], Y[i-1, j]])
            p_right = np.array([X[i, j+1], Y[i, j+1]])
            p_down = np.array([X[i+1, j], Y[i+1, j]])
            p_left = np.array([X[i, j-1], Y[i, j-1]])

            # Vectors from the center to neighbors
            v_up = p_up - p
            v_right = p_right - p
            v_down = p_down - p
            v_left = p_left - p

            # Compute angles between consecutive vectors in clockwise order
            theta_up_right = angle_between_vectors(v_up, v_right)
            theta_right_down = angle_between_vectors(v_right, v_down)
            theta_down_left = angle_between_vectors(v_down, v_left)
            theta_left_up = angle_between_vectors(v_left, v_up)

            # Kawasaki's condition: sum of alternating angles = 180 degrees
            dev1 = (theta_up_right + theta_down_left) - 180.0
            dev2 = (theta_right_down + theta_left_up) - 180.0

            deviations.extend([dev1, dev2])

    return np.array(deviations)

def cost_function_with_deviation_and_rotation(y_deviation, m, n, alpha, delta, theta_max, lambda_reg=1.0):
    """
    Cost function to minimize deviations from flat foldability and Y-deviation magnitude.

    Parameters:
    - y_deviation (array-like): Current Y-deviations for each column.
    - m (int): Number of rows.
    - n (int): Number of columns.
    - alpha (float): Folding angle in radians.
    - delta (float): Scaling factor per column.
    - theta_max (float): Maximum rotation angle in radians.
    - lambda_reg (float): Regularization coefficient for Y-deviation.

    Returns:
    - cost (float): Combined cost of angle deviations and Y-deviation magnitude.
    """
    X, Y = compute_coordinates_with_deviation_and_rotation(m, n, alpha, delta, y_deviation, theta_max)
    deviations = compute_angle_deviations(m, n, X, Y)
    # Sum of squared deviations from flat foldability
    angle_cost = np.sum(deviations**2)
    # Regularization to minimize Y-deviation magnitude
    reg_cost = lambda_reg * np.sum(y_deviation**2)
    # Total cost
    cost = angle_cost + reg_cost
    return cost

def optimize_y_deviation_and_rotation(m, n, alpha, delta, theta_max, lambda_reg=1.0, delta_max=0.1, initial_guess=None):
    """
    Optimize Y-deviations to achieve flat foldability with controlled skew.

    Parameters:
    - m (int): Number of rows.
    - n (int): Number of columns.
    - alpha (float): Folding angle in radians.
    - delta (float): Scaling factor per column.
    - theta_max (float): Maximum rotation angle in radians.
    - lambda_reg (float): Regularization coefficient for Y-deviation.
    - delta_max (float): Maximum allowed Y-deviation.
    - initial_guess (array-like, optional): Initial guess for Y-deviations.

    Returns:
    - result.x (1D numpy array): Optimized Y-deviations.
    - result (OptimizeResult): Full optimization result.
    """
    if initial_guess is None:
        # Initialize Y-deviations based on standard Miura-ori without skew
        y_offsets = [0.0]
        a = 1.0  # Fixed initial vertical side length
        for j in range(1, n+1):
            a_j = a * (1 - delta * (j / n))
            a_j = max(a_j, 1e-3)
            b_j = a_j * np.tan(alpha)
            y_new = y_offsets[j-1] + b_j * np.sin(alpha)
            y_offsets.append(y_new)
        initial_guess = np.array(y_offsets)

    # Define bounds: y_deviation[j] should be within [-delta_max, delta_max]
    bounds = [(-delta_max, delta_max) for _ in range(n+1)]

    # Define constraints: y_deviation[j] - y_deviation[j-1] >= epsilon
    # To ensure that y_deviation is monotonically increasing
    constraints = []
    epsilon = 1e-4  # Small positive value to ensure strict inequality
    for j in range(1, n+1):
        constraints.append({
            'type': 'ineq',
            'fun': lambda y, j=j: y[j] - y[j-1] - epsilon  # y[j] > y[j-1] + epsilon
        })

    # Optimization using SLSQP
    result = minimize(
        cost_function_with_deviation_and_rotation,
        initial_guess,
        args=(m, n, alpha, delta, theta_max, lambda_reg),
        method='SLSQP',
        bounds=bounds,
        constraints=constraints,
        options={'maxiter': 10000, 'ftol': 1e-9, 'disp': False}
    )

    # Validate constraints
    def validate_constraints(y_deviation, n, epsilon=1e-4):
        violations = []
        for j in range(1, n+1):
            if y_deviation[j] - y_deviation[j-1] < epsilon:
                violations.append((j, y_deviation[j]))
        if violations:
            print("Constraint violations detected:")
            for violation in violations:
                print(f"Column {violation[0]}: y_deviation = {violation[1]:.6f}")
        else:
            print("All constraints satisfied.")

    validate_constraints(result.x, n)

    return result.x, result

def plot_miura_ori_with_deviation_and_rotation(m, n, alpha, delta, y_deviation, theta_max, show_plot=True, save_svg=False):
    """
    Plot the Miura-ori crease pattern with Y-deviation and alternating rotation per column.

    Parameters:
    - m (int): Number of rows.
    - n (int): Number of columns.
    - alpha (float): Folding angle in radians (applied to vertical parallels).
    - delta (float): Scaling factor per column.
    - y_deviation (array-like): Y-deviations for each column.
    - theta_max (float): Maximum rotation angle in radians.
    - show_plot (bool): Whether to display the plot.
    - save_svg (bool): Whether to save the plot as an SVG file.
    """
    X, Y = compute_coordinates_with_deviation_and_rotation(m, n, alpha, delta, y_deviation, theta_max)
    i_values = np.arange(m+1)
    j_values = np.arange(n+1)
    
    fig, ax = plt.subplots(figsize=(8, 8))
    
    # Plot longitudinal folds (vertical lines)
    for j in j_values:
        for i in range(m):
            fold_color = 'blue' if j % 2 == 0 else 'red'
            ax.plot([X[i, j], X[i+1, j]], [Y[i, j], Y[i+1, j]], color=fold_color, lw=2)
    
    # Plot transversal folds (horizontal lines)
    for i in i_values:
        for j in range(n):
            fold_color = 'red' if (i + j) % 2 == 0 else 'blue'
            ax.plot([X[i, j], X[i, j+1]], [Y[i, j], Y[i, j+1]], color=fold_color, lw=2)
    
    # Plot outer boundaries
    for i in range(m):
        # Left boundary
        ax.plot([X[i, 0], X[i+1, 0]], [Y[i, 0], Y[i+1, 0]], color='black', lw=2)
        # Right boundary
        ax.plot([X[i, n], X[i+1, n]], [Y[i, n], Y[i+1, n]], color='black', lw=2)
    for j in range(n):
        # Top boundary
        ax.plot([X[0, j], X[0, j+1]], [Y[0, j], Y[0, j+1]], color='black', lw=2)
        # Bottom boundary
        ax.plot([X[m, j], X[m, j+1]], [Y[m, j], Y[m, j+1]], color='black', lw=2)
    
    # Final plot adjustments
    ax.axis('off')
    
    if show_plot:
        plt.show()
    if save_svg:
        fig.savefig(f'skewed_miura_ori_{m}_{n}_{alpha:.2f}_{delta:.2f}.svg', format='svg', dpi=300)
        print(f'SVG file saved as "skewed_miura_ori_{m}_{n}_{alpha:.2f}_{delta:.2f}.svg"')
    plt.close(fig)

def interactive_miura_ori_simplified():
    """
    Create an interactive widget to adjust Miura-ori parameters with Y-deviation and alternating rotation per column and visualize the pattern.
    """
    def update_plot(m=5, n=4, alpha=1.05, delta=0.25, theta_max=0.05, show_plot=True, save_svg=False):
        # Optimize Y-deviations with the given theta_max
        y_deviation_opt, result = optimize_y_deviation_and_rotation(
            m=m,
            n=n,
            alpha=alpha,
            delta=delta,
            theta_max=theta_max,
            lambda_reg=1.0,
            delta_max=0.1
        )
        
        # Display optimization status
        print(f"Optimization successful: {result.success}")
        print(f"Message: {result.message}")
        print(f"Final cost (angle deviations + regularization): {result.fun:.6f}")
        
        if result.success:
            # Plot the crease pattern with Y-deviation and rotation
            plot_miura_ori_with_deviation_and_rotation(
                m=m,
                n=n,
                alpha=alpha,
                delta=delta,
                y_deviation=y_deviation_opt,
                theta_max=theta_max,
                show_plot=show_plot,
                save_svg=save_svg
            )
        else:
            print("Optimization did not converge. Please try adjusting the parameters.")

    interactive_plot = interactive(
        update_plot,
        m=IntSlider(min=3, max=20, step=1, value=5, description='Rows (m)'),
        n=IntSlider(min=2, max=20, step=1, value=4, description='Cols (n)'),
        alpha=FloatSlider(min=0.1, max=np.pi-0.1, step=0.05, value=1.05, description='alpha (rad)'),
        delta=FloatSlider(min=0.0, max=0.8, step=0.05, value=0.25, description='delta'),
        theta_max=FloatSlider(min=0.0, max=np.pi/4, step=0.01, value=0.05, description='X Rotation (rad)'),
        show_plot=Checkbox(value=True, description='Show Plot'),
        save_svg=Checkbox(value=False, description='Save SVG')
    )
    
    display(interactive_plot)

# Initialize the simplified interactive interface
interactive_miura_ori_simplified()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interactive, IntSlider, FloatSlider, Checkbox
from IPython.display import display

def plot_pattern(num_segments=5, converge_angle=5.0, inner_angle_dev=2.0, show_plot=True):
    """
    Plot a simplified origami-like pattern with two rows converging and angles alternating around 90° (top) and 270° (bottom).
    
    Parameters:
    - num_segments (int): Number of zig-zag segments.
    - converge_angle (float): Angle in degrees at which top and bottom lines converge.
    - inner_angle_dev (float): Deviation in degrees from 90° and 270°.
    - show_plot (bool): Whether to display the plot.
    """
    # Central horizontal line at y=0
    # We'll lay out segments along the x-axis, starting at x=0.
    
    # Converging lines: We assume top line slopes downward and bottom line slopes upward.
    # Let's say top line starts at some height H and converges by 'converge_angle' over total length.
    # For simplicity, define a length scale:
    length = num_segments * 1.0  # each segment 1 unit in x for simplicity
    
    # Convert converge_angle to radians:
    ca_rad = np.radians(converge_angle)
    
    # The top line at x=0: y_top_start = H
    # The bottom line at x=0: y_bottom_start = -H
    # After length in x, they would be closer. Let's just define H:
    H = 1.0  # initial half-height
    # We will linearly decrease height as we go along x.
    # At x=length, let's say height reduces by some amount related to converge_angle.
    # Suppose converge_angle is how many degrees top line goes down relative to horizontal:
    # Let's interpret converge_angle as how much top and bottom lines tilt.
    # top line angle = -converge_angle (downwards)
    # bottom line angle = converge_angle (upwards)
    
    # Top line: y = H + tan(-ca_rad)*x = H - (tan(ca_rad)*x)
    # Bottom line: y = -H + tan(ca_rad)*x
    
    top_slope = -np.tan(ca_rad)
    bottom_slope = np.tan(ca_rad)
    
    # Angles per segment for top row:
    # top angles alternate: start with 90°+inner_angle_dev, then 90°-inner_angle_dev, etc.
    # bottom angles alternate similarly around 270°.
    
    # Let's build arrays of segment angles:
    top_angles = []
    bottom_angles = []
    for i in range(num_segments):
        if i % 2 == 0:
            # even segment index
            top_angles.append(90 + inner_angle_dev)
            bottom_angles.append(270 - inner_angle_dev)
        else:
            # odd segment index
            top_angles.append(90 - inner_angle_dev)
            bottom_angles.append(270 + inner_angle_dev)
    
    # Each segment: We have a vertical center line at x_i = i (for i in range(num_segments+1))
    # We'll place a joint at x_i for i=0..num_segments.
    # The top and bottom points at x_i:
    # top_y_i = H + top_slope*x_i
    # bottom_y_i = -H + bottom_slope*x_i
    
    x_coords = np.arange(num_segments+1) * 1.0
    top_y = H + top_slope*x_coords
    bottom_y = -H + bottom_slope*x_coords
    
    # We'll draw lines from (x_i,0) to top line at angle close to these angles?
    # Actually, the description: We have 2 rows and a central horizontal line. 
    # The segments are formed how?
    # Let's interpret: We have a central horizontal line at y=0. From each point on the central line:
    #   top segment goes up at chosen angle from horizontal
    #   bottom segment goes down at chosen angle from horizontal
    
    # For each segment i, we have a start point at (x_i,0). 
    # top segment line from (x_i,0) going up at angle top_angles[i]:
    # top angle measured from horizontal line: 
    # angle in degrees from horizontal: top_angles[i]
    # angle_rad = np.radians(top_angles[i])
    # direction vector top: (cos(angle_rad), sin(angle_rad))
    # We want it to meet the top boundary line.
    
    # Solve intersection with top line:
    # top_line eq: y = top_y[i] + slope*(x-x_i) but we have top_y already defined as top boundary at each x_i?
    # Wait, we have top_y at discrete points. Let's approximate top line as linear between x_i and x_{i+1}.
    # We'll find intersection with top line linearly between x_i and x_{i+1}.
    
    # For simplicity, let's just draw short segments that approximate the angle without intersection for demonstration.
    # The user asked for simpler approach, so let's just show lines that represent angles, not intersection with top line:
    
    # We'll draw a fixed length line to show the direction.
    segment_length = 0.5
    fig, ax = plt.subplots(figsize=(8,8))
    # Draw top and bottom boundary lines:
    ax.plot(x_coords, top_y, color='black', lw=2)
    ax.plot(x_coords, bottom_y, color='black', lw=2)
    
    # Draw central line:
    ax.axhline(y=0, color='gray', lw=1, linestyle='--')
    
    for i in range(num_segments):
        x_i = x_coords[i]
        # top angle:
        angle_top_rad = np.radians(top_angles[i])
        dx_top = segment_length * np.cos(angle_top_rad)
        dy_top = segment_length * np.sin(angle_top_rad)
        
        # bottom angle:
        # bottom angles are around horizontal but going downward?
        # 270° means pointing down, 270 - dev means slightly angled differently.
        angle_bottom_rad = np.radians(bottom_angles[i])
        dx_bottom = segment_length * np.cos(angle_bottom_rad)
        dy_bottom = segment_length * np.sin(angle_bottom_rad)
        
        # Top segment start at (x_i,0)
        ax.plot([x_i, x_i+dx_top], [0, dy_top], color='red', lw=2)
        # Bottom segment start at (x_i,0)
        ax.plot([x_i, x_i+dx_bottom], [0, dy_bottom], color='blue', lw=2)
    
    ax.set_aspect('equal', 'box')
    ax.set_xlim(-0.5, num_segments)
    ax.set_ylim(-H-1, H+1)
    ax.axis('off')
    
    if show_plot:
        plt.show()

def interactive_pattern():
    widget = interactive(
        plot_pattern,
        num_segments=IntSlider(min=1, max=20, step=1, value=5, description='Segments'),
        converge_angle=FloatSlider(min=0.0, max=45.0, step=1.0, value=5.0, description='Converge Angle (deg)'),
        inner_angle_dev=FloatSlider(min=0.0, max=10.0, step=0.5, value=2.0, description='Inner Angle Dev (deg)'),
        show_plot=Checkbox(value=True, description='Show Plot')
    )
    display(widget)

# Call the function to display the interactive widget
interactive_pattern()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interactive, IntSlider, FloatSlider, Text, Checkbox
from IPython.display import display

def line_intersection(x0, y0, angle_deg, x_start, x_end, y_start, y_end):
    """
    Find intersection of a line starting at (x0,y0) at angle_deg from horizontal
    with a boundary line defined between (x_start,y_start) and (x_end,y_end).
    
    We'll do a binary search in x between x_start and x_end to find where:
    line_y(t) = boundary_y(x).
    
    The line: x_line(t) = x0 + t*cos(angle), y_line(t) = y0 + t*sin(angle)
    We want to find t where x_line(t) between x_start and x_end and:
    y_line(t) == linear interpolation of boundary_y at that x.
    """
    angle_rad = np.radians(angle_deg)
    dx = np.cos(angle_rad)
    dy = np.sin(angle_rad)
    
    # We'll do a binary search on t, because x = x0 + t*dx
    # We know intersection must lie roughly between x0 and x0+(x_end-x_start) if well-formed.
    # Start with a search interval for t:
    # We know we must reach at least x_end to ensure we pass through boundary segment
    # We'll just try up to t_max as x_end-x0 in dx direction:
    # Actually, dx could be negative if angle > 90 and < 270. Handle that:
    
    # We'll try a certain range for t until we find intersection
    # The boundary line: For a given x, y_bound = y_start + ((y_end - y_start)/(x_end - x_start))*(x - x_start)
    
    # Let's ensure dx>0 or adjust direction:
    # If dx == 0, line vertical - handle separately
    if abs(dx) < 1e-9:
        # Line is vertical: x = x0 always
        # Check if x0 is between x_start and x_end
        if x0 < min(x_start, x_end) or x0 > max(x_start, x_end):
            return None, None
        # then just solve for t from y direction (this might not be needed if well-defined)
        # We'll do linear interpolation for y at x0:
        ratio = (x0 - x_start)/(x_end - x_start)
        y_bound = y_start + (y_end - y_start)*ratio
        # y_line(t)=y0+t*dy => t = (y_bound - y0)/dy
        if abs(dy) < 1e-9:
            return None, None
        t = (y_bound - y0)/dy
        if t < 0:
            return None, None
        return x0, y_bound
    
    # For dx != 0:
    # We'll pick a large range for t:
    t_min = 0.0
    t_max = (x_end - x0)/dx if dx>0 else (x_start - x0)/dx
    if t_max < 0:
        # If t_max <0 means line goes opposite direction from boundary segment
        # we can also try the other direction:
        # Let's just set t_max = abs(x_end - x0)/abs(dx)*2 as a large range
        t_max = abs(x_end - x0)/abs(dx)*2
    if t_max < 0: t_max = -t_max
    if t_max ==0: t_max = 1.0
    
    for _ in range(50):  # binary search steps
        t_mid = 0.5*(t_min + t_max)
        x_line = x0 + t_mid*dx
        y_line = y0 + t_mid*dy
        # Interpolate boundary at x_line
        if x_end != x_start:
            ratio = (x_line - x_start)/(x_end - x_start)
            if ratio < 0 or ratio > 1:
                # x_line not within segment horizontally,
                # shift search range
                if dx>0:
                    if x_line > x_end:
                        t_max = t_mid
                    else:
                        t_min = t_mid
                else:
                    if x_line < x_start:
                        t_max = t_mid
                    else:
                        t_min = t_mid
                continue
            y_bound = y_start + (y_end - y_start)*ratio
        else:
            # Vertical boundary segment?
            # If x_start==x_end, line vertical:
            y_bound = y_start # or handle multiple points?
        
        diff = y_line - y_bound
        if abs(diff)<1e-6:
            # close enough
            return x_line, y_line
        if diff>0:
            # line is above boundary
            # we want to reduce y_line, if dy>0 means we need less t
            # Actually, let's think:
            # If line is above boundary and dy>0, to get closer to boundary we might need smaller t or bigger t?
            # If dy>0 and line above boundary, we might have overshoot t.
            # Let's just do a simple approach:
            t_max = t_mid if dy>0 else t_min
            if dy<0:
                t_min = t_mid
        else:
            # line below boundary
            if dy>0:
                t_min = t_mid
            else:
                t_max = t_mid
    
    # If no intersection found after attempts:
    return None, None

def plot_pattern(num_segments=5, converge_angle=5.0, inner_angle_dev=2.0, inner_line_color='red', show_plot=True):
    """
    Improved pattern plotting:
    - Two converging lines (top and bottom).
    - Central horizontal line at y=0.
    - Angled segments intersect with top/bottom boundaries to form closed shapes.
    - Inner lines in user-selected color.
    - Middle horizontal line alternating blue/red segments.
    """
    # Setup
    length = num_segments * 1.0
    ca_rad = np.radians(converge_angle)
    H = 1.0
    top_slope = -np.tan(ca_rad)
    bottom_slope = np.tan(ca_rad)
    
    x_coords = np.arange(num_segments+1)*1.0
    top_y = H + top_slope*x_coords
    bottom_y = -H + bottom_slope*x_coords
    
    # Angles
    top_angles = []
    bottom_angles = []
    for i in range(num_segments):
        if i % 2 == 0:
            top_angles.append(90 + inner_angle_dev)
            bottom_angles.append(270 - inner_angle_dev)
        else:
            top_angles.append(90 - inner_angle_dev)
            bottom_angles.append(270 + inner_angle_dev)
    
    fig, ax = plt.subplots(figsize=(8,8))
    
    # Draw top and bottom boundary lines:
    ax.plot(x_coords, top_y, color='black', lw=2)
    ax.plot(x_coords, bottom_y, color='black', lw=2)
    
    # Middle horizontal line at y=0, alternate blue/red segments:
    for i in range(num_segments):
        color = 'blue' if i%2==0 else 'red'
        ax.plot([x_coords[i], x_coords[i+1]], [0,0], color=color, lw=2)
    
    # For each segment, find intersection with top and bottom lines:
    # We'll consider each segment defined between x_i and x_{i+1}.
    for i in range(num_segments):
        x_i = x_coords[i]
        x_ip1 = x_coords[i+1]
        
        # top line segment from (x_i, top_y[i]) to (x_ip1, top_y[i+1])
        # bottom line segment from (x_i, bottom_y[i]) to (x_ip1, bottom_y[i+1])
        
        # top angle line: start (x_i,0) angle = top_angles[i]
        tx, ty = line_intersection(
            x_i, 0, top_angles[i],
            x_i, x_ip1, top_y[i], top_y[i+1]
        )
        
        # bottom angle line: start (x_i,0) angle = bottom_angles[i]
        bx, by = line_intersection(
            x_i, 0, bottom_angles[i],
            x_i, x_ip1, bottom_y[i], bottom_y[i+1]
        )
        
        if tx is None or bx is None:
            # If no intersection found, skip drawing shape
            continue
        
        # Draw inner lines:
        # top angled line
        ax.plot([x_i, tx], [0, ty], color=inner_line_color, lw=2)
        # bottom angled line
        ax.plot([x_i, bx], [0, by], color=inner_line_color, lw=2)
        
        # Close the shape by drawing line between (tx,ty) and (bx,by)
        # This forms a trapezoid-like shape
        ax.plot([tx, bx], [ty, by], color=inner_line_color, lw=2, linestyle=':')
    
    ax.set_aspect('equal', 'box')
    ax.set_xlim(-0.5, num_segments+0.5)
    ax.set_ylim(-H-1, H+1)
    ax.axis('off')
    
    if show_plot:
        plt.show()

def interactive_pattern():
    widget = interactive(
        plot_pattern,
        num_segments=IntSlider(min=1, max=20, step=1, value=5, description='Segments'),
        converge_angle=FloatSlider(min=0.0, max=45.0, step=1.0, value=5.0, description='Converge Angle (deg)'),
        inner_angle_dev=FloatSlider(min=0.0, max=10.0, step=0.5, value=2.0, description='Inner Angle Dev (deg)'),
        inner_line_color=Text(value='red', description='Inner Line Color'),
        show_plot=Checkbox(value=True, description='Show Plot')
    )
    display(widget)

# Call the function to display the interactive widget
interactive_pattern()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interactive, IntSlider, FloatSlider, Text, Checkbox
from IPython.display import display

def solve_intersection(x0, y0, angle_deg, H, ca_rad, top=True):
    """
    Solve intersection of an angled line from (x0,y0) at angle_deg (degrees) with either top or bottom line.
    top_line: y = H - tan(ca_rad)*x
    bottom_line: y = -H + tan(ca_rad)*x
    """
    angle_rad = np.radians(angle_deg)
    dx = np.cos(angle_rad)
    dy = np.sin(angle_rad)
    
    # Choose line equation based on top or bottom
    if top:
        # y_top = H - tan(ca_rad)*x
        slope = -np.tan(ca_rad)
        intercept = H
    else:
        # y_bottom = -H + tan(ca_rad)*x
        slope = np.tan(ca_rad)
        intercept = -H
    
    # We want to solve:
    # y0 + t*dy = intercept + slope*(x0 + t*dx)
    # y0 + t*dy = intercept + slope*x0 + slope*t*dx
    # t*dy - slope*t*dx = intercept + slope*x0 - y0
    # t(dy - slope*dx) = intercept + slope*x0 - y0
    denom = (dy - slope*dx)
    if abs(denom) < 1e-12:
        return None, None  # No intersection (parallel lines)
    
    rhs = intercept + slope*x0 - y0
    t = rhs/denom
    if t < 0:
        # Intersection behind start point if t<0, may indicate no forward intersection
        return None, None
    
    x_int = x0 + t*dx
    y_int = y0 + t*dy
    return x_int, y_int

def plot_pattern(num_segments=5, converge_angle=5.0, inner_angle_dev=2.0, inner_line_color='red', show_plot=True):
    # Setup parameters
    H = 1.0
    ca_rad = np.radians(converge_angle)
    
    # We'll lay out segments along x-axis at x=0,...,num_segments for the center line
    x_coords = np.arange(num_segments+1)*1.0
    
    # Draw top/bottom lines:
    # For visualization, let's pick a range:
    # Just show from x=-0.5 to x=num_segments+0.5
    x_range = [x_coords[0]-0.5, x_coords[-1]+0.5]
    # top line: y = H - tan(ca_rad)*x
    # bottom line: y = -H + tan(ca_rad)*x
    
    fig, ax = plt.subplots(figsize=(8,8))
    x_line = np.linspace(x_range[0], x_range[1], 100)
    y_top_line = H - np.tan(ca_rad)*x_line
    y_bottom_line = -H + np.tan(ca_rad)*x_line
    
    ax.plot(x_line, y_top_line, color='black', lw=2)
    ax.plot(x_line, y_bottom_line, color='black', lw=2)
    
    # Middle horizontal line y=0: alternate segments blue/red
    for i in range(num_segments):
        color = 'blue' if i%2==0 else 'red'
        ax.plot([x_coords[i], x_coords[i+1]], [0,0], color=color, lw=2)
    
    # Angles for top and bottom line
    top_angles = []
    bottom_angles = []
    for i in range(num_segments):
        if i%2==0:
            top_angles.append(90 + inner_angle_dev)
            bottom_angles.append(270 - inner_angle_dev)
        else:
            top_angles.append(90 - inner_angle_dev)
            bottom_angles.append(270 + inner_angle_dev)
    
    # Compute intersections and draw shapes
    for i in range(num_segments):
        x_i = x_coords[i]
        # top angled line from (x_i,0) at angle top_angles[i]
        tx, ty = solve_intersection(x_i, 0, top_angles[i], H, ca_rad, top=True)
        # bottom angled line from (x_i,0) at angle bottom_angles[i]
        bx, by = solve_intersection(x_i, 0, bottom_angles[i], H, ca_rad, top=False)
        
        if tx is None or bx is None:
            continue
        
        # Draw inner angled lines:
        # top angled line
        ax.plot([x_i, tx], [0, ty], color=inner_line_color, lw=2)
        # bottom angled line
        ax.plot([x_i, bx], [0, by], color=inner_line_color, lw=2)
        
        # Close shape by connecting (tx,ty) and (bx,by)
        ax.plot([tx, bx], [ty, by], color=inner_line_color, lw=2, linestyle=':')
    
    ax.set_aspect('equal', 'box')
    ax.set_xlim(x_range)
    ax.set_ylim(-H-1, H+1)
    ax.axis('off')
    
    if show_plot:
        plt.show()

def interactive_pattern():
    widget = interactive(
        plot_pattern,
        num_segments=IntSlider(min=1, max=20, step=1, value=5, description='Segments'),
        converge_angle=FloatSlider(min=0.0, max=45.0, step=1.0, value=5.0, description='Converge Angle (deg)'),
        inner_angle_dev=FloatSlider(min=0.0, max=10.0, step=0.5, value=2.0, description='Inner Angle Dev (deg)'),
        inner_line_color=Text(value='red', description='Inner Line Color'),
        show_plot=Checkbox(value=True, description='Show Plot')
    )
    display(widget)

# Call the function to display the interactive widget
interactive_pattern()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interactive, IntSlider, FloatSlider, Text, Checkbox
from IPython.display import display

def solve_intersection(x0, y0, angle_deg, H, ca_rad, top=True):
    """
    Solve intersection of an angled line from (x0,y0) at angle_deg (degrees) with either top or bottom line.
    top_line: y = H - tan(ca_rad)*x
    bottom_line: y = -H + tan(ca_rad)*x
    """
    angle_rad = np.radians(angle_deg)
    dx = np.cos(angle_rad)
    dy = np.sin(angle_rad)
    
    # Choose line equation based on top or bottom
    if top:
        # y_top = H - tan(ca_rad)*x
        slope = -np.tan(ca_rad)
        intercept = H
    else:
        # y_bottom = -H + tan(ca_rad)*x
        slope = np.tan(ca_rad)
        intercept = -H
    
    # We want to solve:
    # y0 + t*dy = intercept + slope*(x0 + t*dx)
    # t*dy - slope*t*dx = intercept + slope*x0 - y0
    # t(dy - slope*dx) = intercept + slope*x0 - y0
    denom = (dy - slope*dx)
    if abs(denom) < 1e-12:
        return None, None  # No intersection (parallel lines)
    
    rhs = intercept + slope*x0 - y0
    t = rhs/denom
    if t < 0:
        # Intersection behind start point if t<0, may indicate no forward intersection
        return None, None
    
    x_int = x0 + t*dx
    y_int = y0 + t*dy
    return x_int, y_int

def plot_pattern(num_segments=5, converge_angle=5.0, inner_angle_dev=2.0, inner_line_color='red', show_plot=True):
    # Setup parameters
    H = 1.0
    ca_rad = np.radians(converge_angle)
    
    # We'll lay out segments along x-axis at x=0,...,num_segments for the center line
    x_coords = np.arange(num_segments+1)*1.0
    
    # Angles for top and bottom line
    top_angles = []
    bottom_angles = []
    for i in range(num_segments):
        if i%2==0:
            top_angles.append(90 + inner_angle_dev)
            bottom_angles.append(270 - inner_angle_dev)
        else:
            top_angles.append(90 - inner_angle_dev)
            bottom_angles.append(270 + inner_angle_dev)
    
    # Compute intersections for all segments to determine boundary limits
    top_intersections = []
    bottom_intersections = []
    for i in range(num_segments):
        x_i = x_coords[i]
        # top angled line intersection
        tx, ty = solve_intersection(x_i, 0, top_angles[i], H, ca_rad, top=True)
        # bottom angled line intersection
        bx, by = solve_intersection(x_i, 0, bottom_angles[i], H, ca_rad, top=False)
        if tx is not None and bx is not None:
            top_intersections.append((tx, ty))
            bottom_intersections.append((bx, by))
    
    if len(top_intersections) == 0 or len(bottom_intersections) == 0:
        # No valid intersections found, just show empty plot
        fig, ax = plt.subplots(figsize=(8,8))
        if show_plot:
            plt.show()
        return
    
    # Determine leftmost and rightmost x-limits from intersections
    # The top and bottom lines will be limited to these ranges
    all_x = [p[0] for p in top_intersections] + [p[0] for p in bottom_intersections]
    x_min = min(all_x)
    x_max = max(all_x)
    
    # We'll sample points on top and bottom lines from x_min to x_max
    x_line = np.linspace(x_min, x_max, 100)
    y_top_line = H - np.tan(ca_rad)*x_line
    y_bottom_line = -H + np.tan(ca_rad)*x_line
    
    fig, ax = plt.subplots(figsize=(8,8))
    # Draw top and bottom lines limited to x_min and x_max
    ax.plot(x_line, y_top_line, color='black', lw=2)
    ax.plot(x_line, y_bottom_line, color='black', lw=2)
    
    # Middle horizontal line y=0: alternate segments blue/red
    for i in range(num_segments):
        color = 'blue' if i%2==0 else 'red'
        ax.plot([x_coords[i], x_coords[i+1]], [0,0], color=color, lw=2)
    
    # Draw inner angled segments:
    for i in range(num_segments):
        x_i = x_coords[i]
        tx, ty = solve_intersection(x_i, 0, top_angles[i], H, ca_rad, top=True)
        bx, by = solve_intersection(x_i, 0, bottom_angles[i], H, ca_rad, top=False)
        
        if tx is None or bx is None:
            continue
        
        # Draw inner angled lines:
        ax.plot([x_i, tx], [0, ty], color=inner_line_color, lw=2)
        ax.plot([x_i, bx], [0, by], color=inner_line_color, lw=2)
        
        # Previously, we closed the shape with a dotted line between (tx, ty) and (bx, by).
        # Now we do NOT draw that dotted line, as requested.
    
    ax.set_aspect('equal', 'box')
    ax.set_xlim(x_min-0.5, x_max+0.5)
    # Compute vertical bounds:
    y_all = [p[1] for p in top_intersections] + [p[1] for p in bottom_intersections] + [0]
    y_min = min(y_all) - 0.5
    y_max = max(y_all) + 0.5
    ax.set_ylim(y_min, y_max)
    ax.axis('off')
    
    if show_plot:
        plt.show()

def interactive_pattern():
    widget = interactive(
        plot_pattern,
        num_segments=IntSlider(min=1, max=20, step=1, value=5, description='Segments'),
        converge_angle=FloatSlider(min=0.0, max=45.0, step=1.0, value=5.0, description='Converge Angle (deg)'),
        inner_angle_dev=FloatSlider(min=0.0, max=10.0, step=0.5, value=2.0, description='Inner Angle Dev (deg)'),
        inner_line_color=Text(value='red', description='Inner Line Color'),
        show_plot=Checkbox(value=True, description='Show Plot')
    )
    display(widget)

# Display the interactive widget
interactive_pattern()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interactive, IntSlider, FloatSlider, Text, Checkbox
from IPython.display import display

def solve_intersection(x0, y0, angle_deg, H, ca_rad, top=True):
    """
    Solve intersection of an angled line from (x0,y0) at angle_deg (degrees) with either top or bottom line.
    top_line: y = H - tan(ca_rad)*x
    bottom_line: y = -H + tan(ca_rad)*x
    """
    angle_rad = np.radians(angle_deg)
    dx = np.cos(angle_rad)
    dy = np.sin(angle_rad)
    
    # Choose line equation based on top or bottom
    if top:
        # y_top = H - tan(ca_rad)*x
        slope = -np.tan(ca_rad)
        intercept = H
    else:
        # y_bottom = -H + tan(ca_rad)*x
        slope = np.tan(ca_rad)
        intercept = -H
    
    denom = (dy - slope*dx)
    if abs(denom) < 1e-12:
        return None, None  # No intersection (parallel lines)
    
    rhs = intercept + slope*x0 - y0
    t = rhs/denom
    if t < 0:
        # Intersection behind start point if t<0
        return None, None
    
    x_int = x0 + t*dx
    y_int = y0 + t*dy
    return x_int, y_int

def plot_pattern(num_segments=5, converge_angle=5.0, inner_angle_dev=2.0, inner_line_color='red', 
                 total_width=5.0, total_height=2.0, show_plot=True):
    """
    Plot the pattern with:
    - Specified number of segments (num_segments)
    - Converge angle (converge_angle)
    - Inner angle deviation (inner_angle_dev)
    - Inner line color (inner_line_color)
    - Total width and height of the pattern
    - No last horizontal segment
    - First and last angled lines are black
    """
    # Set up parameters
    # total_width is the horizontal span of the pattern
    # total_height is the vertical span, so H = total_height/2
    H = total_height/2
    ca_rad = np.radians(converge_angle)
    
    # x-coordinates for segments
    x_coords = np.linspace(0, total_width, num_segments+1)
    
    # Define angles for top and bottom lines per segment
    top_angles = []
    bottom_angles = []
    for i in range(num_segments):
        if i%2==0:
            top_angles.append(90 + inner_angle_dev)
            bottom_angles.append(270 - inner_angle_dev)
        else:
            top_angles.append(90 - inner_angle_dev)
            bottom_angles.append(270 + inner_angle_dev)
    
    # Compute intersections for boundary limit
    top_intersections = []
    bottom_intersections = []
    for i in range(num_segments):
        x_i = x_coords[i]
        # top angled line intersection
        tx, ty = solve_intersection(x_i, 0, top_angles[i], H, ca_rad, top=True)
        # bottom angled line intersection
        bx, by = solve_intersection(x_i, 0, bottom_angles[i], H, ca_rad, top=False)
        if tx is not None and bx is not None:
            top_intersections.append((tx, ty))
            bottom_intersections.append((bx, by))
    
    fig, ax = plt.subplots(figsize=(8,8))
    
    if len(top_intersections)==0 or len(bottom_intersections)==0:
        # If no intersections, just show empty
        if show_plot:
            plt.show()
        return
    
    # Determine x-limits from intersections
    all_x = [p[0] for p in top_intersections] + [p[0] for p in bottom_intersections]
    x_min = min(all_x)
    x_max = max(all_x)
    
    # Sample points on top/bottom lines
    x_line = np.linspace(x_min, x_max, 100)
    y_top_line = H - np.tan(ca_rad)*x_line
    y_bottom_line = -H + np.tan(ca_rad)*x_line
    
    # Draw top and bottom lines within these limits
    ax.plot(x_line, y_top_line, color='black', lw=2)
    ax.plot(x_line, y_bottom_line, color='black', lw=2)
    
    # Middle horizontal line y=0: draw one fewer than num_segments lines
    # We have num_segments+1 points in x_coords, to draw one fewer segment, do range(num_segments-1)
    for i in range(num_segments-1):
        color = 'blue' if i%2==0 else 'red'
        ax.plot([x_coords[i], x_coords[i+1]], [0,0], color=color, lw=2)
    
    # Draw inner angled segments:
    # If i=0 or i=num_segments-1, angled lines are black
    # Otherwise use inner_line_color
    for i in range(num_segments):
        x_i = x_coords[i]
        tx, ty = solve_intersection(x_i, 0, top_angles[i], H, ca_rad, top=True)
        bx, by = solve_intersection(x_i, 0, bottom_angles[i], H, ca_rad, top=False)
        
        if tx is None or bx is None:
            continue
        
        # Determine line color
        line_col = 'black' if (i==0 or i==num_segments-1) else inner_line_color
        
        # Top angled line
        ax.plot([x_i, tx], [0, ty], color=line_col, lw=2)
        # Bottom angled line
        ax.plot([x_i, bx], [0, by], color=line_col, lw=2)
        # No dotted line connecting top and bottom intersections as requested
    
    ax.set_aspect('equal', 'box')
    # Compute vertical bounds:
    y_all = [p[1] for p in top_intersections] + [p[1] for p in bottom_intersections] + [0]
    y_min = min(y_all) - 0.5
    y_max = max(y_all) + 0.5
    ax.set_xlim(x_min-0.5, x_max+0.5)
    ax.set_ylim(y_min, y_max)
    ax.axis('off')
    
    if show_plot:
        plt.show()

def interactive_pattern():
    widget = interactive(
        plot_pattern,
        num_segments=IntSlider(min=1, max=20, step=1, value=5, description='Segments'),
        converge_angle=FloatSlider(min=0.0, max=45.0, step=1.0, value=5.0, description='Converge Angle (deg)'),
        inner_angle_dev=FloatSlider(min=0.0, max=35.0, step=1.0, value=2.0, description='Inner Angle Dev (deg)'),
        inner_line_color=Text(value='red', description='Inner Line Color'),
        total_width=FloatSlider(min=1.0, max=20.0, step=1.0, value=5.0, description='Total Width'),
        total_height=FloatSlider(min=1.0, max=20.0, step=0.2, value=2.0, description='Total Height'),
        show_plot=Checkbox(value=True, description='Show Plot')
    )
    display(widget)

# Display the interactive widget
interactive_pattern()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interactive, IntSlider, FloatSlider, Text, Checkbox
from IPython.display import display

def solve_intersection(x0, y0, angle_deg, H, ca_rad, top=True):
    """
    Solve intersection of an angled line from (x0,y0) at angle_deg (degrees) 
    with either the top or bottom line.
    top_line:    y = H - tan(ca_rad)*x
    bottom_line: y = -H + tan(ca_rad)*x
    """
    angle_rad = np.radians(angle_deg)
    dx = np.cos(angle_rad)
    dy = np.sin(angle_rad)
    
    # Choose line equation based on top or bottom
    if top:
        # y_top = H - tan(ca_rad)*x
        slope = -np.tan(ca_rad)
        intercept = H
    else:
        # y_bottom = -H + tan(ca_rad)*x
        slope = np.tan(ca_rad)
        intercept = -H
    
    denom = (dy - slope*dx)
    if abs(denom) < 1e-12:
        return None, None  # No intersection (parallel lines)
    
    rhs = intercept + slope*x0 - y0
    t = rhs / denom
    if t < 0:
        # Intersection behind start point if t < 0
        return None, None
    
    x_int = x0 + t*dx
    y_int = y0 + t*dy
    return x_int, y_int

def plot_pattern(num_segments=5, 
                 converge_angle=5.0, 
                 inner_angle_dev=2.0, 
                 inner_line_color='red', 
                 total_height=3.0, 
                 start_width=1.0, 
                 delta_factor=0.9, 
                 angle_factor=0.9,       # <-- New parameter for multiplicative angle changes
                 show_plot=True, 
                 save_svg=False, 
                 save_filename=''):
    """
    Plot the pattern with:
    - num_segments: number of segments (horizontal lines + angled lines).
    - converge_angle: angle (in degrees) for the top/bottom lines convergence.
    - inner_angle_dev: base deviation for the inner angled lines.
    - angle_factor: multiplicative factor for each subsequent inner angle.
    - inner_line_color: color of inner angled lines (not the first/last).
    - total_height: total height from top to bottom lines.
    - start_width: length of first horizontal segment.
    - delta_factor: factor by which each subsequent horizontal segment length is multiplied.
    - show_plot: if True, display the plot.
    - save_svg: if True, save the figure in SVG format.
    - save_filename: name/path for saving the SVG (only if save_svg is True).
    """
    H = total_height / 2
    ca_rad = np.radians(converge_angle)
    
    # Compute segment lengths
    # segment_length[i] = start_width * (delta_factor^i)
    segment_lengths = [start_width * (delta_factor**i) for i in range(num_segments)]
    
    # X-coordinates (cumulative sums of segment lengths)
    x_coords = [0.0]
    for length in segment_lengths:
        x_coords.append(x_coords[-1] + length)
    x_coords = np.array(x_coords)
    
    # Compute top/bottom angles with multiplicative factor
    top_angles = []
    bottom_angles = []
    for i in range(num_segments):
        # Each successive angle is multiplied by (angle_factor)^i
        current_dev = inner_angle_dev * (angle_factor ** (i*2))
        if i == num_segments-1:
            if i % 2 == 0:
                top_angles.append(90 - current_dev)
                bottom_angles.append(270 + current_dev)
            else:
                top_angles.append(90 + current_dev)
                bottom_angles.append(270 - current_dev)
                
        elif i % 2 == 0:
            top_angles.append(90 + current_dev)
            bottom_angles.append(270 - current_dev)
        else:
            top_angles.append(90 - current_dev)
            bottom_angles.append(270 + current_dev)
    
    # Solve intersections for boundary lines
    top_intersections = []
    bottom_intersections = []
    for i in range(num_segments):
        x_i = x_coords[i]
        tx, ty = solve_intersection(x_i, 0, top_angles[i], H, ca_rad, top=True)
        bx, by = solve_intersection(x_i, 0, bottom_angles[i], H, ca_rad, top=False)
        
        if tx is not None and bx is not None:
            top_intersections.append((tx, ty))
            bottom_intersections.append((bx, by))
    
    # Prepare figure
    fig, ax = plt.subplots(figsize=(8, 8))
    
    # If no intersections, do nothing
    if len(top_intersections) == 0 or len(bottom_intersections) == 0:
        if show_plot:
            plt.show()
        if save_svg and save_filename.strip():
            fig.savefig(save_filename.strip(), format='svg')
        return
    
    # Determine x-limits from intersections
    all_x = [p[0] for p in top_intersections] + [p[0] for p in bottom_intersections]
    x_min = min(all_x)
    x_max = max(all_x)
    
    # Sample points on top/bottom lines
    x_line = np.linspace(x_min, x_max, 100)
    y_top_line = H - np.tan(ca_rad) * x_line
    y_bottom_line = -H + np.tan(ca_rad) * x_line
    
    # Draw top and bottom lines
    ax.plot(x_line, y_top_line, color='black', lw=2)
    ax.plot(x_line, y_bottom_line, color='black', lw=2)
    
    # Middle horizontal lines (one fewer than num_segments)
    for i in range(num_segments - 1):
        color = 'blue' if i % 2 == 0 else 'red'
        ax.plot([x_coords[i], x_coords[i+1]], [0, 0], color=color, lw=2)
    
    # Draw angled lines
    for i in range(num_segments):
        x_i = x_coords[i]
        tx, ty = solve_intersection(x_i, 0, top_angles[i], H, ca_rad, top=True)
        bx, by = solve_intersection(x_i, 0, bottom_angles[i], H, ca_rad, top=False)
        
        if tx is None or bx is None:
            continue
        
        # First and last angled lines are black; others use inner_line_color
        line_col = 'black' if (i == 0 or i == num_segments - 1) else inner_line_color
        
        # Top angled line
        ax.plot([x_i, tx], [0, ty], color=line_col, lw=2)
        # Bottom angled line
        ax.plot([x_i, bx], [0, by], color=line_col, lw=2)
    
    ax.set_aspect('equal', 'box')
    y_all = [p[1] for p in top_intersections] + [p[1] for p in bottom_intersections] + [0]
    y_min = min(y_all) - 0.5
    y_max = max(y_all) + 0.5
    ax.set_xlim(x_min - 0.5, x_max + 0.5)
    ax.set_ylim(y_min, y_max)
    ax.axis('off')
    
    if show_plot:
        plt.show()
    if save_svg and save_filename.strip():
        fig.savefig(save_filename.strip(), format='svg')

def interactive_pattern():
    """
    Create an interactive widget for exploring pattern plotting 
    with multiplicative angle reduction.
    """
    widget = interactive(
        plot_pattern,
        num_segments=IntSlider(min=1, max=20, step=1, value=5, description='Segments'),
        converge_angle=FloatSlider(min=0.0, max=45.0, step=0.5, value=5.0, description='Converge Angle (deg)'),
        inner_angle_dev=FloatSlider(min=-45.0, max=45.0, step=1.0, value=45.0, description='Inner Angle Dev (deg)'),
        inner_line_color=Text(value='red', description='Inner Line Color'),
        total_height=FloatSlider(min=1.0, max=20.0, step=1.0, value=3.0, description='Total Height'),
        start_width=FloatSlider(min=0.1, max=30.0, step=0.1, value=1.0, description='Start Width'),
        delta_factor=FloatSlider(min=0.1, max=1.0, step=0.05, value=0.9, description='Delta Factor'),
        angle_factor=FloatSlider(
            min=0.1, max=2.0, step=0.02, 
            value=0.9, 
            description='Angle Factor'
        ),
        show_plot=Checkbox(value=True, description='Show Plot'),
        save_svg=Checkbox(value=False, description='Save SVG'),
        save_filename=Text(value='', description='SVG Filename')
    )
    display(widget)

# Display the interactive widget
interactive_pattern()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interactive, IntSlider, FloatSlider, Text, Checkbox
from IPython.display import display

def solve_intersection(x0, y0, angle_deg, H, ca_rad, top=True):
    """
    Solve intersection of an angled line from (x0,y0) at angle_deg (degrees) 
    with either the top or bottom line.
    top_line:    y = H - tan(ca_rad)*x
    bottom_line: y = -H + tan(ca_rad)*x
    """
    angle_rad = np.radians(angle_deg)
    dx = np.cos(angle_rad)
    dy = np.sin(angle_rad)
    
    # Choose line equation based on top or bottom
    if top:
        # y_top = H - tan(ca_rad)*x
        slope = -np.tan(ca_rad)
        intercept = H
    else:
        # y_bottom = -H + tan(ca_rad)*x
        slope = np.tan(ca_rad)
        intercept = -H
    
    denom = (dy - slope*dx)
    if abs(denom) < 1e-12:
        return None, None  # No intersection (parallel lines)
    
    rhs = intercept + slope*x0 - y0
    t = rhs / denom
    if t < 0:
        # Intersection behind start point if t < 0
        return None, None
    
    x_int = x0 + t*dx
    y_int = y0 + t*dy
    return x_int, y_int

def plot_pattern(num_segments=5, 
                 converge_angle=5.0, 
                 inner_angle_dev=2.0, 
                 inner_line_color='red', 
                 total_height=3.0, 
                 start_width=1.0, 
                 delta_factor=0.9, 
                 angle_factor=0.9,       # <-- New parameter for multiplicative angle changes
                 show_plot=True, 
                 save_svg=False, 
                 save_filename=''):
    """
    Plot the pattern with:
    - num_segments: number of segments (horizontal lines + angled lines).
    - converge_angle: angle (in degrees) for the top/bottom lines convergence.
    - inner_angle_dev: base deviation for the inner angled lines.
    - angle_factor: multiplicative factor for each subsequent inner angle.
    - inner_line_color: color of inner angled lines (not the first/last).
    - total_height: total height from top to bottom lines.
    - start_width: length of first horizontal segment.
    - delta_factor: factor by which each subsequent horizontal segment length is multiplied.
    - show_plot: if True, display the plot.
    - save_svg: if True, save the figure in SVG format.
    - save_filename: name/path for saving the SVG (only if save_svg is True).
    """
    H = total_height / 2
    ca_rad = np.radians(converge_angle)
    
    # Compute segment lengths
    # segment_length[i] = start_width * (delta_factor^i)
    segment_lengths = [start_width * (delta_factor**i) for i in range(num_segments)]
    
    # X-coordinates (cumulative sums of segment lengths)
    x_coords = [0.0]
    for length in segment_lengths:
        x_coords.append(x_coords[-1] + length)
    x_coords = np.array(x_coords)
    
    # Compute top/bottom angles with multiplicative factor
    top_angles = []
    bottom_angles = []
    for i in range(num_segments):
        # Each successive angle is multiplied by (angle_factor)^i
        current_dev = inner_angle_dev * (angle_factor ** (i*2))
        print(i)
        if i == num_segments-2:
            if i % 2 == 0:
                top_angles.append(90 - current_dev)
                bottom_angles.append(270 + current_dev)
            else:
                top_angles.append(90 + current_dev)
                bottom_angles.append(270 - current_dev)
                
        elif i % 2 == 0:
            top_angles.append(90 + current_dev)
            bottom_angles.append(270 - current_dev)
        else:
            top_angles.append(90 - current_dev)
            bottom_angles.append(270 + current_dev)
    
    # Solve intersections for boundary lines
    top_intersections = []
    bottom_intersections = []
    for i in range(num_segments):
        x_i = x_coords[i]
        tx, ty = solve_intersection(x_i, 0, top_angles[i], H, ca_rad, top=True)
        bx, by = solve_intersection(x_i, 0, bottom_angles[i], H, ca_rad, top=False)
        
        if tx is not None and bx is not None:
            top_intersections.append((tx, ty))
            bottom_intersections.append((bx, by))
    
    # Prepare figure
    fig, ax = plt.subplots(figsize=(8, 8))
    
    # If no intersections, do nothing
    if len(top_intersections) == 0 or len(bottom_intersections) == 0:
        if show_plot:
            plt.show()
        if save_svg and save_filename.strip():
            fig.savefig(save_filename.strip(), format='svg')
        return
    
    # Determine x-limits from intersections
    all_x = [p[0] for p in top_intersections] + [p[0] for p in bottom_intersections]
    x_min = min(all_x)
    x_max = max(all_x)
    
    # Sample points on top/bottom lines
    x_line = np.linspace(x_min, x_max, 100)
    y_top_line = H - np.tan(ca_rad) * x_line
    y_bottom_line = -H + np.tan(ca_rad) * x_line
    
    # Draw top and bottom lines
    ax.plot(x_line, y_top_line, color='black', lw=2)
    ax.plot(x_line, y_bottom_line, color='black', lw=2)
    
    # Middle horizontal lines (one fewer than num_segments)
    for i in range(num_segments - 1):
        color = 'blue' if i % 2 == 0 else 'red'
        # if i == num_segments - 2:
        #     color = 'red' if i % 2 == 0 else 'blue'
        ax.plot([x_coords[i], x_coords[i+1]], [0, 0], color=color, lw=2)
    
    # Draw angled lines
    for i in range(num_segments):
        x_i = x_coords[i]
        tx, ty = solve_intersection(x_i, 0, top_angles[i], H, ca_rad, top=True)
        bx, by = solve_intersection(x_i, 0, bottom_angles[i], H, ca_rad, top=False)
        
        if tx is None or bx is None:
            continue
        
        # First and last angled lines are black; others use inner_line_color
        line_col = 'black' if (i == 0 or i == num_segments - 1) else inner_line_color
        if i == num_segments - 2:
            line_col = 'blue'
        # Top angled line
        ax.plot([x_i, tx], [0, ty], color=line_col, lw=2)
        # Bottom angled line
        ax.plot([x_i, bx], [0, by], color=line_col, lw=2)
    
    ax.set_aspect('equal', 'box')
    y_all = [p[1] for p in top_intersections] + [p[1] for p in bottom_intersections] + [0]
    y_min = min(y_all) - 0.5
    y_max = max(y_all) + 0.5
    ax.set_xlim(x_min - 0.5, x_max + 0.5)
    ax.set_ylim(y_min, y_max)
    ax.axis('off')
    
    if show_plot:
        plt.show()
    if save_svg and save_filename.strip():
        fig.savefig(save_filename.strip(), format='svg')

def interactive_pattern():
    """
    Create an interactive widget for exploring pattern plotting 
    with multiplicative angle reduction.
    """
    widget = interactive(
        plot_pattern,
        num_segments=IntSlider(min=1, max=20, step=1, value=5, description='Segments'),
        converge_angle=FloatSlider(min=0.0, max=45.0, step=0.5, value=5.0, description='Converge Angle (deg)'),
        inner_angle_dev=FloatSlider(min=-45.0, max=45.0, step=1.0, value=45.0, description='Inner Angle Dev (deg)'),
        inner_line_color=Text(value='red', description='Inner Line Color'),
        total_height=FloatSlider(min=1.0, max=20.0, step=1.0, value=3.0, description='Total Height'),
        start_width=FloatSlider(min=0.1, max=30.0, step=0.1, value=1.0, description='Start Width'),
        delta_factor=FloatSlider(min=0.1, max=1.0, step=0.05, value=0.9, description='Delta Factor'),
        angle_factor=FloatSlider(
            min=0.1, max=2.0, step=0.02, 
            value=0.9, 
            description='Angle Factor'
        ),
        show_plot=Checkbox(value=True, description='Show Plot'),
        save_svg=Checkbox(value=False, description='Save SVG'),
        save_filename=Text(value='', description='SVG Filename')
    )
    display(widget)

# Display the interactive widget
interactive_pattern()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interactive, IntSlider, FloatSlider, Checkbox
from IPython.display import display

def plot_kresling_net(num_sides=6,
                      top_radius=3.0,
                      bottom_radius=3.0,
                      twist_deg=15.0,
                      height=2.0,
                      show_plot=True,
                      save_svg=False,
                      save_filename=''):
    """
    Plots a 2D net of a Kresling-like cylindrical shell with trapezoidal facets.

    Parameters:
    -----------
    num_sides : int
        Number of sides (facets).
    top_radius : float
        Radius of the top polygon (distance from center to each top vertex).
    bottom_radius : float
        Radius of the bottom polygon.
    twist_deg : float
        Twist angle in degrees between the top and bottom polygons.
        If 0, the top and bottom polygons align with no twist.
    height : float
        Vertical height of each trapezoidal facet (distance between top and bottom edges).
    show_plot : bool
        If True, displays the plot within the notebook.
    save_svg : bool
        If True, saves the figure as an SVG file with the name given by save_filename.
    save_filename : str
        Filename (and path, if desired) for the SVG output. Only used if save_svg=True.

    Theory/References:
    ------------------
    [1] Kresling, B. “Natural twist buckling in shells,” Proceedings of the Royal Society A, 2008.
    [2] Lang, R. J. Origami Design Secrets, 2nd ed., CRC Press, 2011.
    [3] Tachi, T. “Freeform Origami,” Journal of Mechanical Design, 2013.

    A real 3D Kresling pattern has angled side edges and can buckle/twist.
    This function merely draws the 2D net of trapezoids.

    """
    # Convert twist to radians
    twist_rad = np.radians(twist_deg)
    
    # Angles for each side (0 to 2π in N segments)
    angles = np.linspace(0, 2*np.pi, num_sides+1)  # num_sides+1 to close the loop
    
    # For each panel, top edge goes between angles[i], angles[i+1]
    # The bottom edge is the same, but offset by twist_rad

    # Precompute top/bottom X/Y
    x_top = top_radius * np.cos(angles)
    y_top = top_radius * np.sin(angles)
    
    # Shift bottom polygon by twist
    # Each angle for bottom is angles[i] + twist_rad
    x_bot = bottom_radius * np.cos(angles + twist_rad)
    y_bot = bottom_radius * np.sin(angles + twist_rad)
    
    # We'll unroll the net in the 2D plane:
    # The net is made up of trapezoids. Each trapezoid has 4 corners:
    #   (x_top[i], y_top[i]) ---- (x_top[i+1], y_top[i+1])
    #   (x_bot[i], y_bot[i]) ---- (x_bot[i+1], y_bot[i+1])
    #
    # But to flatten them in 2D, we place them side-by-side along the "bottom" edges
    # or "top" edges. For simplicity, we can do a linear cumulative distance along the top edges.
    
    # Cumulative arc length along top edges (for net “x” coordinate)
    # This is the distance along the top polygon from i=0 to i.
    def segment_length(xa, ya, xb, yb):
        return np.sqrt((xb - xa)**2 + (yb - ya)**2)
    
    top_arc_lengths = [0.0]
    for i in range(num_sides):
        seg_len = segment_length(x_top[i], y_top[i], x_top[i+1], y_top[i+1])
        top_arc_lengths.append(top_arc_lengths[-1] + seg_len)
    
    # The total length of the top polygon's perimeter
    total_top_perim = top_arc_lengths[-1] if len(top_arc_lengths) > 1 else 0.0
    
    # Flatten (unroll) each trapezoid in the net:
    # We'll put the top edge in the XY-plane at y=0, with x spanning from
    # top_arc_lengths[i] to top_arc_lengths[i+1].
    # Then the bottom edge of the trapezoid is drawn at y=-height.
    
    # We'll store lists of (x, y) for lines to plot
    lines_top = []    # top edges
    lines_bottom = [] # bottom edges
    lines_sides = []  # left & right edges of each trapezoid
    
    # For each trapezoid i
    for i in range(num_sides):
        # X-coordinates in the net
        xA_top = top_arc_lengths[i]
        xB_top = top_arc_lengths[i+1]
        
        # The length of the top edge in the net
        top_edge_length = xB_top - xA_top
        
        # The actual top edge in 2D (original shape) has length 'top_edge_length'.
        # The bottom edge has length ~ segment_length on the bottom polygon:
        seg_len_bot = segment_length(x_bot[i], y_bot[i], x_bot[i+1], y_bot[i+1])
        
        # We assume no scale difference (the net is just 'unrolled'), so
        # the bottom edge in the net will also measure 'seg_len_bot', placed at y = -height.
        # We'll place the left corner of the bottom edge exactly below xA_top.
        
        xA_bot = xA_top
        xB_bot = xA_bot + seg_len_bot
        
        # Store the edges
        # Top edge from (xA_top, 0) to (xB_top, 0)
        lines_top.append( ((xA_top, 0.0), (xB_top, 0.0)) )
        # Bottom edge from (xA_bot, -height) to (xB_bot, -height)
        lines_bottom.append( ((xA_bot, -height), (xB_bot, -height)) )
        
        # Side edges:
        #   left side from (xA_top, 0) to (xA_bot, -height)
        #   right side from (xB_top, 0) to (xB_bot, -height)
        lines_sides.append( ((xA_top, 0.0), (xA_bot, -height)) )
        lines_sides.append( ((xB_top, 0.0), (xB_bot, -height)) )
    
    # Plot the net
    fig, ax = plt.subplots(figsize=(10, 4))
    
    # Plot top edges (blue), bottom edges (red), sides (black dashed)
    for tseg in lines_top:
        ax.plot([tseg[0][0], tseg[1][0]], [tseg[0][1], tseg[1][1]], color='blue', lw=2)
    for bseg in lines_bottom:
        ax.plot([bseg[0][0], bseg[1][0]], [bseg[0][1], bseg[1][1]], color='red', lw=2)
    for sseg in lines_sides:
        ax.plot([sseg[0][0], sseg[1][0]], [sseg[0][1], sseg[1][1]], color='black', lw=1, ls='--')
    
    # Clean up axes
    ax.set_aspect('equal', 'box')
    ax.set_xlabel("Unrolled X (arbitrary units)")
    ax.set_ylabel("Y (arbitrary units)")
    ax.grid(True, alpha=0.3)
    
    if show_plot:
        plt.show()
    
    if save_svg and save_filename.strip():
        fig.savefig(save_filename.strip(), format='svg')

def interactive_kresling():
    """
    Creates an interactive widget to explore a parametric Kresling net
    by adjusting the number of sides, radii, twist angle, height, etc.
    """
    w = interactive(
        plot_kresling_net,
        num_sides=IntSlider(min=3, max=20, step=1, value=6, description='Sides'),
        top_radius=FloatSlider(min=0.5, max=10.0, step=0.5, value=3.0, description='Top Radius'),
        bottom_radius=FloatSlider(min=0.5, max=10.0, step=0.5, value=3.0, description='Bottom Radius'),
        twist_deg=FloatSlider(min=-60, max=60, step=5, value=15, description='Twist (deg)'),
        height=FloatSlider(min=0.1, max=10.0, step=0.5, value=2.0, description='Height'),
        show_plot=Checkbox(value=True, description='Show Plot'),
        save_svg=Checkbox(value=False, description='Save SVG'),
        save_filename=''
    )
    display(w)

# Run this cell in a Jupyter notebook to create the interactive Kresling net:
interactive_kresling()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interactive, IntSlider, FloatSlider, Checkbox, Text
from IPython.display import display

def plot_kresling_net_with_MV(num_sides=6,
                              top_radius=3.0,
                              bottom_radius=3.0,
                              twist_deg=30.0,
                              height=2.0,
                              show_plot=True,
                              save_svg=False,
                              save_filename=''):
    """
    Plot a single-level Kresling net (unrolled) with a standard
    alternating Mountain/Valley diagonal assignment.
    
    Parameters
    ----------
    num_sides : int
        Number of polygon sides (n).
    top_radius : float
        Radius of the top n-gon.
    bottom_radius : float
        Radius of the bottom n-gon.
    twist_deg : float
        Twist angle (in degrees) between top and bottom polygons.
    height : float
        Axial distance (vertical) between top and bottom rings.
    show_plot : bool
        If True, show the plot inline.
    save_svg : bool
        If True, save the resulting figure in SVG format.
    save_filename : str
        Filename for saving, used only if save_svg == True.
    
    Mountain/Valley Assignment
    --------------------------
    - We label the diagonal fold in each trapezoid as M or V,
      alternating around the ring.
    - 'M' lines are colored red dashed; 'V' lines are blue dashed.
    
    References
    ----------
    - Misseroni et al. "Origami engineering," Nature Reviews Methods Primers,
      (2024) 4:40. 
    - Kresling, B. "Natural twist buckling in shells," Proc. Roy. Soc. A, 2008.
    - Chen, Y. et al. "Origami of thick panels," Science, 349, 2015.
    """
    # Convert twist to radians
    twist_rad = np.radians(twist_deg)
    
    # Angles for each side (0..2π in n segments + closure)
    angles = np.linspace(0, 2*np.pi, num_sides+1)
    
    # X/Y for top polygon
    x_top = top_radius * np.cos(angles)
    y_top = top_radius * np.sin(angles)
    
    # X/Y for bottom polygon (shifted by twist_rad)
    x_bot = bottom_radius * np.cos(angles + twist_rad)
    y_bot = bottom_radius * np.sin(angles + twist_rad)
    
    # Compute arc length along top edges for flattening
    def seg_len(xa, ya, xb, yb):
        return np.sqrt((xb - xa)**2 + (yb - ya)**2)
    
    # Cumulative arc length around top ring
    top_arclen = [0.0]
    for i in range(num_sides):
        s = seg_len(x_top[i], y_top[i], x_top[i+1], y_top[i+1])
        top_arclen.append(top_arclen[-1] + s)
    total_perim = top_arclen[-1]
    
    # We'll flatten the top ring at y=0, from x=0 to x=total_perim
    # Each trapezoid i goes from xA_top to xB_top on the top edge
    # and from xA_bot to xB_bot on the bottom edge, placed at y = -height
    
    # Generate lists of line segments for plotting
    top_segments = []
    bottom_segments = []
    diagonal_segments = []  # We'll store ( (xA_top,0),(xA_bot,-height), i ) so we know which index
    
    for i in range(num_sides):
        # top edge:
        xA_top = top_arclen[i]
        xB_top = top_arclen[i+1]
        top_segments.append( ((xA_top,0), (xB_top,0)) )
        
        # bottom edge length in real shape:
        b_len = seg_len(x_bot[i], y_bot[i], x_bot[i+1], y_bot[i+1])
        # place it in net from xA_bot to xB_bot at y=-height
        xA_bot = xA_top
        xB_bot = xA_bot + b_len
        bottom_segments.append( ((xA_bot, -height), (xB_bot, -height)) )
        
        # Diagonals: left side => (xA_top,0)->(xA_bot, -height)
        #            right side => (xB_top,0)->(xB_bot, -height)
        # Kresling typically has one diagonal fold in each trapezoid
        # We'll choose the diagonal from top-right to bottom-left or vice versa
        # that forms the "twisted" shape. Let's pick the diagonal from:
        #   (xB_top, 0) to (xA_bot, -height)
        # This is a standard approach, but can be reversed if you want.
        
        diag1 = ((xB_top, 0), (xA_bot, -height))
        # We'll store the index i so we can assign M or V in an alternating pattern
        diagonal_segments.append( (diag1, i) )
    
    # PLOT
    fig, ax = plt.subplots(figsize=(10,5))
    
    # Draw top edges (solid black)
    for seg in top_segments:
        ax.plot([seg[0][0], seg[1][0]], [seg[0][1], seg[1][1]],
                color='black', lw=2, solid_capstyle='round')
    
    # Draw bottom edges (solid black)
    for seg in bottom_segments:
        ax.plot([seg[0][0], seg[1][0]], [seg[0][1], seg[1][1]],
                color='black', lw=2, solid_capstyle='round')
    
    # Draw diagonals with an alternating M/V assignment
    # Let's say even i => "mountain" (red dashed), odd i => "valley" (blue dashed).
    # Feel free to invert or adopt any pattern needed.
    for diag, idx in diagonal_segments:
        # If idx is even => Mountain, else => Valley
        if idx % 2 == 0:
            # Mountain
            c_color = 'red'
            style = (0,(5,5))  # dashed
            labeltext = 'M'
        else:
            # Valley
            c_color = 'blue'
            style = (0,(3,5,1,5))  # dotted dash
            labeltext = 'V'
        
        # Plot the diagonal
        ax.plot([diag[0][0], diag[1][0]],
                [diag[0][1], diag[1][1]],
                color=c_color, lw=2, ls=style)
        
        # (Optional) Place a small label near the midpoint
        mx = 0.5*(diag[0][0] + diag[1][0])
        my = 0.5*(diag[0][1] + diag[1][1])
        # Shift label slightly to avoid overlap
        ax.text(mx, my + 0.1, labeltext, color=c_color, fontsize=12,
                ha='center', va='center', fontweight='bold')
    
    ax.set_aspect('equal', 'box')
    ax.set_xlabel("Unrolled X (arbitrary units)")
    ax.set_ylabel("Y (arbitrary units)")
    ax.set_title(f"Kresling Net (n={num_sides}), Twist={twist_deg}°, "
                 f"M/V with standard alternating pattern")
    
    ax.grid(True, alpha=0.2)
    
    if show_plot:
        plt.show()
    if save_svg and save_filename.strip():
        fig.savefig(save_filename.strip(), format='svg')

def interactive_kresling_MV():
    """
    Create an interactive widget to explore a parametric single-level
    Kresling net with M/V assignment.
    """
    w = interactive(
        plot_kresling_net_with_MV,
        num_sides=IntSlider(min=3, max=20, step=1, value=6, description='Sides'),
        top_radius=FloatSlider(min=0.5, max=10.0, step=0.5, value=3.0, description='Top R'),
        bottom_radius=FloatSlider(min=0.5, max=10.0, step=0.5, value=3.0, description='Bot R'),
        twist_deg=FloatSlider(min=-90, max=90, step=5, value=30, description='Twist (deg)'),
        height=FloatSlider(min=0.1, max=5.0, step=0.5, value=2.0, description='Height'),
        show_plot=Checkbox(value=True, description='Show Plot'),
        save_svg=Checkbox(value=False, description='Save SVG'),
        save_filename=Text(value='', description='SVG Filename')
    )
    display(w)

# When run in a notebook, call this to show the interactive widget:
interactive_kresling_MV()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interactive, IntSlider, FloatSlider, Checkbox, Text
from IPython.display import display

def plot_multilevel_kresling_net(num_sides=6,
                                 num_levels=2,
                                 top_radius=3.0,
                                 bottom_radius=3.0,
                                 twist_deg=30.0,
                                 height=2.0,
                                 show_plot=True,
                                 save_svg=False,
                                 save_filename=''):
    """
    Plots a multi-level Kresling net with:
      - num_sides : # of polygon sides per ring.
      - num_levels: number of vertical 'segments' (so total rings = num_levels+1).
      - top_radius, bottom_radius: radii of the top ring (level 0) and bottom ring (level num_levels).
                                   The intermediate rings are linearly interpolated in radius.
      - twist_deg : base twist angle (°). Even-indexed levels get +twist_deg, odd-indexed get -twist_deg.
      - height    : vertical spacing between consecutive rings (the total net extends num_levels * height).
    
    Custom features:
      - No dash or M/V labeling (all angled lines are solid).
      - The first and last angled lines per level are black, the rest are red (or your chosen color).
      - Each trapezoid’s top/bottom edges are truncated to the boundary set by the angled lines,
        so there's no ‘extra horizontal space’ at each ring.
      - If show_plot = True, displays the figure. If save_svg = True and a filename is provided,
        saves as an SVG file.

    References
    ----------
    - Misseroni et al., "Origami Engineering", Nature Reviews Methods Primers (2024) 4:40.
    - Kresling, B. “Natural twist buckling in shells,” Proceedings of the Royal Society A, 2008.
    - Chen, Y. et al., “Origami of thick panels,” Science, 349, 2015.
    """

    # We will create ring i = 0..num_levels, each ring having:
    #   radius_i = linear interpolation from top_radius to bottom_radius
    #   twist_i  = +twist_deg if i is even, -twist_deg if odd
    # The ring is "flattened" in the net at y = -i*height.

    if num_levels < 1:
        # degenerate case => just 1 ring
        num_levels = 1

    # 1) Compute ring-specific radius & twist
    ring_indices = np.arange(num_levels + 1)
    # Linear interpolation for radius
    radii = top_radius + (ring_indices/float(num_levels))*(bottom_radius - top_radius)
    # Alternate sign for twist
    def ring_twist(i):
        return twist_deg if (i % 2 == 0) else -twist_deg

    # 2) For each ring i in [0..num_levels], compute polygon coords, then flatten them
    #    We store: ring_coords[i] = (x[i][k], y[i][k]) for k in 0..num_sides
    #    Then we do arc-length flattening so each ring i has "flat_x" from 0..something at y=-i*height

    # We'll store: ring_arclengths[i][k], the cumulative arc length from side 0..k
    ring_coords = []
    ring_arclengths = []
    angles = np.linspace(0, 2*np.pi, num_sides+1)  # closes the ring

    def seglen(xa, ya, xb, yb):
        return np.sqrt((xb - xa)**2 + (yb - ya)**2)

    for i in range(num_levels+1):
        twist_rad = np.radians(ring_twist(i))
        r = radii[i]
        # polygon
        x_r = r * np.cos(angles + twist_rad)
        y_r = r * np.sin(angles + twist_rad)
        ring_coords.append((x_r, y_r))

        # compute arc length
        arc_len = [0.0]
        for k in range(num_sides):
            arc_len.append( arc_len[-1] + seglen(x_r[k], y_r[k], x_r[k+1], y_r[k+1]) )
        ring_arclengths.append(arc_len)

    # 3) Build line segments for each "level" i in [0..num_levels-1],
    #    connecting ring i to ring i+1 with trapezoids. We'll store them in a list for plotting.
    #    Each trapezoid is formed by:
    #       top edge:    from (xA_top, y_i)   to (xB_top, y_i)
    #       bottom edge: from (xA_bot, y_i+1) to (xB_bot, y_i+1)
    #       angled edges: from top to bottom => left side, right side

    # Flatten ring i at y=-i*height; ring i+1 at y= -(i+1)*height

    # We'll define: xA_top = ring_arclengths[i][k], xB_top = ring_arclengths[i][k+1]
    #               xA_bot = ring_arclengths[i+1][k], xB_bot= ring_arclengths[i+1][k+1]
    # So each trapezoid is bounded horizontally by min and max of the angled edges at top/bottom.

    # We'll store them in a structure trapezoids[i][k] = dict with keys for top, bottom, left_angle, right_angle
    # Then we can plot them.

    trapezoids = []  # will be a list of lists of dict
    for i in range(num_levels):
        level_traps = []
        y_top = -i*height
        y_bot = -(i+1)*height

        arc_top = ring_arclengths[i]
        arc_bot = ring_arclengths[i+1]

        for k in range(num_sides):
            xA_top = arc_top[k]
            xB_top = arc_top[k+1]
            xA_bot = arc_bot[k]
            xB_bot = arc_bot[k+1]

            # left angled line: (xA_top, y_top) -> (xA_bot, y_bot)
            # right angled line: (xB_top, y_top)-> (xB_bot, y_bot)
            # We'll find the horizontal min/max at top and bottom
            #   top_min = min(xA_top, xB_top), top_max = max(xA_top, xB_top)
            #   bottom_min = min(xA_bot, xB_bot), etc.
            # But we want to define the trapezoid edges so that the top edge goes from xA_top..xB_top
            # and the bottom edge from xA_bot..xB_bot. We'll not reorder them, or else we might mix up
            # the "first angled line" vs. "last angled line" notion.

            top_edge = ((xA_top, y_top), (xB_top, y_top))
            bottom_edge = ((xA_bot, y_bot), (xB_bot, y_bot))
            left_angle = ((xA_top, y_top), (xA_bot, y_bot))
            right_angle = ((xB_top, y_top), (xB_bot, y_bot))

            level_traps.append({
                "top": top_edge,
                "bottom": bottom_edge,
                "left_angle": left_angle,
                "right_angle": right_angle,
                "index_k": k
            })

        trapezoids.append(level_traps)

    # 4) Plot
    fig, ax = plt.subplots(figsize=(10, 5))

    # We'll loop over each level i, then each trapezoid k, and draw the four edges:
    #    - top edge (solid)
    #    - bottom edge (solid)
    #    - left angled edge (solid)
    #    - right angled edge (solid)
    # The user wants "the first and last inner angle black" => means for each trapezoid k=0 => left angled line black,
    # for trapezoid k=(num_sides-1) => right angled line black, the rest are red. We'll interpret it carefully.

    # Note that each trapezoid has two angled edges: left_angle and right_angle.
    # The "first angled line" in that entire ring is the left_angle of k=0,
    # The "last angled line" in that ring is the right_angle of k=(num_sides-1).
    # The other angled edges in between we color red (or your choice).

    for i in range(num_levels):
        for trap in trapezoids[i]:
            # top edge
            (x1, y1), (x2, y2) = trap["top"]
            ax.plot([x1, x2], [y1, y2], color='black', lw=2)

            # bottom edge
            (x3, y3), (x4, y4) = trap["bottom"]
            ax.plot([x3, x4], [y3, y4], color='black', lw=2)

            # angled edges
            k = trap["index_k"]

            # left angled edge
            (lx1, ly1), (lx2, ly2) = trap["left_angle"]
            if k == 0:
                # first angled line in ring => black
                ax.plot([lx1, lx2], [ly1, ly2], color='black', lw=2)
            else:
                # middle lines => red
                ax.plot([lx1, lx2], [ly1, ly2], color='red', lw=2)

            # right angled edge
            (rx1, ry1), (rx2, ry2) = trap["right_angle"]
            if k == (num_sides - 1):
                # last angled line in ring => black
                ax.plot([rx1, rx2], [ry1, ry2], color='black', lw=2)
            else:
                # middle lines => red
                ax.plot([rx1, rx2], [ry1, ry2], color='red', lw=2)

    ax.set_aspect('equal', 'box')
    ax.set_xlabel("Unrolled X (arbitrary units)")
    ax.set_ylabel("Y (arbitrary units)")
    ax.set_title(f"Multi-Level Kresling Net: n={num_sides}, levels={num_levels}\n"
                 f"Top→Bottom radii = {top_radius}→{bottom_radius}, ±{twist_deg}° alternate")

    ax.grid(True, alpha=0.3)

    if show_plot:
        plt.show()
    if save_svg and save_filename.strip():
        fig.savefig(save_filename.strip(), format='svg')

def interactive_multilevel_kresling():
    """
    Creates an interactive widget for exploring multi-level Kresling nets
    with the custom features described.
    """
    w = interactive(
        plot_multilevel_kresling_net,
        num_sides=IntSlider(min=3, max=20, step=1, value=6, description='Sides'),
        num_levels=IntSlider(min=1, max=6, step=1, value=2, description='Levels'),
        top_radius=FloatSlider(min=0.5, max=10.0, step=0.5, value=3.0, description='Top R'),
        bottom_radius=FloatSlider(min=0.5, max=10.0, step=0.5, value=3.0, description='Bot R'),
        twist_deg=FloatSlider(min=0, max=90, step=5, value=30, description='Twist (deg)'),
        height=FloatSlider(min=0.1, max=5.0, step=0.5, value=2.0, description='Height'),
        show_plot=Checkbox(value=True, description='Show Plot'),
        save_svg=Checkbox(value=False, description='Save SVG'),
        save_filename=Text(value='', description='SVG Filename')
    )
    display(w)

# When run in a Jupyter notebook, call:
interactive_multilevel_kresling()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interactive, IntSlider, FloatSlider, Checkbox, Text
from IPython.display import display

def plot_multilevel_kresling_net(num_sides=6,
                                 num_levels=2,
                                 top_radius=3.0,
                                 bottom_radius=3.0,
                                 twist_deg=30.0,
                                 height=2.0,
                                 show_plot=True,
                                 save_svg=False,
                                 save_filename=''):
    """
    Plot a multi-level Kresling net with:
      - num_sides: number of polygon edges per ring.
      - num_levels: number of stacked "levels" (so total rings = num_levels+1).
      - top_radius, bottom_radius: radii of top & bottom rings; intermediate rings are interpolated.
      - twist_deg: base twist angle in degrees (even rings: +twist_deg, odd rings: -twist_deg).
      - height: vertical spacing between each adjacent ring.

    Key customizations:
      - No M/V labels or dashed lines (all folds are solid).
      - First/last angled lines in each ring are black; other angled lines are red.
      - Horizontal edges are truncated to the angled folds (no extra space).
      - If show_plot=True, display the plot; if save_svg=True and save_filename != '',
        save the figure as an SVG.

    References:
      - Misseroni et al., "Origami Engineering", Nature Reviews Methods Primers (2024) 4:40.
      - Kresling, B. "Natural twist buckling in shells," Proc. Roy. Soc. A, 2008.
      - Chen et al., "Origami of thick panels," Science, 349, 2015.
    """

    # Ensure at least 1 level if the user sets num_levels < 1
    if num_levels < 1:
        num_levels = 1

    # Helper to alternate +twist or -twist
    def ring_twist(i):
        return twist_deg if (i % 2 == 0) else -twist_deg

    # We'll have ring i = 0..num_levels
    # Each ring i has a radius computed by linear interpolation
    ring_indices = np.arange(num_levels + 1)
    radii = top_radius + (ring_indices / float(num_levels)) * (bottom_radius - top_radius)

    # We'll store polygon coordinates for each ring in ring_coords[i] = (x_i, y_i),
    # plus arc-length for flattening in ring_arclengths[i].
    ring_coords = []
    ring_arclengths = []

    angles = np.linspace(0, 2*np.pi, num_sides+1)  # closes the polygon

    def seg_len(xa, ya, xb, yb):
        return np.sqrt((xb - xa)**2 + (yb - ya)**2)

    # Build ring polygon coords, compute arc-length flattening
    for i in range(num_levels+1):
        tw_rad = np.radians(ring_twist(i))
        r = radii[i]
        x_ring = r * np.cos(angles + tw_rad)
        y_ring = r * np.sin(angles + tw_rad)
        ring_coords.append((x_ring, y_ring))

        # Compute arc lengths
        arc_len = [0.0]
        for k in range(num_sides):
            arc_len.append(arc_len[-1] + seg_len(x_ring[k], y_ring[k],
                                                x_ring[k+1], y_ring[k+1]))
        ring_arclengths.append(arc_len)

    # Now build trapezoids that connect ring i to ring i+1
    # trapezoids[i][k] holds top, bottom, left_angle, right_angle for the segment k at level i
    trapezoids = []
    for i in range(num_levels):
        level_traps = []
        y_top = -i * height
        y_bot = -(i+1) * height

        arc_top = ring_arclengths[i]
        arc_bot = ring_arclengths[i+1]

        for k in range(num_sides):
            # Flattened top edge
            xA_top = arc_top[k]
            xB_top = arc_top[k+1]
            top_edge = ((xA_top, y_top), (xB_top, y_top))

            # Flattened bottom edge
            xA_bot = arc_bot[k]
            xB_bot = arc_bot[k+1]
            bottom_edge = ((xA_bot, y_bot), (xB_bot, y_bot))

            # Angled edges
            left_angle = ((xA_top, y_top), (xA_bot, y_bot))
            right_angle = ((xB_top, y_top), (xB_bot, y_bot))

            level_traps.append({
                "top": top_edge,
                "bottom": bottom_edge,
                "left_angle": left_angle,
                "right_angle": right_angle,
                "index_k": k
            })
        trapezoids.append(level_traps)

    # Plot
    fig, ax = plt.subplots(figsize=(10, 5))

    for i in range(num_levels):
        for trap in trapezoids[i]:
            # top edge
            (x1, y1), (x2, y2) = trap["top"]
            ax.plot([x1, x2], [y1, y2], color='black', lw=2)

            # bottom edge
            (x3, y3), (x4, y4) = trap["bottom"]
            ax.plot([x3, x4], [y3, y4], color='black', lw=2)

            # angled edges
            k = trap["index_k"]

            # left angled edge
            (lx1, ly1), (lx2, ly2) = trap["left_angle"]
            if k == 0:
                # first angled line => black
                ax.plot([lx1, lx2], [ly1, ly2], color='black', lw=2)
            else:
                # otherwise => red
                ax.plot([lx1, lx2], [ly1, ly2], color='red', lw=2)

            # right angled edge
            (rx1, ry1), (rx2, ry2) = trap["right_angle"]
            if k == (num_sides - 1):
                # last angled line => black
                ax.plot([rx1, rx2], [ry1, ry2], color='black', lw=2)
            else:
                ax.plot([rx1, rx2], [ry1, ry2], color='red', lw=2)

    ax.set_aspect('equal', 'box')
    ax.set_xlabel("Unrolled X (arbitrary units)")
    ax.set_ylabel("Y (arbitrary units)")
    ax.set_title(f"Multi-Level Kresling Net (n={num_sides}, levels={num_levels})\n"
                 f"Top→Bottom radii = {top_radius}→{bottom_radius}, Twist ±{twist_deg}° alt., height={height}")

    ax.grid(True, alpha=0.3)

    if show_plot:
        plt.show()
    if save_svg and save_filename.strip():
        fig.savefig(save_filename.strip(), format='svg')

def interactive_multilevel_kresling():
    """
    Creates an interactive widget to explore the multi-level Kresling net
    with all the features described above.
    """
    w = interactive(
        plot_multilevel_kresling_net,
        num_sides=IntSlider(min=3, max=20, step=1, value=6, description='Sides'),
        num_levels=IntSlider(min=1, max=6, step=1, value=2, description='Levels'),
        top_radius=FloatSlider(min=0.5, max=10.0, step=0.5, value=3.0, description='Top R'),
        bottom_radius=FloatSlider(min=0.5, max=10.0, step=0.5, value=3.0, description='Bot R'),
        twist_deg=FloatSlider(min=0, max=90, step=5, value=30, description='Twist (deg)'),
        height=FloatSlider(min=0.1, max=5.0, step=0.5, value=2.0, description='Height'),
        show_plot=Checkbox(value=True, description='Show Plot'),
        save_svg=Checkbox(value=False, description='Save SVG'),
        save_filename=Text(value='', description='SVG Filename')
    )
    display(w)

# To see the interactive widget in a Jupyter notebook, simply run:
interactive_multilevel_kresling()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interactive, IntSlider, FloatSlider, Checkbox, Text
from IPython.display import display

def plot_multilevel_kresling_net(num_sides=6,
                                 num_levels=2,
                                 top_radius=3.0,
                                 bottom_radius=3.0,
                                 twist_deg=30.0,
                                 height=2.0,
                                 show_plot=True,
                                 save_svg=False,
                                 save_filename=''):
    """
    Plots a multi-level Kresling net satisfying:

      - Ring 0 => 0° twist
      - Ring i>0 => 
          if i odd => +twist_deg   (/ shape in the net)
          if i even => -twist_deg  (\ shape in the net)
      - Each trapezoid corners = actual intersection of
        top/bottom edges with left/right angled lines,
        so no extra horizontal margin.
      - For / ring (odd i): 
          MOUNTAIN (black) = top-left  -> bottom-right
          VALLEY   (red)   = top-right -> bottom-left
      - For \ ring (even i>0):
          MOUNTAIN (black) = top-right -> bottom-left
          VALLEY   (red)   = top-left  -> bottom-right
      - All lines solid, no M/V text labels.

    Usage
    -----
    In a Jupyter notebook, run:
       interactive_multilevel_kresling()
    and adjust sliders for sides, levels, twist, etc.
    """

    # Ensure at least 1 level if user sets num_levels < 1
    if num_levels < 1:
        num_levels = 1

    # ring_twist(i): ring 0 => 0°, ring i>0 => ±twist_deg
    def ring_twist(i):
        if i == 0:
            return 0.0
        return ( twist_deg if (i % 2 == 1) else -twist_deg )

    # We'll have ring i in [0..num_levels]
    ring_indices = np.arange(num_levels + 1)
    # Linear interpolation for radius
    radii = top_radius + (ring_indices / float(num_levels)) * (bottom_radius - top_radius)

    ring_coords = []
    ring_arclengths = []

    angles = np.linspace(0, 2*np.pi, num_sides+1)  # closes the ring

    def seg_len(xa, ya, xb, yb):
        return np.sqrt((xb - xa)**2 + (yb - ya)**2)

    # 1) Build coordinates for each ring (polygon) + arc-length flattening
    for i in range(num_levels+1):
        tw_rad = np.radians(ring_twist(i))
        r = radii[i]
        x_ring = r * np.cos(angles + tw_rad)
        y_ring = r * np.sin(angles + tw_rad)
        ring_coords.append((x_ring, y_ring))

        # Arc length for flattening
        arc_len = [0.0]
        for k in range(num_sides):
            arc_len.append(arc_len[-1] + seg_len(x_ring[k], y_ring[k],
                                                x_ring[k+1], y_ring[k+1]))
        ring_arclengths.append(arc_len)

    # 2) Build trapezoids between ring i and ring i+1
    # Each trapezoid is formed by 4 intersection corners:
    #   top-left, top-right, bottom-left, bottom-right
    trapezoids = []
    for i in range(num_levels):
        y_top = -(i+1) * height
        y_bot = -(i) * height

        arc_top = ring_arclengths[i+i]
        arc_bot = ring_arclengths[i]

        level_traps = []
        for k in range(num_sides):
            # Intersection points => direct flattening
            xA_top = arc_top[k]
            xB_top = arc_top[k+1]
            xA_bot = arc_bot[k]
            xB_bot = arc_bot[k+1]

            # corners
            top_left     = (xA_top, y_top)   # intersection: top line & left angled line
            top_right    = (xB_top, y_top)   # intersection: top line & right angled line
            bottom_left  = (xA_bot, y_bot)
            bottom_right = (xB_bot, y_bot)

            level_traps.append({
                "i": i, "k": k,
                "top_left": top_left,
                "top_right": top_right,
                "bottom_left": bottom_left,
                "bottom_right": bottom_right
            })
        trapezoids.append(level_traps)

    # 3) Plot
    fig, ax = plt.subplots(figsize=(10, 5))

    for i in range(num_levels):
        ring_is_odd = (i % 2 == 0)  # ring 1,3,5 => / shape; ring 2,4,6 => \ shape

        for trap in trapezoids[i]:
            tl = trap["top_left"]
            tr = trap["top_right"]
            bl = trap["bottom_left"]
            br = trap["bottom_right"]

            # top edge in black
            ax.plot([tl[0], tr[0]], [tl[1], tr[1]], color='black', lw=2)
            # bottom edge in black
            ax.plot([bl[0], br[0]], [bl[1], br[1]], color='black', lw=2)

            # If i is odd => / shape:
            #    MOUNTAIN (black) = top-left -> bottom-right
            #    VALLEY   (red)   = top-right-> bottom-left
            #
            # If i is even => \ shape:
            #    MOUNTAIN (black) = top-right-> bottom-left
            #    VALLEY   (red)   = top-left-> bottom-right
            if ring_is_odd:
                # / shape
                mountain_start, mountain_end = tr, bl
                valley_start, valley_end     = tl, br
            else:
                # \ shape
                mountain_start, mountain_end = tl, br
                valley_start, valley_end     = tr, bl


            # draw mountain (black) and valley (red)
            ax.plot([mountain_start[0], mountain_end[0]],
                    [mountain_start[1], mountain_end[1]],
                    color='red', lw=2)

            ax.plot([valley_start[0], valley_end[0]],
                    [valley_start[1], valley_end[1]],
                    color='blue', lw=2)

    ax.set_aspect('equal', 'box')
    ax.set_xlabel("Unrolled X (arbitrary units)")
    ax.set_ylabel("Y (arbitrary units)")
    ax.set_title(
        f"Multi-Level Kresling Net\n"
        f"Ring 0=0°, odd i => +{twist_deg}°, even i>0 => -{twist_deg}°, / vs. \\ shapes.\n"
        f"Mountains=black, Valleys=red (all solid)."
    )

    ax.grid(True, alpha=0.3)

    if show_plot:
        plt.show()
    if save_svg and save_filename.strip():
        fig.savefig(save_filename.strip(), format='svg')

def interactive_multilevel_kresling():
    """
    Creates an interactive widget for the multi-level Kresling net with
    ring 0=0°, odd i => +twist, even i>0 => -twist,
    parallelograms labeled with black (mountain) and red (valley) diagonals.
    """
    w = interactive(
        plot_multilevel_kresling_net,
        num_sides=IntSlider(min=3, max=20, step=1, value=6, description='Sides'),
        num_levels=IntSlider(min=1, max=6, step=1, value=2, description='Levels'),
        top_radius=FloatSlider(min=0.5, max=10.0, step=0.5, value=3.0, description='Top R'),
        bottom_radius=FloatSlider(min=0.5, max=10.0, step=0.5, value=3.0, description='Bot R'),
        twist_deg=FloatSlider(min=0, max=90, step=5, value=30, description='Twist (deg)'),
        height=FloatSlider(min=0.1, max=5.0, step=0.5, value=2.0, description='Height'),
        show_plot=Checkbox(value=True, description='Show Plot'),
        save_svg=Checkbox(value=False, description='Save SVG'),
        save_filename=Text(value='', description='SVG Filename')
    )
    display(w)

# Usage in a Jupyter notebook:
# 1) Run this entire cell
interactive_multilevel_kresling()
# 3) Adjust parameters to see the net update
```

```python
import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interactive, IntSlider, FloatSlider, Checkbox, Text
from IPython.display import display

def plot_multilevel_kresling_net(num_sides=6,
                                 num_levels=2,
                                 top_radius=3.0,
                                 bottom_radius=3.0,
                                 twist_deg=30.0,
                                 height=2.0,
                                 show_plot=True,
                                 save_svg=False,
                                 save_filename=''):
    """
    Multi-level Kresling net with:
      - ring 0 => 0° twist
      - ring i>0 => if i is odd => +twist_deg (/ shape), 
                    if i is even => -twist_deg (\\ shape)
      - Each trapezoid corners = intersection of top/bottom edges 
        with left/right angled lines, so no extra margin.
      - For / ring (odd i): MOUNTAIN (black)=top-left->bottom-right,
                            VALLEY (red)=top-right->bottom-left
      - For \\ ring (even i>0): MOUNTAIN (black)=top-right->bottom-left,
                                VALLEY (red)=top-left->bottom-right
      - All lines are solid.

    ring i is placed at y = -i * height (flattened).
    The radius of ring i is linearly interpolated from
      top_radius (ring 0) to bottom_radius (ring num_levels).
    """

    # Safeguard
    if num_levels < 1:
        num_levels = 1

    # Define the twist for ring i
    def ring_twist(i):
        if i == 0:
            return 0.0
        return (twist_deg if (i % 2 == 1) else -twist_deg)

    # Build ring i in [0..num_levels]
    ring_indices = np.arange(num_levels + 1)
    # Interpolate radius
    radii = top_radius + (ring_indices / float(num_levels)) * (bottom_radius - top_radius)

    # We'll store each ring's polygon coords and arc-length
    ring_coords = []
    ring_arclengths = []

    angles = np.linspace(0, 2*np.pi, num_sides+1)  # closes the ring

    def seg_len(xa, ya, xb, yb):
        return np.sqrt((xb - xa)**2 + (yb - ya)**2)

    # 1) Build ring polygons and flattening arc-length
    for i in range(num_levels+1):
        tw = np.radians(ring_twist(i))
        r = radii[i]
        x_ring = r * np.cos(angles + tw)
        y_ring = r * np.sin(angles + tw)
        ring_coords.append((x_ring, y_ring))

        # Arc length
        arc_len = [0.0]
        for k in range(num_sides):
            arc_len.append(arc_len[-1] + seg_len(x_ring[k], y_ring[k],
                                                x_ring[k+1], y_ring[k+1]))
        ring_arclengths.append(arc_len)

    # 2) Build trapezoids connecting ring i (top) to ring i+1 (bottom)
    # ring i is at y=-i*height, ring i+1 at y=-(i+1)*height
    trapezoids = []
    for i in range(num_levels):
        y_top = -i * height
        y_bot = -(i+1) * height

        # Flattening arcs
        arc_top = ring_arclengths[i]   # ring i
        arc_bot = ring_arclengths[i+1] # ring i+1

        level_traps = []
        for k in range(num_sides):
            # Intersection points => direct flattening
            xA_top = arc_top[k]
            xB_top = arc_top[k+1]
            xA_bot = arc_bot[k]
            xB_bot = arc_bot[k+1]

            # corners
            top_left     = (xA_top, y_top)   # top edge left
            top_right    = (xB_top, y_top)   # top edge right
            bottom_left  = (xA_bot, y_bot)   # bottom edge left
            bottom_right = (xB_bot, y_bot)   # bottom edge right

            level_traps.append({
                "ring_idx": i, "k": k,
                "top_left": top_left,
                "top_right": top_right,
                "bottom_left": bottom_left,
                "bottom_right": bottom_right
            })
        trapezoids.append(level_traps)

    # 3) Plot
    fig, ax = plt.subplots(figsize=(10, 5))

    for i in range(num_levels):
        # if i is odd => / shape, if i is even => \ shape
        # (But user specifically said ring i>0 => odd => /, even => \. 
        #  That implies ring 1 => /, ring 2 => \, etc.)
        # ring 0 => no twist but let's treat as "even" or "odd"?
        # We'll interpret i>0 logic here. For i=0 => no twist anyway, so let's just unify:
        is_odd_ring = (i % 2 == 1)

        for trap in trapezoids[i]:
            tl = trap["top_left"]
            tr = trap["top_right"]
            bl = trap["bottom_left"]
            br = trap["bottom_right"]

            # Draw top and bottom edges in black
            ax.plot([tl[0], tr[0]], [tl[1], tr[1]], color='red', lw=2)
            if i == 0 or i == num_levels + 1:
                ax.plot([tl[0], tr[0]], [tl[1], tr[1]], color='black', lw=2)
            ax.plot([bl[0], br[0]], [bl[1], br[1]], color='black', lw=2)

            # According to the docstring for ring i>0:
            #  / shape (odd i):   MOUNTAIN= tl->br (black), VALLEY= tr->bl (red)
            #  \ shape (even i>0):MOUNTAIN= tr->bl (black), VALLEY= tl->br (red)
            # If i=0 => no twist, but let's just define a fallback (we can treat i=0 as even).
            
            if is_odd_ring:
                # / shape
                mountain_start, mountain_end = tl, bl
                valley_start, valley_end     = tr, bl
            else:
                # \ shape
                mountain_start, mountain_end = tl, bl
                valley_start, valley_end     = tl, br
            if 
            # Mountain in red
            ax.plot([mountain_start[0], mountain_end[0]],
                    [mountain_start[1], mountain_end[1]],
                    color='red', lw=2)

            # Valley in blue
            ax.plot([valley_start[0], valley_end[0]],
                    [valley_start[1], valley_end[1]],
                    color='blue', lw=2)

    ax.set_aspect('equal', 'box')
    
    ax.grid(True, alpha=0.3)

    if show_plot:
        plt.show()

    if save_svg and save_filename.strip():
        ax.axis('off')
        fig.savefig(save_filename.strip(), format='svg')

def interactive_multilevel_kresling():
    """
    Interactive widget for the refactored multi-level Kresling net.
    """
    w = interactive(
        plot_multilevel_kresling_net,
        num_sides=IntSlider(min=3, max=20, step=1, value=6, description='Sides'),
        num_levels=IntSlider(min=1, max=6, step=1, value=2, description='Levels'),
        top_radius=FloatSlider(min=0.5, max=10.0, step=0.5, value=3.0, description='Top R'),
        bottom_radius=FloatSlider(min=0.5, max=10.0, step=0.5, value=3.0, description='Bot R'),
        twist_deg=FloatSlider(min=0, max=90, step=5, value=30, description='Twist (deg)'),
        height=FloatSlider(min=0.1, max=5.0, step=0.5, value=2.0, description='Height'),
        show_plot=Checkbox(value=True, description='Show Plot'),
        save_svg=Checkbox(value=False, description='Save SVG'),
        save_filename=Text(value='', description='SVG Filename')
    )
    display(w)


# Uncomment if you want it to auto-run in the notebook:
interactive_multilevel_kresling()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interactive, IntSlider, FloatSlider, Checkbox, Text
from IPython.display import display

def plot_multilevel_kresling_net(num_sides=6,
                                 num_levels=2,
                                 top_radius=3.0,
                                 bottom_radius=3.0,
                                 twist_deg=30.0,
                                 height=2.0,
                                 show_plot=True,
                                 save_svg=False,
                                 save_filename=''):
    """
    Multi-level Kresling net with:
      - ring 0 => 0° twist
      - ring i>0 => if i is odd => +twist_deg (/ shape), 
                    if i is even => -twist_deg (\\ shape)
      - Each trapezoid corners = intersection of top/bottom edges 
        with left/right angled lines, so no extra margin.
      - For / ring (odd i): 
          MOUNTAIN (black) = top-left->bottom-right
          VALLEY   (red)   = top-right->bottom-left
      - For \\ ring (even i>0): 
          MOUNTAIN (black) = top-right->bottom-left
          VALLEY   (red)   = top-left->bottom-right
      - All lines are solid.

    ring i is placed at y = -i*height (flattened).
    The radius of ring i is linearly interpolated from
      top_radius (ring 0) to bottom_radius (ring num_levels).
    """

    # Safeguard
    if num_levels < 1:
        num_levels = 1

    # ring 0 => 0°, ring i>0 => ±twist_deg
    def ring_twist(i):
        if i == 0:
            return 0.0
        return (twist_deg if (i % 2 == 1) else -twist_deg)

    # Build ring i in [0..num_levels]
    ring_indices = np.arange(num_levels + 1)
    # Interpolate radius
    radii = top_radius + (ring_indices / float(num_levels)) * (bottom_radius - top_radius)

    ring_coords = []
    ring_arclengths = []

    angles = np.linspace(0, 2*np.pi, num_sides+1)  # closes the ring

    def seg_len(xa, ya, xb, yb):
        return np.sqrt((xb - xa)**2 + (yb - ya)**2)

    # 1) Build ring polygons + flattening arc-length
    for i in range(num_levels+1):
        tw = np.radians(ring_twist(i))
        r = radii[i]
        x_ring = r * np.cos(angles + tw)
        y_ring = r * np.sin(angles + tw)
        ring_coords.append((x_ring, y_ring))

        arc_len = [0.0]
        for k in range(num_sides):
            arc_len.append(arc_len[-1] + seg_len(x_ring[k], y_ring[k],
                                                x_ring[k+1], y_ring[k+1]))
        ring_arclengths.append(arc_len)

    # 2) Build trapezoids connecting ring i to ring i+1
    # ring i => y=-i*height, ring i+1 => y=-(i+1)*height
    trapezoids = []
    for i in range(num_levels):
        y_top = -i * height
        y_bot = -(i+1) * height

        arc_top = ring_arclengths[i]
        arc_bot = ring_arclengths[i+1]

        level_traps = []
        for k in range(num_sides):
            xA_top = arc_top[k]
            xB_top = arc_top[k+1]
            xA_bot = arc_bot[k]
            xB_bot = arc_bot[k+1]

            top_left     = (xA_top, y_top)
            top_right    = (xB_top, y_top)
            bottom_left  = (xA_bot, y_bot)
            bottom_right = (xB_bot, y_bot)

            level_traps.append({
                "ring_idx": i, "k": k,
                "top_left": top_left,
                "top_right": top_right,
                "bottom_left": bottom_left,
                "bottom_right": bottom_right
            })
        trapezoids.append(level_traps)

    # 3) Plot
    fig, ax = plt.subplots(figsize=(10, 5))

    for i in range(num_levels):
        # ring i>0 => if i odd => / shape, if i even => \ shape
        # ring 0 => no twist => treat as "even"
        ring_is_odd = (i % 2 == 1)

        for trap in trapezoids[i]:
            k = trap["k"]
            tl = trap["top_left"]
            tr = trap["top_right"]
            bl = trap["bottom_left"]
            br = trap["bottom_right"]

            # Top & bottom edges always black
            ax.plot([tl[0], tr[0]], [tl[1], tr[1]], color='red', lw=2)
            if i == 0 or i == num_levels + 1:
                ax.plot([tl[0], tr[0]], [tl[1], tr[1]], color='black', lw=2)
            ax.plot([bl[0], br[0]], [bl[1], br[1]], color='black', lw=2)

            # FIRST vertical line if k==0 => black
            # LAST  vertical line if k==(num_sides-1) => black
            # otherwise, let's keep them black as well for consistent outlining
            ax.plot([tl[0], bl[0]], [tl[1], bl[1]], color='red', lw=2)
            if k == 0:
                ax.plot([tl[0], bl[0]], [tl[1], bl[1]], color='black', lw=2)
            ax.plot([tr[0], br[0]], [tr[1], br[1]], color='black', lw=2)

            # MOUNTAIN/VALLEY lines
            if ring_is_odd:
                # / shape
                # MOUNTAIN= top-left->bottom-right (black)
                # VALLEY=   top-right->bottom-left (red)
                mt_start, mt_end = tl, br
                v_start, v_end   = tr, bl
            else:
                # \ shape
                # MOUNTAIN= top-right->bottom-left (black)
                # VALLEY=   top-left->bottom-right (red)
                mt_start, mt_end = tr, bl
                v_start, v_end   = tl, br

            # draw them
            # ax.plot([mt_start[0], mt_end[0]], [mt_start[1], mt_end[1]],
            #         color='black', lw=2)
            ax.plot([v_start[0], v_end[0]], [v_start[1], v_end[1]],
                    color='blue', lw=2)

    ax.set_aspect('equal', 'box')
    ax.grid(True, alpha=0.3)

    if show_plot:
        plt.show()

    if save_svg and save_filename.strip():
        # Optional: hide axes if you want a purely outline net
        ax.axis('off')
        fig.savefig(save_filename.strip(), format='svg')

def interactive_multilevel_kresling():
    """
    Interactive widget for the multi-level Kresling net
    with correct top/bottom edges, first/last vertical lines in black,
    mountain/valley lines, etc.
    """
    w = interactive(
        plot_multilevel_kresling_net,
        num_sides=IntSlider(min=3, max=20, step=1, value=6, description='Sides'),
        num_levels=IntSlider(min=1, max=6, step=1, value=2, description='Levels'),
        top_radius=FloatSlider(min=0.5, max=10.0, step=0.5, value=3.0, description='Top R'),
        bottom_radius=FloatSlider(min=0.5, max=10.0, step=0.5, value=3.0, description='Bot R'),
        twist_deg=FloatSlider(min=0, max=90, step=5, value=30, description='Twist (deg)'),
        height=FloatSlider(min=0.1, max=5.0, step=0.5, value=2.0, description='Height'),
        show_plot=Checkbox(value=True, description='Show Plot'),
        save_svg=Checkbox(value=False, description='Save SVG'),
        save_filename=Text(value='', description='SVG Filename')
    )
    display(w)

# If you want it to auto-run in the notebook:
interactive_multilevel_kresling()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interactive, IntSlider, FloatSlider, Checkbox, Text
from IPython.display import display

def plot_multilevel_kresling_net(num_sides=6,
                                 num_levels=2,
                                 radius=3.0,             # <-- Single radius for all levels
                                 twist_deg=30.0,
                                 height=2.0,
                                 show_plot=True,
                                 save_svg=False,
                                 save_filename=''):
    """
    Multi-level Kresling net with:
      - ring i => i * twist_deg  (0°, twist_deg, 2*twist_deg, ...)
      - A single radius for all rings
      - ring i is placed at y = -i*height (flattened).
    """

    # Safeguard: ensure at least 1 level
    if num_levels < 1:
        num_levels = 1

    # ring_twist(i) = i * twist_deg
    def ring_twist(i):
        return i * twist_deg

    # Prepare angles for each side
    angles = np.linspace(0, 2*np.pi, num_sides+1)  # closes the ring

    # Helper for segment length
    def seg_len(xa, ya, xb, yb):
        return np.sqrt((xb - xa)**2 + (yb - ya)**2)

    # 1) Build coordinates + flattened arc-length for rings 0..num_levels
    ring_coords = []
    ring_arclengths = []
    for i in range(num_levels+1):
        tw = np.radians(ring_twist(i))
        # Single radius for all rings
        r = radius
        x_ring = r * np.cos(angles + tw)
        y_ring = r * np.sin(angles + tw)
        ring_coords.append((x_ring, y_ring))

        # Compute cumulative arc length around the ring
        arc_len = [0.0]
        for k in range(num_sides):
            arc_len.append(arc_len[-1] + seg_len(x_ring[k], y_ring[k],
                                                x_ring[k+1], y_ring[k+1]))
        ring_arclengths.append(arc_len)

    # 2) Build trapezoids connecting ring i to ring i+1
    trapezoids = []
    for i in range(num_levels):
        y_top = -i * height
        y_bot = -(i+1) * height

        arc_top = ring_arclengths[i]
        arc_bot = ring_arclengths[i+1]

        level_traps = []
        for k in range(num_sides):
            xA_top = arc_top[k]
            xB_top = arc_top[k+1]
            xA_bot = arc_bot[k]
            xB_bot = arc_bot[k+1]

            top_left     = (xA_top, y_top)
            top_right    = (xB_top, y_top)
            bottom_left  = (xA_bot, y_bot)
            bottom_right = (xB_bot, y_bot)

            level_traps.append({
                "ring_idx": i, "k": k,
                "top_left": top_left,
                "top_right": top_right,
                "bottom_left": bottom_left,
                "bottom_right": bottom_right
            })
        trapezoids.append(level_traps)

    # 3) Plot
    fig, ax = plt.subplots(figsize=(10, 5))

    for i in range(num_levels):
        # ring 0 => twist=0°, ring 1 => twist=twist_deg, ring 2 => 2*twist_deg, etc.
        ring_tw = ring_twist(i)
        ring_is_odd = (i % 2 == 1)

        for trap in trapezoids[i]:
            k = trap["k"]
            tl = trap["top_left"]
            tr = trap["top_right"]
            bl = trap["bottom_left"]
            br = trap["bottom_right"]

            # Top & bottom edges
            ax.plot([tl[0], tr[0]], [tl[1], tr[1]], color='red', lw=2)
            # Optionally highlight ring 0 or the final ring boundary
            if i == 0 or i == num_levels + 1:
                ax.plot([tl[0], tr[0]], [tl[1], tr[1]], color='black', lw=2)
            ax.plot([bl[0], br[0]], [bl[1], br[1]], color='black', lw=2)

            # Vertical edges
            ax.plot([tl[0], bl[0]], [tl[1], bl[1]], color='red', lw=2)
            if k == 0:
                ax.plot([tl[0], bl[0]], [tl[1], bl[1]], color='black', lw=2)
            ax.plot([tr[0], br[0]], [tr[1], br[1]], color='black', lw=2)

            # MOUNTAIN/VALLEY lines
            if ring_is_odd:
                # For ring i=odd => / shape in original code
                mt_start, mt_end = tl, br  # black line was commented out
                v_start, v_end   = tr, bl
            else:
                # For ring i=even => \ shape
                mt_start, mt_end = tr, bl
                v_start, v_end   = tl, br

            # Optionally draw the valley line in blue
            ax.plot([v_start[0], v_end[0]], [v_start[1], v_end[1]],
                    color='blue', lw=2)
            # If you want to restore the mountain line, uncomment:
            ax.plot([mt_start[0], mt_end[0]], [mt_start[1], mt_end[1]],
                    color='black', lw=2)

    ax.set_aspect('equal', 'box')
    ax.grid(True, alpha=0.3)

    if show_plot:
        plt.show()

    # Save as SVG if requested
    if save_svg and save_filename.strip():
        ax.axis('off')
        fig.savefig(save_filename.strip(), format='svg')

def interactive_multilevel_kresling():
    """
    Interactive widget for the multi-level Kresling net:
      - ring i => i * twist_deg
      - Single radius for all levels
      - ring i at y = -i * height
    """
    w = interactive(
        plot_multilevel_kresling_net,
        num_sides=IntSlider(min=3, max=20, step=1, value=6, description='Sides'),
        num_levels=IntSlider(min=1, max=6, step=1, value=2, description='Levels'),
        radius=FloatSlider(min=0.5, max=10.0, step=0.5, value=3.0, description='Radius'),
        twist_deg=FloatSlider(min=0, max=90, step=5, value=30, description='Twist (deg)'),
        height=FloatSlider(min=0.1, max=5.0, step=0.5, value=2.0, description='Height'),
        show_plot=Checkbox(value=True, description='Show Plot'),
        save_svg=Checkbox(value=False, description='Save SVG'),
        save_filename=Text(value='', description='SVG Filename')
    )
    display(w)

# To auto-run in a notebook environment, simply call:
interactive_multilevel_kresling()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interactive, IntSlider, FloatSlider, Checkbox, Text
from IPython.display import display

def plot_multilevel_kresling_net(num_sides=6,
                                 num_levels=2,
                                 radius=3.0,             # Single radius for all rings
                                 twist_deg=30.0,
                                 height=2.0,
                                 show_plot=True,
                                 save_svg=False,
                                 save_filename=''):
    """
    Multi-level Kresling net generator with:
      - ring i => twist = i * twist_deg
      - ring i at y = -i * height
      - single radius (constant across levels)
      - output is a 2D "net" (flattened) connecting each consecutive pair of rings.

    Parameters
    ----------
    num_sides : int
        Number of polygon sides in each ring.
    num_levels : int
        Number of cylindrical (or "frustum-like") segments stacked vertically.
    radius : float
        The radius for every ring (no top vs. bottom difference).
    twist_deg : float
        Base twist in degrees. Ring i has i*twist_deg rotation.
    height : float
        Vertical spacing between consecutive rings.
    show_plot : bool
        If True, shows the Matplotlib figure.
    save_svg : bool
        If True, exports the figure as an SVG (requires save_filename).
    save_filename : str
        Path/filename for saving the SVG output (only used if save_svg is True).

    Notes
    -----
    - The net is constructed by flattening each ring's circumference
      into an approximate straight arc using cumulative edge lengths.
    - Each trapezoid in the net corresponds to a side of ring i joined
      to the same side of ring i+1.

    References
    ----------
    - Tachi, T. (2013). "Freeform Origami Tessellations by Generalized Resch’s Patterns."
      Journal of Mechanical Design.
    - Demaine, E. D. & O'Rourke, J. (2007). 'Geometric Folding Algorithms: Linkages, 
      Origami, Polyhedra.' Cambridge University Press.
    - Python documentation on defining functions: 
      https://docs.python.org/3/tutorial/controlflow.html#defining-functions
    """

    if num_levels < 1:
        num_levels = 1  # Safeguard against invalid input

    # Function to compute how many degrees of twist ring i has
    def ring_twist(i):
        return i * twist_deg

    # For each ring i, we build a polygon of num_sides, 
    # rotated by ring_twist(i) and of radius = `radius`.
    angles = np.linspace(0, 2*np.pi, num_sides + 1)  # ensures closure

    # Helper for Euclidean distance
    def seg_len(xa, ya, xb, yb):
        return np.sqrt((xb - xa)**2 + (yb - ya)**2)

    # 1) Build ring coordinates + flatten them into arc-length for i in [0..num_levels]
    ring_coords = []       # holds (x[], y[]) for each ring
    ring_arclengths = []   # holds arc-length array for each ring

    for i in range(num_levels + 1):
        # Twist for ring i (in radians)
        tw = np.radians(ring_twist(i))

        # Coordinates of ring i
        x_ring = radius * np.cos(angles + tw)
        y_ring = radius * np.sin(angles + tw)
        ring_coords.append((x_ring, y_ring))

        # Compute cumulative arc-length
        arc_len = [0.0]
        for k in range(num_sides):
            arc_len.append(
                arc_len[-1] + seg_len(x_ring[k], y_ring[k],
                                      x_ring[k+1], y_ring[k+1])
            )
        ring_arclengths.append(arc_len)

    # 2) Construct trapezoids bridging ring i and ring i+1
    #    Each ring is placed at y=-i*height (flattened), 
    #    so top of trapezoid is at y_top = -i*height
    #    bottom is y_bot = -(i+1)*height
    trapezoids = []
    for i in range(num_levels):
        y_top = -i * height
        y_bot = -(i + 1) * height
        arc_top = ring_arclengths[i]
        arc_bot = ring_arclengths[i + 1]

        level_traps = []
        for k in range(num_sides):
            xA_top = arc_top[k]
            xB_top = arc_top[k+1]
            xA_bot = arc_bot[k]
            xB_bot = arc_bot[k+1]

            trap_data = {
                "ring_idx": i,
                "k": k,
                "top_left":     (xA_top, y_top),
                "top_right":    (xB_top, y_top),
                "bottom_left":  (xA_bot, y_bot),
                "bottom_right": (xB_bot, y_bot)
            }
            level_traps.append(trap_data)

        trapezoids.append(level_traps)

    # 3) Plot the net
    fig, ax = plt.subplots(figsize=(10, 5))

    for i in range(num_levels):
        # "Odd ring" vs. "even ring" logic for mountain/valley lines
        ring_is_odd = (i % 2 == 1)

        for trap in trapezoids[i]:
            tl = trap["top_left"]
            tr = trap["top_right"]
            bl = trap["bottom_left"]
            br = trap["bottom_right"]
            k  = trap["k"]

            # Draw top edge (red), bottom edge (black)
            ax.plot([tl[0], tr[0]], [tl[1], tr[1]], color='red', lw=2)
            ax.plot([bl[0], br[0]], [bl[1], br[1]], color='black', lw=2)

            # Draw vertical edges
            ax.plot([tl[0], bl[0]], [tl[1], bl[1]], color='red', lw=2)
            ax.plot([tr[0], br[0]], [tr[1], br[1]], color='black', lw=2)
            # Optional extra highlighting for first side k=0
            if k == 0:
                ax.plot([tl[0], bl[0]], [tl[1], bl[1]], color='black', lw=2)

            # Diagonals (Mountain/Valley)
            if ring_is_odd:
                # / shape
                # MOUNTAIN: top-left->bottom-right
                # VALLEY:   top-right->bottom-left
                mt_start, mt_end = tl, br
                v_start, v_end   = tr, bl
            else:
                # \ shape
                # MOUNTAIN: top-right->bottom-left
                # VALLEY:   top-left->bottom-right
                mt_start, mt_end = tr, bl
                v_start, v_end   = tl, br

            # Example: only draw the valley line in blue
            ax.plot([v_start[0], v_end[0]], [v_start[1], v_end[1]],
                    color='blue', lw=2)
            # If you want the mountain line, uncomment below:
            # ax.plot([mt_start[0], mt_end[0]], [mt_start[1], mt_end[1]],
            #         color='black', lw=2)

    # Configure and optionally display
    ax.set_aspect('equal', 'box')
    ax.grid(True, alpha=0.3)

    if show_plot:
        plt.show()

    # If user wants to save to SVG
    if save_svg and save_filename.strip():
        ax.axis('off')
        fig.savefig(save_filename.strip(), format='svg')

def interactive_multilevel_kresling():
    """
    Creates an interactive widget for exploring the multi-level Kresling net.
    Adjust num_sides, num_levels, radius, twist_deg, height, and see changes live.

    References
    ----------
    - ipywidgets: https://ipywidgets.readthedocs.io/en/stable/
    """
    w = interactive(
        plot_multilevel_kresling_net,
        num_sides=IntSlider(min=3, max=20, step=1, value=6, description='Sides'),
        num_levels=IntSlider(min=1, max=6, step=1, value=2, description='Levels'),
        radius=FloatSlider(min=0.5, max=10.0, step=0.5, value=3.0, description='Radius'),
        twist_deg=FloatSlider(min=0, max=90, step=5, value=30, description='Twist (deg)'),
        height=FloatSlider(min=0.1, max=5.0, step=0.5, value=2.0, description='Height'),
        show_plot=Checkbox(value=True, description='Show Plot'),
        save_svg=Checkbox(value=False, description='Save SVG'),
        save_filename=Text(value='', description='SVG Filename')
    )
    display(w)
interactive_multilevel_kresling()
```

```python
# test_kresling.py
import numpy as np
import pytest

def test_ring_twist_calculation():
    """
    Check ring_twist(i) logic. For small i in [0..3], 
    we expect ring_twist(i) = i*twist_deg.
    """
    twist_deg = 30
    num_levels = 3
    # We'll capture the ring_twist calculation by hooking 
    # into a direct function or by analyzing ring 0..num_levels
    # Suppose we define ring_twist inline to mimic the code:
    def ring_twist(i):
        return i * twist_deg

    for i in range(num_levels+1):
        assert ring_twist(i) == i*twist_deg, f"Ring {i} twist should be i*{twist_deg}."

@pytest.mark.parametrize("num_sides,num_levels,twist_deg,height", [
    (6, 2, 30, 2.0),
    (5, 3, 20, 1.0),
])
def test_plot_runs_without_error(num_sides, num_levels, twist_deg, height):
    """
    Smoke test: ensure plot function runs without raising exceptions,
    and optionally check 'validate_twist' warnings are not thrown.
    """
    try:
        plot_multilevel_kresling_net(num_sides=num_sides,
                                     num_levels=num_levels,
                                     radius=3.0,
                                     twist_deg=twist_deg,
                                     height=height,
                                     show_plot=False,
                                     validate_twist=True)  # triggers validation
    except Exception as e:
        pytest.fail(f"Plot function raised an exception: {e}")
test_ring_twist_calculation()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interactive, IntSlider, FloatSlider, Checkbox, Text, Dropdown
from IPython.display import display

def plot_multilevel_kresling_net(num_sides=6,
                                 num_levels=2,
                                 radius=3.0,
                                 twist_deg=30.0,
                                 height=2.0,
                                 pattern_type='helical',   # <-- NEW parameter
                                 show_plot=True,
                                 save_svg=False,
                                 save_filename='',
                                 validate_twist=False):
    """
    Multi-level Kresling net generator with a single radius for all rings,
    plus an option to choose between 'helical' or 'zigzag' twist patterns.

    Parameters
    ----------
    num_sides : int
        Number of sides in each ring polygon.
    num_levels : int
        Number of stacked rings (levels).
    radius : float
        Radius for all rings (constant).
    twist_deg : float
        The base twist angle (in degrees).
    height : float
        Vertical spacing between consecutive rings.
    pattern_type : str
        'helical' => ring i has i * twist_deg;
        'zigzag'  => ring 0=0, ring 1=+twist_deg, ring 2=-twist_deg, ring 3=+twist_deg, etc.
    show_plot : bool
        If True, displays the resulting matplotlib figure.
    save_svg : bool
        If True, saves the figure as an SVG (requires save_filename).
    save_filename : str
        Path (or filename) to save the SVG if save_svg is True.
    validate_twist : bool
        If True, prints a simple validation check for the ring twist angles.

    Returns
    -------
    None
    """

    if num_levels < 1:
        num_levels = 1

    # Decide how ring i is twisted based on pattern_type
    def ring_twist(i):
        if pattern_type == 'helical':
            return i * twist_deg
        elif pattern_type == 'zigzag':
            if i == 0:
                return 0.0
            # odd i => +twist_deg, even i => -twist_deg
            return twist_deg if (i % 2 == 1) else -twist_deg
        else:
            # fallback to helical if unknown pattern
            return i * twist_deg

    # (Optional) validation check on ring i's first vertex
    def check_angle(i, x, y, deg_tolerance=1.0):
        angle = np.degrees(np.arctan2(y, x))  # range ~ [-180, 180]
        expected = ring_twist(i)
        diff = angle - expected
        while diff > 180:
            diff -= 360
        while diff < -180:
            diff += 360
        if abs(diff) > deg_tolerance:
            print(f"[Validate Twist] Ring {i}: first vertex angle ~ {angle:.2f}°, "
                  f"expected ~ {expected:.2f}° (±{deg_tolerance}°). diff={diff:.2f}°")

    angles = np.linspace(0, 2*np.pi, num_sides + 1)

    def seg_len(xa, ya, xb, yb):
        return np.sqrt((xb - xa)**2 + (yb - ya)**2)

    ring_coords = []
    ring_arclengths = []

    # 1) Build each ring
    for i in range(num_levels + 1):
        tw = np.radians(ring_twist(i))
        x_ring = radius * np.cos(angles + tw)
        y_ring = radius * np.sin(angles + tw)

        if validate_twist and len(x_ring) > 0:
            check_angle(i, x_ring[0], y_ring[0])

        ring_coords.append((x_ring, y_ring))

        # Flattening arc length
        arc_len = [0.0]
        for k in range(num_sides):
            arc_len.append(
                arc_len[-1] + seg_len(x_ring[k], y_ring[k], x_ring[k+1], y_ring[k+1])
            )
        ring_arclengths.append(arc_len)

    # 2) Build trapezoids bridging ring i and ring i+1
    trapezoids = []
    for i in range(num_levels):
        y_top = -i * height
        y_bot = -(i + 1) * height
        arc_top = ring_arclengths[i]
        arc_bot = ring_arclengths[i+1]

        level_traps = []
        for k in range(num_sides):
            xA_top = arc_top[k]
            xB_top = arc_top[k+1]
            xA_bot = arc_bot[k]
            xB_bot = arc_bot[k+1]
            trap_data = {
                "ring_idx": i, "k": k,
                "top_left": (xA_top, y_top),
                "top_right": (xB_top, y_top),
                "bottom_left": (xA_bot, y_bot),
                "bottom_right": (xB_bot, y_bot)
            }
            level_traps.append(trap_data)
        trapezoids.append(level_traps)

    # 3) Plotting
    fig, ax = plt.subplots(figsize=(10, 5))
    for i in range(num_levels):
        ring_is_odd = (i % 2 == 1)
        for trap in trapezoids[i]:
            tl, tr = trap["top_left"], trap["top_right"]
            bl, br = trap["bottom_left"], trap["bottom_right"]
            k      = trap["k"]

            # Edges
            ax.plot([tl[0], tr[0]], [tl[1], tr[1]], color='red', lw=2)
            ax.plot([bl[0], br[0]], [bl[1], br[1]], color='black', lw=2)
            ax.plot([tl[0], bl[0]], [tl[1], bl[1]], color='red', lw=2)
            ax.plot([tr[0], br[0]], [tr[1], br[1]], color='black', lw=2)
            # highlight k=0 edge
            if k == 0:
                ax.plot([tl[0], bl[0]], [tl[1], bl[1]], color='black', lw=2)

            # Diagonals (mountain/valley)
            if ring_is_odd:
                # / shape
                mt_start, mt_end = tl, br
                v_start, v_end   = tr, bl
            else:
                # \ shape
                mt_start, mt_end = tr, bl
                v_start, v_end   = tl, br

            # Valley line in blue
            ax.plot([v_start[0], v_end[0]], [v_start[1], v_end[1]],
                    color='blue', lw=2)
            # If you want mountain lines as well, uncomment:
            # ax.plot([mt_start[0], mt_end[0]], [mt_start[1], mt_end[1]],
            #         color='black', lw=2)

    ax.set_aspect('equal', 'box')
    ax.grid(True, alpha=0.3)

    if show_plot:
        plt.show()

    if save_svg and save_filename.strip():
        ax.axis('off')
        fig.savefig(save_filename.strip(), format='svg')

def interactive_multilevel_kresling():
    """
    Interactive widget to explore helical vs. zigzag Kresling patterns.
    """
    w = interactive(
        plot_multilevel_kresling_net,
        num_sides=IntSlider(min=3, max=20, step=1, value=6, description='Sides'),
        num_levels=IntSlider(min=1, max=6, step=1, value=2, description='Levels'),
        radius=FloatSlider(min=0.5, max=10.0, step=0.5, value=3.0, description='Radius'),
        twist_deg=FloatSlider(min=0, max=90, step=5, value=30, description='Twist (deg)'),
        height=FloatSlider(min=0.1, max=5.0, step=0.5, value=2.0, description='Height'),
        pattern_type=Dropdown(options=['helical','zigzag'], value='helical', description='Pattern'),
        show_plot=Checkbox(value=True, description='Show Plot'),
        save_svg=Checkbox(value=False, description='Save SVG'),
        save_filename=Text(value='', description='SVG Filename'),
        validate_twist=Checkbox(value=False, description='Validate Twist')
    )
    display(w)
interactive_multilevel_kresling()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interactive, FloatSlider, IntSlider, Checkbox
from IPython.display import display

# Enable inline plotting if in a Jupyter notebook
# %matplotlib inline  # (Uncomment if needed in your environment)

def compute_coordinates_with_rotation_and_height(m, n, alpha, row_spacing, y_deviation, theta_max):
    """
    Compute X/Y coordinates for a simplified Miura-ori-like grid that has:
      - No scaling factor (delta) across rows.
      - Each row i is placed at y = i * row_spacing + y_deviation[i].
      - Each row i is horizontally skewed by an alternating rotation angle ±theta_max.

    Parameters
    ----------
    m : int
        Number of rows (there will be m+1 horizontal lines).
    n : int
        Number of columns (there will be n+1 vertical lines).
    alpha : float
        Folding angle in radians (used to set horizontal side length = a * tan(alpha)).
    row_spacing : float
        A base vertical distance between consecutive rows.
    y_deviation : array-like of length (m+1)
        Additional per-row offsets in the vertical direction.
    theta_max : float
        The maximum rotation angle in radians, alternating sign per row.

    Returns
    -------
    X, Y : (m+1) x (n+1) arrays
        Coordinates of the grid vertices.
    """
    # Base vertical side length
    a = 1.0

    # Initialize coordinate arrays
    X = np.zeros((m+1, n+1))
    Y = np.zeros((m+1, n+1))

    for i in range(m+1):
        # Horizontal side length for row i
        b_i = a * np.tan(alpha)

        # Alternating sign for rotation angle
        rotation_angle = theta_max * ((-1)**i)

        # This row's base vertical coordinate
        row_y_base = i * row_spacing + y_deviation[i]

        for j in range(n+1):
            # Horizontal position if no skew
            base_x = j * b_i

            # Shift horizontally on odd rows to create a typical Miura-like tiling
            if i % 2 == 1:
                base_x += b_i / 2.0

            # Skew offset due to row i's rotation
            skew_offset = y_deviation[i] * np.tan(rotation_angle)

            # Assign final coordinates
            X[i, j] = base_x + skew_offset
            Y[i, j] = row_y_base

    # Optionally center the pattern horizontally
    X -= np.mean(X)

    return X, Y

def plot_miura_ori_with_diagonals_and_height(m, n, alpha, row_spacing, y_deviation, theta_max,
                                             show_plot=True, save_svg=False):
    """
    Plot a Miura-ori grid that has row-based height spacing and parallelogram diagonals.

    Parameters
    ----------
    m, n : int
        Grid dimensions.
    alpha : float
        Folding angle in radians.
    row_spacing : float
        Vertical spacing between consecutive rows.
    y_deviation : array-like
        Additional vertical offset for each row i.
    theta_max : float
        Alternating rotation amplitude (radians).
    show_plot : bool
        If True, display the Matplotlib figure.
    save_svg : bool
        If True, save the figure to an SVG file.
    """
    X, Y = compute_coordinates_with_rotation_and_height(m, n, alpha, row_spacing, y_deviation, theta_max)

    fig, ax = plt.subplots(figsize=(8, 8))

    # Draw vertical edges
    for irow in range(m):
        for jcol in range(n+1):
            x1, y1 = X[irow, jcol],   Y[irow, jcol]
            x2, y2 = X[irow+1, jcol], Y[irow+1, jcol]
            ax.plot([x1, x2], [y1, y2], color='red', lw=2)
            if jcol == 0 or jcol == n:
                ax.plot([x1, x2], [y1, y2], color='black', lw=2)

    # Draw horizontal edges
    for irow in range(m+1):
        for jcol in range(n):
            x1, y1 = X[irow, jcol],   Y[irow, jcol]
            x2, y2 = X[irow, jcol+1], Y[irow, jcol+1]
            ax.plot([x1, x2], [y1, y2], color='red', lw=2)
            if irow == 0 or irow == m:
                ax.plot([x1, x2], [y1, y2], color='black', lw=2)

    # Parallelogram diagonals in each cell
    for irow in range(m):
        for jcol in range(n):
            p00 = (X[irow, jcol],     Y[irow, jcol])
            p01 = (X[irow, jcol+1],   Y[irow, jcol+1])
            p10 = (X[irow+1, jcol],   Y[irow+1, jcol])
            p11 = (X[irow+1, jcol+1], Y[irow+1, jcol+1])
            # Diagonal #1 (top-left to bottom-right)
            if irow % 2 == 0:
                ax.plot([p00[0], p11[0]], [p00[1], p11[1]], color='blue', lw=1.5)
            # Diagonal #2 (top-right to bottom-left)
            else:
                ax.plot([p01[0], p10[0]], [p00[1], p11[1]], color='blue',  lw=1.5)

    # Aesthetics
    ax.set_aspect('equal', 'box')
    ax.axis('off')

    if show_plot:
        plt.show()

    if save_svg:
        filename = f"miura_simplified_m{m}_n{n}_a{alpha}_r{row_spacing}.svg"
        fig.savefig(filename, format='svg', dpi=300)
        print(f"Saved SVG as '{filename}'")

    plt.close(fig)

def interactive_miura_simplified_with_height():
    """
    Interactive widget that includes a 'row_spacing' parameter to give 
    the pattern visible height between rows.
    """
    def update_plot(m=4, n=5, alpha=1.0, row_spacing=1.0, theta_max=0.1,
                    show_plot=True, save_svg=False):
        # Example: zero y_deviation (or you can modify to set your own row-by-row offsets)
        y_dev = np.zeros(m+1)

        plot_miura_ori_with_diagonals_and_height(
            m, n, alpha, row_spacing, y_dev, theta_max,
            show_plot=show_plot,
            save_svg=save_svg
        )

    iw = interactive(
        update_plot,
        m=IntSlider(min=2, max=20, step=1, value=4, description='Rows (m)'),
        n=IntSlider(min=2, max=20, step=1, value=5, description='Cols (n)'),
        alpha=FloatSlider(min=0.1, max=np.pi/2, step=0.05, value=1.0, description='alpha (rad)'),
        row_spacing=FloatSlider(min=0.1, max=5.0, step=0.1, value=1.0, description='Row Spacing'),
        theta_max=FloatSlider(min=-np.pi/4, max=np.pi/4, step=0.01, value=0.1, description='±Rot. (rad)'),
        show_plot=Checkbox(value=True, description='Show Plot'),
        save_svg=Checkbox(value=False, description='Save SVG')
    )
    display(iw)

# Uncomment to run the interactive widget in a Jupyter environment:
interactive_miura_simplified_with_height()
```

```python
import matplotlib.pyplot as plt
import random
from mpl_toolkits.mplot3d import Axes3D  # noqa: F401 unused import

import re

# Step 1: Parse the DSL
def parse_dsl(dsl_text):
    # Extract inputs
    input_line = re.search(r"INPUT\s+(.*)", dsl_text)
    inputs = []
    if input_line:
        inputs = [x.strip() for x in input_line.group(1).split(',')]
    
    # Extract outputs
    output_line = re.search(r"OUTPUT\s+(.*)", dsl_text)
    outputs = []
    if output_line:
        outputs = [x.strip() for x in output_line.group(1).split(',')]
    
    # Extract logic definitions, e.g. "Z = (A & B) | (!C)"
    logic_pattern = re.compile(r"([\w]+)\s*=\s*(.*)")
    logic_lines = logic_pattern.findall(dsl_text)
    
    return inputs, outputs, logic_lines

# Step 2: Convert expressions to a netlist (very simplified).
# E.g., "Z = (A & B) | (!C)" -> 
# Gate1: G1 = A AND B
# Gate2: G2 = NOT C
# Gate3: Z = G1 OR G2

def synthesize_netlist(logic_defs):
    gate_count = 0
    netlist = []
    
    # A helper function to create a unique gate name
    def new_gate():
        nonlocal gate_count
        gate_count += 1
        return f"G{gate_count}"
    
    # Simple token-based approach for AND, OR, NOT
    for lhs, rhs in logic_defs:
        # Remove outer parentheses
        expr = rhs.strip()
        
        # We'll do a naive parse: split by OR first, then AND, then NOT
        # For a robust approach, you'd build an actual parser or use an existing library.
        # Here, we handle only a single-level expression for demo:
        
        # Example: (A & B) | (!C) -> two sub-expressions: "(A & B)" and "(!C)"
        or_parts = [p.strip(" ()") for p in expr.split('|')]
        
        if len(or_parts) == 1:
            # No OR => either AND or NOT or single var
            # We'll check for AND or NOT inside
            sub = or_parts[0]
            final_output = handle_subexpr(sub, netlist, new_gate, lhs)
            netlist.append((lhs, "OUT", final_output))
        else:
            # We have multiple sub-expressions joined by OR
            or_outputs = []
            for p in or_parts:
                sub_expr = p.strip()
                out_var = handle_subexpr(sub_expr, netlist, new_gate)
                or_outputs.append(out_var)
            
            # Now we need an OR gate to combine all sub-expressions
            if len(or_outputs) == 1:
                netlist.append((lhs, "OUT", or_outputs[0]))
            else:
                gate_name = new_gate()
                netlist.append((gate_name, "OR", or_outputs))
                netlist.append((lhs, "OUT", gate_name))
                
    return netlist

def handle_subexpr(expr, netlist, new_gate, final_name=None):
    """Parses a sub-expression that may contain AND, NOT, or a single variable."""
    
    # Check for NOT:
    if expr.startswith('!'):
        var = expr[1:].strip("() ")
        gate_name = new_gate()
        netlist.append((gate_name, "NOT", var))
        return gate_name
    
    # Check for AND:
    if '&' in expr:
        parts = [p.strip("() ") for p in expr.split('&')]
        if len(parts) == 2:
            gate_name = new_gate()
            netlist.append((gate_name, "AND", parts))
            return gate_name
        else:
            # More advanced logic needed if there's more than two operands
            pass
    
    # If it’s just a single variable
    return expr

# Example usage:
dsl = """
# Define inputs
INPUT A, B, C

# Define outputs
OUTPUT Z

# Define logic
Z = (A & B) | (!C)
"""

inputs, outputs, logic_defs = parse_dsl(dsl)
netlist = synthesize_netlist(logic_defs)

print("Inputs:", inputs)
print("Outputs:", outputs)
print("Netlist:")
for gate in netlist:
    print(" ", gate)
    
def assign_3d_positions(netlist):
    """
    Assigns each unique node in the netlist a 3D coordinate (x,y,z).
    Returns a dict: node_name -> (x, y, z)
    """
    # Collect all gate names + input signals that appear in the netlist
    # Example netlist item: ('G1', 'AND', ['A','B']) or ('Z','OUT','G3')
    unique_nodes = set()
    edges = []
    for item in netlist:
        src, gate_type, inputs = item
        unique_nodes.add(src)
        if isinstance(inputs, list):
            for i in inputs:
                unique_nodes.add(i)
                edges.append((i, src))
        else:
            # single input (e.g. 'C' or 'G2')
            unique_nodes.add(inputs)
            edges.append((inputs, src))
    
    # Simple layering approach: group by gate_type or netlist order
    # We'll just place them in “layers” along x-axis, random y,z
    node_positions = {}
    layer_counter = 0
    # to keep track of whether we encountered a gate or not
    gate_types = {"AND","OR","NOT","NAND","NOR","OUT","FLIPFLOP","REGISTER","TIMER","EDGEDET"}
    
    for node in unique_nodes:
        if node.startswith("G"): 
            # assume it's an internal gate
            x = layer_counter * 5.0
            y = random.uniform(-5, 5)
            z = random.uniform(-2, 2)
            layer_counter += 1
        elif node in gate_types:
            # if the node is actually the string "AND" or "OUT", skip
            # (the netlist might store them differently—depends on your code)
            x = 999; y = 999; z = 999  # ignore these placeholders
        else:
            # likely an input or output variable
            x = 0
            y = random.uniform(-10, 10)
            z = random.uniform(-2, 2)
        node_positions[node] = (x, y, z)
    
    return node_positions, edges

def plot_netlist_3D(netlist):
    """
    Plots the netlist as a 3D graph in Matplotlib.
    """
    node_positions, edges = assign_3d_positions(netlist)

    fig = plt.figure()
    ax = fig.add_subplot(111, projection='3d')
    ax.set_title("Fluidic Logic Netlist - 3D Layout")

    # Plot each node
    for node, (x, y, z) in node_positions.items():
        # We skip placeholder coords used for gate_type strings
        if x == 999:
            continue
        ax.scatter(x, y, z, marker='o', s=50, label=node)

        # Label the node
        ax.text(x, y, z, node, size=8, zorder=2, color='black')

    # Plot edges as lines
    for (src, dst) in edges:
        if (src in node_positions) and (dst in node_positions):
            x_vals = [node_positions[src][0], node_positions[dst][0]]
            y_vals = [node_positions[src][1], node_positions[dst][1]]
            z_vals = [node_positions[src][2], node_positions[dst][2]]
            # Draw a thin line
            ax.plot(x_vals, y_vals, z_vals, color='gray', linewidth=1)

    # Tweak visualization
    ax.set_xlabel("X Axis (layer)")
    ax.set_ylabel("Y Axis")
    ax.set_zlabel("Z Axis")
    plt.show()
dsl = """
# Example DSL
INPUT A, B, C
OUTPUT Z
Z = (A & B) | (!C)
"""

inputs, outputs, logic_defs = parse_dsl(dsl)   # existing from previous code
netlist = synthesize_netlist(logic_defs)       # existing from previous code

# Now visualize in 3D
plot_netlist_3D(netlist)
```
